# ARRGO: Dimensional Generalization

This notebook generalizes the ARRGO framework from one-dimensional bounded optimization to bounded optimization in arbitrary finite dimensions.

The previous development formulated, implemented, and experimentally evaluated ARRGO on one-dimensional search domains. In that formulation, the optimization problem is defined over a bounded interval:

$$
\Omega=[a,b]\subset\mathbb{R}.
$$

The objective of this notebook is to determine whether the fundamental ARRGO framework can be formulated independently of dimensionality.

The generalized search domain is defined as:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j]
\subset\mathbb{R}^{d},
$$

where \(d\geq1\) is a finite dimension and

$$
a_j<b_j
\qquad
\forall j\in\{1,\ldots,d\}.
$$

Under this formulation, a point is represented by a vector:

$$
x=(x_1,x_2,\ldots,x_d)\in\mathbb{R}^{d},
$$

and the optimization problem becomes:

$$
\max_{x\in\Omega} f(x).
$$

The purpose of the generalization is not simply to replace a scalar variable with a vector. Each essential component of ARRGO must be reconsidered in the multidimensional setting.

The analysis therefore examines the generalization of:

- the optimization domain;
- region representation;
- objective evaluation;
- sampling;
- region analysis;
- coverage;
- behavioral information;
- uncertainty;
- optimization potential;
- candidate generation;
- structural refinement;
- region splitting;
- region hierarchy;
- global region selection;
- termination;
- certified bounds;
- convergence mechanisms.

The central question is:

> Can the core ARRGO refinement principle be formulated for arbitrary finite-dimensional bounded domains while preserving its deterministic, information-driven, and region-based structure?

The answer will be developed progressively through mathematical definitions, algorithmic rules, implementation requirements, and multidimensional validation.

The one-dimensional formulation will be treated as the special case \(d=1\). Therefore, the goal is not to replace the original formulation, but to establish a more general framework in which the existing one-dimensional ARRGO algorithm appears as a special case.

The distinction between empirical optimization behavior and mathematically certified guarantees is also preserved throughout this generalization.

Empirical results will be interpreted as evidence about observed algorithmic behavior, whereas certified guarantees will remain conditional on their stated mathematical assumptions.

## 1. Generalization Objectives

The dimensional generalization of ARRGO is evaluated at both the mathematical and algorithmic levels.

A valid generalization must preserve the fundamental structure of ARRGO while adapting its geometric and computational components to finite-dimensional search spaces.

The main objectives are the following.

### 1.1 Mathematical Generality

The definitions used by ARRGO should remain valid for an arbitrary finite dimension:

$$
d\geq1.
$$

The one-dimensional formulation must be recoverable as the special case:

$$
d=1.
$$

The generalized formulation should therefore not depend on properties that are exclusive to one-dimensional intervals.

### 1.2 Region-Based Refinement

The region remains the fundamental structural unit of ARRGO.

The generalized framework must replace one-dimensional intervals with multidimensional regions while preserving the principle that the search domain is progressively refined through a hierarchy of regions.

The refinement process should remain based on information obtained from evaluated points and regional analysis.

### 1.3 Information-Driven Behavior

The choice of where and how to refine should continue to depend on information extracted from the objective function.

The generalization should preserve the distinction between:

- information acquisition through sampling;
- structural refinement through splitting;
- regional analysis;
- global region selection.

The algorithm should not reduce to a fixed geometric partitioning scheme merely because the dimensionality increases.

### 1.4 Geometric Validity

All generalized regions and refinement operations must remain geometrically valid.

In particular:

- every region must remain inside the global domain;
- every generated candidate must be feasible;
- every child region must be a valid subset of its parent;
- refinement must provide measurable spatial contraction;
- the region hierarchy must remain consistent.

### 1.5 Preservation of Global Optimization State

The generalized framework must preserve the global optimization state used by ARRGO.

In particular, the incumbent solution must continue to be defined from evaluated objective values:

$$
x_{\mathrm{best}}^{(t)}
=
\operatorname*{arg\,max}_{x_i\in D_t}
f(x_i),
$$

with corresponding best observed value:

$$
f_{\mathrm{best}}^{(t)}
=
\max_{x_i\in D_t}f(x_i).
$$

The best observed objective value should remain monotone non-decreasing as new evaluations are added.

### 1.6 Preservation of Certified Reasoning

Where the required assumptions are available, the certified formulation must also generalize to higher dimensions.

For example, if the objective satisfies a valid global Lipschitz condition:

$$
|f(x)-f(y)|
\leq
L\|x-y\|
\qquad
\forall x,y\in\Omega,
$$

the generalized framework should be able to construct valid regional upper and lower information bounds.

The corresponding global relationship should remain:

$$
f_{\mathrm{best}}
\leq
f^\ast
\leq
P_{\mathrm{global}},
$$

where \(P_{\mathrm{global}}\) is a valid global upper bound on the unknown optimum.

### 1.7 Preservation of Determinism

The generalized ARRGO algorithm should remain deterministic under identical:

- objective functions;
- domains;
- configurations;
- numerical tolerances;
- initial conditions.

No stochastic component is introduced merely as a consequence of increasing the dimension.

### 1.8 Scalability of the Formulation

The mathematical formulation should apply to arbitrary finite dimensions without requiring a separate theoretical definition for each dimension.

Thus, the target is not to construct independent 2D, 3D, and 5D versions of ARRGO.

Instead, the goal is to establish a single formulation parameterized by \(d\):

$$
\mathrm{ARRGO}_d,
\qquad
d\geq1.
$$

### 1.9 Validation Requirement

A successful mathematical formulation does not by itself establish that the implementation behaves correctly.

The generalized framework must therefore be validated experimentally in multiple dimensions.

The validation should include at least:

- two-dimensional problems;
- three-dimensional problems;
- higher-dimensional problems;
- deterministic reproducibility;
- geometric consistency;
- refinement behavior;
- optimization accuracy;
- and, where applicable, certified bounds.

Therefore, the generalization process follows the sequence:

$$
\boxed{
\text{Mathematical Generalization}
\rightarrow
\text{Algorithmic Generalization}
\rightarrow
\text{Implementation}
\rightarrow
\text{Validation}
}
$$

A multidimensional implementation will be considered consistent with the generalized ARRGO framework only when these stages agree with one another.

## 2. Generalized Optimization Problem

The one-dimensional ARRGO formulation considers the bounded optimization problem:

$$
\max_{x\in[a,b]} f(x).
$$

To generalize ARRGO to arbitrary finite dimensions, the decision variable is represented by a vector:

$$
x=(x_1,x_2,\ldots,x_d)\in\mathbb{R}^{d}.
$$

The bounded search domain is defined as the Cartesian product:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j],
$$

where:

$$
a_j<b_j,
\qquad
j=1,\ldots,d.
$$

Therefore, the feasible domain can equivalently be written as:

$$
\Omega
=
\left\{
x\in\mathbb{R}^{d}
:
a_j\leq x_j\leq b_j
\quad
\forall j=1,\ldots,d
\right\}.
$$

The generalized ARRGO optimization problem is:

$$
\boxed{
\max_{x\in\Omega} f(x)
}
$$

where the objective function is:

$$
f:\Omega\rightarrow\mathbb{R}.
$$

The objective is treated as a deterministic black-box function.

This means that ARRGO assumes that it can evaluate:

$$
y=f(x)
$$

at a feasible point \(x\), but does not require analytical access to:

- the mathematical expression of \(f\);
- its gradient;
- its Hessian;
- or any other analytical derivative.

The algorithm therefore interacts with the objective through function evaluations.

### Global Optimum

The global optimal value is defined as:

$$
f^\ast
=
\max_{x\in\Omega}f(x).
$$

Any point satisfying:

$$
f(x^\ast)=f^\ast
$$

is a global maximizer.

The set of all global maximizers is:

$$
X^\ast
=
\left\{
x\in\Omega
:
f(x)=f^\ast
\right\}.
$$

This set may contain one point or multiple points.

Therefore, the generalized framework does not assume that the global optimizer is unique.

### Existence of a Global Optimum

The domain \(\Omega\) is a finite Cartesian product of closed and bounded intervals.

Consequently, \(\Omega\) is compact in \(\mathbb{R}^{d}\).

If the objective function is continuous:

$$
f\in C(\Omega),
$$

the extreme value theorem guarantees that the global maximum is attained.

Therefore:

$$
\exists x^\ast\in\Omega
\quad\text{such that}\quad
f(x^\ast)=f^\ast.
$$

This establishes the existence of at least one global maximizer under the stated continuity assumption.

### Reduction to One Dimension

For \(d=1\), the generalized formulation becomes:

$$
x=x_1,
$$

and:

$$
\Omega=[a_1,b_1].
$$

The optimization problem consequently reduces to:

$$
\max_{x\in[a_1,b_1]}f(x),
$$

which is the original one-dimensional ARRGO formulation.

Thus, the one-dimensional problem is a special case of the generalized finite-dimensional formulation.

### Generalization Principle

The dimensional parameter \(d\) changes the representation of the search space and the geometry of its regions, but it does not change the fundamental optimization objective.

The generalized ARRGO framework therefore seeks to solve:

$$
\max_{x\in\Omega}f(x)
$$

for arbitrary finite \(d\geq1\), while preserving the original region-based and information-driven refinement principles.

## 3. Generalized Objective Evaluation

ARRGO interacts with the objective function exclusively through feasible function evaluations.

In the one-dimensional formulation, an evaluation is associated with a scalar point:

$$
x_i\in\Omega
$$

and its corresponding objective value:

$$
y_i=f(x_i).
$$

In the generalized \(d\)-dimensional formulation, each evaluation is associated with a vector-valued point:

$$
x_i
=
(x_{i,1},x_{i,2},\ldots,x_{i,d})
\in\Omega,
$$

together with the objective value:

$$
y_i=f(x_i).
$$

An objective evaluation can therefore be represented as:

$$
e_i=(x_i,y_i),
$$

where:

$$
x_i\in\Omega,
\qquad
y_i=f(x_i)\in\mathbb{R}.
$$

An evaluation identifier may additionally be associated with each evaluation for implementation and provenance purposes:

$$
e_i=(x_i,y_i,\operatorname{Id}_i).
$$

### Feasibility of an Evaluation

An evaluation is valid only when its point belongs to the search domain:

$$
x_i\in\Omega.
$$

Equivalently:

$$
a_j
\leq
x_{i,j}
\leq
b_j
\qquad
\forall j=1,\ldots,d.
$$

The objective value is then obtained by evaluating:

$$
y_i=f(x_i).
$$

ARRGO does not assume that the objective provides derivatives or other analytical information at \(x_i\).

### Evaluation History

After \(t\) objective evaluations, the evaluation history is defined as:

$$
D_t
=
\left\{
e_1,e_2,\ldots,e_t
\right\}.
$$

Equivalently:

$$
D_t
=
\left\{
(x_i,f(x_i))
\right\}_{i=1}^{t}.
$$

The history contains the information that ARRGO has explicitly acquired from the objective function.

The algorithm may use this information to analyze regions, generate candidates, update the incumbent, and determine subsequent refinement actions.

### Global Incumbent

The best observed solution after \(t\) evaluations is defined by:

$$
x_{\mathrm{best}}^{(t)}
\in
\operatorname*{arg\,max}_{x_i\in D_t} f(x_i).
$$

The corresponding best observed objective value is:

$$
f_{\mathrm{best}}^{(t)}
=
\max_{1\leq i\leq t}f(x_i).
$$

If multiple evaluated points have the same best objective value, the set of best observed points may contain more than one element.

The incumbent therefore represents the best solution known to ARRGO at the current stage of execution.

### Regional Evaluation Set

For a region \(R\subseteq\Omega\), the evaluations associated with that region are:

$$
D_R
=
\left\{
e_i\in D_t
:
x_i\in R
\right\}.
$$

This regional view allows ARRGO to analyze the information available inside individual regions without requiring the entire evaluation history to be treated as a single undifferentiated set.

### Evaluation Information

Each evaluation provides two types of information:

1. **Spatial information**

$$
x_i\in\Omega
$$

identifies where the objective was evaluated.

2. **Objective information**

$$
y_i=f(x_i)
$$

identifies the observed objective value at that location.

ARRGO combines these two forms of information when analyzing and refining regions.

### Deterministic Evaluation Assumption

The generalized formulation assumes deterministic objective evaluation.

Therefore, repeated evaluation of the same point is expected to produce the same objective value:

$$
f(x_i)=f(x_i).
$$

Stochastic and noisy objective functions are outside the current scope of the theoretical formulation.

### Reduction to One Dimension

When \(d=1\), the evaluation point becomes:

$$
x_i=x_{i,1}\in[a_1,b_1],
$$

and the generalized evaluation:

$$
e_i=(x_i,f(x_i),\operatorname{Id}_i)
$$

reduces to the original one-dimensional representation.

Thus, the evaluation model is dimension-independent; only the representation of the decision variable changes from a scalar to a vector.

## 4. Generalized Region

In one dimension, ARRGO represents a region by an interval:

$$
R=[l,r].
$$

In \(d\) dimensions, this interval is generalized to a hyper-rectangle.

For each dimension \(j\), let:

$$
l_j<r_j.
$$

The region is then defined as:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j].
$$

Equivalently:

$$
R
=
\left\{
x\in\mathbb{R}^{d}
:
l_j\leq x_j\leq r_j
\quad
\forall j=1,\ldots,d
\right\}.
$$

The lower and upper boundaries of the region are represented by two vectors:

$$
l=(l_1,\ldots,l_d),
$$

and:

$$
r=(r_1,\ldots,r_d).
$$

### Region Feasibility

A valid region must be contained inside the global search domain:

$$
R\subseteq\Omega.
$$

Therefore:

$$
a_j
\leq
l_j
<
r_j
\leq
b_j
\qquad
\forall j=1,\ldots,d.
$$

This condition guarantees that every point generated inside the region is feasible with respect to the global optimization problem.

### Region Side Lengths

The side length of region \(R\) along dimension \(j\) is:

$$
w_j(R)=r_j-l_j.
$$

The complete vector of side lengths is:

$$
w(R)
=
(w_1(R),\ldots,w_d(R)).
$$

The maximum side length is:

$$
w_{\max}(R)
=
\max_{1\leq j\leq d}w_j(R).
$$

The Euclidean diameter of the hyper-rectangle is:

$$
\operatorname{diam}(R)
=
\sqrt{
\sum_{j=1}^{d}
w_j(R)^2
}.
$$

These quantities provide different measures of the spatial size and resolution of a region.

### Region Center

The center of \(R\) is defined coordinate-wise by:

$$
c(R)
=
\left(
\frac{l_1+r_1}{2},
\ldots,
\frac{l_d+r_d}{2}
\right).
$$

Because:

$$
l_j
\leq
\frac{l_j+r_j}{2}
\leq
r_j,
$$

the center satisfies:

$$
c(R)\in R.
$$

The region center is therefore a feasible candidate sampling location.

However, the center is only a candidate point and is not assumed to be the maximizer of the objective within the region.

### Regional Evaluation Information

The evaluations associated with a region are defined as:

$$
D_R
=
\left\{
e_i\in D_t
:
x_i\in R
\right\}.
$$

Thus, each region has access to the objective information obtained from evaluated points contained in that region.

This regional information is used to construct the local information state of ARRGO.

### Region as the Fundamental Refinement Unit

The region remains the fundamental structural unit of ARRGO after dimensional generalization.

In one dimension:

$$
\text{Region}=\text{interval}.
$$

In \(d\) dimensions:

$$
\text{Region}=\text{hyper-rectangle}.
$$

The underlying principle remains unchanged:

> ARRGO progressively refines the search domain by analyzing and refining regions according to the information available within them.

The dimensional generalization therefore changes the geometry of the region while preserving its role in the optimization framework.

### Reduction to One Dimension

For \(d=1\):

$$
R
=
[l_1,r_1],
$$

which is exactly the original one-dimensional region representation.

Therefore, the one-dimensional region model is a special case of the generalized multidimensional formulation.

## 5. Generalized Region Hierarchy

ARRGO organizes regions into a hierarchical structure created through successive refinement operations.

In the one-dimensional formulation, a region is an interval and refinement produces smaller child intervals.

In the multidimensional formulation, the same principle is preserved, but regions are hyper-rectangles.

### Root Region

The root region is the complete optimization domain:

$$
R_0=\Omega.
$$

Therefore:

$$
R_0
=
\prod_{j=1}^{d}[a_j,b_j].
$$

The root region contains the entire feasible search space.

### Parent-Child Relationship

Suppose a region \(R_p\) is structurally refined.

The refinement operation partitions \(R_p\) into one or more child regions:

$$
R_{c_1},R_{c_2},\ldots,R_{c_k}.
$$

Each child must satisfy:

$$
R_{c_i}\subseteq R_p.
$$

The child regions must also remain feasible:

$$
R_{c_i}\subseteq\Omega.
$$

For a valid partition, the children collectively cover the portion of the parent assigned to the refinement:

$$
R_p
=
\bigcup_{i=1}^{k}R_{c_i},
$$

subject to the chosen boundary-ownership convention.

### Region Hierarchy

The complete hierarchy can therefore be represented as a directed tree:

$$
\mathcal{H}
=
(\mathcal{R},\mathcal{E}),
$$

where:

- \(\mathcal{R}\) is the set of regions;
- \(\mathcal{E}\) contains parent-child relationships.

If \(R_p\) is the parent of \(R_c\), then:

$$
(R_p,R_c)\in\mathcal{E}.
$$

The root region has no parent.

Every non-root region has exactly one structural parent.

### Parent Retention

ARRGO does not delete a parent region after refinement.

If:

$$
R_p
\rightarrow
\{R_{c_1},\ldots,R_{c_k}\},
$$

then \(R_p\) remains part of the hierarchy.

The children are added to the hierarchy while the parent retains its historical identity and relationship to the children.

This property is referred to as **parent retention**.

### No-Pruning Principle

The generalized framework preserves the no-pruning principle:

> A region is never removed from the global hierarchy merely because it has been refined or because another region currently appears more informative.

Therefore, the hierarchy stores both:

- refined regions;
- their descendants.

This preserves the structural history of the search process.

A region may become inactive or stable, but this does not imply that it has been deleted from the hierarchy.

### Active and Non-Active Regions

At a given execution stage, regions may have different lifecycle states.

For example:

$$
\operatorname{State}(R)
\in
\{
\mathrm{ACTIVE},
\mathrm{STABLE},
\mathrm{REFINED}
\}.
$$

An active region may still be considered for further refinement.

A stable region may temporarily require no refinement under the current information state.

A refined region has already undergone structural refinement and has one or more descendants.

The lifecycle state is distinct from the existence of the region in the hierarchy.

### Inherited Structure

When a region is refined, its descendants inherit structural information from the parent-child relationship.

However, the hierarchy itself does not imply that objective evaluations are automatically duplicated across all descendants.

Evaluations remain global objects with explicit spatial locations:

$$
e_i=(x_i,f(x_i),\operatorname{Id}_i).
$$

A regional evaluation set is obtained by testing whether:

$$
x_i\in R.
$$

Thus, the hierarchy stores structural relationships, while the evaluation history stores acquired objective information.

### Hierarchy Depth

The depth of a region is the number of parent-child transitions between the root and that region.

For the root:

$$
\operatorname{depth}(R_0)=0.
$$

If \(R_c\) is a direct child of \(R_p\):

$$
\operatorname{depth}(R_c)
=
\operatorname{depth}(R_p)+1.
$$

The hierarchy depth provides a measure of how many successive structural refinements have occurred along a particular spatial path.

### Dimensional Independence

The hierarchy definition does not depend on the dimension of the search space.

Only the geometric representation of each region changes:

$$
R\subset\mathbb{R}
\quad\longrightarrow\quad
R\subset\mathbb{R}^{d}.
$$

The parent-child relationship, parent retention, and no-pruning principles remain unchanged.

### Reduction to One Dimension

For \(d=1\), each hyper-rectangle becomes an interval and the generalized hierarchy reduces to the original one-dimensional ARRGO hierarchy.

Therefore, the multidimensional hierarchy is a direct structural generalization of the one-dimensional formulation.

## 6. Generalized Sampling

Sampling is the information-acquisition mechanism of ARRGO.

A sampling operation evaluates the objective function at a selected feasible point in the current search space.

In one dimension, a sampling point is a scalar:

$$
x\in[l,r].
$$

In \(d\) dimensions, a sampling point is a vector:

$$
x=(x_1,\ldots,x_d)\in R,
$$

where \(R\) is a valid hyper-rectangular region.

### Feasible Sampling Point

For a region:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j],
$$

a point \(x\) is feasible for sampling inside \(R\) if:

$$
l_j\leq x_j\leq r_j
\qquad
\forall j=1,\ldots,d.
$$

Therefore:

$$
x\in R.
$$

Since:

$$
R\subseteq\Omega,
$$

every point sampled inside a valid region is also feasible for the global optimization problem.

### Sampling as Information Acquisition

Given a selected point \(x\), ARRGO evaluates:

$$
y=f(x).
$$

The resulting evaluation is added to the global evaluation history:

$$
D_{t+1}
=
D_t
\cup
\{(x,f(x),\operatorname{Id})\}.
$$

Sampling therefore increases the amount of objective information available to the algorithm.

Unlike structural refinement, sampling does not change the geometric boundaries of the region.

Thus:

$$
\boxed{
\text{Sampling}
=
\text{Information Refinement}
}
$$

whereas:

$$
\boxed{
\text{Splitting}
=
\text{Structural Refinement}
}
$$

### Candidate Sampling Points

ARRGO may generate multiple feasible sampling candidates inside a region.

Let the candidate set be:

$$
C_{\mathrm{sample}}(R)
=
\{x^{(1)},\ldots,x^{(m)}\},
$$

where:

$$
x^{(k)}\in R
\qquad
\forall k.
$$

Each candidate represents a possible location at which additional objective information can be acquired.

Candidate generation itself does not evaluate the objective.

Only the selected candidate consumes an objective evaluation.

### Region Center as a Candidate

The region center:

$$
c(R)
=
\left(
\frac{l_1+r_1}{2},
\ldots,
\frac{l_d+r_d}{2}
\right)
$$

is always feasible and can therefore be included among possible sampling candidates.

However, ARRGO does not assume that the center is always the most informative sampling location.

Other candidates may be generated according to the unresolved information state of the region.

### Avoiding Redundant Evaluations

Because objective evaluations are the primary computational resource, ARRGO should avoid evaluating the same point repeatedly when the objective is deterministic.

For a candidate \(x\), the algorithm can determine whether an equivalent point already exists in the evaluation history according to the configured numerical tolerance.

Conceptually:

$$
x
\approx
x_i
\quad\Longrightarrow\quad
\text{avoid redundant evaluation}.
$$

The exact equality criterion is an implementation-level numerical robustness decision.

### Sampling and Regional Information

After a new evaluation is obtained, the regional information state may change.

For a region \(R\), the regional evaluation set becomes:

$$
D_R^{(t+1)}
=
\left\{
e_i\in D_{t+1}
:
x_i\in R
\right\}.
$$

The new evaluation may therefore change:

- coverage information;
- behavioral information;
- uncertainty estimates;
- optimization-potential estimates;
- refinement priorities.

Sampling can consequently affect future structural refinement decisions without directly changing the region geometry.

### Sampling Does Not Imply Refinement

A sampling operation does not automatically imply that the region should be split.

After sampling, the updated regional information state is re-evaluated.

The region may then be:

- sampled again;
- structurally refined;
- considered sufficiently resolved;
- or remain active for future global consideration.

Therefore, sampling and splitting are separate decisions in ARRGO.

### Global Sampling

Although sampling is associated with a selected region, the resulting evaluation belongs to the global evaluation history.

Thus:

$$
D_t
=
\bigcup_R D_R
$$

is interpreted through spatial membership rather than through duplicated ownership.

A single evaluation corresponds to one spatial point and can be relevant to any region that contains that point according to the hierarchy and boundary convention.

### Reduction to One Dimension

For \(d=1\):

$$
x=(x_1)
$$

becomes a scalar point and:

$$
R=[l_1,r_1].
$$

The generalized sampling definition therefore reduces to the original one-dimensional sampling process.

### Generalization Principle

The dimensional generalization changes the representation of candidate locations from scalars to vectors, but preserves the fundamental role of sampling:

> Sampling acquires new objective information without directly changing the geometric structure of the search space.

## 7. Generalized Spatial Coverage

Spatial coverage describes how well the evaluated points represent the spatial extent of a region.

In the one-dimensional formulation, coverage can be analyzed using the distances between neighboring evaluation points and the boundaries of the region.

In higher dimensions, neighboring-point gaps alone are not sufficient to characterize spatial coverage.

For a region:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j],
$$

and its regional evaluation set:

$$
D_R
=
\{(x_i,f(x_i),\operatorname{Id}_i):x_i\in R\},
$$

coverage must describe how effectively the evaluated points represent the geometry of \(R\).

### Coordinate-Wise Coverage

For each dimension \(j\), consider the coordinates of the evaluated points:

$$
X_{R,j}
=
\{x_{i,j}:x_i\in R\}.
$$

After including the corresponding region boundaries, the coordinate-wise coverage gaps can be analyzed through the ordered set:

$$
l_j
\leq
z_{j,1}
\leq
z_{j,2}
\leq
\cdots
\leq
z_{j,n_j}
\leq
r_j.
$$

The maximum coordinate-wise gap is:

$$
g_j(R)
=
\max_k
\left(
z_{j,k+1}-z_{j,k}
\right).
$$

A normalized coordinate-wise coverage measure can then be defined as:

$$
\gamma_j(R)
=
\frac{g_j(R)}
{r_j-l_j},
$$

provided that \(r_j-l_j>0\).

Smaller values of \(\gamma_j(R)\) indicate finer coordinate-wise spatial resolution.

### Multidimensional Coverage Resolution

Because coverage is inherently multidimensional, the vector:

$$
\Gamma(R)
=
(\gamma_1(R),\ldots,\gamma_d(R))
$$

provides a coordinate-wise coverage profile.

A conservative scalar summary may be obtained using:

$$
\gamma_{\max}(R)
=
\max_{1\leq j\leq d}\gamma_j(R).
$$

This quantity identifies the least-resolved coordinate direction.

However, the complete vector \(\Gamma(R)\) should be retained when possible because different dimensions may have substantially different coverage resolutions.

### Geometric Interpretation

Coordinate-wise coverage is useful but does not completely characterize the geometry of a multidimensional point set.

Two evaluation sets can have similar coordinate-wise distributions while leaving large unsampled spatial regions.

Therefore, coordinate-wise coverage is treated as an information signal rather than a complete geometric coverage certificate.

A stronger geometric characterization may use the distance from points in the region to the nearest evaluated point.

For \(x\in R\), define:

$$
d_R(x,D_R)
=
\min_{x_i\in D_R}
\|x-x_i\|_2.
$$

The corresponding worst-case spatial distance is:

$$
h(R)
=
\max_{x\in R}
d_R(x,D_R).
$$

This quantity is a covering-radius measure of the evaluated points within the region.

Smaller values of \(h(R)\) indicate that every point in the region is spatially closer to at least one evaluated point.

### Coverage as an Unresolved Information Signal

The coverage state of a region can therefore be represented by spatial-resolution information such as:

$$
Q_{\mathrm{coverage}}(R)
=
\left(
\Gamma(R),
h(R)
\right).
$$

This quantity is not itself a probability or a guarantee of objective-function accuracy.

It describes the spatial resolution of the available evaluations.

### Sampling Implication

If a region has poor spatial coverage, additional sampling may provide useful information without requiring immediate structural splitting.

Therefore:

$$
\text{poor coverage}
\quad\Longrightarrow\quad
\text{sampling may be informative}.
$$

The actual sampling decision remains dependent on the complete regional information state.

### Boundary Coverage

The boundaries of a region are part of its spatial geometry.

Therefore, coverage analysis should account for the distance between evaluated points and the region boundaries.

This prevents a region from appearing well covered solely because its interior evaluations are close to one another while large portions near the boundary remain unresolved.

### Empty Regions

If:

$$
D_R=\varnothing,
$$

no objective information has yet been acquired inside the region.

Such a region has unresolved spatial information and requires an appropriate sampling action before meaningful objective-based regional analysis can be performed.

### One-Dimensional Reduction

For \(d=1\), the coordinate-wise coverage profile contains a single component:

$$
\Gamma(R)=(\gamma_1(R)).
$$

The multidimensional formulation therefore reduces to the familiar one-dimensional analysis of gaps between evaluations and region boundaries.

### Generalization Principle

The essential role of coverage remains unchanged:

> Coverage measures how much spatial information ARRGO has acquired about a region.

The multidimensional formulation extends this concept from one-dimensional gaps to coordinate-wise and geometric spatial resolution while avoiding the assumption that a single scalar gap completely describes multidimensional coverage.

## 8. Generalized Regional Behavior

Regional behavior describes the variation of the objective function within a region based on the available evaluations.

For a region:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j],
$$

the regional evaluation set is:

$$
D_R
=
\{(x_i,f(x_i),\operatorname{Id}_i):x_i\in R\}.
$$

The purpose of behavioral analysis is to characterize how the observed objective values vary spatially within \(R\).

### Objective Variation

For two evaluated points:

$$
x_i,x_k\in R,
$$

the observed objective difference is:

$$
\Delta f_{ik}
=
f(x_i)-f(x_k).
$$

The corresponding spatial displacement is:

$$
\Delta x_{ik}
=
x_i-x_k.
$$

These quantities provide local information about how objective values change across the region.

### Directional Variation

In multidimensional space, objective variation depends on direction.

For two distinct evaluated points, define the normalized direction:

$$
u_{ik}
=
\frac{x_i-x_k}
{\|x_i-x_k\|_2},
$$

provided that:

$$
x_i\neq x_k.
$$

The observed directional rate of change between the two evaluations can then be characterized by:

$$
s_{ik}
=
\frac{f(x_i)-f(x_k)}
{\|x_i-x_k\|_2}.
$$

This quantity is an observed finite-difference rate along the direction connecting the two evaluated points.

It should not be interpreted as an analytical gradient.

### Directional Behavior Profile

The set of observed directional differences can be represented as a behavioral profile:

$$
B_R
=
\left\{
(u_{ik},s_{ik})
:
x_i,x_k\in D_R,\;
x_i\neq x_k
\right\}.
$$

This profile contains information about the observed variation of the objective across different directions inside the region.

### Coordinate-Wise Behavior

The behavior can also be examined separately along each coordinate direction.

For dimension \(j\), consider two evaluations whose displacement is primarily associated with coordinate \(j\).

The corresponding coordinate-wise variation can be represented by:

$$
s_{ik}^{(j)}
=
\frac{
f(x_i)-f(x_k)
}{
|x_{i,j}-x_{k,j}|
},
$$

provided that:

$$
x_{i,j}\neq x_{k,j}.
$$

These coordinate-wise quantities can be used as auxiliary information for identifying dimensions in which the objective exhibits stronger observed variation.

### Behavior Variation

ARRGO is interested not only in the magnitude of observed variation but also in how that variation changes across a region.

For two observed directional rates:

$$
s_{ik}
\quad\text{and}\quad
s_{pq},
$$

their difference can be expressed as:

$$
\Delta s
=
s_{ik}-s_{pq}.
$$

Large variation in observed rates may indicate that the objective exhibits heterogeneous local behavior within the region.

This information can contribute to the refinement decision.

### Behavior as an Information Signal

The regional behavior representation is therefore an information signal derived from evaluated points.

It does not constitute a complete model of the objective function.

In particular, finite evaluations cannot generally determine the exact unseen behavior of an arbitrary black-box function.

Therefore:

$$
B_R
\neq
\text{exact objective behavior over }R.
$$

Instead:

$$
B_R
=
\text{observed behavioral information available to ARRGO}.
$$

### Refinement Implication

If the observed behavior within a region varies substantially, additional refinement may be useful because a coarse region may contain heterogeneous objective behavior.

Conceptually:

$$
\text{high observed behavioral variation}
\quad\Longrightarrow\quad
\text{potential need for refinement}.
$$

However, this signal does not independently determine the refinement action.

The final decision must also consider coverage, uncertainty, optimization potential, and global relevance.

### Insufficient Data

If a region contains fewer than two distinct evaluated points, pairwise directional variation cannot be computed reliably.

For:

$$
|D_R|<2,
$$

the behavioral information is therefore considered insufficient.

Such a region may require additional sampling before meaningful behavioral analysis can be performed.

### Deterministic Behavior

Because ARRGO assumes a deterministic objective function, the same evaluated point has a unique objective value.

Consequently, behavioral quantities derived from the same evaluation history are deterministic.

### One-Dimensional Reduction

When \(d=1\), the direction between two distinct points is one-dimensional and the Euclidean distance becomes:

$$
\|x_i-x_k\|_2
=
|x_i-x_k|.
$$

The directional finite difference therefore reduces to the familiar one-dimensional slope:

$$
s_{ik}
=
\frac{f(x_i)-f(x_k)}
{|x_i-x_k|}.
$$

Thus, the one-dimensional behavioral analysis is recovered as a special case of the generalized formulation.

### Generalization Principle

The essential role of behavioral analysis remains unchanged:

> ARRGO uses observed objective variation across space to identify regions whose local behavior may require additional information or structural refinement.

The multidimensional formulation extends this concept from a single spatial direction to multiple directions and coordinate-wise variation.

## 9. Generalized Regional Uncertainty

Regional uncertainty represents the amount of unresolved information about the objective function within a region.

In ARRGO, uncertainty has two distinct interpretations:

1. empirical uncertainty derived from the available evaluations;
2. certified uncertainty derived from mathematically valid bounds.

These two concepts must not be treated as equivalent.

### 9.1 Empirical Regional Uncertainty

For a region:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j],
$$

let:

$$
D_R
=
\{(x_i,f(x_i),\operatorname{Id}_i):x_i\in R\}.
$$

The available evaluations provide only finite information about the objective.

Therefore, empirical uncertainty describes the amount of objective behavior that remains unresolved by the available samples.

Possible sources of empirical uncertainty include:

- sparse spatial coverage;
- large distances between evaluated points;
- strong observed variation;
- heterogeneous local behavior;
- insufficient sampling density.

A regional empirical uncertainty profile can therefore be represented abstractly as:

$$
Q_{\mathrm{uncertainty}}^{\mathrm{emp}}(R).
$$

This profile is an algorithmic information signal rather than a mathematical bound on the unknown objective.

In particular:

$$
Q_{\mathrm{uncertainty}}^{\mathrm{emp}}(R)
\neq
f_{\max}(R)-f_{\min}(R)
$$

in general, because finite evaluations cannot determine the exact maximum and minimum of an arbitrary black-box objective over the entire region.

### 9.2 Certified Regional Uncertainty

A mathematically rigorous uncertainty measure requires additional assumptions about the objective.

Assume that \(f\) satisfies a known valid Lipschitz condition:

$$
|f(x)-f(y)|
\leq
L\|x-y\|_2
\qquad
\forall x,y\in\Omega.
$$

For every evaluated point:

$$
(x_i,f(x_i))\in D_R,
$$

the Lipschitz condition implies:

$$
f(x)
\geq
f(x_i)-L\|x-x_i\|_2,
$$

and:

$$
f(x)
\leq
f(x_i)+L\|x-x_i\|_2.
$$

Combining all available evaluations gives the regional lower envelope:

$$
L_R(x)
=
\max_{x_i\in D_R}
\left[
f(x_i)-L\|x-x_i\|_2
\right],
$$

and the regional upper envelope:

$$
U_R(x)
=
\min_{x_i\in D_R}
\left[
f(x_i)+L\|x-x_i\|_2
\right].
$$

For every \(x\in R\), provided that \(L\) is valid:

$$
L_R(x)
\leq
f(x)
\leq
U_R(x).
$$

The pointwise certified uncertainty is therefore:

$$
u_R(x)
=
U_R(x)-L_R(x).
$$

The worst-case regional certified uncertainty is:

$$
u_{\max}(R)
=
\max_{x\in R}u_R(x).
$$

This quantity provides a mathematically meaningful measure of the remaining uncertainty under the Lipschitz assumption.

### 9.3 Certified Uncertainty and Region Diameter

The Lipschitz condition also gives a geometric relationship between spatial resolution and objective uncertainty.

For any two points \(x,y\in R\):

$$
|f(x)-f(y)|
\leq
L\|x-y\|_2.
$$

Since:

$$
\|x-y\|_2
\leq
\operatorname{diam}(R),
$$

we obtain:

$$
|f(x)-f(y)|
\leq
L\operatorname{diam}(R).
$$

Therefore, shrinking the region diameter reduces the maximum possible variation allowed by the Lipschitz bound.

This establishes a direct connection between structural refinement and certified uncertainty.

### 9.4 Empty or Insufficiently Sampled Regions

If:

$$
D_R=\varnothing,
$$

the regional objective is completely unsampled.

A certified envelope based on regional evaluations cannot therefore be constructed from \(D_R\) alone.

Similarly, empirical behavioral uncertainty may be high because no local objective information is available.

Such regions require information acquisition before detailed regional analysis can be performed.

### 9.5 Uncertainty as a Refinement Signal

High empirical uncertainty may indicate that additional sampling or structural refinement is useful.

High certified uncertainty may indicate that the mathematical bounds for the region remain too loose.

Conceptually:

$$
\text{high uncertainty}
\quad\Longrightarrow\quad
\text{additional refinement may be required}.
$$

However, uncertainty alone does not determine the refinement action.

ARRGO combines uncertainty with other regional and global information.

### 9.6 Empirical and Certified Interpretations

The distinction can be summarized as:

$$
\boxed{
\text{Empirical Uncertainty}
=
\text{unresolved observed information}
}
$$

whereas:

$$
\boxed{
\text{Certified Uncertainty}
=
\text{valid mathematical bound width}
}
$$

Empirical uncertainty supports heuristic or information-driven decisions.

Certified uncertainty supports mathematical statements about the unknown objective, but only under the assumptions required to construct valid bounds.

### 9.7 One-Dimensional Reduction

For \(d=1\):

$$
\|x-x_i\|_2
=
|x-x_i|,
$$

so the certified envelopes reduce to:

$$
L_R(x)
=
\max_{x_i\in D_R}
\left[
f(x_i)-L|x-x_i|
\right],
$$

and:

$$
U_R(x)
=
\min_{x_i\in D_R}
\left[
f(x_i)+L|x-x_i|
\right].
$$

Thus, the one-dimensional certified uncertainty formulation is recovered exactly.

### Generalization Principle

The fundamental role of uncertainty remains unchanged:

> ARRGO uses unresolved information to determine whether a region requires additional refinement.

In the multidimensional framework, empirical uncertainty remains an information signal, while certified uncertainty provides rigorous bounds when the necessary assumptions are satisfied.

## 10. Generalized Optimization Potential

Optimization potential represents the remaining possibility that a region contains a solution whose objective value exceeds the current incumbent.

For a region:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j],
$$

ARRGO uses the available regional information to estimate or bound the best objective value that may still be attainable inside \(R\).

### 10.1 Empirical Optimization Potential

In Empirical Mode, finite evaluations provide observed evidence about a region but do not generally provide a rigorous upper bound on its unknown objective values.

Therefore, an empirical potential signal may be defined from the available regional information as:

$$
Q_{\mathrm{potential}}^{\mathrm{emp}}(R).
$$

This quantity represents the observed or estimated relevance of the region to future optimization.

It must not be interpreted as a mathematical upper bound on the unknown regional optimum.

In particular:

$$
Q_{\mathrm{potential}}^{\mathrm{emp}}(R)
\neq
\max_{x\in R}f(x)
$$

in general.

The true regional optimum is:

$$
f_R^\ast
=
\max_{x\in R}f(x),
$$

but this value is generally unknown to ARRGO.

### 10.2 Certified Regional Potential

In Certified Mode, assume that \(f\) satisfies a known valid Lipschitz condition:

$$
|f(x)-f(y)|
\leq
L\|x-y\|_2
\qquad
\forall x,y\in\Omega.
$$

For an evaluated point \(x_i\in R\):

$$
f(x)
\leq
f(x_i)+L\|x-x_i\|_2.
$$

Combining the upper constraints from all evaluations in \(D_R\) gives:

$$
U_R(x)
=
\min_{x_i\in D_R}
\left[
f(x_i)+L\|x-x_i\|_2
\right].
$$

The certified regional potential is then defined as:

$$
P(R)
=
\max_{x\in R}U_R(x).
$$

Because \(U_R(x)\) is a valid upper bound on \(f(x)\):

$$
f_R^\ast
\leq
P(R).
$$

Therefore, \(P(R)\) is a valid upper bound on the best objective value that can still occur inside region \(R\), under the validity of the Lipschitz constant.

### 10.3 Global Potential

Let \(\mathcal{R}_t\) denote the set of regions considered in the global region structure.

The global certified potential is:

$$
P_{\mathrm{global}}^{(t)}
=
\max_{R\in\mathcal{R}_t}P(R).
$$

If the regional potentials are valid upper bounds, then:

$$
f^\ast
\leq
P_{\mathrm{global}}^{(t)}.
$$

Together with the best observed value:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast,
$$

we obtain:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
\leq
f^\ast
\leq
P_{\mathrm{global}}^{(t)}
}
$$

This inequality is the basis for the certified global optimality gap.

### 10.4 Global Certification Gap

The certified global gap is defined as:

$$
\Delta_{\mathrm{global}}^{(t)}
=
P_{\mathrm{global}}^{(t)}
-
f_{\mathrm{best}}^{(t)}.
$$

Since:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast
\leq
P_{\mathrm{global}}^{(t)},
$$

the actual optimization error satisfies:

$$
0
\leq
f^\ast-f_{\mathrm{best}}^{(t)}
\leq
\Delta_{\mathrm{global}}^{(t)}.
$$

Therefore, if:

$$
\Delta_{\mathrm{global}}^{(t)}
\leq
\varepsilon,
$$

then:

$$
f_{\mathrm{best}}^{(t)}
\geq
f^\ast-\varepsilon.
$$

Thus, the incumbent is epsilon-optimal under the stated assumptions.

### 10.5 Potential as a Global Selection Signal

A region with a larger potential may represent a more important unresolved portion of the search space.

Conceptually:

$$
P(R_1)>P(R_2)
$$

indicates that the current certified information permits a higher possible objective value in \(R_1\) than in \(R_2\).

Potential can therefore contribute to global region selection.

However, potential should not be considered in isolation.

A region with high potential but poor spatial coverage may require sampling, while another region with similar potential may benefit more from structural refinement.

Therefore, ARRGO combines potential with other regional information.

### 10.6 Potential and the Incumbent

The relevance of a region can also be interpreted relative to the current incumbent.

Define the regional potential excess:

$$
E(R)
=
P(R)-f_{\mathrm{best}}.
$$

If:

$$
E(R)\leq0,
$$

then the certified upper bound of the region does not exceed the current best observed value.

If the regional bound is valid, this implies that the region cannot contain a solution with objective value greater than the incumbent.

This observation can provide a rigorous basis for identifying regions that no longer have potential to improve the incumbent.

Any such interpretation must respect the global no-pruning design of ARRGO: a region may become non-improving for the current incumbent without being deleted from the hierarchy.

### 10.7 Potential and Region Refinement

Structural refinement can reduce the geometric diameter of regions.

Under a valid Lipschitz bound, smaller regions restrict the maximum possible objective variation:

$$
|f(x)-f(y)|
\leq
L\operatorname{diam}(R).
$$

Consequently, sufficiently refined regions can produce tighter upper bounds when enough local evaluation information is available.

Thus, potential connects global optimization relevance with local spatial refinement.

### 10.8 Empty Regions

If:

$$
D_R=\varnothing,
$$

the regional upper envelope cannot be constructed from regional evaluations alone.

Additional information is therefore required before a finite evaluation-based certified potential can be obtained for that region.

### 10.9 One-Dimensional Reduction

For \(d=1\):

$$
\|x-x_i\|_2
=
|x-x_i|.
$$

The certified potential therefore becomes:

$$
P(R)
=
\max_{x\in R}
\min_{x_i\in D_R}
\left[
f(x_i)+L|x-x_i|
\right],
$$

which is the original one-dimensional certified potential formulation.

### Generalization Principle

The role of optimization potential remains unchanged:

> Potential measures how much room a region may still have to contain an improved solution.

In Empirical Mode, potential is an information signal.

In Certified Mode, potential can become a mathematically valid upper bound when the required assumptions are satisfied.

## 11. Generalized Candidate Generation

Candidate generation produces feasible alternatives that ARRGO may evaluate or use for structural refinement.

Candidates are generated from the current information state of a region and do not themselves consume objective evaluations.

The generalized framework distinguishes between two main candidate types:

1. sampling candidates;
2. structural refinement candidates.

### 11.1 Sampling Candidates

For a region:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j],
$$

a sampling candidate is a point:

$$
x^{(k)}\in R.
$$

The sampling candidate set is:

$$
C_{\mathrm{sample}}(R)
=
\left\{
x^{(1)},\ldots,x^{(m)}
\right\},
$$

where:

$$
x^{(k)}
=
(x_1^{(k)},\ldots,x_d^{(k)}).
$$

Every candidate must satisfy:

$$
l_j
\leq
x_j^{(k)}
\leq
r_j
\qquad
\forall j.
$$

Therefore:

$$
C_{\mathrm{sample}}(R)\subseteq R.
$$

Candidate generation may use the geometry of the region and the currently available evaluations.

Examples of possible sampling locations include:

- the region center;
- points near poorly covered areas;
- points associated with unresolved behavioral variation;
- points suggested by regional potential information.

The specific candidate-generation mechanism is an algorithmic design decision and must remain deterministic.

### 11.2 Structural Refinement Candidates

A structural refinement candidate represents a possible way to partition a region.

Unlike a sampling candidate, it is not itself an objective-evaluation point.

A structural candidate can therefore be represented abstractly as:

$$
s=(j,\xi),
$$

where:

- \(j\in\{1,\ldots,d\}\) identifies a coordinate dimension;
- \(\xi\) identifies a feasible partition location along that dimension.

For a region \(R\), the partition location must satisfy:

$$
l_j<\xi<r_j.
$$

The remaining coordinates of the region remain unchanged.

A structural candidate therefore describes both:

$$
\text{which dimension to refine}
$$

and:

$$
\text{where to refine that dimension}.
$$

### 11.3 Candidate Feasibility

All generated candidates must satisfy the geometric constraints of their corresponding region.

For sampling:

$$
x\in R.
$$

For structural refinement:

$$
l_j<\xi<r_j.
$$

This guarantees that candidate generation cannot directly produce an invalid region or an infeasible sampling point.

### 11.4 Candidate Profiles

A candidate may be associated with a profile describing its expected or estimated structural value according to the information currently available.

Conceptually:

$$
Q(c)
=
\left(
Q_{\mathrm{coverage}},
Q_{\mathrm{behavior}},
Q_{\mathrm{uncertainty}},
Q_{\mathrm{potential}},
Q_{\mathrm{structural}}
\right).
$$

These components are information signals rather than probabilities unless a probabilistic model has explicitly been introduced.

Therefore, ARRGO does not interpret such profiles as probability distributions.

### 11.5 Candidate Evaluation Versus Candidate Generation

Candidate generation and objective evaluation are separate operations.

For a sampling candidate:

$$
x^{(k)}
\rightarrow
f(x^{(k)})
$$

only occurs if that candidate is selected for actual evaluation.

A generated candidate that is not selected does not consume an objective evaluation.

Similarly, structural candidates do not consume objective evaluations merely by being generated or compared.

### 11.6 Candidate Provenance

Each candidate should retain sufficient information to identify its origin.

For a sampling candidate, this may include:

$$
\operatorname{Source}(c)=R.
$$

For a structural candidate:

$$
\operatorname{Source}(c)=(R,j,\xi).
$$

Candidate provenance allows the algorithm to determine which region and which refinement decision produced the candidate.

### 11.7 Deterministic Candidate Generation

Given identical:

- region geometry;
- evaluation history;
- configuration;
- numerical tolerances;

candidate generation should produce the same candidate set.

This preserves deterministic behavior in the generalized framework.

### 11.8 Candidate Selection

Candidate generation does not determine the final action.

After candidates are generated, ARRGO evaluates their available information profiles and selects an appropriate candidate according to the refinement decision mechanism.

Therefore:

$$
\text{Candidate Generation}
\neq
\text{Candidate Selection}.
$$

This separation allows the candidate-generation mechanism to remain independent from the final global decision policy.

### 11.9 Dimensional Generalization

The essential generalization is:

$$
x\in\mathbb{R}
\quad\longrightarrow\quad
x\in\mathbb{R}^{d}
$$

for sampling candidates, and:

$$
\text{split location}
\quad\longrightarrow\quad
(\text{split dimension},\text{split location})
$$

for structural candidates.

Thus, multidimensional candidate generation introduces an additional structural decision: selecting the coordinate direction in which refinement should occur.

### 11.10 One-Dimensional Reduction

For \(d=1\), the only available coordinate dimension is:

$$
j=1.
$$

Therefore, a structural candidate reduces to:

$$
s=(1,\xi),
$$

which is equivalent to selecting a scalar split location \(\xi\).

The generalized candidate representation therefore reduces to the original one-dimensional formulation.

### Generalization Principle

The fundamental role of candidate generation remains unchanged:

> ARRGO generates feasible alternatives from the current regional information state and separates the generation of alternatives from the decision to evaluate or structurally refine them.

The multidimensional formulation adds coordinate-direction information to structural candidates while preserving the original deterministic candidate-driven refinement process.

## 12. Generalized Structural Splitting

Structural splitting is the mechanism through which ARRGO increases the spatial resolution of the search domain.

In one dimension, a region:

$$
R=[l,r]
$$

is divided at a scalar split location.

In \(d\) dimensions, a region is a hyper-rectangle:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j].
$$

A structural split selects one coordinate direction and divides the region along that direction.

### 12.1 Split Dimension

Let:

$$
j\in\{1,\ldots,d\}
$$

denote the selected split dimension.

Let the split location satisfy:

$$
l_j<s_j<r_j.
$$

The split therefore acts only on coordinate \(j\).

### 12.2 Binary Split

A binary split produces two child regions.

The first child is:

$$
R_{\mathrm{left}}
=
[l_1,r_1]
\times\cdots\times
[l_j,s_j]
\times\cdots\times
[l_d,r_d].
$$

The second child is:

$$
R_{\mathrm{right}}
=
[l_1,r_1]
\times\cdots\times
[s_j,r_j]
\times\cdots\times
[l_d,r_d].
$$

The two children therefore differ only in their bounds along the selected coordinate direction.

### 12.3 Child Containment

Every child must remain inside the parent:

$$
R_{\mathrm{left}}\subseteq R,
$$

and:

$$
R_{\mathrm{right}}\subseteq R.
$$

This follows directly from:

$$
l_j<s_j<r_j.
$$

All other coordinate bounds are inherited unchanged from the parent.

### 12.4 Coverage Preservation

The two children collectively cover the parent:

$$
R
=
R_{\mathrm{left}}
\cup
R_{\mathrm{right}}.
$$

If the split boundary \(x_j=s_j\) is included in both children, then:

$$
R_{\mathrm{left}}
\cap
R_{\mathrm{right}}
\neq
\varnothing.
$$

Their intersection is the splitting hyperplane inside the parent:

$$
R_{\mathrm{left}}
\cap
R_{\mathrm{right}}
=
\left\{
x\in R:x_j=s_j
\right\}.
$$

This overlap is a boundary representation issue rather than an overlap of positive volume.

For implementation, a consistent boundary-ownership convention must be defined when assigning evaluations to regions.

### 12.5 Volume

The volume of a \(d\)-dimensional region is:

$$
\operatorname{Vol}(R)
=
\prod_{k=1}^{d}(r_k-l_k).
$$

After splitting along dimension \(j\):

$$
\operatorname{Vol}(R_{\mathrm{left}})
=
(s_j-l_j)
\prod_{k\neq j}(r_k-l_k),
$$

and:

$$
\operatorname{Vol}(R_{\mathrm{right}})
=
(r_j-s_j)
\prod_{k\neq j}(r_k-l_k).
$$

Therefore:

$$
\operatorname{Vol}(R_{\mathrm{left}})
+
\operatorname{Vol}(R_{\mathrm{right}})
=
\operatorname{Vol}(R).
$$

Thus, binary splitting preserves the total volume of the parent.

### 12.6 Maximum Side Length

Structural refinement should reduce the spatial scale of the refined region.

For region \(R\), define:

$$
w_{\max}(R)
=
\max_{1\leq k\leq d}(r_k-l_k).
$$

After splitting along dimension \(j\), the new side lengths are:

$$
w_k(R_{\mathrm{child}})
=
r_k-l_k,
\qquad
k\neq j,
$$

while along the selected dimension:

$$
w_j(R_{\mathrm{left}})
=
s_j-l_j,
$$

and:

$$
w_j(R_{\mathrm{right}})
=
r_j-s_j.
$$

Therefore, the split directly refines the selected coordinate direction.

### 12.7 Geometric Diameter

The Euclidean diameter of a hyper-rectangle is:

$$
\operatorname{diam}(R)
=
\sqrt{
\sum_{k=1}^{d}(r_k-l_k)^2
}.
$$

After a non-degenerate split:

$$
l_j<s_j<r_j,
$$

both child regions have strictly smaller diameter than the parent.

For the left child:

$$
\operatorname{diam}(R_{\mathrm{left}})
=
\sqrt{
(s_j-l_j)^2
+
\sum_{k\neq j}(r_k-l_k)^2
},
$$

and because:

$$
s_j-l_j<r_j-l_j,
$$

we obtain:

$$
\operatorname{diam}(R_{\mathrm{left}})
<
\operatorname{diam}(R).
$$

The same argument applies to the right child.

Therefore, every valid binary split produces spatial contraction of both children.

### 12.8 Balanced Split

A particularly simple deterministic split is the midpoint split:

$$
s_j
=
\frac{l_j+r_j}{2}.
$$

This divides the selected side into two equal lengths:

$$
s_j-l_j
=
r_j-s_j
=
\frac{r_j-l_j}{2}.
$$

A midpoint split therefore provides a guaranteed contraction factor of \(1/2\) along the selected coordinate.

However, ARRGO does not require every split to occur at the midpoint.

The split location may instead be selected according to the structural and information-based candidate evaluation mechanism.

### 12.9 General Contraction Condition

Let the contraction factor be:

$$
\rho,
\qquad
0<\rho<1.
$$

A split along dimension \(j\) satisfies the coordinate-wise contraction condition if:

$$
\max
\left(
\frac{s_j-l_j}{r_j-l_j},
\frac{r_j-s_j}{r_j-l_j}
\right)
\leq
\rho.
$$

Equivalently:

$$
\max
\left(
s_j-l_j,
r_j-s_j
\right)
\leq
\rho(r_j-l_j).
$$

For a binary split to satisfy this condition, necessarily:

$$
\rho\geq\frac12.
$$

When:

$$
\rho=\frac12,
$$

the midpoint is the only valid split location.

When:

$$
\rho>\frac12,
$$

a non-empty interval of valid split locations exists.

### 12.10 Why Contraction Matters

Repeated structural refinement should progressively increase spatial resolution.

If a refinement path repeatedly splits a coordinate direction and each split satisfies the contraction condition, the side length along that direction decreases geometrically.

After \(n\) such refinements:

$$
w_j^{(n)}
\leq
\rho^n w_j^{(0)}.
$$

Since:

$$
0<\rho<1,
$$

we have:

$$
\lim_{n\rightarrow\infty}w_j^{(n)}=0.
$$

This establishes geometric contraction along repeatedly refined directions.

For a region to become arbitrarily small in all dimensions, the refinement process must also ensure that every relevant coordinate direction is refined sufficiently often.

### 12.11 Directional Refinement

A multidimensional algorithm must therefore decide not only:

$$
\text{whether to split},
$$

but also:

$$
\text{which dimension to split}.
$$

This introduces a new structural decision compared with the one-dimensional case.

The dimension-selection mechanism may use information such as:

- directional behavioral variation;
- coordinate-wise coverage;
- uncertainty;
- optimization potential;
- geometric resolution.

The final decision remains deterministic and information-driven.

### 12.12 No-Pruning During Splitting

Structural splitting does not remove the parent region from the hierarchy.

If:

$$
R
\rightarrow
\{R_{\mathrm{left}},R_{\mathrm{right}}\},
$$

then:

$$
R
\in\mathcal{H}
$$

continues to hold.

Only the structural representation is expanded by adding the children.

This preserves the ARRGO no-pruning principle.

### 12.13 One-Dimensional Reduction

For:

$$
d=1,
$$

there is only one possible split dimension:

$$
j=1.
$$

The hyper-rectangle becomes:

$$
R=[l_1,r_1],
$$

and the binary split reduces to:

$$
[l_1,s_1]
\cup
[s_1,r_1].
$$

The generalized contraction condition becomes:

$$
\max
\left(
\frac{s_1-l_1}{r_1-l_1},
\frac{r_1-s_1}{r_1-l_1}
\right)
\leq
\rho,
$$

which is exactly the one-dimensional contraction formulation.

### Generalization Principle

The multidimensional splitting mechanism preserves the essential ARRGO structure:

> A region is structurally refined by selecting a coordinate direction and a feasible split location, producing smaller child regions while preserving feasibility, coverage, hierarchy, and deterministic contraction.

The additional multidimensional decision is the selection of the refinement direction.

## 13. Multidimensional Refinement Direction Selection

In one-dimensional ARRGO, there is only one possible spatial refinement direction.

In \(d\) dimensions, a region can be refined along any of its coordinate dimensions.

Therefore, structural refinement requires an additional decision:

$$
\text{Which coordinate dimension should be refined?}
$$

This decision is referred to as **refinement direction selection**.

### 13.1 Candidate Refinement Directions

For a region:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j],
$$

the set of coordinate refinement directions is:

$$
\mathcal{J}_R
=
\{1,\ldots,d\}.
$$

Each candidate direction \(j\) corresponds to a possible structural split along coordinate \(x_j\).

A directional refinement candidate can therefore be represented as:

$$
c_j=(j,s_j),
$$

where:

$$
l_j<s_j<r_j.
$$

### 13.2 Why Direction Selection Is Necessary

A multidimensional region may exhibit different levels of unresolved information along different dimensions.

For example, one dimension may have:

- poor spatial coverage;
- strong observed objective variation;
- large uncertainty;
- substantial potential variation.

Another dimension may already be well resolved.

Applying the same refinement strategy to every dimension would ignore this difference.

Therefore, ARRGO should use available information to determine which refinement direction is most relevant.

### 13.3 Coordinate-Wise Resolution

For each dimension \(j\), define a spatial resolution signal:

$$
Q_{\mathrm{resolution}}^{(j)}(R).
$$

This signal may incorporate the current side length:

$$
w_j(R)=r_j-l_j,
$$

as well as coordinate-wise coverage information.

A dimension with a relatively large unresolved spatial scale may be a candidate for refinement.

However, geometric size alone does not determine the refinement direction.

### 13.4 Directional Behavioral Information

The behavioral profile introduced previously provides information about how objective values vary across space.

For dimension \(j\), an auxiliary directional behavior signal may be constructed from observed finite differences associated with that coordinate.

Conceptually:

$$
Q_{\mathrm{behavior}}^{(j)}(R).
$$

A larger observed variation along a dimension may indicate that additional spatial resolution in that direction could reveal useful objective structure.

This remains an empirical information signal rather than a proof that the optimum lies in that direction.

### 13.5 Directional Coverage Information

Coordinate-wise coverage provides another signal.

For each dimension:

$$
\gamma_j(R)
$$

describes the relative size of the largest coordinate-wise coverage gap.

A dimension with a larger unresolved coverage gap may benefit from additional sampling or structural refinement.

Thus:

$$
Q_{\mathrm{coverage}}^{(j)}(R)
=
\gamma_j(R)
$$

can contribute to the directional refinement profile.

### 13.6 Directional Uncertainty

If certified bounds are available, the effect of refining a particular dimension can be considered through its contribution to regional uncertainty.

Conceptually, for candidate direction \(j\):

$$
Q_{\mathrm{uncertainty}}^{(j)}(R)
$$

describes the unresolved uncertainty associated with refining that coordinate.

In Empirical Mode, the corresponding quantity is an information signal.

In Certified Mode, rigorous uncertainty reduction must be derived from valid mathematical bounds rather than inferred solely from observed samples.

### 13.7 Directional Potential

A refinement direction may also be evaluated according to its effect on optimization potential.

For a candidate split:

$$
c_j=(j,s_j),
$$

the resulting child regions can be analyzed separately.

Their potential information can then be compared with the parent's current potential.

Conceptually:

$$
Q_{\mathrm{potential}}^{(j)}(R)
=
\text{potential resolution associated with splitting along }j.
$$

A direction that produces more informative regional separation may be preferred.

### 13.8 Directional Information Profile

The information available for each coordinate direction can therefore be represented as:

$$
Q_j(R)
=
\left(
Q_{\mathrm{coverage}}^{(j)},
Q_{\mathrm{behavior}}^{(j)},
Q_{\mathrm{uncertainty}}^{(j)},
Q_{\mathrm{potential}}^{(j)},
Q_{\mathrm{resolution}}^{(j)}
\right).
$$

The collection of directional profiles is:

$$
\mathcal{Q}(R)
=
\{Q_1(R),\ldots,Q_d(R)\}.
$$

This representation preserves the multidimensional structure of the information instead of immediately collapsing all signals into an arbitrary weighted scalar.

### 13.9 Dominance Between Directions

Because different directions may be favorable according to different criteria, a simple weighted sum is not required.

Instead, directional candidates can be compared using non-dominance.

Direction \(j_1\) dominates direction \(j_2\) if it is at least as favorable according to all relevant directional criteria and strictly more favorable according to at least one criterion.

Conceptually:

$$
Q_{j_1}(R)
\succeq
Q_{j_2}(R)
$$

with strict improvement in at least one component.

The set of non-dominated refinement directions is then:

$$
\mathcal{J}_{\mathrm{ND}}(R)
\subseteq
\mathcal{J}_R.
$$

This prevents an arbitrary fixed weighting scheme from determining the refinement direction.

### 13.10 Final Direction Selection

If multiple non-dominated directions remain, ARRGO requires a deterministic tie-breaking policy.

Possible deterministic criteria include:

1. stronger unresolved spatial resolution;
2. stronger observed directional variation;
3. stronger potential separation;
4. larger valid contraction opportunity;
5. deterministic dimension identifier as the final tie-break.

The final rule must be explicitly specified by the implementation.

The dimension identifier should be used only as a final deterministic tie-break, not as evidence that one dimension is intrinsically more important than another.

### 13.11 Direction Selection Is Not a Guarantee

A selected refinement direction is based on the information currently available to ARRGO.

It does not imply that the global optimum must lie along that dimension.

In particular:

$$
\text{selected direction}
\neq
\text{known optimal direction}.
$$

The purpose of direction selection is to allocate structural refinement toward regions and dimensions where the available information indicates that additional resolution may be useful.

### 13.12 Fairness Requirement

A purely local directional rule could repeatedly favor the same dimensions.

For convergence in multiple dimensions, this creates an important requirement.

Every dimension that remains spatially unresolved must receive sufficient refinement over the execution process.

Therefore, a multidimensional convergence argument requires a fairness or completeness condition ensuring that no relevant coordinate direction is permanently ignored.

Conceptually:

$$
\text{unresolved direction}
\quad\Longrightarrow\quad
\text{eventual refinement opportunity}.
$$

This is a theoretical condition, not something that can be assumed merely from the existence of multiple candidate directions.

### 13.13 Interaction With Sampling

Direction selection does not necessarily imply immediate splitting.

If the dominant unresolved information is spatial coverage rather than structural heterogeneity, additional sampling may be more appropriate.

Thus, ARRGO first considers the regional refinement objective and then determines whether:

$$
\text{Sampling}
$$

or:

$$
\text{Structural Splitting}
$$

provides the more appropriate information refinement.

If splitting is selected, directional selection determines the coordinate axis along which the region is refined.

### 13.14 One-Dimensional Reduction

For:

$$
d=1,
$$

the set of possible directions is:

$$
\mathcal{J}_R=\{1\}.
$$

Therefore, no directional choice is required.

The only available refinement direction is automatically selected, and the generalized mechanism reduces to the original one-dimensional refinement process.

### Generalization Principle

The multidimensional extension introduces a new decision layer:

$$
\boxed{
\text{Region}
\rightarrow
\text{Refinement Direction}
\rightarrow
\text{Split Location}
}
$$

The direction is selected from the currently available regional information, while preserving deterministic behavior and avoiding arbitrary fixed weighting of heterogeneous information signals.

For convergence, directional selection must additionally satisfy a fairness condition so that unresolved dimensions cannot be ignored indefinitely.

## 14. Generalized Refinement Decision

ARRGO must determine the appropriate refinement action for each region according to the information currently available.

The three fundamental actions are:

1. sampling;
2. structural splitting;
3. stable state.

This decision is preserved from the one-dimensional formulation, but the multidimensional setting introduces additional spatial directions and geometric choices.

### 14.1 Regional Information State

For a region \(R\), the available information can be represented by:

$$
Q(R)
=
\left(
Q_{\mathrm{coverage}}(R),
Q_{\mathrm{behavior}}(R),
Q_{\mathrm{uncertainty}}(R),
Q_{\mathrm{potential}}(R)
\right).
$$

In the multidimensional formulation, these quantities may themselves contain directional or vector-valued information.

For example:

$$
Q_{\mathrm{coverage}}(R)
=
(\gamma_1,\ldots,\gamma_d).
$$

Therefore, the regional information state should not be assumed to be a single scalar.

### 14.2 Sampling Decision

Sampling is appropriate when the dominant unresolved information is associated with insufficient objective observations.

Typical conditions include:

- insufficient evaluations inside the region;
- poor spatial coverage;
- unresolved local behavior;
- a candidate sampling point that can provide new information.

Conceptually:

$$
\text{Sampling Need}(R)
\quad\Longrightarrow\quad
\text{Sample}.
$$

The selected sampling point must satisfy:

$$
x_{\mathrm{sample}}\in R.
$$

After evaluation:

$$
D_{t+1}
=
D_t
\cup
\{(x_{\mathrm{sample}},f(x_{\mathrm{sample}}),\operatorname{Id})\}.
$$

The regional information state is then recomputed.

### 14.3 Structural Splitting Decision

Structural splitting is appropriate when the region contains sufficiently informative structure that additional spatial separation can improve resolution.

A split candidate is:

$$
c=(j,s_j),
$$

where:

$$
j\in\{1,\ldots,d\},
$$

and:

$$
l_j<s_j<r_j.
$$

The candidate must satisfy the required contraction condition.

If structural refinement is selected, the chosen candidate determines both:

- the refinement direction \(j\);
- the split location \(s_j\).

### 14.4 Stable Decision

A region may be considered stable when the currently available information does not justify additional refinement under the configured resolution criteria.

Conceptually:

$$
\text{Resolved}(R)
\quad\Longrightarrow\quad
\text{Stable}.
$$

Stability is therefore an information-state decision.

It does not mean that the true objective inside the region is known exactly.

In Empirical Mode, stability does not constitute a mathematical optimality guarantee.

In Certified Mode, a stronger statement may be possible when valid bounds establish that further refinement cannot improve the required certificate beyond the specified tolerance.

### 14.5 Action Selection Hierarchy

The refinement decision can be organized conceptually as:

$$
\boxed{
\text{Regional Information}
\rightarrow
\text{Resolution Assessment}
\rightarrow
\text{Action Selection}
}
$$

The resolution assessment determines which type of information remains unresolved.

If spatial objective information is insufficient, sampling may be preferred.

If the region contains sufficiently resolved information but requires greater structural separation, splitting may be preferred.

If the relevant resolution criteria are satisfied, the region may be considered stable.

### 14.6 Sampling Versus Splitting

Sampling and splitting provide different forms of refinement.

Sampling changes:

$$
D_t
\rightarrow
D_{t+1},
$$

while preserving the region geometry.

Splitting changes the structural representation:

$$
R
\rightarrow
\{R_1,R_2\},
$$

while not directly adding a new objective evaluation.

Therefore:

$$
\boxed{
\text{Sampling}
=
\text{Information Acquisition}
}
$$

and:

$$
\boxed{
\text{Splitting}
=
\text{Spatial Resolution}
}
$$

The algorithm should select between these actions according to the dominant unresolved information.

### 14.7 Multidimensional Split Selection

If structural splitting is selected, ARRGO evaluates candidate directions:

$$
j\in\{1,\ldots,d\},
$$

and candidate split locations:

$$
s_j\in(l_j,r_j).
$$

Each candidate can therefore be represented by:

$$
c_j=(j,s_j,Q_j),
$$

where \(Q_j\) contains the information profile associated with that candidate.

The candidate set is:

$$
C_{\mathrm{split}}(R)
=
\{c_1,\ldots,c_m\}.
$$

Only feasible candidates satisfying the contraction and numerical constraints are retained.

### 14.8 Multi-Criteria Decision

The action decision may depend on several distinct criteria.

For example:

$$
Q_{\mathrm{action}}(R)
=
\left(
Q_{\mathrm{coverage}},
Q_{\mathrm{behavior}},
Q_{\mathrm{uncertainty}},
Q_{\mathrm{potential}}
\right).
$$

These criteria represent different aspects of unresolved information.

ARRGO should avoid interpreting them as probabilities or combining them using arbitrary fixed weights without mathematical justification.

Instead, the decision can use:

- feasibility constraints;
- resolution criteria;
- dominance relationships;
- deterministic priority rules.

### 14.9 Global Context

A region is not evaluated completely independently from the rest of the search space.

The current global incumbent:

$$
f_{\mathrm{best}}^{(t)}
$$

affects the relevance of regional potential.

In Certified Mode, if:

$$
P(R)\leq f_{\mathrm{best}}^{(t)},
$$

then the region cannot contain a point that improves the current incumbent, provided that \(P(R)\) is a valid upper bound.

This information can influence the regional action decision.

However, the region remains stored in the hierarchy under the no-pruning principle.

### 14.10 Stable Does Not Mean Deleted

If a region is classified as stable:

$$
\operatorname{State}(R)=\mathrm{STABLE},
$$

the region remains part of the hierarchy.

A later global information update may change its relevance.

Therefore:

$$
\mathrm{STABLE}
\neq
\mathrm{DELETED}.
$$

This distinction is essential for preserving the no-pruning architecture.

### 14.11 Progress Requirement

Every refinement action must produce meaningful progress.

For sampling, progress means acquiring a new valid objective evaluation.

For splitting, progress means producing valid child regions with improved spatial resolution.

A split that produces numerically indistinguishable or degenerate children must not be accepted.

Similarly, the algorithm must not repeatedly select an action that cannot change the information state.

This prevents refinement loops without effective progress.

### 14.12 Budget Constraint

Only objective evaluations consume the evaluation budget.

Therefore, if:

$$
N_t=N_{\max},
$$

no further sampling operation may be performed.

Structural operations may not consume objective evaluations, but they must still be bounded by valid refinement and termination rules to prevent unproductive structural expansion.

### 14.13 Deterministic Decision

Given identical:

- global state;
- regional information;
- configuration;
- candidate set;
- numerical tolerances;

the same refinement action should be selected.

Therefore:

$$
\operatorname{Action}(R,D_t,\Theta)
$$

is deterministic.

This property is preserved in the multidimensional framework.

### 14.14 One-Dimensional Reduction

For:

$$
d=1,
$$

there is only one possible refinement direction.

The multidimensional action decision therefore reduces to the original one-dimensional choice between:

$$
\text{Sample},
\qquad
\text{Split},
\qquad
\text{Stable}.
$$

The additional directional-selection layer disappears because:

$$
\mathcal{J}_R=\{1\}.
$$

### Generalization Principle

The generalized ARRGO refinement decision preserves the original three-way action structure:

$$
\boxed{
\text{Sample}
\;\;|\;\;
\text{Split}
\;\;|\;\;
\text{Stable}
}
$$

while adding multidimensional direction selection when splitting is chosen.

The decision is based on unresolved information, geometric validity, global relevance, and deterministic rules rather than arbitrary fixed weighting.

Most importantly, stability represents a current information state, not deletion or a universal proof of optimality.

## 15. Generalized Global Region Selection

ARRGO maintains multiple regions simultaneously as part of its global search hierarchy.

At each refinement stage, the algorithm must determine which region should receive the next refinement opportunity.

This decision is referred to as **global region selection**.

### 15.1 Global Region Set

Let the current hierarchy contain the set of regions:

$$
\mathcal{R}_t.
$$

Not every region necessarily has the same lifecycle state.

For global refinement, ARRGO considers the regions that are currently eligible for further action.

Define the eligible region set as:

$$
\mathcal{R}_{\mathrm{eligible}}^{(t)}
\subseteq
\mathcal{R}_t.
$$

A region may be eligible when its current state and information indicate that additional refinement is possible.

### 15.2 Global Relevance

The relevance of a region depends on the information currently available about that region.

A generalized regional relevance profile can be represented as:

$$
Q_{\mathrm{global}}(R)
=
\left(
Q_{\mathrm{potential}}(R),
Q_{\mathrm{coverage}}(R),
Q_{\mathrm{behavior}}(R),
Q_{\mathrm{uncertainty}}(R),
Q_{\mathrm{resolution}}(R)
\right).
$$

These components describe different aspects of the region's current importance to the global search.

They should not automatically be interpreted as a single scalar score.

### 15.3 Certified Global Relevance

In Certified Mode, regional potential provides a particularly strong global signal.

For region \(R\):

$$
P(R)
=
\max_{x\in R}U_R(x).
$$

If the upper bound is valid, then:

$$
f_R^\ast
\leq
P(R).
$$

A region with larger \(P(R)\) has a larger currently permitted upper bound on its regional optimum.

Such a region may therefore deserve priority for further investigation.

### 15.4 Potential Relative to the Incumbent

The current incumbent is:

$$
f_{\mathrm{best}}^{(t)}.
$$

The regional potential excess is:

$$
E(R)
=
P(R)-f_{\mathrm{best}}^{(t)}.
$$

A positive value indicates that the current certified information still permits the region to contain an improvement over the incumbent.

If:

$$
E(R)\leq0,
$$

then, under valid certified bounds, the region cannot contain a point with objective value greater than the current incumbent.

This is a mathematical relevance criterion in Certified Mode.

### 15.5 Empirical Global Relevance

In Empirical Mode, no universal upper bound is assumed.

Therefore, global region selection must rely on observed information such as:

- objective values already observed in the region;
- spatial coverage;
- behavioral variation;
- regional resolution;
- unresolved information.

These quantities provide guidance but do not establish that a region cannot contain a better unseen solution.

Therefore:

$$
\text{low empirical relevance}
\neq
\text{proof of global irrelevance}.
$$

### 15.6 Exploration and Exploitation

Global region selection must balance two competing requirements.

**Exploration** requires continued attention to spatially unresolved regions.

**Exploitation** requires allocating additional refinement toward regions that currently show strong evidence of containing high-quality solutions.

This distinction can be expressed conceptually as:

$$
\text{Global Search}
=
\text{Exploration}
+
\text{Exploitation}.
$$

ARRGO does not introduce randomness to achieve this balance.

Instead, the balance is determined by the current information state and deterministic selection rules.

### 15.7 Global Region Priority

A deterministic global priority can be defined as an ordered collection of criteria.

For example:

1. unresolved refinement information;
2. certified potential, when available;
3. coverage resolution;
4. behavioral resolution;
5. geometric resolution;
6. deterministic identifier as a final tie-break.

The exact priority policy is an algorithmic choice.

It is not itself a mathematical theorem about which region contains the optimum.

### 15.8 Non-Dominance

Because the global information state is multi-criteria, ARRGO may first identify non-dominated regions.

A region \(R_1\) dominates \(R_2\) if it is at least as favorable according to all relevant criteria and strictly more favorable according to at least one criterion.

The set of non-dominated regions is:

$$
\mathcal{R}_{\mathrm{ND}}
\subseteq
\mathcal{R}_{\mathrm{eligible}}^{(t)}.
$$

The final region selection can then apply a deterministic policy to the remaining non-dominated candidates.

This avoids introducing arbitrary fixed weights between fundamentally different information dimensions.

### 15.9 Global Selection Is Different From Local Selection

Two distinct decisions must be maintained:

**Global region selection**

$$
\mathcal{R}_{\mathrm{eligible}}
\rightarrow
R^\ast
$$

determines which region receives attention.

**Local action selection**

$$
R^\ast
\rightarrow
\{\text{Sample},\text{Split},\text{Stable}\}
$$

determines what operation should be performed inside that region.

These decisions must not be conflated.

### 15.10 Interaction With Direction Selection

If the selected region requires structural refinement, the algorithm performs another decision:

$$
R^\ast
\rightarrow
(j,s_j).
$$

Therefore, the complete multidimensional decision chain is:

$$
\boxed{
\text{Global Region}
\rightarrow
\text{Local Action}
\rightarrow
\text{Refinement Direction}
\rightarrow
\text{Split Location}
}
$$

This hierarchy separates global search allocation from local geometric refinement.

### 15.11 Fairness

A global selection mechanism must not permanently ignore regions that remain unresolved and relevant.

For convergence, the selection process therefore requires a fairness condition.

Conceptually:

$$
R\text{ remains relevant and unresolved}
\quad\Longrightarrow\quad
R\text{ receives future refinement opportunities}.
$$

This condition is especially important in higher dimensions because a region can be locally competitive along one direction while other parts of the search space remain unresolved.

Fairness prevents the algorithm from indefinitely concentrating all refinement on one spatial path.

### 15.12 No-Pruning

Global selection does not imply deletion.

If a region is not selected at the current iteration:

$$
R\not=R^\ast,
$$

it remains in the hierarchy.

Likewise, if a region has already been refined:

$$
\operatorname{State}(R)=\mathrm{REFINED},
$$

the region remains stored together with its descendants.

Thus:

$$
\text{not selected}
\neq
\text{removed}.
$$

### 15.13 Global Incumbent Update

Whenever a new objective evaluation is obtained, the global incumbent is updated.

After \(t\) evaluations:

$$
f_{\mathrm{best}}^{(t)}
=
\max_{1\leq i\leq t}f(x_i).
$$

Therefore:

$$
f_{\mathrm{best}}^{(t+1)}
\geq
f_{\mathrm{best}}^{(t)}.
$$

The best observed objective value is consequently monotone non-decreasing.

The incumbent provides global information that can affect subsequent regional priorities.

### 15.14 Certified Global Gap

When valid certified potentials are available, global selection can also be interpreted through the remaining global gap:

$$
\Delta_{\mathrm{global}}^{(t)}
=
P_{\mathrm{global}}^{(t)}
-
f_{\mathrm{best}}^{(t)}.
$$

A region contributing strongly to:

$$
P_{\mathrm{global}}^{(t)}
$$

is directly relevant to reducing the remaining certification uncertainty.

Thus, global region selection can connect local refinement decisions to the global certification objective.

### 15.15 Termination Interaction

Global region selection continues only while a valid refinement action exists and the termination conditions have not been satisfied.

Possible termination conditions include:

- certified gap below tolerance;
- evaluation budget exhausted;
- numerical limit reached;
- no valid refinement action;
- configuration error.

If:

$$
\Delta_{\mathrm{global}}
\leq
\varepsilon
$$

in a valid Certified Mode configuration, the required certification objective has been reached and further refinement is unnecessary for that certificate.

### 15.16 One-Dimensional Reduction

For:

$$
d=1,
$$

every region has only one spatial refinement direction.

The global region-selection mechanism therefore reduces to selecting among one-dimensional intervals according to their regional information and global relevance.

The additional directional-selection layer disappears.

### Generalization Principle

Global region selection preserves the fundamental ARRGO principle:

> The algorithm allocates refinement opportunities across the search hierarchy according to the currently available regional information while maintaining global exploration, deterministic behavior, and fairness.

The multidimensional formulation adds no fundamentally new global objective; it extends the existing region-selection mechanism to a collection of hyper-rectangular regions with multidimensional information profiles.

## 16. Generalized Certified Bounds

Certified bounds provide a mathematically valid characterization of the unknown objective when the required assumptions are satisfied.

The multidimensional generalization preserves the Lipschitz-based certification mechanism used in the one-dimensional formulation.

### 16.1 Multidimensional Lipschitz Assumption

Assume that the objective:

$$
f:\Omega\rightarrow\mathbb{R}
$$

satisfies the global Lipschitz condition:

$$
|f(x)-f(y)|
\leq
L\|x-y\|_2
\qquad
\forall x,y\in\Omega,
$$

where:

$$
L\geq0
$$

is a valid global Lipschitz constant.

The validity of this constant is an assumption of the certified formulation.

### 16.2 Point-Wise Upper Bound

For an evaluated point:

$$
(x_i,f(x_i)),
$$

the Lipschitz condition gives:

$$
f(x)
\leq
f(x_i)+L\|x-x_i\|_2.
$$

Therefore, every evaluated point provides an upper bound on the objective at any other point in the domain.

### 16.3 Point-Wise Lower Bound

Similarly:

$$
f(x)
\geq
f(x_i)-L\|x-x_i\|_2.
$$

Thus, every evaluation also provides a lower bound.

### 16.4 Regional Upper Envelope

For a region:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j],
$$

with regional evaluation set:

$$
D_R,
$$

the combined upper envelope is:

$$
U_R(x)
=
\min_{x_i\in D_R}
\left[
f(x_i)+L\|x-x_i\|_2
\right].
$$

For every:

$$
x\in R,
$$

the Lipschitz assumption guarantees:

$$
f(x)
\leq
U_R(x).
$$

Therefore, \(U_R\) is a valid regional upper envelope.

### 16.5 Regional Lower Envelope

Similarly:

$$
L_R(x)
=
\max_{x_i\in D_R}
\left[
f(x_i)-L\|x-x_i\|_2
\right].
$$

For every:

$$
x\in R,
$$

we have:

$$
L_R(x)
\leq
f(x).
$$

Therefore:

$$
\boxed{
L_R(x)
\leq
f(x)
\leq
U_R(x)
}
$$

for every \(x\in R\), under the stated Lipschitz assumption.

### 16.6 Regional Certified Potential

The regional optimum is:

$$
f_R^\ast
=
\max_{x\in R}f(x).
$$

Since:

$$
f(x)
\leq
U_R(x),
$$

we obtain:

$$
f_R^\ast
\leq
\max_{x\in R}U_R(x).
$$

Define:

$$
P(R)
=
\max_{x\in R}U_R(x).
$$

Therefore:

$$
\boxed{
f_R^\ast
\leq
P(R)
}
$$

and \(P(R)\) is a valid upper bound on the regional optimum.

### 16.7 Global Certified Potential

Let:

$$
\mathcal{R}_t
$$

denote the set of regions used to represent the current global search coverage.

The global potential is:

$$
P_{\mathrm{global}}^{(t)}
=
\max_{R\in\mathcal{R}_t}P(R).
$$

If the regions collectively cover the search domain, then:

$$
f^\ast
\leq
P_{\mathrm{global}}^{(t)}.
$$

Together with:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast,
$$

the certified relationship becomes:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
\leq
f^\ast
\leq
P_{\mathrm{global}}^{(t)}
}
$$

### 16.8 Certified Global Gap

The remaining certified uncertainty is:

$$
\Delta_{\mathrm{global}}^{(t)}
=
P_{\mathrm{global}}^{(t)}
-
f_{\mathrm{best}}^{(t)}.
$$

The actual objective-value error therefore satisfies:

$$
0
\leq
f^\ast-f_{\mathrm{best}}^{(t)}
\leq
\Delta_{\mathrm{global}}^{(t)}.
$$

Consequently, if:

$$
\Delta_{\mathrm{global}}^{(t)}
\leq
\varepsilon,
$$

then:

$$
f_{\mathrm{best}}^{(t)}
\geq
f^\ast-\varepsilon.
$$

Thus, the incumbent is epsilon-optimal under the certified assumptions.

### 16.9 Geometric Upper-Bound Relation

For any region \(R\) and any evaluated point \(x_i\in R\):

$$
\|x-x_i\|_2
\leq
\operatorname{diam}(R)
\qquad
\forall x\in R.
$$

Therefore:

$$
f(x)
\leq
f(x_i)+L\operatorname{diam}(R).
$$

This gives a simple geometric upper-bound relation.

As the region diameter decreases, the maximum variation allowed by the Lipschitz condition also decreases:

$$
L\operatorname{diam}(R)
\rightarrow0
\qquad
\text{as}
\qquad
\operatorname{diam}(R)\rightarrow0.
$$

This connects structural refinement directly to the tightening of possible objective variation.

### 16.10 Multiple Evaluations Improve Bounds

The upper envelope uses the minimum over all regional evaluations:

$$
U_R(x)
=
\min_{x_i\in D_R}
\left[
f(x_i)+L\|x-x_i\|_2
\right].
$$

Adding a new evaluation cannot increase this upper envelope.

If:

$$
D_R^{(t)}
\subseteq
D_R^{(t+1)},
$$

then:

$$
U_R^{(t+1)}(x)
\leq
U_R^{(t)}(x).
$$

Similarly, adding evaluations cannot decrease the lower envelope:

$$
L_R^{(t+1)}(x)
\geq
L_R^{(t)}(x).
$$

Therefore, additional valid evaluations can only tighten the certified interval:

$$
[L_R(x),U_R(x)].
$$

### 16.11 Refinement and Certification

Suppose a region is structurally refined into children:

$$
R
\rightarrow
\{R_1,\ldots,R_k\}.
$$

Each child satisfies:

$$
R_i\subseteq R.
$$

If the children collectively cover the parent, then:

$$
f_R^\ast
=
\max_i f_{R_i}^\ast.
$$

Since each child has a valid certified upper bound:

$$
f_{R_i}^\ast
\leq
P(R_i),
$$

we obtain:

$$
f_R^\ast
\leq
\max_iP(R_i).
$$

Thus, structural refinement preserves the validity of the global upper-bound reasoning.

### 16.12 Parent Retention and Certified Bounds

ARRGO retains parent regions after splitting.

Therefore, the hierarchy may contain both:

$$
R
$$

and its children:

$$
R_1,\ldots,R_k.
$$

The parent region remains part of the structural history.

However, care must be taken when constructing the global certified potential.

If both parent and child regions are included as independent coverage regions, the same spatial domain may be represented multiple times.

This can produce a conservative but potentially loose global bound.

Therefore, a rigorous implementation must explicitly define which regions constitute the current coverage representation used for certification.

The hierarchy and the certification frontier are conceptually distinct objects.

### 16.13 Certification Frontier

Let:

$$
\mathcal{F}_t
\subseteq
\mathcal{R}_t
$$

denote a set of regions that collectively covers the domain without relying on duplicated parent coverage.

This set may be referred to as the **certification frontier**.

The global certified potential should then be defined over a valid covering frontier:

$$
P_{\mathrm{global}}^{(t)}
=
\max_{R\in\mathcal{F}_t}P(R).
$$

The exact construction of \(\mathcal{F}_t\) is an implementation and algorithm-design requirement.

The essential requirement is:

$$
\Omega
\subseteq
\bigcup_{R\in\mathcal{F}_t}R.
$$

When the frontier forms a partition up to shared boundaries, this becomes:

$$
\Omega
=
\bigcup_{R\in\mathcal{F}_t}R.
$$

This distinction becomes particularly important in the no-pruning architecture.

### 16.14 Certification Assumptions

The certified relationship is valid only when all required assumptions hold.

In particular:

1. the objective evaluations are valid;
2. the objective satisfies the stated Lipschitz condition;
3. the supplied \(L\) is a valid global Lipschitz constant;
4. the regions used for global certification cover the complete search domain;
5. the regional upper-bound calculations are mathematically valid;
6. numerical errors do not invalidate the intended inequalities.

If any of these assumptions fail, the resulting quantity must not be presented as a rigorous global certificate.

### 16.15 One-Dimensional Reduction

For:

$$
d=1,
$$

the Euclidean distance becomes:

$$
\|x-x_i\|_2
=
|x-x_i|.
$$

The multidimensional certified envelopes therefore reduce exactly to the original one-dimensional formulas.

Thus, the certified framework is dimension-independent under the same Lipschitz assumption.

### Generalization Principle

The multidimensional certified framework preserves the central ARRGO guarantee mechanism:

$$
\boxed{
\text{Valid Lipschitz Assumption}
+
\text{Valid Coverage}
+
\text{Valid Bounds}
\Rightarrow
f_{\mathrm{best}}
\leq
f^\ast
\leq
P_{\mathrm{global}}
}
$$

The mathematical certification mechanism therefore does not depend fundamentally on one-dimensional geometry.

What changes in higher dimensions is the geometry and computational difficulty of evaluating the envelopes and maintaining a valid coverage frontier.

The distinction between the complete historical hierarchy and the current certification frontier is essential for maintaining rigor under the no-pruning principle.

## 17. Multidimensional Convergence Conditions

The previous sections generalized the geometric and information-based components of ARRGO from one dimension to arbitrary finite dimensions.

The next requirement is to determine whether the convergence mechanism can also be generalized.

The objective is not to claim that every execution of ARRGO must converge under arbitrary conditions.

Instead, the goal is to identify sufficient conditions under which the sequence of ARRGO refinements can progressively resolve the search domain and approach the global optimum.

### 17.1 Global Optimization Setting

Consider:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j]
\subset\mathbb{R}^{d},
$$

with:

$$
a_j<b_j
\qquad
\forall j=1,\ldots,d.
$$

Assume:

$$
f\in C(\Omega).
$$

Because \(\Omega\) is compact and \(f\) is continuous, a global maximum exists:

$$
f^\ast
=
\max_{x\in\Omega}f(x).
$$

### 17.2 Spatial Refinement

Let:

$$
R_t
=
\prod_{j=1}^{d}[l_{j,t},r_{j,t}]
$$

denote a region selected along a refinement path.

Define its side lengths:

$$
w_{j,t}
=
r_{j,t}-l_{j,t}.
$$

The Euclidean diameter is:

$$
\operatorname{diam}(R_t)
=
\sqrt{
\sum_{j=1}^{d}w_{j,t}^{2}
}.
$$

For the diameter of a refinement path to converge to zero, every coordinate direction must eventually receive sufficient refinement.

### 17.3 Coordinate-Wise Contraction

Suppose a split along dimension \(j\) satisfies:

$$
w_j^{\mathrm{child}}
\leq
\rho\,w_j^{\mathrm{parent}},
$$

where:

$$
0<\rho<1.
$$

If dimension \(j\) is refined \(n_j\) times along a refinement path, then:

$$
w_j^{(n_j)}
\leq
\rho^{n_j}w_j^{(0)}.
$$

Therefore:

$$
n_j\rightarrow\infty
\quad\Longrightarrow\quad
w_j^{(n_j)}\rightarrow0.
$$

### 17.4 Fair Refinement Across Dimensions

A central requirement of multidimensional convergence is therefore:

> Every coordinate direction that remains spatially unresolved must receive sufficient refinement opportunities.

Formally, along any refinement path that remains relevant to the global search, let:

$$
n_j(t)
$$

denote the number of effective refinements performed along dimension \(j\) up to stage \(t\).

A sufficient directional fairness condition is:

$$
n_j(t)\rightarrow\infty
\qquad
\forall j=1,\ldots,d.
$$

Under this condition:

$$
w_j(t)\rightarrow0
\qquad
\forall j.
$$

Consequently:

$$
\operatorname{diam}(R_t)
\rightarrow0.
$$

This condition is stronger than merely requiring that the algorithm perform infinitely many splits.

It requires refinement to be distributed across all unresolved coordinate directions.

### 17.5 Why Fairness Is Necessary

Suppose a two-dimensional region has side lengths:

$$
w_1>0,
\qquad
w_2>0.
$$

If the algorithm repeatedly refines only dimension \(1\), then:

$$
w_1(t)\rightarrow0,
$$

but:

$$
w_2(t)=w_2(0)
$$

may remain unchanged.

Therefore:

$$
\operatorname{diam}(R_t)
\not\rightarrow0.
$$

The region becomes narrow but does not collapse to a point.

Thus, repeated refinement without directional fairness is insufficient for a general multidimensional convergence argument.

### 17.6 Spatial Resolution and Objective Resolution

If the objective is continuous on the compact domain, it is uniformly continuous.

Therefore, for every:

$$
\varepsilon>0,
$$

there exists:

$$
\delta>0
$$

such that:

$$
\|x-y\|_2<\delta
\quad\Longrightarrow\quad
|f(x)-f(y)|<\varepsilon
$$

for all:

$$
x,y\in\Omega.
$$

Consequently, if:

$$
\operatorname{diam}(R_t)\rightarrow0,
$$

then the variation of the objective within the region also tends to zero:

$$
\sup_{x,y\in R_t}
|f(x)-f(y)|
\rightarrow0.
$$

This provides the fundamental connection between spatial contraction and objective-value resolution.

### 17.7 Sampling Requirement

Spatial contraction alone does not guarantee that ARRGO evaluates informative points near the global optimum.

The refinement process must also acquire objective information from relevant regions.

Therefore, a second condition is required:

> Regions that remain globally relevant must receive sufficient objective evaluations.

This may be expressed through an information-acquisition or sampling fairness condition.

Conceptually, for every relevant refinement path:

$$
|D_{R_t}|
\rightarrow
\text{sufficient information}
$$

as the region is progressively refined.

The exact formal definition depends on the sampling policy.

### 17.8 Incumbent Improvement

The best observed objective value is:

$$
f_{\mathrm{best}}^{(t)}
=
\max_{1\leq i\leq t}f(x_i).
$$

Because the evaluation history only grows:

$$
f_{\mathrm{best}}^{(t+1)}
\geq
f_{\mathrm{best}}^{(t)}.
$$

Thus, the incumbent sequence is monotone non-decreasing.

Since:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast,
$$

the sequence is bounded above.

Therefore, the scalar sequence:

$$
\{f_{\mathrm{best}}^{(t)}\}
$$

has a finite limit:

$$
\lim_{t\rightarrow\infty}
f_{\mathrm{best}}^{(t)}
=
\bar f
\leq
f^\ast.
$$

To establish:

$$
\bar f=f^\ast,
$$

additional conditions on spatial refinement and information acquisition are required.

### 17.9 Convergence of Objective Values

Assume that a sequence of evaluated points:

$$
x_{i_t}
$$

approaches a global maximizer:

$$
x_{i_t}\rightarrow x^\ast,
$$

where:

$$
f(x^\ast)=f^\ast.
$$

By continuity:

$$
f(x_{i_t})
\rightarrow
f(x^\ast)
=
f^\ast.
$$

Since the incumbent is at least as good as every previously evaluated point:

$$
f_{\mathrm{best}}^{(t)}
\geq
f(x_{i_t}),
$$

and:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast,
$$

it follows that:

$$
f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast.
$$

Thus, sufficiently strong spatial refinement and evaluation near a global maximizer imply objective-value convergence.

### 17.10 Certified Convergence

Under the additional assumption of a valid global Lipschitz constant \(L\), the certified framework provides a direct convergence mechanism.

If:

$$
\operatorname{diam}(R_t)\rightarrow0
$$

and the relevant regions acquire sufficient evaluations, the regional upper envelopes can become increasingly tight.

Consequently, under valid coverage and bound construction:

$$
P_{\mathrm{global}}^{(t)}
-
f_{\mathrm{best}}^{(t)}
\rightarrow0.
$$

Therefore:

$$
\Delta_{\mathrm{global}}^{(t)}
\rightarrow0.
$$

This statement remains conditional on valid certified assumptions and sufficient refinement and information acquisition.

### 17.11 Generalized Convergence Principle

The multidimensional convergence mechanism can therefore be summarized as:

$$
\boxed{
\begin{aligned}
&\text{Spatial Contraction}\\
&+\text{Directional Fairness}\\
&+\text{Sufficient Information Acquisition}\\
&+\text{Global Relevance}\\
&+\text{Continuity}
\\
&\Longrightarrow
\text{Global Objective-Value Convergence}
\end{aligned}
}
$$

For Certified Mode, valid Lipschitz bounds and valid domain coverage provide an additional route to certified convergence of the global gap.

### 17.12 What Is Actually Guaranteed

The generalized framework does not claim:

$$
\text{arbitrary execution}
\Longrightarrow
\text{global convergence}.
$$

Instead, the convergence statement is conditional:

$$
\boxed{
\text{Assumptions}
+
\text{Valid Refinement}
+
\text{Fairness}
+
\text{Sufficient Information}
\Longrightarrow
\text{Convergence}
}
$$

This distinction is essential for mathematical correctness.

### 17.13 One-Dimensional Reduction

For:

$$
d=1,
$$

there is only one coordinate direction.

Therefore, directional fairness becomes automatic because:

$$
n_1(t)\rightarrow\infty
$$

is the only required refinement-direction condition.

The multidimensional convergence conditions consequently reduce to the original one-dimensional convergence mechanism.

### Generalization Principle

The convergence argument is fundamentally dimension-independent.

The additional requirement introduced by higher dimensions is directional completeness:

> A globally relevant refinement process must not permanently ignore any unresolved coordinate direction.

Once spatial contraction, sufficient information acquisition, global relevance, and the required regularity assumptions are satisfied, the original ARRGO convergence mechanism extends naturally from intervals to hyper-rectangular regions in arbitrary finite dimensions.

## 18. Generalized Global Convergence Mechanism

The previous section established conditions under which a relevant multidimensional refinement path can become spatially arbitrarily small.

Global optimization requires a stronger property.

It is not sufficient for one refinement path to converge to a point.

The search process must ensure that globally relevant portions of the domain continue to receive refinement and objective information.

This section formulates the generalized global convergence mechanism of ARRGO.

### 18.1 Global Search Coverage

Let the current certification or refinement frontier be:

$$
\mathcal{F}_t.
$$

The frontier must collectively represent the complete search domain:

$$
\Omega
\subseteq
\bigcup_{R\in\mathcal{F}_t}R.
$$

For a partition up to shared boundaries:

$$
\Omega
=
\bigcup_{R\in\mathcal{F}_t}R.
$$

This ensures that the global search does not permanently exclude a portion of the feasible domain.

### 18.2 Global Region Relevance

At each stage, ARRGO evaluates the available information associated with regions in the current search structure.

A region may remain globally relevant because:

- its objective information is incomplete;
- its spatial resolution is insufficient;
- its observed behavior is unresolved;
- its certified potential remains above the incumbent;
- or its global status cannot yet be ruled out.

The global selection mechanism must preserve opportunities for such regions to receive future refinement.

### 18.3 Global Fairness

Let:

$$
R\in\mathcal{F}_t
$$

be a region that remains globally relevant and unresolved.

A global fairness condition requires that \(R\) cannot be ignored indefinitely.

Conceptually:

$$
R\text{ remains relevant and unresolved}
\quad\Longrightarrow\quad
R\text{ receives future refinement opportunities}.
$$

This condition prevents the global search from becoming permanently concentrated on a single region.

### 18.4 Refinement of the Global Frontier

Suppose a frontier region:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j]
$$

is selected for structural refinement.

A valid split produces:

$$
R_1,\ldots,R_k
\subseteq R.
$$

The parent remains in the historical hierarchy, while the frontier representation is updated so that the children provide the refined coverage.

Thus:

$$
\mathcal{F}_{t+1}
=
\left(
\mathcal{F}_t\setminus\{R\}
\right)
\cup
\{R_1,\ldots,R_k\}.
$$

The complete hierarchy does not undergo this replacement; only the certification/refinement frontier does.

### 18.5 Spatial Resolution of the Frontier

For each frontier region, define:

$$
\delta(R)
=
\operatorname{diam}(R).
$$

Global spatial refinement can be characterized by:

$$
\sup_{R\in\mathcal{F}_t}\delta(R)
\rightarrow0
$$

along a sufficiently refined globally relevant process.

This means that the largest unresolved spatial scale of the frontier tends toward zero.

Such a condition is stronger than shrinking only one selected region.

### 18.6 Why Global Coverage Matters

Suppose ARRGO refined one small region around a candidate optimum while leaving another large region almost completely unresolved.

Even if:

$$
\operatorname{diam}(R_{\mathrm{selected}})\rightarrow0,
$$

the existence of a large unresolved region means that a potentially better global optimum may remain undiscovered.

Therefore:

$$
\text{local contraction}
\not\Rightarrow
\text{global convergence}.
$$

Global convergence requires sufficient resolution of all globally relevant portions of the search space.

### 18.7 Interaction With the Global Incumbent

The incumbent value is:

$$
f_{\mathrm{best}}^{(t)}
=
\max_{1\leq i\leq t}f(x_i).
$$

The incumbent is monotone:

$$
f_{\mathrm{best}}^{(t+1)}
\geq
f_{\mathrm{best}}^{(t)}.
$$

As new regions are sampled and refined, the incumbent may improve.

The global selection mechanism therefore operates relative to an evolving objective-value reference.

### 18.8 Certified Global Relevance

In Certified Mode, the regional potential provides a mathematically meaningful relevance measure.

For every frontier region:

$$
f_R^\ast
\leq
P(R).
$$

Therefore:

$$
P(R)
\leq
f_{\mathrm{best}}^{(t)}
$$

implies:

$$
f_R^\ast
\leq
f_{\mathrm{best}}^{(t)}.
$$

Such a region cannot improve the incumbent under valid certified assumptions.

Regions satisfying:

$$
P(R)>f_{\mathrm{best}}^{(t)}
$$

remain potentially relevant to the certification objective.

### 18.9 Certified Global Gap

The global potential is:

$$
P_{\mathrm{global}}^{(t)}
=
\max_{R\in\mathcal{F}_t}P(R).
$$

The certified gap is:

$$
\Delta_{\mathrm{global}}^{(t)}
=
P_{\mathrm{global}}^{(t)}
-
f_{\mathrm{best}}^{(t)}.
$$

If:

$$
\Delta_{\mathrm{global}}^{(t)}
\rightarrow0,
$$

then:

$$
f^\ast-f_{\mathrm{best}}^{(t)}
\rightarrow0.
$$

Thus, the certified gap provides a direct global convergence criterion.

### 18.10 Objective-Value Convergence

Under the continuity assumption and sufficient global refinement and sampling, the incumbent satisfies:

$$
f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast.
$$

This follows because globally relevant regions cannot remain unresolved indefinitely and sufficiently refined regions have diminishing spatial diameter.

By continuity, objective variation within increasingly small regions becomes arbitrarily small.

Therefore, evaluations obtained sufficiently close to global maximizers can approach the global optimum value.

### 18.11 Location Convergence

Objective-value convergence does not necessarily imply convergence to a unique point.

If the global maximizer is unique:

$$
X^\ast=\{x^\ast\},
$$

then sufficiently accurate objective-value convergence together with spatial refinement can support convergence of relevant evaluated points toward:

$$
x^\ast.
$$

If multiple global maximizers exist:

$$
|X^\ast|>1,
$$

there is no unique target location.

In that case, convergence should be interpreted as approaching the set:

$$
X^\ast.
$$

This distinction is important for flat or plateau-like objectives.

### 18.12 Global Convergence Mechanism

The generalized ARRGO mechanism can therefore be represented as:

$$
\boxed{
\begin{aligned}
&\text{Global Coverage}\\
&+\text{Global Fairness}\\
&+\text{Directional Refinement}\\
&+\text{Spatial Contraction}\\
&+\text{Sufficient Sampling}\\
&+\text{Objective Regularity}
\\
&\Longrightarrow
\text{Global Objective-Value Convergence}
\end{aligned}
}
$$

For Certified Mode:

$$
\boxed{
\text{Valid Bounds}
+
\text{Global Coverage}
+
\text{Progressive Refinement}
\Longrightarrow
\Delta_{\mathrm{global}}\rightarrow0
}
$$

under the stated assumptions.

### 18.13 No-Pruning Compatibility

Global convergence does not require deleting regions from the hierarchy.

The complete hierarchy:

$$
\mathcal{H}_t
$$

stores the historical search structure.

The current frontier:

$$
\mathcal{F}_t
$$

represents the current spatial coverage used for refinement or certification.

Therefore:

$$
\mathcal{F}_t
\subseteq
\mathcal{H}_t.
$$

This separation allows ARRGO to preserve the no-pruning principle while maintaining a mathematically meaningful current search frontier.

### 18.14 One-Dimensional Reduction

For:

$$
d=1,
$$

hyper-rectangles become intervals and there is only one spatial refinement direction.

The multidimensional global convergence mechanism therefore reduces to the original one-dimensional mechanism involving:

- region coverage;
- interval contraction;
- sufficient sampling;
- global region selection;
- incumbent improvement;
- and, where applicable, certified global bounds.

### Generalization Principle

The global convergence mechanism is fundamentally independent of the number of dimensions.

The essential requirement introduced by multidimensional geometry is that spatial refinement must be sufficiently distributed across all unresolved coordinate directions and globally relevant regions.

Therefore, ARRGO remains a region-based global optimization framework in arbitrary finite dimensions, provided that the stated coverage, fairness, refinement, sampling, and regularity assumptions are satisfied.

## 19. Multidimensional Evaluation Density and Information Acquisition

Spatial refinement reduces the geometric size of regions, but convergence also depends on the acquisition of objective information.

A region can become geometrically small while still containing insufficient objective evaluations.

Therefore, ARRGO requires a separate condition describing how objective information is acquired during refinement.

### 19.1 Evaluation Set

At stage \(t\), let the global evaluation history be:

$$
D_t
=
\{(x_i,f(x_i),\operatorname{Id}_i)\}_{i=1}^{t}.
$$

For a region:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j],
$$

the regional evaluations are:

$$
D_R
=
\{e_i\in D_t:x_i\in R\}.
$$

The amount and spatial distribution of \(D_R\) determine the local information resolution available to ARRGO.

### 19.2 Spatial Resolution of Evaluations

For a sequence of evaluated points that becomes increasingly dense in a region, the maximum distance from an arbitrary point in the region to its nearest evaluation decreases.

Define:

$$
h(R,D_R)
=
\sup_{x\in R}
\min_{x_i\in D_R}
\|x-x_i\|_2.
$$

This quantity is the covering radius of the evaluated points over \(R\).

When:

$$
h(R,D_R)\rightarrow0,
$$

the evaluation set becomes increasingly spatially resolving over the region.

### 19.3 Information Acquisition

ARRGO acquires information through new objective evaluations:

$$
D_t
\rightarrow
D_{t+1}.
$$

Each valid new evaluation provides:

- a new spatial observation;
- a new objective value;
- additional information for regional analysis;
- potentially tighter certified bounds.

Therefore, information acquisition is distinct from structural refinement.

### 19.4 Sampling Does Not Automatically Imply Density

Increasing the number of evaluations does not necessarily imply that the evaluation set becomes spatially dense.

For example, repeatedly sampling within a small portion of a region may increase:

$$
|D_R|
$$

while leaving:

$$
h(R,D_R)
$$

large elsewhere.

Therefore:

$$
\text{evaluation count}
\neq
\text{spatial resolution}.
$$

The spatial distribution of evaluations is essential.

### 19.5 Global Information Acquisition

For global convergence, the evaluation process must sufficiently resolve regions that remain globally relevant.

Conceptually, if:

$$
R_t
$$

remains globally relevant, then ARRGO must continue acquiring information in or sufficiently near \(R_t\).

A suitable asymptotic requirement is that the relevant regions do not remain permanently devoid of informative evaluations.

### 19.6 Relation to Continuity

Assume:

$$
f\in C(\Omega).
$$

Since \(\Omega\) is compact, \(f\) is uniformly continuous.

Therefore, for every:

$$
\varepsilon>0,
$$

there exists:

$$
\delta>0
$$

such that:

$$
\|x-y\|_2<\delta
\quad\Longrightarrow\quad
|f(x)-f(y)|<\varepsilon.
$$

If evaluations sufficiently resolve a region so that:

$$
h(R,D_R)<\delta,
$$

then every point in \(R\) lies within \(\delta\) of at least one evaluated point.

Consequently, the objective variation between each point and a nearby evaluation is smaller than \(\varepsilon\).

This connects spatial evaluation density with objective-value resolution.

### 19.7 Relation to Certified Bounds

Under the Lipschitz assumption:

$$
|f(x)-f(y)|
\leq
L\|x-y\|_2,
$$

the covering radius provides a direct bound.

For every \(x\in R\), let \(x_i\) be a nearest evaluated point.

Then:

$$
\|x-x_i\|_2
\leq
h(R,D_R).
$$

Therefore:

$$
f(x)
\leq
f(x_i)+Lh(R,D_R),
$$

and:

$$
f(x)
\geq
f(x_i)-Lh(R,D_R).
$$

Hence, spatially denser evaluations can produce tighter local objective bounds.

### 19.8 Information Acquisition Near a Global Maximizer

Let:

$$
x^\ast\in X^\ast
$$

be a global maximizer.

If ARRGO produces a sequence of evaluations:

$$
x_{i_t}
\rightarrow
x^\ast,
$$

then continuity gives:

$$
f(x_{i_t})
\rightarrow
f(x^\ast)
=
f^\ast.
$$

Therefore, obtaining evaluations sufficiently close to a global maximizer is sufficient for the best observed objective value to approach the global optimum.

This does not require the optimizer itself to be evaluated exactly.

### 19.9 Global Fairness and Information Acquisition

A multidimensional convergence argument therefore requires two related fairness properties.

**Spatial refinement fairness**

Every relevant unresolved coordinate direction must eventually receive sufficient structural refinement.

**Information acquisition fairness**

Every globally relevant unresolved region must eventually receive sufficient objective information.

Together:

$$
\boxed{
\text{Spatial Fairness}
+
\text{Information Fairness}
}
$$

prevent the algorithm from permanently ignoring a relevant portion of the search space.

### 19.10 Information Refinement Versus Structural Refinement

The two mechanisms can be summarized as:

$$
\boxed{
\text{Sampling}
\rightarrow
\text{Information Resolution}
}
$$

and:

$$
\boxed{
\text{Splitting}
\rightarrow
\text{Spatial Resolution}
}
$$

Neither mechanism alone is sufficient in the general case.

A useful global optimization process must coordinate both.

### 19.11 Empty Regions

If:

$$
D_R=\varnothing,
$$

then:

$$
h(R,D_R)
$$

is undefined under the nearest-evaluation definition.

Such a region has no local objective information.

Therefore, an appropriate sampling action is required before local evaluation density can be assessed.

### 19.12 Finite Evaluation Budgets

In practical execution, ARRGO operates under a finite evaluation budget:

$$
N_{\max}<\infty.
$$

Therefore, the asymptotic condition:

$$
h(R,D_R)\rightarrow0
$$

describes a theoretical convergence mechanism rather than a requirement that must be reached during every finite execution.

A finite-budget run produces the best solution supported by the information acquired before termination.

### 19.13 One-Dimensional Reduction

For:

$$
d=1,
$$

the Euclidean distance becomes absolute distance:

$$
\|x-x_i\|_2=|x-x_i|.
$$

The covering-radius definition therefore reduces to the one-dimensional maximum distance from a point in the interval to the nearest evaluation.

Thus, the generalized information-acquisition formulation contains the original one-dimensional formulation as a special case.

### Generalization Principle

The multidimensional convergence framework requires more than geometric contraction.

The essential information condition is:

> Globally relevant regions must receive sufficiently distributed objective evaluations as their spatial resolution increases.

This establishes the second major component of the convergence mechanism:

$$
\boxed{
\text{Spatial Refinement}
+
\text{Information Acquisition}
\rightarrow
\text{Objective Resolution}
}
$$

Together with global fairness and the required regularity assumptions, this provides the information-theoretic foundation for multidimensional convergence.

## 20. Generalized Termination Criteria

Termination determines when ARRGO should stop the optimization process.

The termination mechanism must remain valid for arbitrary finite dimensions while distinguishing between:

- successful certified termination;
- resource-limited termination;
- numerical termination;
- absence of valid refinement actions;
- invalid configuration.

### 20.1 Global Execution State

At stage \(t\), the relevant global state includes:

$$
G_t
=
\left(
D_t,
\mathcal{H}_t,
\mathcal{F}_t,
x_{\mathrm{best}}^{(t)},
f_{\mathrm{best}}^{(t)}
\right).
$$

The state represents:

- acquired objective information;
- complete region hierarchy;
- current refinement or certification frontier;
- incumbent location;
- best observed objective value.

Termination is determined from this global state and the configured constraints.

### 20.2 Certified Tolerance Termination

In Certified Mode, suppose a valid global certified potential exists:

$$
P_{\mathrm{global}}^{(t)}
=
\max_{R\in\mathcal{F}_t}P(R).
$$

The certified gap is:

$$
\Delta_{\mathrm{global}}^{(t)}
=
P_{\mathrm{global}}^{(t)}
-
f_{\mathrm{best}}^{(t)}.
$$

If:

$$
\Delta_{\mathrm{global}}^{(t)}
\leq
\varepsilon,
$$

then:

$$
f_{\mathrm{best}}^{(t)}
\geq
f^\ast-\varepsilon.
$$

Therefore, ARRGO may terminate with a mathematically justified epsilon-optimality certificate.

This termination condition is valid only when the certified assumptions hold.

### 20.3 Evaluation Budget Termination

Let:

$$
N_{\max}
$$

be the maximum number of objective evaluations.

If:

$$
|D_t|=N_{\max},
$$

then no additional objective evaluation may be performed.

ARRGO terminates with:

$$
\operatorname{Reason}
=
\mathrm{BudgetExhausted}.
$$

In this case, the result is the best solution found under the available budget.

Budget termination does not imply global optimality in Empirical Mode.

### 20.4 Numerical Limit Termination

Numerical precision can eventually prevent meaningful refinement.

Examples include:

- a candidate point becoming numerically indistinguishable from an existing evaluation;
- a split location becoming indistinguishable from a region boundary;
- child regions becoming numerically degenerate;
- floating-point operations failing to produce distinguishable spatial states.

If no numerically meaningful progress can be made, ARRGO may terminate with:

$$
\operatorname{Reason}
=
\mathrm{NumericalLimit}.
$$

This termination reason indicates a numerical limitation rather than an optimization certificate.

### 20.5 No Valid Action Termination

A region may have no feasible refinement action remaining under the current configuration.

More generally, if:

$$
\mathcal{A}_{\mathrm{valid}}^{(t)}
=
\varnothing,
$$

where \(\mathcal{A}_{\mathrm{valid}}^{(t)}\) is the set of valid global actions, ARRGO terminates with:

$$
\operatorname{Reason}
=
\mathrm{NoValidAction}.
$$

This prevents the algorithm from entering an infinite loop in which it repeatedly selects an action that cannot produce meaningful progress.

### 20.6 Configuration Error

A configuration is invalid if the assumptions required for execution are violated.

Examples include:

- invalid domain bounds;
- invalid evaluation budget;
- invalid contraction factor;
- missing required certified parameters;
- inconsistent numerical tolerances.

Such a condition should be detected before or during execution and represented separately from normal optimization termination.

Conceptually:

$$
\operatorname{Reason}
=
\mathrm{ConfigurationError}.
$$

A configuration error is not evidence about the objective function.

### 20.7 Stability and Termination

A region becoming stable does not necessarily terminate the global optimization process.

A stable region remains part of the hierarchy:

$$
R\in\mathcal{H}_t.
$$

Other regions may remain unresolved and globally relevant.

Therefore:

$$
\text{Stable Region}
\neq
\text{Global Termination}.
$$

Global termination occurs only when the complete termination criteria are satisfied.

### 20.8 Multidimensional Resolution

In \(d\) dimensions, a region may be highly refined along one coordinate while remaining coarse along another.

Therefore, a local resolution condition must account for the complete geometry.

For example:

$$
\operatorname{diam}(R)
=
\sqrt{
\sum_{j=1}^{d}(r_j-l_j)^2
}
$$

provides a global geometric measure of the region size.

A region should not be considered spatially fully resolved merely because one coordinate width has become small.

### 20.9 Directional Resolution

For each dimension:

$$
w_j(R)=r_j-l_j.
$$

A multidimensional resolution assessment can therefore examine:

$$
\left(
w_1(R),\ldots,w_d(R)
\right).
$$

If a convergence condition requires all directions to become sufficiently resolved, then the relevant criterion must ensure:

$$
w_j(R)\rightarrow0
\qquad
\forall j.
$$

This is consistent with the directional fairness requirement established earlier.

### 20.10 Global Termination Versus Local Resolution

The distinction between local and global termination is essential.

A local region may satisfy:

$$
\operatorname{Resolved}(R)=\mathrm{True},
$$

while another region remains unresolved.

Therefore:

$$
\exists R:
\operatorname{Resolved}(R)=\mathrm{True}
$$

does not imply:

$$
\operatorname{Terminate}=\mathrm{True}.
$$

The global termination decision must consider the complete eligible frontier.

### 20.11 Certified Global Termination

In Certified Mode, a particularly strong termination condition is:

$$
\Delta_{\mathrm{global}}^{(t)}
\leq
\varepsilon.
$$

Because:

$$
0
\leq
f^\ast-f_{\mathrm{best}}^{(t)}
\leq
\Delta_{\mathrm{global}}^{(t)},
$$

this condition provides a direct bound on the remaining objective-value error.

Thus, unlike empirical convergence indicators, certified tolerance termination has a mathematical interpretation.

### 20.12 Empirical Termination

Empirical Mode does not generally provide a rigorous unseen-objective bound.

Therefore, practical empirical termination may use:

- evaluation budget;
- numerical limits;
- absence of valid actions;
- explicitly configured empirical resolution criteria.

Such termination indicates that ARRGO has completed its configured execution.

It does not by itself imply:

$$
f_{\mathrm{best}}=f^\ast.
$$

### 20.13 Termination Priority

If multiple termination conditions become true simultaneously, the implementation should use an explicit deterministic priority.

A possible priority is:

1. configuration validity;
2. certified tolerance;
3. evaluation budget;
4. numerical limit;
5. no valid action.

The exact ordering is an implementation policy and should be documented.

The important requirement is that the policy is deterministic and unambiguous.

### 20.14 Termination Correctness

Every normal execution should terminate for a clearly identifiable reason.

The termination state should therefore include:

$$
\operatorname{TerminationReason}.
$$

This prevents a result from being interpreted without knowing whether ARRGO stopped because:

- the required certificate was achieved;
- the evaluation budget ended;
- numerical progress became impossible;
- no valid action remained;
- or the configuration was invalid.

### 20.15 One-Dimensional Reduction

For:

$$
d=1,
$$

the region diameter becomes:

$$
\operatorname{diam}(R)=r_1-l_1.
$$

Directional resolution has only one component.

Therefore, the generalized termination mechanism reduces to the original one-dimensional termination logic.

### Generalization Principle

Termination remains a global decision in the multidimensional ARRGO framework.

The key additional requirement is that spatial resolution must account for all relevant coordinate directions.

Therefore:

$$
\boxed{
\text{Local Stability}
\neq
\text{Global Termination}
}
$$

and:

$$
\boxed{
\text{Certified Gap}
\leq
\varepsilon
\Rightarrow
\text{Certified Termination}
}
$$

when the required assumptions and valid global bounds are satisfied.

The generalized termination mechanism therefore preserves the distinction between best-found empirical results, numerical stopping conditions, and mathematically certified optimization results.

## 21. One-Dimensional to Multidimensional Mapping

The multidimensional formulation is intended as a generalization of the original one-dimensional ARRGO framework.

The relationship between the two formulations can therefore be expressed component by component.

The one-dimensional case corresponds to:

$$
d=1.
$$

The generalized formulation allows:

$$
d\geq1.
$$

### 21.1 Optimization Domain

One-dimensional formulation:

$$
\Omega=[a,b].
$$

Multidimensional formulation:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j].
$$

For:

$$
d=1,
$$

the generalized domain becomes:

$$
\Omega=[a_1,b_1].
$$

Therefore, the original bounded interval is recovered exactly.

### 21.2 Decision Variable

One-dimensional formulation:

$$
x\in\mathbb{R}.
$$

Multidimensional formulation:

$$
x=(x_1,\ldots,x_d)\in\mathbb{R}^{d}.
$$

For:

$$
d=1,
$$

the vector contains a single coordinate and becomes the original scalar decision variable.

### 21.3 Region Representation

One-dimensional formulation:

$$
R=[l,r].
$$

Multidimensional formulation:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j].
$$

For:

$$
d=1,
$$

this becomes:

$$
R=[l_1,r_1].
$$

Thus, the interval is a special case of the hyper-rectangular representation.

### 21.4 Region Width

One-dimensional formulation:

$$
w(R)=r-l.
$$

Multidimensional formulation:

$$
w_j(R)=r_j-l_j.
$$

The generalized formulation additionally provides:

$$
w_{\max}(R)
=
\max_j w_j(R)
$$

and:

$$
\operatorname{diam}(R)
=
\sqrt{\sum_{j=1}^{d}w_j(R)^2}.
$$

For:

$$
d=1,
$$

we obtain:

$$
\operatorname{diam}(R)=r_1-l_1.
$$

### 21.5 Region Center

One-dimensional formulation:

$$
c(R)=\frac{l+r}{2}.
$$

Multidimensional formulation:

$$
c(R)
=
\left(
\frac{l_1+r_1}{2},
\ldots,
\frac{l_d+r_d}{2}
\right).
$$

For:

$$
d=1,
$$

the generalized center becomes the original interval midpoint.

### 21.6 Objective Evaluation

One-dimensional evaluation:

$$
e_i=(x_i,f(x_i),\operatorname{Id}_i).
$$

Multidimensional evaluation:

$$
e_i=(x_i,f(x_i),\operatorname{Id}_i),
\qquad
x_i\in\mathbb{R}^{d}.
$$

The data structure is conceptually unchanged.

Only the type of the spatial coordinate changes from scalar to vector.

### 21.7 Regional Evaluation Set

The general definition is:

$$
D_R
=
\{e_i\in D_t:x_i\in R\}.
$$

For one dimension:

$$
x_i\in[l,r].
$$

For multiple dimensions:

$$
l_j\leq x_{i,j}\leq r_j
\qquad
\forall j.
$$

Therefore, the same regional-membership principle is preserved.

### 21.8 Spatial Coverage

One-dimensional coverage can be described through ordered gaps between evaluations and region boundaries.

The multidimensional formulation extends this to:

- coordinate-wise coverage;
- geometric distance to evaluations;
- covering radius.

The one-dimensional covering radius becomes:

$$
h(R,D_R)
=
\sup_{x\in R}
\min_{x_i\in D_R}|x-x_i|.
$$

In \(d\) dimensions:

$$
h(R,D_R)
=
\sup_{x\in R}
\min_{x_i\in D_R}\|x-x_i\|_2.
$$

Thus, the one-dimensional spatial-distance concept is directly generalized through the Euclidean norm.

### 21.9 Behavioral Information

In one dimension, local behavior can be analyzed along the scalar coordinate.

In \(d\) dimensions, behavior may depend on multiple coordinate directions.

The generalized framework therefore allows directional information such as:

$$
B_R
=
(B_1,\ldots,B_d),
$$

where each component represents information about the objective along a particular coordinate direction.

The exact implementation may use additional cross-direction information when required.

For:

$$
d=1,
$$

the behavioral representation reduces to a single directional component.

### 21.10 Uncertainty

The Lipschitz-based multidimensional envelopes are:

$$
U_R(x)
=
\min_{x_i\in D_R}
\left[
f(x_i)+L\|x-x_i\|_2
\right],
$$

and:

$$
L_R(x)
=
\max_{x_i\in D_R}
\left[
f(x_i)-L\|x-x_i\|_2
\right].
$$

For:

$$
d=1,
$$

the Euclidean distance becomes absolute distance:

$$
\|x-x_i\|_2=|x-x_i|.
$$

Therefore, the original one-dimensional envelopes are recovered exactly.

### 21.11 Optimization Potential

The regional potential remains:

$$
P(R)
=
\max_{x\in R}U_R(x).
$$

The definition does not depend on dimension.

Only the geometry of \(R\) and the distance function change.

Therefore, the same conceptual potential mechanism applies to arbitrary finite dimensions.

### 21.12 Structural Splitting

In one dimension, a split is defined by:

$$
s\in(l,r).
$$

The interval becomes:

$$
[l,s]
\cup
[s,r].
$$

In \(d\) dimensions, a coordinate split is represented by:

$$
(j,s_j),
$$

where:

$$
j\in\{1,\ldots,d\},
$$

and:

$$
s_j\in(l_j,r_j).
$$

The resulting children are obtained by separating the parent along coordinate \(j\).

For:

$$
d=1,
$$

there is only one possible coordinate direction.

Therefore:

$$
j=1
$$

and the multidimensional split reduces to the original interval split.

### 21.13 Contraction

The one-dimensional contraction condition can be expressed using interval length.

The multidimensional formulation requires spatial contraction of the resulting child regions.

A coordinate split reduces the side length of the selected coordinate.

To guarantee convergence of the full region diameter, sufficient refinement must eventually occur in every unresolved coordinate direction.

Thus, the generalized condition introduces directional fairness in addition to local contraction.

For:

$$
d=1,
$$

there is only one direction, so the additional directional requirement becomes automatic.

### 21.14 Sampling

One-dimensional sampling selects:

$$
x\in R.
$$

Multidimensional sampling selects:

$$
x=(x_1,\ldots,x_d)\in R.
$$

The underlying operation remains:

$$
y=f(x).
$$

Therefore, sampling is dimension-independent at the conceptual level.

### 21.15 Global Region Selection

The global region-selection mechanism remains based on regional information.

The principal difference is that the regional information profile becomes multidimensional.

The global decision remains:

$$
\mathcal{R}_{\mathrm{eligible}}
\rightarrow
R^\ast.
$$

The selected region may then undergo:

$$
R^\ast
\rightarrow
\{\mathrm{Sample},\mathrm{Split},\mathrm{Stable}\}.
$$

If splitting is selected:

$$
R^\ast
\rightarrow
(j,s_j).
$$

For:

$$
d=1,
$$

the direction \(j\) is fixed.

### 21.16 Hierarchy

The hierarchy definition is unchanged:

$$
\mathcal{H}
=
(\mathcal{R},\mathcal{E}).
$$

Only the geometric representation of each region changes.

Parent retention and no-pruning remain valid in both cases.

### 21.17 Incumbent

The global incumbent is dimension-independent:

$$
f_{\mathrm{best}}^{(t)}
=
\max_{1\leq i\leq t}f(x_i).
$$

The corresponding location changes from a scalar to a vector:

$$
x_{\mathrm{best}}^{(t)}
\in\mathbb{R}^{d}.
$$

For:

$$
d=1,
$$

the original scalar incumbent is recovered.

### 21.18 Termination

The global termination mechanism remains based on:

- certified gap;
- evaluation budget;
- numerical limits;
- valid-action availability;
- configuration validity.

The geometric resolution criteria become multidimensional through:

$$
\operatorname{diam}(R)
$$

and the coordinate-wise widths:

$$
w_j(R).
$$

Thus, termination remains dimension-independent at the conceptual level.

### 21.19 Summary Mapping

The complete correspondence can be summarized as follows:

$$
\boxed{
\begin{array}{c|c}
\text{One Dimension} & \text{Finite Dimensions}\\
\hline
[a,b] &
\prod_{j=1}^{d}[a_j,b_j]
\\
x &
(x_1,\ldots,x_d)
\\
[l,r] &
\prod_{j=1}^{d}[l_j,r_j]
\\
r-l &
(w_1,\ldots,w_d)
\\
\frac{l+r}{2} &
\left(\frac{l_1+r_1}{2},\ldots,\frac{l_d+r_d}{2}\right)
\\
|x-x_i| &
\|x-x_i\|_2
\\
s &
(j,s_j)
\end{array}
}
$$

The algorithmic principles themselves remain unchanged.

### 21.20 Generalization Result

The analysis establishes that the principal ARRGO concepts have dimension-independent definitions.

The changes required by increasing dimension are primarily geometric and representational:

$$
\boxed{
\text{Scalar Geometry}
\rightarrow
\text{Vector Geometry}
}
$$

while the optimization logic remains:

$$
\boxed{
\text{Analyze}
\rightarrow
\text{Select}
\rightarrow
\text{Refine}
\rightarrow
\text{Update}
}
$$

The one-dimensional ARRGO framework is therefore embedded within the generalized formulation as the special case:

$$
d=1.
$$

The remaining question is whether the generalized formulation is sufficient to establish a formal convergence result under explicit assumptions.

That question is addressed in the multidimensional convergence theorem and proof sketch.

## 22. Multidimensional Global Convergence Theorem

The previous sections established that the fundamental ARRGO components can be formulated for an arbitrary finite-dimensional bounded domain.

The next step is to state the conditions under which the generalized ARRGO refinement process provides a global convergence result.

The purpose of this theorem is not to claim that every implementation of ARRGO automatically converges.

Instead, the theorem identifies the assumptions that connect:

- bounded search domains;
- continuity of the objective;
- valid region refinement;
- spatial contraction;
- information acquisition;
- global selection fairness;
- and, when certification is required, valid Lipschitz bounds.

### 22.1 Theorem Statement

Let:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j]
\subset\mathbb{R}^{d},
$$

where:

$$
d<\infty,
\qquad
a_j<b_j
\quad
\forall j.
$$

Let:

$$
f:\Omega\rightarrow\mathbb{R}
$$

be continuous.

Assume that ARRGO satisfies the following conditions.

#### A1. Compact Feasible Domain

The search domain is closed and bounded:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j].
$$

Therefore, \(\Omega\) is compact.

#### A2. Deterministic Objective Evaluation

For every feasible point \(x\in\Omega\), the evaluation returns the deterministic value:

$$
y=f(x).
$$

#### A3. Valid Region Representation

Every region generated by ARRGO is a hyper-rectangle:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j],
$$

satisfying:

$$
R\subseteq\Omega.
$$

#### A4. Valid Structural Refinement

Every structural refinement produces child regions contained in the parent and preserves the required spatial coverage of the refined region.

Thus, for a parent \(R_p\):

$$
R_{c_i}\subseteq R_p
$$

for every child \(R_{c_i}\), and the children collectively represent the refined portion of \(R_p\).

#### A5. Spatial Contraction

For every refinement path that continues to receive structural refinement, the spatial diameter of its regions converges to zero:

$$
\operatorname{diam}(R_t)
\longrightarrow
0
\qquad
\text{as }t\rightarrow\infty.
$$

Equivalently, sufficiently persistent refinement cannot leave a region with a non-vanishing spatial diameter.

#### A6. Information Acquisition

ARRGO acquires objective information in unresolved regions sufficiently often so that regions containing potentially optimal points are not permanently left without informative evaluations.

#### A7. Global Selection Fairness

No region that remains globally relevant can be permanently excluded from future consideration.

In particular, if a region continues to contain a point whose objective value can exceed the current incumbent by a non-vanishing amount, that region must remain eligible for future refinement or information acquisition.

#### A8. Exact Objective Evaluation

Whenever ARRGO evaluates a point \(x\), the stored objective value is exactly:

$$
f(x).
$$

Numerical implementation errors are treated separately from the mathematical convergence statement.

---

### 22.2 Existence of a Global Optimum

Because \(\Omega\) is compact and \(f\) is continuous, the extreme value theorem gives:

$$
\exists x^\ast\in\Omega
$$

such that:

$$
f(x^\ast)
=
f^\ast
=
\max_{x\in\Omega}f(x).
$$

Therefore, the global optimum exists.

---

### 22.3 The Global Incumbent

After \(t\) evaluations, define:

$$
f_{\mathrm{best}}^{(t)}
=
\max_{1\leq i\leq t}f(x_i).
$$

Because each new evaluation is added to the history:

$$
D_t
\subseteq
D_{t+1},
$$

we have:

$$
f_{\mathrm{best}}^{(t)}
\leq
f_{\mathrm{best}}^{(t+1)}.
$$

Therefore:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
\text{ is monotone non-decreasing.}
}
$$

Since every evaluated point belongs to \(\Omega\):

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast.
$$

Thus, the incumbent objective value is bounded above by the global optimum.

---

### 22.4 Refinement Near a Global Maximizer

Consider a global maximizer:

$$
x^\ast\in X^\ast.
$$

Because ARRGO maintains a hierarchy of valid regions covering the relevant search space, there exists a sequence of regions:

$$
R_0
\supseteq
R_1
\supseteq
R_2
\supseteq
\cdots
$$

such that:

$$
x^\ast\in R_t
$$

for the regions on the corresponding refinement path.

Under the spatial-contraction assumption:

$$
\operatorname{diam}(R_t)
\rightarrow
0.
$$

Therefore, the region containing \(x^\ast\) becomes arbitrarily small.

---

### 22.5 Consequence of Continuity

Because \(f\) is continuous at \(x^\ast\), for every:

$$
\varepsilon>0,
$$

there exists:

$$
\delta>0
$$

such that:

$$
\|x-x^\ast\|_2<\delta
\quad\Longrightarrow\quad
|f(x)-f(x^\ast)|<\varepsilon.
$$

Since:

$$
\operatorname{diam}(R_t)\rightarrow0,
$$

there exists a sufficiently large \(t\) for which:

$$
\operatorname{diam}(R_t)<\delta.
$$

Consequently, points sufficiently close to \(x^\ast\) can be identified within the refined region.

If ARRGO acquires informative evaluations along this refinement process, there exists an evaluated point \(x_i\in R_t\) satisfying:

$$
\|x_i-x^\ast\|_2<\delta.
$$

Therefore:

$$
f(x_i)
>
f(x^\ast)-\varepsilon.
$$

Since:

$$
f(x^\ast)=f^\ast,
$$

we obtain:

$$
f(x_i)>f^\ast-\varepsilon.
$$

The incumbent is at least as good as this evaluation:

$$
f_{\mathrm{best}}^{(t)}
\geq
f(x_i).
$$

Hence:

$$
f_{\mathrm{best}}^{(t)}
>
f^\ast-\varepsilon.
$$

---

### 22.6 Global Objective-Value Convergence

From the previous result:

$$
f^\ast-\varepsilon
<
f_{\mathrm{best}}^{(t)}
\leq
f^\ast
$$

for sufficiently large \(t\).

Therefore:

$$
0
\leq
f^\ast-f_{\mathrm{best}}^{(t)}
<
\varepsilon.
$$

Since this holds for every:

$$
\varepsilon>0,
$$

we obtain:

$$
\boxed{
\lim_{t\rightarrow\infty}
f_{\mathrm{best}}^{(t)}
=
f^\ast.
}
$$

Thus, under assumptions A1–A8 and the stated refinement and information-acquisition conditions, ARRGO achieves global convergence in objective value.

---

### 22.7 Location Convergence Is Different

The theorem establishes convergence of the best observed objective value.

It does not automatically imply that:

$$
x_{\mathrm{best}}^{(t)}
\rightarrow
x^\ast
$$

for a particular global maximizer.

If the set of global maximizers contains multiple points:

$$
|X^\ast|>1,
$$

the sequence of incumbent locations may move between different optimal or near-optimal regions.

Therefore, objective-value convergence is the primary global convergence statement.

Stronger location-convergence results require additional assumptions, such as uniqueness or an isolated global maximizer.

---

### 22.8 Multidimensional Nature of the Result

The proof does not depend on a special property of one-dimensional intervals.

The essential geometric condition is:

$$
\operatorname{diam}(R_t)\rightarrow0.
$$

The argument therefore remains valid for:

$$
R_t\subset\mathbb{R}^{d}
$$

for every finite:

$$
d\geq1.
$$

The one-dimensional case is recovered when:

$$
d=1.
$$

---

### 22.9 Certified Version

If, in addition, the objective satisfies a valid global Lipschitz condition:

$$
|f(x)-f(y)|
\leq
L\|x-y\|_2,
\qquad
\forall x,y\in\Omega,
$$

then ARRGO can construct valid regional upper envelopes:

$$
U_R(x)
=
\min_{x_i\in D_R}
\left[
f(x_i)+L\|x-x_i\|_2
\right].
$$

The regional optimization potential is:

$$
P(R)
=
\max_{x\in R}U_R(x).
$$

If the global potential is defined as a valid upper bound over the relevant region representation:

$$
P_{\mathrm{global}}
=
\max_R P(R),
$$

then:

$$
f_{\mathrm{best}}
\leq
f^\ast
\leq
P_{\mathrm{global}}.
$$

Consequently, the global certification gap:

$$
\Delta_{\mathrm{global}}
=
P_{\mathrm{global}}
-
f_{\mathrm{best}}
$$

satisfies:

$$
0
\leq
f^\ast-f_{\mathrm{best}}
\leq
\Delta_{\mathrm{global}}.
$$

Therefore, if:

$$
\Delta_{\mathrm{global}}
\leq
\varepsilon,
$$

then:

$$
\boxed{
f_{\mathrm{best}}
\geq
f^\ast-\varepsilon.
}
$$

The incumbent is consequently epsilon-optimal under the validity of the Lipschitz assumption and the correctness of the computed bounds.

---

### 22.10 Interpretation of the Theorem

The convergence result can be summarized as the following chain:

$$
\boxed{
\begin{aligned}
&\text{Compact Domain}
+
\text{Continuity}
\\
&\quad\Downarrow
\\
&\text{Global Maximizer Exists}
\\
&\quad\Downarrow
\\
&\text{Valid Refinement}
+
\text{Spatial Contraction}
\\
&\quad\Downarrow
\\
&\text{Regions Become Arbitrarily Small}
\\
&\quad\Downarrow
\\
&\text{Information Acquisition Near Relevant Regions}
\\
&\quad\Downarrow
\\
&\text{Evaluations Approach Global-Optimal Objective Value}
\\
&\quad\Downarrow
\\
&\boxed{
f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast
}
\end{aligned}
}
$$

The theorem therefore identifies the essential mechanism behind multidimensional ARRGO convergence:

> ARRGO converges globally in objective value when valid region refinement contracts the search around globally relevant points and the information-acquisition process does not permanently neglect those regions.

The result is conditional on the stated assumptions and should not be interpreted as an unconditional guarantee for arbitrary implementations or configurations.

## 23. One-Dimensional Recovery Corollary

The multidimensional convergence theorem was formulated for arbitrary finite dimensions:

$$
d\geq1.
$$

The original ARRGO framework was developed for the one-dimensional case.

This section formally establishes that the one-dimensional formulation is recovered by setting:

$$
d=1.
$$

### 23.1 Domain Recovery

The generalized domain is:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j].
$$

For:

$$
d=1,
$$

this becomes:

$$
\Omega=[a_1,b_1].
$$

Renaming:

$$
a_1=a,
\qquad
b_1=b,
$$

gives:

$$
\Omega=[a,b].
$$

Therefore, the original one-dimensional search domain is recovered exactly.

### 23.2 Region Recovery

The generalized region is:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j].
$$

For:

$$
d=1,
$$

we obtain:

$$
R=[l_1,r_1].
$$

Renaming:

$$
l_1=l,
\qquad
r_1=r,
$$

gives:

$$
R=[l,r].
$$

Thus, the original interval-based region representation is recovered.

### 23.3 Point Representation Recovery

The generalized decision variable is:

$$
x=(x_1,\ldots,x_d).
$$

For:

$$
d=1,
$$

the point becomes:

$$
x=(x_1).
$$

Identifying the single coordinate with the scalar variable:

$$
x=x_1,
$$

recovers the original one-dimensional representation.

### 23.4 Region Center Recovery

The generalized region center is:

$$
c(R)
=
\left(
\frac{l_1+r_1}{2},
\ldots,
\frac{l_d+r_d}{2}
\right).
$$

For:

$$
d=1,
$$

this becomes:

$$
c(R)
=
\frac{l_1+r_1}{2}.
$$

Therefore:

$$
c(R)=\frac{l+r}{2},
$$

which is the original interval midpoint.

### 23.5 Spatial Size Recovery

The generalized maximum side length is:

$$
w_{\max}(R)
=
\max_{1\leq j\leq d}(r_j-l_j).
$$

For:

$$
d=1,
$$

there is only one side length:

$$
w_{\max}(R)=r_1-l_1.
$$

Therefore:

$$
w_{\max}(R)=r-l.
$$

Similarly, the generalized Euclidean diameter:

$$
\operatorname{diam}(R)
=
\sqrt{
\sum_{j=1}^{d}(r_j-l_j)^2
}
$$

becomes:

$$
\operatorname{diam}(R)
=
\sqrt{(r-l)^2}
=
r-l,
$$

because:

$$
r>l.
$$

Thus, the multidimensional diameter exactly recovers the one-dimensional interval length.

### 23.6 Distance Recovery

The generalized spatial distance is:

$$
\|x-y\|_2.
$$

For:

$$
d=1,
$$

this becomes:

$$
\|x-y\|_2
=
|x-y|.
$$

Therefore, all distance-based regional quantities reduce to their original one-dimensional forms.

### 23.7 Lipschitz Bounds Recovery

The multidimensional Lipschitz condition is:

$$
|f(x)-f(y)|
\leq
L\|x-y\|_2.
$$

For:

$$
d=1,
$$

this becomes:

$$
|f(x)-f(y)|
\leq
L|x-y|.
$$

The generalized upper envelope:

$$
U_R(x)
=
\min_{x_i\in D_R}
\left[
f(x_i)+L\|x-x_i\|_2
\right]
$$

therefore becomes:

$$
U_R(x)
=
\min_{x_i\in D_R}
\left[
f(x_i)+L|x-x_i|
\right].
$$

Similarly, the lower envelope becomes:

$$
L_R(x)
=
\max_{x_i\in D_R}
\left[
f(x_i)-L|x-x_i|
\right].
$$

These are exactly the one-dimensional Lipschitz envelopes used by the original certified formulation.

### 23.8 Structural Split Recovery

In the generalized formulation, a coordinate split is represented by:

$$
(j,s_j),
$$

where:

$$
j\in\{1,\ldots,d\}.
$$

For:

$$
d=1,
$$

the only possible coordinate is:

$$
j=1.
$$

Therefore, the split is represented only by:

$$
s_1\in(l_1,r_1).
$$

Renaming:

$$
s_1=s,
$$

gives the original interval split:

$$
[l,r]
\rightarrow
[l,s]\cup[s,r].
$$

Thus, the generalized structural refinement mechanism contains the original one-dimensional split as a special case.

### 23.9 Contraction Recovery

The generalized refinement condition requires the spatial size of persistent refinement paths to converge to zero:

$$
\operatorname{diam}(R_t)\rightarrow0.
$$

For:

$$
d=1,
$$

the diameter is simply the interval width:

$$
\operatorname{diam}(R_t)
=
r_t-l_t.
$$

Therefore:

$$
r_t-l_t
\rightarrow
0.
$$

This is precisely the original one-dimensional spatial contraction condition.

### 23.10 Incumbent Recovery

The generalized incumbent is:

$$
x_{\mathrm{best}}^{(t)}
\in
\operatorname*{arg\,max}_{x_i\in D_t}f(x_i).
$$

For:

$$
d=1,
$$

the incumbent location becomes a scalar.

The best observed objective value remains:

$$
f_{\mathrm{best}}^{(t)}
=
\max_{1\leq i\leq t}f(x_i).
$$

Therefore, the global optimization state is unchanged at the conceptual level.

### 23.11 Convergence Result

The multidimensional theorem established:

$$
\lim_{t\rightarrow\infty}
f_{\mathrm{best}}^{(t)}
=
f^\ast
$$

under the stated assumptions.

Setting:

$$
d=1
$$

therefore gives:

$$
\boxed{
\lim_{t\rightarrow\infty}
f_{\mathrm{best}}^{(t)}
=
f^\ast
\qquad
\text{for the original one-dimensional ARRGO formulation.}
}
$$

Hence, the previous one-dimensional convergence result is not a separate theory.

It is the \(d=1\) instance of the generalized finite-dimensional result.

### 23.12 Corollary

**Corollary.**

If the assumptions of the multidimensional convergence theorem hold for:

$$
d=1,
$$

then the original one-dimensional ARRGO formulation is globally convergent in objective value.

More generally, the one-dimensional ARRGO framework is a special case of the generalized finite-dimensional formulation:

$$
\boxed{
\mathrm{ARRGO}_1
\subset
\mathrm{ARRGO}_d,
\qquad
d\geq1.
}
$$

The notation above expresses structural inclusion: the one-dimensional algorithm is obtained from the generalized formulation by fixing the dimension to one.

### 23.13 Significance

This recovery result is important because the dimensional generalization does not introduce a disconnected algorithm.

Instead, it establishes a single framework parameterized by dimension:

$$
\mathrm{ARRGO}_d.
$$

The original implementation and experiments can therefore be interpreted as validation of the \(d=1\) case, while future multidimensional implementations and experiments evaluate the same theoretical framework for:

$$
d>1.
$$

The generalization is consequently a unification of the existing one-dimensional theory rather than a replacement for it.

## 24. Multidimensional Algorithmic Requirements

The convergence theorem establishes the mathematical conditions under which the ARRGO refinement process can converge in arbitrary finite dimensions.

The next requirement is to translate the generalized mathematical formulation into concrete algorithmic requirements.

The purpose of this section is to distinguish between:

1. components that remain conceptually unchanged;
2. components that require geometric generalization;
3. components that require new multidimensional decision logic.

This distinction is important because increasing the dimension should not change the fundamental ARRGO optimization philosophy.

### 24.1 Dimension Parameter

The generalized algorithm must explicitly represent the search-space dimension:

$$
d\in\mathbb{N},
\qquad
d\geq1.
$$

The dimension should be derived from the domain representation rather than manually duplicated throughout the algorithm.

For a domain:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j],
$$

the algorithm must know:

$$
d=\dim(\Omega).
$$

The one-dimensional implementation corresponds to:

$$
d=1.
$$

### 24.2 Domain Representation

The scalar lower and upper bounds of the one-dimensional implementation:

$$
a,b
$$

must be generalized to vectors:

$$
a=(a_1,\ldots,a_d),
$$

and:

$$
b=(b_1,\ldots,b_d).
$$

The feasibility condition becomes:

$$
a_j\leq x_j\leq b_j
\qquad
\forall j.
$$

Every algorithmic operation that creates a point must satisfy these coordinate-wise constraints.

### 24.3 Point Representation

A point must be represented as a vector:

$$
x\in\mathbb{R}^{d}.
$$

The algorithm must therefore treat point coordinates as a single spatial object rather than as independent unrelated scalar values.

For:

$$
d=1,
$$

the vector representation reduces to the existing scalar representation.

### 24.4 Region Representation

Every region must store two vectors:

$$
l=(l_1,\ldots,l_d),
$$

and:

$$
r=(r_1,\ldots,r_d).
$$

The region is:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j].
$$

The implementation must be able to compute:

$$
w_j=r_j-l_j,
$$

for every coordinate direction.

It must also support:

$$
w_{\max}(R)
=
\max_j w_j
$$

and:

$$
\operatorname{diam}(R)
=
\sqrt{\sum_j w_j^2}.
$$

### 24.5 Region Membership

A point belongs to a region when:

$$
x\in R
$$

if and only if:

$$
l_j\leq x_j\leq r_j
\qquad
\forall j.
$$

Therefore, regional evaluation assignment must become a vectorized coordinate-wise membership operation.

This operation is fundamental because regional analysis, coverage, uncertainty, and refinement all depend on identifying:

$$
D_R
=
\{e_i\in D_t:x_i\in R\}.
$$

### 24.6 Sampling Candidate Generation

Sampling candidates must be generated as feasible vectors:

$$
x^{(k)}\in R.
$$

The candidate-generation mechanism should be able to represent information from multiple coordinate directions.

For example, candidates may be generated around:

- the region center;
- existing informative evaluations;
- unresolved coordinate directions;
- spatially sparse locations.

The exact candidate set remains an algorithmic design choice, but every candidate must satisfy:

$$
x^{(k)}\in R.
$$

### 24.7 Sampling Selection

The selection mechanism must evaluate the information value of candidate points using the multidimensional regional information state.

A candidate should not be selected solely because it is geometrically central.

Instead, the decision should consider whether the candidate can improve unresolved information such as:

$$
Q_R
=
(Q_{\mathrm{coverage}},
Q_{\mathrm{behavior}},
Q_{\mathrm{uncertainty}},
Q_{\mathrm{potential}}).
$$

The resulting selection remains deterministic.

### 24.8 Behavioral Analysis

Behavioral analysis must become directional.

For each coordinate direction:

$$
j\in\{1,\ldots,d\},
$$

the algorithm may analyze how the objective changes along that direction.

A directional behavioral profile can be represented as:

$$
B_R
=
(B_1,\ldots,B_d).
$$

This allows ARRGO to distinguish between regions that are:

- locally flat;
- strongly varying;
- directionally sensitive;
- or anisotropic.

The multidimensional formulation should not assume that all directions have equal importance.

### 24.9 Structural Split Selection

A multidimensional split requires two decisions:

1. which coordinate direction should be refined;
2. where the split should occur along that direction.

Therefore, a split candidate is represented as:

$$
c_{\mathrm{split}}
=
(j,s_j).
$$

The coordinate must satisfy:

$$
j\in\{1,\ldots,d\},
$$

and the split location must satisfy:

$$
l_j<s_j<r_j.
$$

The resulting child regions must remain valid hyper-rectangles.

### 24.10 Direction Selection

Direction selection is a new requirement introduced by multidimensional geometry.

ARRGO must determine which coordinate direction provides the most informative structural refinement.

Possible information sources include:

- directional objective variation;
- directional slope variation;
- directional coverage;
- directional uncertainty;
- directional optimization potential;
- spatial resolution.

No fixed direction should be preferred merely because of its coordinate index.

A deterministic tie-breaking rule may be required when multiple directions are equally informative.

### 24.11 Split-Location Selection

After selecting a coordinate direction, ARRGO must select a valid split location:

$$
s_j\in(l_j,r_j).
$$

The selected split must satisfy the configured contraction condition.

For example, if the refinement rule requires:

$$
\max
\left(
\frac{s_j-l_j}{r_j-l_j},
\frac{r_j-s_j}{r_j-l_j}
\right)
\leq\rho,
$$

then the split is geometrically valid only when:

$$
\rho\geq\frac12.
$$

The split-selection mechanism must therefore respect both information quality and geometric validity.

### 24.12 Multidimensional Refinement Action

The refinement decision remains:

$$
A_R
\in
\{
\mathrm{Sample},
\mathrm{Split},
\mathrm{Stable}
\}.
$$

However, a split action now contains additional structural information:

$$
A_R
=
(\mathrm{Split},j,s_j).
$$

A sampling action contains:

$$
A_R
=
(\mathrm{Sample},x).
$$

Thus, the action representation becomes explicitly multidimensional.

### 24.13 Global Region Selection

The global selection mechanism remains responsible for choosing which region should receive attention.

The selected region must be determined from its unresolved information profile and global relevance.

Conceptually:

$$
R^\ast
=
\operatorname{SelectRegion}
(\mathcal{R}_{\mathrm{eligible}}).
$$

The selection policy must remain deterministic.

The introduction of multiple dimensions must not cause the algorithm to independently optimize each coordinate.

ARRGO remains a region-based global optimization algorithm.

### 24.14 Global Incumbent

The global incumbent logic remains unchanged:

$$
f_{\mathrm{best}}^{(t)}
=
\max_{1\leq i\leq t}f(x_i).
$$

Only the spatial representation of the incumbent changes:

$$
x_{\mathrm{best}}^{(t)}
\in
\mathbb{R}^{d}.
$$

Therefore, no separate optimization state is required for each dimension.

### 24.15 Information Sharing

Because regions are multidimensional, an evaluation may be geometrically relevant to several structural contexts.

The implementation must therefore distinguish between:

- ownership of an evaluation in the global history;
- spatial relevance of that evaluation to a region.

The global evaluation should exist exactly once in the evaluation history.

Regional views should be derived from spatial membership.

This preserves provenance while allowing regional analysis to use inherited spatial information.

### 24.16 Hierarchy Update

When a region is split along coordinate \(j\), the parent remains in the hierarchy.

The children are created by modifying only the \(j\)-th coordinate bounds.

For example:

$$
R
=
\prod_{k=1}^{d}[l_k,r_k].
$$

A split at \(s_j\) produces:

$$
R_1
=
\left(
\prod_{k\neq j}[l_k,r_k]
\right)
\times
[l_j,s_j],
$$

and:

$$
R_2
=
\left(
\prod_{k\neq j}[l_k,r_k]
\right)
\times
[s_j,r_j].
$$

The parent-child relationship must remain explicit.

### 24.17 Lifecycle Management

The lifecycle states remain:

$$
\operatorname{State}(R)
\in
\{
\mathrm{ACTIVE},
\mathrm{STABLE},
\mathrm{REFINED}
\}.
$$

These states do not depend on dimension.

The meaning of the states remains:

- ACTIVE: eligible for further consideration;
- STABLE: currently sufficiently resolved;
- REFINED: structurally split.

Parent retention remains mandatory.

### 24.18 Termination

Termination must account for multidimensional spatial resolution.

The relevant quantities include:

$$
w_{\max}(R),
$$

and:

$$
\operatorname{diam}(R).
$$

For directional refinement, the algorithm must also detect whether unresolved coordinate directions remain.

A region should not be considered geometrically resolved merely because one coordinate has become very small while another remains large.

### 24.19 Certified Bounds

The certified formulation must replace scalar distance with multidimensional distance:

$$
|x-x_i|
\rightarrow
\|x-x_i\|_2.
$$

The upper envelope becomes:

$$
U_R(x)
=
\min_{x_i\in D_R}
\left[
f(x_i)+L\|x-x_i\|_2
\right].
$$

The lower envelope becomes:

$$
L_R(x)
=
\max_{x_i\in D_R}
\left[
f(x_i)-L\|x-x_i\|_2
\right].
$$

The implementation must preserve the validity of these bounds when operating in higher dimensions.

### 24.20 Numerical Robustness

Multidimensional numerical robustness requires tolerance handling for:

- vector equality;
- point duplication;
- boundary membership;
- region containment;
- split locations;
- zero-width or degenerate regions;
- distance calculations.

A vector tolerance may be implemented using a norm-based criterion such as:

$$
\|x-y\|_2\leq\tau_x.
$$

The exact numerical policy remains an implementation decision, but it must be consistent across all components.

### 24.21 What Does Not Change

The following conceptual principles remain unchanged when moving from one dimension to \(d\) dimensions:

1. objective evaluations are the source of acquired information;
2. regions are the structural units of refinement;
3. sampling acquires information;
4. splitting changes spatial resolution;
5. parent regions are retained;
6. pruning is not performed;
7. the global incumbent is based on observed objective values;
8. refinement decisions are deterministic;
9. global selection remains region-based;
10. empirical results remain best-found results;
11. certified guarantees remain conditional on valid assumptions.

Therefore, the multidimensional generalization changes the geometry and representation without changing the fundamental ARRGO philosophy.

### 24.22 Implementation Principle

The implementation should therefore be organized around dimension-independent abstractions.

Conceptually:

$$
\boxed{
\text{ARRGO Logic}
+
\text{Dimension-Independent Geometry}
}
$$

rather than:

$$
\boxed{
\text{Separate 1D Algorithm}
+
\text{Separate 2D Algorithm}
+
\text{Separate 3D Algorithm}
}
$$

A single generalized implementation parameterized by \(d\) is therefore the target architecture.

The one-dimensional implementation should then be treated as:

$$
\mathrm{ARRGO}_1
=
\mathrm{ARRGO}(d=1).
$$

The multidimensional implementation is consequently a generalization of the same algorithmic framework rather than a collection of independent special cases.

## 25. Multidimensional Structural Splitting

Structural splitting is the mechanism through which ARRGO increases the spatial resolution of the search domain.

In one dimension, a region is divided at a scalar split point.

In multiple dimensions, a region must first determine a coordinate direction and then determine a split location along that direction.

Therefore, multidimensional structural refinement consists of two coupled decisions:

$$
\boxed{
\text{Direction Selection}
\rightarrow
\text{Split-Location Selection}
}
$$

### 25.1 Parent Region

Consider a valid region:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j].
$$

For every coordinate:

$$
l_j<r_j.
$$

The side length in direction \(j\) is:

$$
w_j=r_j-l_j.
$$

### 25.2 Candidate Split Direction

A valid split direction is an index:

$$
j\in\{1,\ldots,d\}.
$$

The direction-selection mechanism determines which coordinate should be structurally refined.

The selection may depend on the regional information profile:

$$
Q_R
=
(Q_{\mathrm{coverage}},
Q_{\mathrm{behavior}},
Q_{\mathrm{uncertainty}},
Q_{\mathrm{potential}}).
$$

In particular, directional information may identify a coordinate along which the region remains insufficiently resolved.

The direction itself is therefore part of the refinement decision.

### 25.3 Candidate Split Location

After selecting direction \(j\), a split location:

$$
s_j
$$

must satisfy:

$$
l_j<s_j<r_j.
$$

The split location divides the side interval:

$$
[l_j,r_j]
$$

into:

$$
[l_j,s_j]
$$

and:

$$
[s_j,r_j].
$$

All other coordinate intervals remain unchanged.

### 25.4 Child Region Construction

The first child is:

$$
R_1
=
\prod_{k=1}^{d}I_k^{(1)},
$$

where:

$$
I_k^{(1)}
=
\begin{cases}
[l_j,s_j], & k=j,\\
[l_k,r_k], & k\neq j.
\end{cases}
$$

The second child is:

$$
R_2
=
\prod_{k=1}^{d}I_k^{(2)},
$$

where:

$$
I_k^{(2)}
=
\begin{cases}
[s_j,r_j], & k=j,\\
[l_k,r_k], & k\neq j.
\end{cases}
$$

Thus:

$$
R_1\cup R_2=R.
$$

Both children satisfy:

$$
R_1\subseteq R,
\qquad
R_2\subseteq R.
$$

Therefore, the structural refinement preserves feasibility.

### 25.5 Boundary Ownership

The split point:

$$
x_j=s_j
$$

belongs to both closed child regions under the mathematical representation above.

This creates a potential double-membership issue when assigning evaluations to regional views.

Therefore, the implementation must define an explicit boundary-ownership convention.

For example, one may use a half-open representation for internal ownership:

$$
[l_j,s_j)
\quad\text{and}\quad
[s_j,r_j],
$$

while retaining closed hyper-rectangles for geometric reasoning.

The precise convention is an implementation decision, but it must be applied consistently.

### 25.6 Spatial Contraction

Let the original side length in direction \(j\) be:

$$
w_j=r_j-l_j.
$$

The two child side lengths in the split direction are:

$$
w_j^{(1)}=s_j-l_j,
$$

and:

$$
w_j^{(2)}=r_j-s_j.
$$

A contraction factor:

$$
0<\rho<1
$$

requires:

$$
\max
\left(
\frac{s_j-l_j}{r_j-l_j},
\frac{r_j-s_j}{r_j-l_j}
\right)
\leq\rho.
$$

Equivalently:

$$
s_j-l_j
\leq
\rho(r_j-l_j),
$$

and:

$$
r_j-s_j
\leq
\rho(r_j-l_j).
$$

Therefore, the split is valid only if:

$$
s_j
\in
\left[
l_j+\left(1-\rho\right)(r_j-l_j),
\;
l_j+\rho(r_j-l_j)
\right].
$$

For a binary split, this interval is non-empty only when:

$$
\rho\geq\frac12.
$$

At:

$$
\rho=\frac12,
$$

the only admissible split is the midpoint.

For:

$$
\rho>\frac12,
$$

a range of valid split locations exists.

### 25.7 Effect on Region Diameter

Before splitting, the diameter is:

$$
\operatorname{diam}(R)
=
\sqrt{
\sum_{k=1}^{d}w_k^2
}.
$$

After splitting direction \(j\), the child diameters are:

$$
\operatorname{diam}(R_1)
=
\sqrt{
(s_j-l_j)^2
+
\sum_{k\neq j}w_k^2
},
$$

and:

$$
\operatorname{diam}(R_2)
=
\sqrt{
(r_j-s_j)^2
+
\sum_{k\neq j}w_k^2
}.
$$

Because:

$$
s_j-l_j<r_j-l_j,
$$

and:

$$
r_j-s_j<r_j-l_j,
$$

both child diameters are strictly smaller than the parent diameter.

Therefore, every valid non-degenerate coordinate split provides local geometric contraction.

### 25.8 Why Directional Refinement Is Necessary

Reducing the size of only one coordinate direction does not by itself guarantee:

$$
\operatorname{diam}(R_t)\rightarrow0.
$$

For example, suppose:

$$
w_1\rightarrow0
$$

while:

$$
w_2,\ldots,w_d
$$

remain bounded away from zero.

Then:

$$
\operatorname{diam}(R_t)
\not\rightarrow0.
$$

Therefore, global convergence requires that persistent refinement eventually resolves every relevant coordinate direction.

This is the multidimensional version of the refinement-fairness requirement.

### 25.9 Directional Fairness

For every persistent region path relevant to a global maximizer, the refinement process must not permanently neglect an unresolved coordinate direction.

Conceptually, if direction \(j\) remains spatially unresolved, then eventually the algorithm must allow a refinement action in that direction.

This condition can be expressed informally as:

$$
w_j(R_t)\not\rightarrow0
\quad\Longrightarrow\quad
\text{direction }j
\text{ remains eligible for future refinement}.
$$

Under sufficient directional fairness:

$$
w_j(R_t)\rightarrow0
\qquad
\forall j.
$$

Consequently:

$$
\operatorname{diam}(R_t)
=
\sqrt{\sum_{j=1}^{d}w_j(R_t)^2}
\rightarrow0.
$$

### 25.10 Split Candidate Representation

A multidimensional split candidate should therefore contain at least:

$$
c_{\mathrm{split}}
=
(j,s_j).
$$

For more complete provenance, it may additionally contain:

- parent region identifier;
- direction;
- split coordinate;
- contraction information;
- structural information profile;
- candidate-generation source.

The mathematical identity of the split, however, is determined by:

$$
(j,s_j).
$$

### 25.11 Structural Value

A split candidate should be evaluated according to the information it is expected to reveal through structural refinement.

The structural information profile may include:

$$
S(j,s_j)
=
\left(
S_{\mathrm{coverage}},
S_{\mathrm{behavior}},
S_{\mathrm{uncertainty}},
S_{\mathrm{potential}}
\right).
$$

This profile should be interpreted as a collection of information signals rather than as a probability distribution.

The split-selection mechanism can then compare feasible candidates according to their unresolved information.

### 25.12 Deterministic Selection

If multiple split candidates are equally preferred under the selected information criterion, a deterministic tie-breaking rule must be applied.

The tie-breaking rule should depend on intrinsic candidate properties, such as:

- information score;
- unresolved directional resolution;
- split balance;
- region identifier.

It should not introduce randomness.

A deterministic tie-breaking policy guarantees that identical inputs produce identical refinement decisions.

### 25.13 No-Pruning Preservation

Structural splitting does not remove the parent region.

After:

$$
R
\rightarrow
\{R_1,R_2\},
$$

the hierarchy becomes:

$$
\mathcal{H}_{t+1}
=
\mathcal{H}_t
\cup
\{R_1,R_2\}
$$

together with the corresponding parent-child relationships.

The parent remains represented as a historical region.

Thus, structural refinement and pruning remain separate concepts.

### 25.14 Reduction to One Dimension

For:

$$
d=1,
$$

there is only one possible direction:

$$
j=1.
$$

Therefore, the multidimensional split candidate:

$$
(j,s_j)
$$

reduces to:

$$
(1,s).
$$

Ignoring the fixed direction index, this is exactly the original scalar split location:

$$
s.
$$

The two hyper-rectangular children consequently become:

$$
[l,s]
\quad\text{and}\quad
[s,r].
$$

Thus, the original one-dimensional structural refinement mechanism is recovered exactly.

### 25.15 Structural Generalization Result

The multidimensional splitting mechanism can therefore be summarized as:

$$
\boxed{
R
\overset{(j,s_j)}{\longrightarrow}
\{R_1,R_2\}
}
$$

where:

1. \(j\) identifies the coordinate direction;
2. \(s_j\) identifies the split location;
3. both children remain inside the parent;
4. the children collectively cover the parent;
5. the selected coordinate contracts;
6. the parent remains in the hierarchy;
7. repeated fair refinement can reduce the region diameter toward zero.

The essential structural principle of ARRGO is therefore preserved:

> Structural refinement increases spatial resolution without removing previously represented regions from the global search hierarchy.

## 26. Multidimensional Information-Driven Refinement

The central principle of ARRGO is that refinement should be driven by information rather than by a predetermined geometric schedule.

In one dimension, the refinement decision is based on information available inside an interval.

In \(d\) dimensions, the same principle must account for multiple spatial directions.

The generalized refinement process therefore consists of three levels:

$$
\boxed{
\text{Regional Analysis}
\rightarrow
\text{Action Generation}
\rightarrow
\text{Action Selection}
}
$$

### 26.1 Regional Information State

For a region:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j],
$$

the regional information state is represented by:

$$
Q_R
=
\left(
Q_{\mathrm{coverage}},
Q_{\mathrm{behavior}},
Q_{\mathrm{uncertainty}},
Q_{\mathrm{potential}}
\right).
$$

Each component describes a different unresolved aspect of the region.

The profile is not interpreted as a probability distribution.

Instead, it represents a collection of deterministic information signals.

### 26.2 Directional Information

Multidimensional regions contain multiple coordinate directions.

For each:

$$
j\in\{1,\ldots,d\},
$$

ARRGO may maintain a directional information profile:

$$
Q_R^{(j)}.
$$

This profile can describe unresolved information specifically associated with coordinate \(j\).

Examples include:

- insufficient spatial coverage along direction \(j\);
- strong objective variation along direction \(j\);
- unresolved uncertainty along direction \(j\);
- large regional potential associated with structures aligned with direction \(j\).

The exact signals may evolve with the implementation, but their purpose remains the same: identify where additional refinement can reveal useful information.

### 26.3 Sampling Versus Splitting

The generalized algorithm retains the distinction between the two refinement mechanisms.

Sampling:

$$
\boxed{
\text{Sampling}
\rightarrow
\text{Acquire Objective Information}
}
$$

Splitting:

$$
\boxed{
\text{Splitting}
\rightarrow
\text{Increase Spatial Resolution}
}
$$

Sampling should therefore be preferred when the primary unresolved issue is insufficient information inside an existing region.

Splitting should be preferred when the existing region is too coarse to represent important spatial structure adequately.

### 26.4 Sampling Candidate Information

A sampling candidate:

$$
x^{(k)}\in R
$$

can be evaluated according to the information it may provide.

A candidate may be useful when it:

- fills a spatially unresolved area;
- distinguishes between competing local behaviors;
- improves the representation of a promising region;
- reduces local uncertainty;
- provides information near a potentially optimal location.

The candidate profile can therefore be represented as:

$$
I_{\mathrm{sample}}(x^{(k)};R).
$$

This is an information profile, not a probability of improvement.

### 26.5 Sampling Candidate Selection

Let:

$$
C_{\mathrm{sample}}(R)
=
\{x^{(1)},\ldots,x^{(m)}\}.
$$

The algorithm selects a feasible candidate using a deterministic comparison rule.

Conceptually:

$$
x^\ast_{\mathrm{sample}}
=
\operatorname{SelectSample}
\left(
C_{\mathrm{sample}}(R),
I_{\mathrm{sample}}
\right).
$$

The selection mechanism must satisfy:

$$
x^\ast_{\mathrm{sample}}\in R.
$$

If several candidates are equally preferred, deterministic tie-breaking is applied.

### 26.6 Split Candidate Information

A multidimensional split candidate is:

$$
c_{\mathrm{split}}
=
(j,s_j).
$$

Its information profile can be represented as:

$$
I_{\mathrm{split}}(j,s_j;R).
$$

The profile may contain:

$$
I_{\mathrm{split}}
=
\left(
I_{\mathrm{coverage}},
I_{\mathrm{behavior}},
I_{\mathrm{uncertainty}},
I_{\mathrm{potential}}
\right).
$$

Again, these are deterministic information signals.

### 26.7 Pareto-Based Candidate Comparison

Because the information profile may contain several objectives, reducing it to an arbitrary weighted scalar can distort the decision.

Therefore, the generalized formulation can use non-dominance.

For two split candidates \(c_1\) and \(c_2\), candidate \(c_1\) dominates \(c_2\) when it is at least as informative in every relevant component and strictly more informative in at least one.

Conceptually:

$$
c_1\succ c_2
$$

if:

$$
I_k(c_1)\geq I_k(c_2)
\qquad
\forall k,
$$

and:

$$
I_q(c_1)>I_q(c_2)
$$

for at least one component \(q\).

The non-dominated candidate set is then:

$$
C_{\mathrm{ND}}
=
\left\{
c\in C:
\nexists c'\in C
\text{ such that }
c'\succ c
\right\}.
$$

This avoids introducing arbitrary fixed weights into the multidimensional decision process.

### 26.8 Direction Selection Through Non-Dominance

The same principle can be applied to coordinate directions.

For each direction \(j\), construct a directional profile:

$$
I_{\mathrm{direction}}(j).
$$

Directions can then be compared according to their unresolved information.

A direction that is dominated by another direction need not be selected.

The final direction is chosen deterministically from the non-dominated set.

This preserves multiple competing information objectives without assuming that one objective is universally more important than another.

### 26.9 Global Region Relevance

Local information alone is not sufficient.

A region may have high local uncertainty but contain little evidence of global relevance.

Conversely, a region may have moderate local uncertainty while containing a potentially superior objective value.

Therefore, regional selection must combine local unresolved information with global relevance.

Conceptually:

$$
\operatorname{Priority}(R)
=
\operatorname{LocalInformation}(R)
+
\operatorname{GlobalRelevance}(R),
$$

where the expression represents a conceptual combination rather than a required arithmetic sum.

The exact comparison rule must be deterministic and must preserve globally relevant regions from permanent neglect.

### 26.9 Refinement Action Selection

For a selected region, ARRGO chooses among:

$$
A_R
\in
\{
\mathrm{Sample},
\mathrm{Split},
\mathrm{Stable}
\}.
$$

The decision depends on the unresolved information state.

Conceptually:

$$
A_R
=
\operatorname{SelectAction}(Q_R).
$$

Possible interpretations are:

$$
Q_{\mathrm{coverage}}
\text{ dominant}
\quad\Longrightarrow\quad
\mathrm{Sample},
$$

and:

$$
Q_{\mathrm{behavior}},
Q_{\mathrm{uncertainty}},
\text{ or structural information}
\text{ unresolved}
\quad\Longrightarrow\quad
\mathrm{Split}.
$$

The word "dominant" refers to the selected deterministic decision policy and does not imply a fixed numerical weighting.

### 26.10 Avoiding Coordinate-Wise Optimization

A critical requirement is that ARRGO must not decompose the original problem into \(d\) independent one-dimensional optimization problems.

The objective remains:

$$
\max_{x\in\Omega}f(x).
$$

All coordinates jointly define the objective point:

$$
x=(x_1,\ldots,x_d).
$$

Directional analysis is used only to determine where structural information should be acquired.

It does not replace the original multidimensional optimization problem with independent coordinate optimization.

### 26.11 Interaction Between Directions

A multidimensional objective may contain interactions between coordinates.

For example:

$$
f(x_1,x_2)
$$

may depend strongly on the combination of \(x_1\) and \(x_2\), even if each coordinate considered separately appears relatively uninformative.

Therefore, coordinate-wise information must not be interpreted as a complete description of the objective.

The generalized framework should allow regional sampling to reveal interactions that cannot be detected from independent directional observations.

### 26.12 Cross-Directional Information

When coordinate interactions are significant, additional regional information may be obtained through multidimensional candidate points.

For example, points near combinations of important coordinate locations can provide information about:

$$
f(x_1,\ldots,x_d)
$$

that cannot be inferred reliably from separate one-dimensional slices.

Thus, directional analysis guides refinement, while actual objective evaluations remain fully multidimensional.

### 26.13 Refinement Progress

After an action is executed, the regional information state must be recomputed.

Therefore:

$$
Q_R^{(t)}
\rightarrow
Q_R^{(t+1)}.
$$

A successful refinement operation should change at least one unresolved information component or spatial-resolution property.

If repeated actions fail to change the relevant state, the implementation should recognize the possibility of numerical stagnation or an invalid action.

### 26.14 Stability

A region may be considered stable only when the selected resolution criteria indicate that no currently necessary refinement remains.

Conceptually:

$$
Q_R
\in
\mathcal{S}_{\mathrm{resolved}}
\quad\Longrightarrow\quad
\operatorname{State}(R)=\mathrm{STABLE}.
$$

Stability is therefore information-dependent.

It is not equivalent to deleting the region or permanently removing it from the hierarchy.

### 26.15 Global Fairness

The multidimensional convergence theorem requires that globally relevant regions and unresolved directions are not permanently ignored.

Therefore, the refinement policy must satisfy a fairness condition.

Informally:

> Every region and coordinate direction that remains globally relevant and unresolved must remain eligible for future information acquisition or structural refinement.

This condition is essential because local greedy decisions alone do not guarantee global convergence.

### 26.16 Determinism

Every stage of information-driven refinement must be deterministic.

Given identical:

- objective function;
- domain;
- evaluation history;
- region hierarchy;
- configuration;
- numerical tolerances;

the algorithm must generate the same:

$$
Q_R,
$$

candidate sets, action comparisons, and selected refinement action.

Therefore:

$$
\boxed{
\text{Same State}
\Rightarrow
\text{Same Decision}
}
$$

### 26.17 Generalized Refinement Principle

The complete multidimensional refinement principle can be summarized as:

$$
\boxed{
\begin{aligned}
&\text{Analyze regional information}
\\
&\Downarrow
\\
&\text{Identify unresolved spatial or objective information}
\\
&\Downarrow
\\
&\text{Generate feasible sampling and split candidates}
\\
&\Downarrow
\\
&\text{Compare candidates using deterministic information criteria}
\\
&\Downarrow
\\
&\text{Select Sample, Split, or Stable}
\\
&\Downarrow
\\
&\text{Update global information}
\\
&\Downarrow
\\
&\text{Repeat}
\end{aligned}
}
$$

The fundamental ARRGO principle is therefore preserved in arbitrary finite dimensions:

> Refinement is driven by the information that remains unresolved in the search space, while global selection ensures that potentially optimal regions are not permanently neglected.

This principle connects the multidimensional algorithmic design directly to the convergence assumptions established earlier.

## 27. Multidimensional Resolution and Progress Criteria

A central requirement of ARRGO is that refinement must produce measurable progress.

In one dimension, spatial progress can be described primarily through interval contraction.

In multiple dimensions, spatial resolution becomes directional because a region may be highly refined along one coordinate while remaining coarse along another.

Therefore, multidimensional resolution must account for the complete geometry of a region.

### 27.1 Directional Resolution

For a region:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j],
$$

the side length in direction \(j\) is:

$$
w_j(R)=r_j-l_j.
$$

A smaller value of \(w_j(R)\) indicates finer spatial resolution along coordinate \(j\).

The directional resolution state is therefore represented by:

$$
W(R)
=
(w_1(R),\ldots,w_d(R)).
$$

This vector should be retained rather than immediately reduced to a single scalar.

### 27.2 Maximum Side Length

A conservative scalar summary is:

$$
w_{\max}(R)
=
\max_{1\leq j\leq d}w_j(R).
$$

This quantity identifies the least-resolved coordinate direction.

If:

$$
w_{\max}(R)\rightarrow0,
$$

then every coordinate width must also converge to zero because:

$$
0\leq w_j(R)\leq w_{\max}(R).
$$

Consequently:

$$
w_j(R)\rightarrow0
\qquad
\forall j.
$$

### 27.3 Euclidean Diameter

The Euclidean diameter of the region is:

$$
\operatorname{diam}(R)
=
\sqrt{
\sum_{j=1}^{d}w_j(R)^2
}.
$$

The diameter provides a coordinate-independent measure of the spatial extent of the region.

Because:

$$
w_{\max}(R)
\leq
\operatorname{diam}(R)
\leq
\sqrt{d}\,w_{\max}(R),
$$

the two measures are equivalent up to a dimension-dependent constant.

Therefore:

$$
w_{\max}(R)\rightarrow0
\quad\Longleftrightarrow\quad
\operatorname{diam}(R)\rightarrow0.
$$

### 27.4 Directional Refinement Progress

Suppose region \(R\) is split along coordinate \(j\).

The selected coordinate width changes from:

$$
w_j=r_j-l_j
$$

to:

$$
w_j^{(1)}=s_j-l_j,
$$

and:

$$
w_j^{(2)}=r_j-s_j.
$$

Both satisfy:

$$
w_j^{(1)}<w_j,
\qquad
w_j^{(2)}<w_j.
$$

The widths of all other coordinates remain unchanged during this split.

Thus, every valid coordinate split produces local spatial progress in at least one direction.

### 27.5 Global Resolution Progress

Local contraction alone does not imply global resolution.

A region may repeatedly split in one direction while other directions remain unresolved.

Therefore, the global refinement process must ensure that persistent relevant regions eventually receive refinement in every unresolved coordinate direction.

Under this condition:

$$
w_j(R_t)\rightarrow0
\qquad
\forall j.
$$

Consequently:

$$
\operatorname{diam}(R_t)\rightarrow0.
$$

### 27.6 Coverage Resolution

Spatial resolution and evaluation coverage are distinct concepts.

A geometrically small region can still contain insufficient objective evaluations.

Therefore, ARRGO must track both:

$$
\text{Geometric Resolution}
$$

and:

$$
\text{Information Resolution}.
$$

The latter may include coordinate-wise coverage:

$$
\Gamma(R)
=
(\gamma_1,\ldots,\gamma_d),
$$

and geometric covering radius:

$$
h(R)
=
\sup_{x\in R}
\min_{x_i\in D_R}
\|x-x_i\|_2.
$$

Thus, a region should not be considered fully resolved solely because its geometric diameter is small.

### 27.7 Behavioral Resolution

A region may also remain unresolved with respect to objective behavior.

Let:

$$
B_R
$$

denote its behavioral information.

Behavioral resolution may improve when additional evaluations allow ARRGO to distinguish between competing local patterns.

The exact representation of \(B_R\) is implementation-dependent, but the conceptual requirement is:

$$
\text{new information}
\rightarrow
\text{improved regional understanding}.
$$

### 27.8 Uncertainty Resolution

In Certified Mode, the regional uncertainty is derived from valid upper and lower envelopes.

For:

$$
x\in R,
$$

define:

$$
u_R(x)
=
U_R(x)-L_R(x).
$$

The worst-case regional uncertainty is:

$$
u_{\max}(R)
=
\sup_{x\in R}u_R(x).
$$

As spatial and informational resolution improve, this quantity can decrease under the validity of the certified model.

However, decreasing uncertainty should be distinguished from merely observing a smaller region.

### 27.9 Potential Resolution

The regional optimization potential is:

$$
P(R)
=
\sup_{x\in R}U_R(x).
$$

A region remains globally relevant when its potential can still support an objective value superior to the current incumbent.

Therefore, potential is not simply another measure of spatial size.

It connects local information to global optimization relevance.

### 27.10 Global Relevance Condition

Let:

$$
f_{\mathrm{best}}
$$

be the current incumbent.

A region with:

$$
P(R)\leq f_{\mathrm{best}}
$$

cannot contain a point with an objective value exceeding the incumbent under the validity of the certified upper bound.

However, because the current ARRGO framework preserves regions and does not prune them, such a region is not deleted.

Instead, its current global relevance is reduced.

This distinction preserves the no-pruning principle while still allowing the algorithm to recognize regions whose current potential is insufficient to justify immediate refinement.

### 27.11 Progress Vector

The complete regional resolution state can therefore be viewed as a vector:

$$
\mathcal{R}(R)
=
\left(
W(R),
\Gamma(R),
B_R,
u_{\max}(R),
P(R)
\right).
$$

This is not intended to be collapsed into an arbitrary weighted scalar.

Different components describe different forms of unresolved information.

### 27.12 Progress After Sampling

A sampling operation changes the evaluation history:

$$
D_t
\rightarrow
D_{t+1}.
$$

Consequently:

$$
D_R^{(t)}
\rightarrow
D_R^{(t+1)}.
$$

The geometry of \(R\) remains unchanged, but information-related quantities may change:

$$
\Gamma(R),
\qquad
B_R,
\qquad
u_{\max}(R),
\qquad
P(R).
$$

Therefore:

$$
\boxed{
\text{Sampling}
\rightarrow
\text{Information Progress}
}
$$

### 27.13 Progress After Splitting

A structural split changes the geometric representation:

$$
R
\rightarrow
\{R_1,R_2\}.
$$

The child regions have smaller spatial extent in the selected direction.

Therefore:

$$
\boxed{
\text{Splitting}
\rightarrow
\text{Geometric Progress}
}
$$

The children may then require additional sampling to achieve sufficient information resolution.

### 27.14 Combined Progress

ARRGO therefore contains two complementary refinement mechanisms:

$$
\boxed{
\begin{aligned}
\text{Sampling}
&\rightarrow
\text{Information Resolution}
\\
\text{Splitting}
&\rightarrow
\text{Spatial Resolution}
\end{aligned}
}
$$

The combination is important because neither mechanism alone is sufficient for the general framework.

Sampling without structural refinement may leave large regions geometrically unresolved.

Splitting without sufficient sampling may create many regions with inadequate objective information.

### 27.15 Stagnation

An action should be considered potentially stagnant when it repeatedly fails to produce meaningful progress in the relevant unresolved information.

For example, repeated sampling of an already sufficiently covered location may provide no useful new spatial information.

Similarly, repeated structural refinement may become numerically invalid when region widths approach machine precision.

The implementation should therefore detect conditions such as:

$$
\Delta\mathcal{R}\approx0,
$$

within the configured numerical tolerance.

Stagnation should be handled explicitly rather than allowing an infinite sequence of ineffective actions.

### 27.16 Resolution-Based Stability

A region can be considered stable when the selected resolution criteria indicate that the currently relevant information is sufficiently resolved.

Conceptually:

$$
\mathcal{R}(R)
\in
\mathcal{S}_{\mathrm{resolved}}
\quad\Longrightarrow\quad
\operatorname{State}(R)=\mathrm{STABLE}.
$$

This does not imply that the region has been removed from the hierarchy.

### 27.17 Relationship to Convergence

The convergence theorem requires spatial contraction of relevant refinement paths:

$$
\operatorname{diam}(R_t)\rightarrow0.
$$

The resolution framework makes this condition operational by explicitly tracking:

$$
w_j(R),
\qquad
w_{\max}(R),
\qquad
\operatorname{diam}(R).
$$

It also separates geometric resolution from information resolution.

Thus, the theorem-level requirement:

$$
\operatorname{diam}(R_t)\rightarrow0
$$

is connected directly to measurable algorithmic quantities.

### 27.18 One-Dimensional Reduction

For:

$$
d=1,
$$

the directional width vector becomes:

$$
W(R)=(r-l).
$$

The maximum width and diameter both become:

$$
w_{\max}(R)
=
\operatorname{diam}(R)
=
r-l.
$$

The multidimensional resolution framework therefore reduces exactly to the original one-dimensional interval-resolution model.

### 27.19 Generalized Progress Principle

The complete multidimensional progress principle is:

$$
\boxed{
\begin{aligned}
&\text{Identify unresolved information}
\\
&\Downarrow
\\
&\text{Choose Sampling or Structural Refinement}
\\
&\Downarrow
\\
&\text{Acquire information or contract geometry}
\\
&\Downarrow
\\
&\text{Recompute regional resolution}
\\
&\Downarrow
\\
&\text{Detect progress or stagnation}
\\
&\Downarrow
\\
&\text{Continue while relevant information remains unresolved}
\end{aligned}
}
$$

This provides the operational connection between the abstract convergence conditions and the eventual multidimensional implementation.

The key requirement is that ARRGO must make measurable progress toward both:

$$
\boxed{
\text{Spatial Resolution}
}
$$

and:

$$
\boxed{
\text{Information Resolution}
}
$$

while preserving global relevance and deterministic refinement.

## 28. Multidimensional Certified Bounds

Certified Mode provides a mathematical upper and lower bound on the unknown objective values inside a region.

The one-dimensional formulation uses absolute distance:

$$
|x-x_i|.
$$

In the multidimensional formulation, this distance is replaced by the Euclidean norm:

$$
\|x-x_i\|_2.
$$

This section establishes the multidimensional certified construction and its relationship to the convergence framework.

### 28.1 Global Lipschitz Assumption

Assume that the objective satisfies a valid global Lipschitz condition:

$$
|f(x)-f(y)|
\leq
L\|x-y\|_2,
\qquad
\forall x,y\in\Omega,
$$

where:

$$
L\geq0.
$$

The constant \(L\) must be a valid upper bound over the entire configured domain.

A value estimated from a finite collection of samples is not automatically a valid global Lipschitz constant.

Therefore, Certified Mode requires the validity of the supplied bound as an assumption.

### 28.2 Regional Evaluation Set

For a region:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j],
$$

let:

$$
D_R
=
\{(x_i,f(x_i),\operatorname{Id}_i):x_i\in R\}.
$$

Each evaluation provides a known objective value:

$$
f(x_i).
$$

The Lipschitz condition allows this known information to constrain the possible objective value at any other point in the region.

### 28.3 Upper Bound From One Evaluation

For an evaluated point \(x_i\), the Lipschitz condition implies:

$$
|f(x)-f(x_i)|
\leq
L\|x-x_i\|_2.
$$

Therefore:

$$
f(x)
\leq
f(x_i)
+
L\|x-x_i\|_2.
$$

Thus, every evaluated point produces a valid upper bound:

$$
U_i(x)
=
f(x_i)
+
L\|x-x_i\|_2.
$$

### 28.4 Regional Upper Envelope

Because every evaluated point provides a valid upper bound, the tightest bound obtained from the regional evaluations is:

$$
U_R(x)
=
\min_{x_i\in D_R}
\left[
f(x_i)
+
L\|x-x_i\|_2
\right].
$$

For every:

$$
x\in R,
$$

we therefore have:

$$
f(x)\leq U_R(x).
$$

This establishes the validity of the regional upper envelope under the Lipschitz assumption.

### 28.5 Lower Bound From One Evaluation

The same Lipschitz condition also gives:

$$
f(x)
\geq
f(x_i)
-
L\|x-x_i\|_2.
$$

Therefore, every evaluated point provides a valid lower bound:

$$
L_i(x)
=
f(x_i)
-
L\|x-x_i\|_2.
$$

### 28.6 Regional Lower Envelope

The strongest lower bound implied by the regional evaluations is:

$$
L_R(x)
=
\max_{x_i\in D_R}
\left[
f(x_i)
-
L\|x-x_i\|_2
\right].
$$

Consequently:

$$
L_R(x)\leq f(x)
\qquad
\forall x\in R.
$$

Therefore:

$$
\boxed{
L_R(x)
\leq
f(x)
\leq
U_R(x)
}
$$

for every point in the region, under the stated assumptions.

### 28.7 Regional Uncertainty

The pointwise uncertainty interval is:

$$
[L_R(x),U_R(x)].
$$

Its width is:

$$
u_R(x)
=
U_R(x)-L_R(x).
$$

The worst-case regional uncertainty is:

$$
u_{\max}(R)
=
\sup_{x\in R}
u_R(x).
$$

A smaller value indicates that the available evaluations constrain the possible objective value more tightly.

This quantity is a mathematical uncertainty bound under the Lipschitz assumption.

### 28.8 Regional Optimization Potential

The regional optimization potential is defined as:

$$
P(R)
=
\sup_{x\in R}U_R(x).
$$

Because:

$$
f(x)\leq U_R(x),
$$

we obtain:

$$
\sup_{x\in R}f(x)
\leq
P(R).
$$

Therefore, \(P(R)\) is a valid upper bound on the best objective value that can occur inside region \(R\).

### 28.9 Global Potential

Suppose the region representation provides a collection of regions whose union covers the search domain:

$$
\Omega
\subseteq
\bigcup_{R\in\mathcal{R}}R.
$$

Then the global potential can be defined as:

$$
P_{\mathrm{global}}
=
\sup_{R\in\mathcal{R}}P(R).
$$

Since every feasible point belongs to at least one represented region:

$$
f^\ast
=
\sup_{x\in\Omega}f(x)
\leq
P_{\mathrm{global}}.
$$

Together with:

$$
f_{\mathrm{best}}
\leq
f^\ast,
$$

we obtain:

$$
\boxed{
f_{\mathrm{best}}
\leq
f^\ast
\leq
P_{\mathrm{global}}.
}
$$

### 28.10 Global Certification Gap

The certified global gap is:

$$
\Delta_{\mathrm{global}}
=
P_{\mathrm{global}}
-
f_{\mathrm{best}}.
$$

From the previous bounds:

$$
0
\leq
f^\ast-f_{\mathrm{best}}
\leq
\Delta_{\mathrm{global}}.
$$

Therefore, the global gap is a valid upper bound on the actual optimality error.

### 28.11 Epsilon-Optimality

If:

$$
\Delta_{\mathrm{global}}
\leq
\varepsilon,
$$

then:

$$
f^\ast-f_{\mathrm{best}}
\leq
\varepsilon.
$$

Equivalently:

$$
\boxed{
f_{\mathrm{best}}
\geq
f^\ast-\varepsilon.
}
$$

Thus, the incumbent is epsilon-optimal under the validity of the certified bounds.

### 28.12 Empty Regions

If:

$$
D_R=\varnothing,
$$

the regional envelope cannot be constructed from regional evaluations alone.

The implementation must therefore handle empty regions explicitly.

Possible policies include:

- requiring an initial sample before certification;
- deriving a valid domain-level bound from additional assumptions;
- marking the region uncertified until an evaluation is available.

The implementation must not silently fabricate a finite certified bound for an empty region.

### 28.13 Evaluation Reuse

An evaluation belongs to the global history exactly once:

$$
e_i=(x_i,f(x_i),\operatorname{Id}_i).
$$

Regional views may reuse that evaluation whenever:

$$
x_i\in R.
$$

This does not duplicate the objective evaluation.

It only allows the same acquired information to contribute to regional analysis.

### 28.14 Geometry of the Upper Envelope

In one dimension, each evaluated point produces a V-shaped upper cone:

$$
f(x_i)+L|x-x_i|.
$$

In \(d\) dimensions, the corresponding object is a cone in the Euclidean metric:

$$
f(x_i)+L\|x-x_i\|_2.
$$

The regional upper envelope is the lower envelope of these cones:

$$
U_R(x)
=
\min_i
\left[
f(x_i)+L\|x-x_i\|_2
\right].
$$

This geometric interpretation remains valid in every finite dimension.

### 28.15 Effect of Additional Sampling

Suppose a new evaluation is added to the regional set:

$$
D_R^{(t)}
\subseteq
D_R^{(t+1)}.
$$

The new upper envelope is:

$$
U_R^{(t+1)}(x)
=
\min
\left(
U_R^{(t)}(x),
f(x_{t+1})+L\|x-x_{t+1}\|_2
\right).
$$

Therefore:

$$
U_R^{(t+1)}(x)
\leq
U_R^{(t)}(x).
$$

Similarly, the lower envelope satisfies:

$$
L_R^{(t+1)}(x)
\geq
L_R^{(t)}(x).
$$

Consequently, additional valid evaluations cannot widen the certified interval:

$$
u_R^{(t+1)}(x)
\leq
u_R^{(t)}(x).
$$

This provides a formal monotonicity property for certified information refinement.

### 28.16 Effect on Global Potential

Because additional evaluations can only decrease the regional upper envelopes:

$$
P(R)
=
\sup_{x\in R}U_R(x)
$$

cannot increase due solely to adding valid evaluations to the same region.

Thus:

$$
P^{(t+1)}(R)
\leq
P^{(t)}(R).
$$

At the global level, this monotonicity requires care when the region representation itself changes.

If the collection of represented regions changes through splitting, the global potential must be computed using a region collection that still provides a valid cover of the domain.

The certificate therefore depends on both:

1. valid regional bounds;
2. valid global coverage.

### 28.17 Splitting and Certification

When a region is split:

$$
R
\rightarrow
\{R_1,R_2\},
$$

each child satisfies:

$$
R_i\subseteq R.
$$

Certified bounds can then be constructed independently over the child regions using the available regional information.

However, splitting alone does not create new objective information.

Therefore:

$$
\boxed{
\text{Splitting}
\neq
\text{New Objective Evaluation}
}
$$

Its primary effect is to improve geometric resolution.

### 28.18 Certification and No-Pruning

A region with low current potential may become globally irrelevant relative to the incumbent:

$$
P(R)\leq f_{\mathrm{best}}.
$$

Nevertheless, ARRGO does not delete the region.

The region remains represented in the hierarchy.

Its current potential may be used to determine that no immediate refinement is necessary while preserving the historical structure.

### 28.19 One-Dimensional Reduction

For:

$$
d=1,
$$

the Euclidean norm becomes:

$$
\|x-x_i\|_2=|x-x_i|.
$$

Therefore:

$$
U_R(x)
=
\min_i
\left[
f(x_i)+L|x-x_i|
\right],
$$

and:

$$
L_R(x)
=
\max_i
\left[
f(x_i)-L|x-x_i|
\right].
$$

These are exactly the original one-dimensional certified envelopes.

### 28.20 Certified Generalization Result

The multidimensional certified framework preserves the original ARRGO certificate structure:

$$
\boxed{
f_{\mathrm{best}}
\leq
f^\ast
\leq
P_{\mathrm{global}}
}
$$

and:

$$
\boxed{
0
\leq
f^\ast-f_{\mathrm{best}}
\leq
\Delta_{\mathrm{global}}.
}
$$

The only fundamental geometric replacement is:

$$
\boxed{
|x-x_i|
\rightarrow
\|x-x_i\|_2.
}
$$

Therefore, Certified Mode generalizes naturally from one dimension to arbitrary finite-dimensional Euclidean domains, provided that the Lipschitz assumption and regional/global coverage conditions remain valid.

## 29. Generalized ARRGO Algorithm

The previous sections established the mathematical and geometric requirements for extending ARRGO to arbitrary finite dimensions.

This section combines those requirements into a single generalized algorithmic specification.

The generalized algorithm is denoted by:

$$
\mathrm{ARRGO}_d,
$$

where:

$$
d\geq1.
$$

The algorithm operates on a bounded domain:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j].
$$

Its objective remains:

$$
\max_{x\in\Omega}f(x).
$$

### 29.1 Inputs

The generalized ARRGO algorithm receives:

1. A bounded search domain:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j].
$$

2. A deterministic objective function:

$$
f:\Omega\rightarrow\mathbb{R}.
$$

3. A maximum evaluation budget:

$$
N_{\max}.
$$

4. A contraction parameter:

$$
\rho.
$$

5. Numerical tolerances required by the implementation.

6. An execution mode:

$$
\mathrm{Mode}
\in
\{
\mathrm{Empirical},
\mathrm{Certified}
\}.
$$

7. If Certified Mode is selected, a valid Lipschitz constant:

$$
L.
$$

8. If certification termination is required, an epsilon tolerance:

$$
\varepsilon.
$$

### 29.2 Initialization

The initial region is:

$$
R_0=\Omega.
$$

The region hierarchy is initialized with:

$$
\mathcal{H}_0=\{R_0\}.
$$

The evaluation history is initially empty:

$$
D_0=\varnothing.
$$

The incumbent is initially undefined until the first objective evaluation is obtained.

The global evaluation count is:

$$
N_0=0.
$$

### 29.3 Initial Sampling

ARRGO acquires an initial set of objective evaluations inside the root region.

Let:

$$
C_{\mathrm{initial}}
\subseteq R_0
$$

be the deterministic initial candidate set.

For every selected initial point \(x_i\):

$$
y_i=f(x_i).
$$

The evaluation is added to:

$$
D_t.
$$

The evaluation count is increased:

$$
N_t
\rightarrow
N_t+1.
$$

Initial sampling must respect the available evaluation budget.

### 29.4 Incumbent Initialization

After at least one valid evaluation:

$$
f_{\mathrm{best}}^{(t)}
=
\max_{e_i\in D_t}f(x_i).
$$

The corresponding incumbent location is:

$$
x_{\mathrm{best}}^{(t)}
\in
\operatorname*{arg\,max}_{e_i\in D_t}f(x_i).
$$

The incumbent is updated whenever a newly evaluated point produces a strictly better observed objective value.

### 29.5 Regional Analysis

For each eligible region:

$$
R\in\mathcal{R}_{\mathrm{eligible}},
$$

ARRGO constructs its current information state:

$$
Q_R
=
\left(
Q_{\mathrm{coverage}},
Q_{\mathrm{behavior}},
Q_{\mathrm{uncertainty}},
Q_{\mathrm{potential}}
\right).
$$

The regional analysis uses the evaluations spatially associated with \(R\).

### 29.6 Global Region Selection

The algorithm determines which region should receive attention.

Conceptually:

$$
R^\ast
=
\operatorname{SelectRegion}
\left(
\mathcal{R}_{\mathrm{eligible}},
Q,
f_{\mathrm{best}}
\right).
$$

The selection must remain deterministic and must preserve global fairness.

A region must not be permanently ignored solely because another region currently appears more informative.

### 29.7 Action Generation

For the selected region \(R^\ast\), ARRGO generates feasible actions.

Sampling candidates:

$$
C_{\mathrm{sample}}(R^\ast)
=
\{x^{(1)},\ldots,x^{(m)}\},
$$

with:

$$
x^{(k)}\in R^\ast.
$$

Split candidates:

$$
C_{\mathrm{split}}(R^\ast)
=
\{(j,s_j)\},
$$

where:

$$
j\in\{1,\ldots,d\},
$$

and:

$$
l_j<s_j<r_j.
$$

Only geometrically valid split candidates are retained.

### 29.8 Action Analysis

Each feasible action receives an information profile.

For sampling:

$$
I_{\mathrm{sample}}(x;R^\ast).
$$

For splitting:

$$
I_{\mathrm{split}}(j,s_j;R^\ast).
$$

The profiles may contain information about:

- spatial coverage;
- objective behavior;
- uncertainty;
- optimization potential;
- structural resolution.

These profiles are deterministic information signals.

### 29.9 Local Action Selection

The selected action is:

$$
A^\ast
=
\operatorname{SelectAction}
\left(
C_{\mathrm{sample}},
C_{\mathrm{split}},
Q_{R^\ast}
\right).
$$

The action belongs to:

$$
A^\ast
\in
\{
\mathrm{Sample},
\mathrm{Split},
\mathrm{Stable}
\}.
$$

For a sampling action:

$$
A^\ast=(\mathrm{Sample},x^\ast).
$$

For a split action:

$$
A^\ast=(\mathrm{Split},j^\ast,s_{j^\ast}).
$$

### 29.10 Sampling Execution

If:

$$
A^\ast=(\mathrm{Sample},x^\ast),
$$

ARRGO evaluates:

$$
y^\ast=f(x^\ast).
$$

The evaluation is added to the global history:

$$
D_{t+1}
=
D_t
\cup
\{(x^\ast,y^\ast,\operatorname{Id})\}.
$$

The incumbent is then updated.

Sampling consumes one unit of the evaluation budget.

### 29.11 Split Execution

If:

$$
A^\ast=(\mathrm{Split},j^\ast,s_{j^\ast}),
$$

the selected region is partitioned along coordinate \(j^\ast\).

The two children are:

$$
R_1
=
\prod_{k=1}^{d}I_k^{(1)},
$$

and:

$$
R_2
=
\prod_{k=1}^{d}I_k^{(2)},
$$

where the selected coordinate is divided at \(s_{j^\ast}\).

No objective evaluation is automatically performed by the split.

Therefore:

$$
N_{t+1}=N_t.
$$

The parent remains in the hierarchy.

### 29.12 Stable Action

If:

$$
A^\ast=\mathrm{Stable},
$$

the region is marked as currently stable.

Its representation in the hierarchy is preserved.

The algorithm must nevertheless ensure that a stable decision does not create an infinite loop in which no state changes.

If no valid action can produce meaningful progress, an appropriate termination reason is recorded.

### 29.13 Global State Update

After every action, ARRGO updates:

- evaluation history;
- incumbent;
- region hierarchy;
- region states;
- regional information;
- global relevance;
- certified quantities when applicable.

Thus:

$$
G_t
\rightarrow
G_{t+1}.
$$

The global best objective remains monotone:

$$
f_{\mathrm{best}}^{(t)}
\leq
f_{\mathrm{best}}^{(t+1)}.
$$

### 29.14 Certified Update

In Certified Mode, ARRGO recomputes valid regional upper bounds:

$$
U_R(x)
=
\min_{x_i\in D_R}
\left[
f(x_i)+L\|x-x_i\|_2
\right].
$$

The regional potential is:

$$
P(R)
=
\sup_{x\in R}U_R(x).
$$

The global potential is:

$$
P_{\mathrm{global}}
=
\sup_R P(R),
$$

provided that the represented region collection maintains valid coverage of the domain.

The certification gap is:

$$
\Delta_{\mathrm{global}}
=
P_{\mathrm{global}}
-
f_{\mathrm{best}}.
$$

### 29.15 Termination Check

After each state update, ARRGO evaluates its termination conditions.

Possible termination conditions include:

$$
\Delta_{\mathrm{global}}
\leq
\varepsilon,
$$

when Certified Mode is active.

The evaluation budget condition is:

$$
N_t\geq N_{\max}.
$$

Additional termination conditions may include:

- numerical limits;
- absence of valid actions;
- configuration errors.

Termination must be associated with an explicit reason.

### 29.16 Main Iteration

If no termination condition is satisfied, ARRGO returns to regional analysis:

$$
\boxed{
\begin{aligned}
&\text{Analyze Regions}
\\
&\Downarrow
\\
&\text{Select Global Region}
\\
&\Downarrow
\\
&\text{Generate Actions}
\\
&\Downarrow
\\
&\text{Select Action}
\\
&\Downarrow
\\
&\text{Sample or Split}
\\
&\Downarrow
\\
&\text{Update Global State}
\\
&\Downarrow
\\
&\text{Check Termination}
\\
&\Downarrow
\\
&\text{Repeat}
\end{aligned}
}
$$

### 29.17 Generalized Algorithm Specification

The complete generalized ARRGO procedure can therefore be represented as:

$$
\boxed{
\begin{aligned}
\mathrm{ARRGO}_d(&f,\Omega,N_{\max},\rho,\mathrm{Mode},L,\varepsilon)
\\[2mm]
&\rightarrow
\text{Initialize }R_0=\Omega
\\
&\rightarrow
\text{Acquire Initial Evaluations}
\\
&\rightarrow
\text{Initialize Incumbent}
\\
&\rightarrow
\text{Analyze Eligible Regions}
\\
&\rightarrow
\text{Select Globally Relevant Region}
\\
&\rightarrow
\text{Generate Feasible Actions}
\\
&\rightarrow
\text{Select Deterministic Action}
\\
&\rightarrow
\begin{cases}
\text{Sample}(x),\\
\text{Split}(j,s_j),\\
\text{Stable}
\end{cases}
\\
&\rightarrow
\text{Update Global State}
\\
&\rightarrow
\text{Update Certified Quantities}
\\
&\rightarrow
\text{Check Termination}
\\
&\rightarrow
\text{Repeat Until Termination}.
\end{aligned}
}
$$

### 29.18 Correctness Requirements

A multidimensional implementation is considered consistent with this specification only if it preserves the following invariants.

#### Invariant 1 — Feasibility

Every evaluated point satisfies:

$$
x_i\in\Omega.
$$

#### Invariant 2 — Region Validity

Every region satisfies:

$$
R\subseteq\Omega.
$$

#### Invariant 3 — Hierarchy Consistency

Every non-root region has exactly one structural parent.

#### Invariant 4 — Parent Retention

Structural refinement never deletes the parent from the hierarchy.

#### Invariant 5 — Incumbent Monotonicity

The best observed objective value satisfies:

$$
f_{\mathrm{best}}^{(t)}
\leq
f_{\mathrm{best}}^{(t+1)}.
$$

#### Invariant 6 — Determinism

Identical inputs and state produce identical decisions.

#### Invariant 7 — Budget Validity

The number of objective evaluations never exceeds the configured budget.

#### Invariant 8 — Certified Validity

When Certified Mode is active and the Lipschitz assumption is valid:

$$
f_{\mathrm{best}}
\leq
f^\ast
\leq
P_{\mathrm{global}}.
$$

### 29.19 One-Dimensional Recovery

For:

$$
d=1,
$$

the generalized algorithm becomes the original ARRGO process.

The domain becomes an interval, regions become intervals, points become scalars, and there is only one structural split direction.

Therefore:

$$
\boxed{
\mathrm{ARRGO}_1
=
\mathrm{ARRGO}(d=1).
}
$$

The generalized algorithm is consequently a single dimension-parameterized formulation rather than a collection of separate algorithms.

### 29.20 Final Generalization Statement

The generalized ARRGO algorithm preserves the original optimization philosophy while extending its geometry to arbitrary finite dimensions.

Its essential structure remains:

$$
\boxed{
\text{Evaluate}
\rightarrow
\text{Analyze}
\rightarrow
\text{Refine}
\rightarrow
\text{Update}
\rightarrow
\text{Repeat}
}
$$

while the spatial representation becomes:

$$
\boxed{
\mathbb{R}
\rightarrow
\mathbb{R}^{d}.
}
$$

The mathematical generalization, convergence theorem, structural refinement rules, certified bounds, and algorithmic specification are therefore mutually consistent.

The next stage is to translate this generalized specification into implementation-level abstractions without prematurely modifying the existing one-dimensional implementation.

## 30. Multidimensional Invariant Set

A generalized implementation of ARRGO must preserve a set of structural, numerical, geometric, informational, and optimization invariants.

These invariants define properties that must remain valid throughout execution, independently of the selected dimension.

They also provide the foundation for the later implementation and validation stages.

### 30.1 Domain Invariant

The configured optimization domain is:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j].
$$

Every coordinate satisfies:

$$
a_j\leq x_j\leq b_j.
$$

Therefore, every evaluated point must satisfy:

$$
x\in\Omega.
$$

No sampling candidate may lie outside the configured domain.

### 30.2 Dimension Invariant

Every point must have exactly \(d\) coordinates:

$$
x\in\mathbb{R}^{d}.
$$

Every region must contain exactly \(d\) lower and upper bounds:

$$
l,r\in\mathbb{R}^{d}.
$$

Thus, dimensionality is an explicit property of the algorithm rather than an implicit assumption of the implementation.

### 30.3 Region Invariant

Every region has the form:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j],
$$

with:

$$
a_j\leq l_j\leq r_j\leq b_j.
$$

Therefore:

$$
R\subseteq\Omega.
$$

No structural operation may create a region outside the configured domain.

### 30.4 Positive-Width Invariant

For every non-degenerate region:

$$
l_j<r_j.
$$

A dimension may become numerically indistinguishable from zero only when an explicit numerical stopping condition is reached.

The implementation must not continue structural refinement indefinitely once further subdivision becomes numerically meaningless.

### 30.5 Evaluation Invariant

Every evaluation has the form:

$$
e_i
=
(x_i,f(x_i),\operatorname{Id}_i),
$$

where:

$$
x_i\in\Omega.
$$

The stored objective value must correspond to the objective evaluation at that point.

The evaluation history is therefore a single source of objective-evaluation truth.

### 30.6 Evaluation Identity Invariant

Every evaluation has a unique identifier:

$$
\operatorname{Id}_i
\neq
\operatorname{Id}_k
\qquad
\text{for }i\neq k.
$$

Evaluation identifiers represent evaluation records and must not be confused with region identifiers.

### 30.7 Evaluation Budget Invariant

Let:

$$
N_t=|D_t|.
$$

Then:

$$
0\leq N_t\leq N_{\max}.
$$

Every new objective evaluation increases the count by exactly one.

Structural operations such as splitting do not consume objective-evaluation budget unless the implementation explicitly performs additional evaluations.

### 30.8 Incumbent Invariant

The incumbent is always one of the evaluated points:

$$
x_{\mathrm{best}}^{(t)}
\in
\{x_i:e_i\in D_t\}.
$$

Its objective value is:

$$
f_{\mathrm{best}}^{(t)}
=
f(x_{\mathrm{best}}^{(t)})
=
\max_{e_i\in D_t}f(x_i).
$$

Consequently:

$$
f_{\mathrm{best}}^{(t)}
\leq
f_{\mathrm{best}}^{(t+1)}.
$$

The best observed objective value can never decrease.

### 30.9 Region Membership Invariant

A regional evaluation set contains exactly those evaluations whose points belong to the region:

$$
D_R
=
\{e_i\in D_t:x_i\in R\}.
$$

Regional membership must therefore be determined from the current geometric region and the evaluation point rather than from duplicated evaluation records.

### 30.10 Hierarchy Invariant

The region hierarchy contains a root region:

$$
R_0=\Omega.
$$

Every non-root region has exactly one structural parent.

The hierarchy therefore forms a rooted tree or an equivalent rooted acyclic structure.

### 30.11 Parent Retention Invariant

When a region is split:

$$
R
\rightarrow
\{R_1,R_2\},
$$

the parent \(R\) remains represented in the hierarchy.

Therefore:

$$
R\in\mathcal{H}
$$

continues to hold after refinement.

ARRGO does not use pruning as a mechanism for controlling the hierarchy.

### 30.12 Child Containment Invariant

For every valid split:

$$
R_1\subseteq R,
\qquad
R_2\subseteq R.
$$

The children must remain geometrically contained inside their parent.

### 30.13 Coverage Invariant

Structural refinement must preserve coverage of the region being refined.

Conceptually:

$$
R
\subseteq
R_1\cup R_2.
$$

For a complete partition, the implementation should additionally ensure that the children do not leave an uncovered portion of the parent.

Boundary ownership must be explicitly defined when closed mathematical regions create overlapping split boundaries.

### 30.14 Contraction Invariant

If coordinate \(j\) is split at \(s_j\), then:

$$
\max
\left(
\frac{s_j-l_j}{r_j-l_j},
\frac{r_j-s_j}{r_j-l_j}
\right)
\leq
\rho,
$$

where:

$$
0<\rho<1.
$$

Therefore, both child widths in the selected direction are strictly smaller than the parent width.

### 30.15 Directional Refinement Invariant

A multidimensional convergence argument requires more than repeated splitting.

The refinement process must ensure that globally relevant regions cannot indefinitely avoid refinement in a particular coordinate direction.

For a region sequence containing a globally relevant point:

$$
R_0\supseteq R_1\supseteq R_2\supseteq\cdots,
$$

directional fairness requires that unresolved coordinate widths are eventually reduced.

Thus, under the appropriate fairness condition:

$$
w_j(R_t)\rightarrow0
\qquad
\text{for all }j=1,\ldots,d.
$$

Consequently:

$$
\operatorname{diam}(R_t)\rightarrow0.
$$

### 30.16 Sampling Invariant

Every sampling action selects a feasible point:

$$
x_{\mathrm{sample}}\in R
\subseteq
\Omega.
$$

Sampling creates new objective information.

Therefore, a successful sampling action must add a valid evaluation to the global history unless the selected point is already known and the implementation explicitly rejects it as redundant.

### 30.17 Splitting Invariant

A splitting action modifies spatial structure without inherently adding objective information.

Thus:

$$
\boxed{
\text{Split}
\Rightarrow
\text{Structural Change}
}
$$

but not necessarily:

$$
\text{Split}
\Rightarrow
\text{New Evaluation}.
$$

This distinction must remain explicit in the implementation.

### 30.18 Action Validity Invariant

Every executed action must belong to the feasible action set:

$$
A_t
\in
\mathcal{A}_{\mathrm{feasible}}(G_t).
$$

An action violating domain, region, contraction, budget, or numerical constraints must never be executed.

### 30.19 Determinism Invariant

Given identical:

- objective function;
- domain;
- configuration;
- evaluation history;
- hierarchy state;
- numerical environment;

ARRGO must produce the same next decision.

Thus:

$$
G_t
\longrightarrow
G_{t+1}
$$

is deterministic.

If multiple candidates are equally valid, the implementation must use an explicit deterministic tie-breaking rule.

### 30.20 Information-Ownership Invariant

Objective evaluations are stored globally exactly once.

Regional information is derived from those evaluations.

Therefore:

$$
\boxed{
\text{Global Evaluation History}
=
\text{Single Source of Evaluation Truth}
}
$$

while regions provide structural views over that information.

### 30.21 Information-Driven Refinement Invariant

The selected refinement action must be determined from the current information state.

The algorithm must not reduce to an unconditional fixed partitioning scheme.

The decision may depend on:

$$
Q_R
=
(Q_{\mathrm{coverage}},
Q_{\mathrm{behavior}},
Q_{\mathrm{uncertainty}},
Q_{\mathrm{potential}}),
$$

together with global relevance and geometric constraints.

No arbitrary weighted sum is required to define the decision.

### 30.22 Joint-Objective Invariant

The objective remains a joint function:

$$
f(x_1,\ldots,x_d).
$$

Multidimensional ARRGO must not implicitly replace the original optimization problem with independent coordinate-wise optimization.

Sampling and analysis must therefore remain capable of representing interactions between coordinates.

### 30.23 Certified-Bound Invariant

In Certified Mode, if the supplied Lipschitz constant is valid:

$$
|f(x)-f(y)|
\leq
L\|x-y\|_2,
$$

then the regional envelopes must satisfy:

$$
L_R(x)
\leq
f(x)
\leq
U_R(x).
$$

Consequently:

$$
f_{\mathrm{best}}
\leq
f^\ast
\leq
P_{\mathrm{global}},
$$

provided that the region collection used for certification covers the complete domain.

### 30.24 Certified-Gap Invariant

The certified gap is:

$$
\Delta_{\mathrm{global}}
=
P_{\mathrm{global}}
-
f_{\mathrm{best}}.
$$

Under valid certification assumptions:

$$
\Delta_{\mathrm{global}}\geq0.
$$

Furthermore:

$$
f^\ast-f_{\mathrm{best}}
\leq
\Delta_{\mathrm{global}}.
$$

### 30.25 No-Pruning Invariant

A region must not be deleted merely because:

$$
P(R)\leq f_{\mathrm{best}}.
$$

Such a region may become currently irrelevant for refinement, but it remains part of the structural hierarchy.

This preserves the complete refinement history.

### 30.26 Stable-State Invariant

A region may be classified as:

$$
\mathrm{ACTIVE},
\qquad
\mathrm{STABLE},
\qquad
\mathrm{REFINED}.
$$

A stable state means that no currently justified local refinement is required.

It does not mean that the region has been removed from the global representation.

### 30.27 Progress Invariant

Every non-terminal iteration must produce meaningful state progress.

Progress may occur through:

- acquisition of a new objective evaluation;
- reduction of a region's spatial scale;
- improvement of the incumbent;
- reduction of certified uncertainty;
- reduction of a certified global potential;
- transition to a meaningful stable or refined state.

An implementation must prevent an infinite sequence of iterations in which no relevant state changes.

### 30.28 Termination Invariant

When ARRGO terminates, exactly one explicit termination reason must be recorded.

Possible reasons include:

$$
\{
\mathrm{CertifiedTolerance},
\mathrm{BudgetExhausted},
\mathrm{NumericalLimit},
\mathrm{NoValidAction},
\mathrm{ConfigurationError}
\}.
$$

The returned result must be interpreted according to that reason.

In particular:

$$
\mathrm{BudgetExhausted}
\not\Rightarrow
\mathrm{GlobalOptimality}.
$$

### 30.29 Empirical-Mode Invariant

In Empirical Mode, the algorithm may return:

$$
x_{\mathrm{best}},
\qquad
f_{\mathrm{best}},
$$

as the best-found solution.

No mathematical optimality certificate is implied solely by the empirical result.

### 30.30 Certified-Mode Invariant

In Certified Mode, an epsilon-optimality claim is permitted only when:

1. the Lipschitz assumption is valid;
2. the regional bounds are valid;
3. the region collection provides valid global coverage;
4. the computed global gap satisfies:

$$
\Delta_{\mathrm{global}}\leq\varepsilon.
$$

Only under these conditions may ARRGO claim:

$$
f_{\mathrm{best}}
\geq
f^\ast-\varepsilon.
$$

### 30.31 One-Dimensional Recovery

For:

$$
d=1,
$$

all multidimensional invariants reduce to their original one-dimensional counterparts.

The domain becomes:

$$
\Omega=[a,b],
$$

the region becomes:

$$
R=[l,r],
$$

and the Euclidean distance becomes:

$$
\|x-x_i\|_2=|x-x_i|.
$$

Therefore, the invariant set is dimension-independent and contains the existing one-dimensional ARRGO formulation as a special case.

### 30.32 Invariant Summary

The generalized implementation must preserve the following core chain:

$$
\boxed{
\begin{aligned}
&\text{Valid Domain}
\\
&\Downarrow
\\
&\text{Valid Regions}
\\
&\Downarrow
\\
&\text{Valid Evaluations}
\\
&\Downarrow
\\
&\text{Valid Information}
\\
&\Downarrow
\\
&\text{Valid Refinement}
\\
&\Downarrow
\\
&\text{Global Incumbent}
\\
&\Downarrow
\\
&\text{Valid Termination}
\end{aligned}
}
$$

For Certified Mode, this chain is extended by:

$$
\boxed{
\text{Valid Lipschitz Assumption}
\Rightarrow
\text{Valid Regional Bounds}
\Rightarrow
\text{Valid Global Bound}
\Rightarrow
\text{Valid Epsilon-Certificate}.
}
$$

These invariants form the contract that the future multidimensional implementation must satisfy.

The implementation should therefore be derived from these invariants rather than modifying the existing one-dimensional code incrementally without first establishing the generalized abstractions.

## 31. Dimension-Independent and Dimension-Dependent Components

The multidimensional generalization does not require every component of ARRGO to be redesigned.

Some components are fundamentally independent of dimensionality, while others depend directly on the geometry of the search space.

Separating these two categories is important for designing a single dimension-parameterized implementation.

### 31.1 Dimension-Independent Components

The following concepts preserve their mathematical structure regardless of the dimension.

#### Objective Evaluation

The evaluation remains:

$$
e_i=(x_i,f(x_i),\operatorname{Id}_i).
$$

Only the representation of \(x_i\) changes from a scalar to a vector.

The conceptual meaning of an evaluation does not change.

#### Evaluation History

The global evaluation history remains:

$$
D_t=\{e_1,\ldots,e_t\}.
$$

Evaluations are still stored globally exactly once.

#### Global Incumbent

The incumbent remains:

$$
f_{\mathrm{best}}^{(t)}
=
\max_{e_i\in D_t}f(x_i).
$$

The best observed value is therefore dimension-independent.

#### Region Hierarchy

The structural hierarchy remains a parent-child hierarchy:

$$
R
\rightarrow
\{R_1,R_2\}.
$$

Parent retention and the no-pruning principle remain unchanged.

#### Region Lifecycle

The region states remain:

$$
\{
\mathrm{ACTIVE},
\mathrm{STABLE},
\mathrm{REFINED}
\}.
$$

Their meaning does not depend on dimension.

#### Sampling vs Splitting

The distinction remains fundamental:

$$
\boxed{
\text{Sampling}
=
\text{Information Refinement}
}
$$

and:

$$
\boxed{
\text{Splitting}
=
\text{Structural Refinement}.
}
$$

This distinction is independent of geometry.

#### Global Region Selection

The purpose remains selecting the region that should receive computational attention.

The selection mechanism may use different geometric signals in higher dimensions, but the global decision concept is unchanged.

#### Action Selection

The possible action categories remain:

$$
\{
\mathrm{Sample},
\mathrm{Split},
\mathrm{Stable}
\}.
$$

Only the representation of the action changes.

#### Termination

The conceptual termination categories remain:

- certified tolerance;
- evaluation budget exhaustion;
- numerical limits;
- no valid action;
- configuration error.

#### Determinism

The deterministic requirement remains:

$$
G_t
\rightarrow
G_{t+1}
$$

must be uniquely determined by the current state and configuration.

#### Empirical vs Certified Mode

The distinction remains unchanged:

$$
\mathrm{Empirical}
\Rightarrow
\text{best-found solution},
$$

while:

$$
\mathrm{Certified}
\Rightarrow
\text{best-found solution + valid mathematical certificate},
$$

under the required assumptions.

### 31.2 Dimension-Dependent Components

Other components depend directly on the geometry of \(\mathbb{R}^d\).

#### Point Representation

In one dimension:

$$
x\in\mathbb{R}.
$$

In \(d\) dimensions:

$$
x=(x_1,\ldots,x_d)\in\mathbb{R}^d.
$$

Therefore, the implementation must replace scalar coordinate handling with vector-aware operations.

#### Domain Representation

The one-dimensional domain:

$$
[a,b]
$$

becomes:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j].
$$

The implementation therefore requires lower and upper bound vectors.

#### Region Representation

The one-dimensional region:

$$
[l,r]
$$

becomes:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j].
$$

A region therefore requires \(d\) lower bounds and \(d\) upper bounds.

#### Region Width

One-dimensional width is:

$$
w=r-l.
$$

The multidimensional representation becomes the width vector:

$$
W(R)
=
(w_1,\ldots,w_d),
$$

where:

$$
w_j=r_j-l_j.
$$

#### Region Diameter

The multidimensional diameter is:

$$
\operatorname{diam}(R)
=
\sqrt{
\sum_{j=1}^{d}w_j^2
}.
$$

This quantity has no nontrivial distinction from width in one dimension because:

$$
\operatorname{diam}(R)=w
\qquad
(d=1).
$$

#### Region Center

The scalar midpoint:

$$
c=\frac{l+r}{2}
$$

becomes:

$$
c(R)
=
\left(
\frac{l_1+r_1}{2},
\ldots,
\frac{l_d+r_d}{2}
\right).
$$

#### Distance

One-dimensional distance:

$$
|x-x_i|
$$

becomes:

$$
\|x-x_i\|_2.
$$

This affects certified bounds, coverage measures, and distance-based candidate generation.

#### Spatial Coverage

In one dimension, coverage can be described using ordered neighboring evaluation gaps.

In multiple dimensions, there is no unique equivalent of a one-dimensional neighboring gap.

Coverage therefore requires multidimensional measures such as:

- coordinate-wise gaps;
- directional coverage;
- geometric covering radius;
- local distance to evaluated points.

#### Behavior Analysis

One-dimensional local behavior can be analyzed along a scalar coordinate.

In multiple dimensions, behavior becomes directional and potentially interaction-dependent.

A regional behavior representation may therefore contain information such as:

$$
B_R
=
(B_R^{(1)},\ldots,B_R^{(d)}),
$$

together with information about interactions between coordinates when required.

#### Structural Splitting

In one dimension there is only one possible splitting direction.

In \(d\) dimensions, a split requires:

$$
(j,s_j),
$$

where:

$$
j\in\{1,\ldots,d\}.
$$

Thus, direction selection becomes a new decision problem.

#### Contraction

In one dimension, contraction is applied to interval length.

In multiple dimensions, contraction must be interpreted along the selected coordinate.

For direction \(j\):

$$
\max
\left(
\frac{s_j-l_j}{r_j-l_j},
\frac{r_j-s_j}{r_j-l_j}
\right)
\leq
\rho.
$$

Global geometric contraction additionally requires directional fairness.

#### Certified Bounds

The one-dimensional Lipschitz distance:

$$
|x-x_i|
$$

is replaced by:

$$
\|x-x_i\|_2.
$$

Therefore:

$$
U_R(x)
=
\min_i
\left[
f(x_i)+L\|x-x_i\|_2
\right],
$$

and:

$$
L_R(x)
=
\max_i
\left[
f(x_i)-L\|x-x_i\|_2
\right].
$$

The mathematical certificate structure itself remains unchanged.

### 31.3 Components Requiring Partial Redesign

Some components preserve their purpose but require a more substantial redesign.

#### Candidate Generation

Sampling candidates must now be vectors:

$$
x^{(k)}
\in
R.
$$

Split candidates must include both a direction and a location:

$$
(j,s_j).
$$

Therefore, candidate generation becomes a geometric search over both direction and position.

#### Action Profiles

In one dimension, candidate profiles can often be represented using scalar signals.

In multiple dimensions, profiles naturally become vectors or structured collections:

$$
I_{\mathrm{sample}}(x;R),
$$

and:

$$
I_{\mathrm{split}}(j,s_j;R).
$$

These profiles should not automatically be collapsed into a weighted scalar score.

#### Pareto Selection

Because multiple dimensions may produce competing refinement signals, non-dominated candidate selection becomes especially important.

For candidate profiles:

$$
p_1,\ldots,p_k,
$$

a candidate may remain feasible when no other candidate is strictly better across every relevant criterion.

A deterministic tie-breaking rule is then applied among the non-dominated candidates.

### 31.4 Components That Must Not Be Changed

The dimensional generalization must not accidentally change the fundamental ARRGO philosophy.

The following principles remain fixed:

$$
\boxed{
\begin{aligned}
&\text{Black-box objective}
\\
&\text{Deterministic execution}
\\
&\text{Region-based search}
\\
&\text{Information-driven refinement}
\\
&\text{Sampling/Splitting distinction}
\\
&\text{Parent retention}
\\
&\text{No pruning}
\\
&\text{Global incumbent}
\\
&\text{Empirical/Certified distinction}
\end{aligned}
}
$$

These are algorithmic principles rather than one-dimensional implementation details.

### 31.5 Design Consequence

The correct implementation strategy is therefore not:

$$
\mathrm{ARRGO}_{1D}
\rightarrow
\mathrm{ARRGO}_{2D}
\rightarrow
\mathrm{ARRGO}_{3D}
\rightarrow
\cdots
$$

as a collection of separate algorithms.

Instead, the target architecture is:

$$
\boxed{
\mathrm{ARRGO}(d)
}
$$

with:

$$
d=1,2,3,\ldots
$$

as a configuration parameter.

The same algorithmic components should operate on dimension-parameterized geometric structures.

### 31.6 Special Case Requirement

The generalized implementation must satisfy:

$$
\mathrm{ARRGO}(1)
=
\mathrm{ARRGO}_{1D}.
$$

This means that moving to higher dimensions must not require a fundamentally different algorithm.

Instead, the one-dimensional behavior should emerge naturally when:

$$
d=1.
$$

### 31.7 Generalization Principle

The dimensional extension can therefore be summarized as:

$$
\boxed{
\text{Preserve the Algorithm}
+
\text{Generalize the Geometry}
}
$$

The algorithmic logic remains conceptually stable, while the geometric layer becomes dimension-aware.

This separation provides the foundation for implementing ARRGO as a general finite-dimensional framework rather than as a collection of dimension-specific variants.

## 33. Multidimensional Sampling Strategy

Sampling is the primary mechanism through which ARRGO acquires new objective information.

In one dimension, sampling candidates can be generated along an interval.

In multiple dimensions, sampling must generate feasible points inside a hyper-rectangular region while preserving:

- deterministic behavior;
- spatial coverage;
- local information acquisition;
- joint objective evaluation;
- numerical validity;
- compatibility with the global evaluation history.

The purpose is not to construct an exhaustive grid.

Instead, sampling should adaptively place evaluations where the current regional information state indicates that additional information is useful.

### 33.1 Sampling Domain

For a selected region:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j],
$$

every sampling candidate must satisfy:

$$
x
=
(x_1,\ldots,x_d)\in R.
$$

Therefore:

$$
l_j\leq x_j\leq r_j,
\qquad
j=1,\ldots,d.
$$

### 33.2 Sampling Candidate Representation

A sampling candidate is a vector:

$$
c_{\mathrm{sample}}
=
x
\in
\mathbb{R}^d.
$$

A complete candidate record may additionally contain its information profile:

$$
C_{\mathrm{sample}}
=
(x,I_{\mathrm{sample}}(x;R)).
$$

The candidate's location and its estimated information value must remain conceptually separate.

### 33.3 Regional Center

The natural geometric center of the region is:

$$
c(R)
=
\left(
\frac{l_1+r_1}{2},
\ldots,
\frac{l_d+r_d}{2}
\right).
$$

The center is a useful candidate because it is maximally symmetric with respect to the hyper-rectangle.

However:

$$
\boxed{
\text{Center Candidate}
\neq
\text{Guaranteed Best Candidate}.
}
$$

ARRGO must not assume that the optimum is located at the region center.

### 33.4 Existing Evaluation Awareness

Before generating a new sampling candidate, ARRGO considers the existing regional evaluations:

$$
D_R.
$$

A candidate that is numerically identical to an already evaluated point should not produce a redundant objective evaluation.

Define a numerical point-equality relation:

$$
x\approx_\tau x_i
$$

when:

$$
\|x-x_i\|_2\leq\tau_x.
$$

Candidates satisfying this relation with an existing evaluation may be rejected or replaced by another feasible candidate.

### 33.5 Coverage-Driven Candidate Generation

Sampling should respond to spatially unresolved areas.

A candidate can be generated in a location where the available evaluations provide weak coverage.

In multiple dimensions, this can be described through the distance to existing evaluations:

$$
d_{\min}(x;D_R)
=
\min_{x_i\in D_R}
\|x-x_i\|_2.
$$

A larger value indicates that the candidate lies farther from the currently observed points.

This quantity is a geometric information signal.

It is not a probability of improvement.

### 33.6 Geometric Exploration Candidate

A deterministic exploration candidate can therefore be selected by maximizing a coverage-related quantity:

$$
x_{\mathrm{coverage}}
\in
\operatorname*{arg\,max}_{x\in R}
d_{\min}(x;D_R),
$$

subject to numerical and candidate-generation constraints.

In practice, this continuous optimization problem may be approximated by a deterministic finite candidate set.

The approximation must be explicitly defined and must not be presented as an exact maximization unless it is solved exactly.

### 33.7 Behavior-Driven Candidate Generation

Sampling may also target areas where the objective exhibits unresolved local behavior.

Suppose nearby evaluations provide substantially different objective values.

This may indicate local variation.

A behavior-oriented candidate may therefore be generated near regions where observed objective differences are informative.

The corresponding signal can be represented conceptually as:

$$
B_{\mathrm{sample}}(x;D_R).
$$

This is an empirical information signal.

It does not constitute a derivative or a proof of local optimality.

### 33.8 Potential-Aware Candidate Generation

In Certified Mode, candidate generation may additionally consider the certified upper envelope:

$$
U_R(x)
=
\min_i
\left[
f(x_i)+L\|x-x_i\|_2
\right].
$$

A candidate may be informative when it lies in a region where the current certified potential remains high.

Thus:

$$
U_R(x)
$$

can contribute to the sampling profile.

The potential itself should not be treated as a probability of finding the optimum.

### 33.9 Candidate Information Profile

The sampling information profile can be represented as:

$$
I_{\mathrm{sample}}(x;R)
=
\left(
I_{\mathrm{coverage}},
I_{\mathrm{behavior}},
I_{\mathrm{potential}},
I_{\mathrm{resolution}}
\right).
$$

Each component represents a distinct information criterion.

The components should retain their original meaning rather than being collapsed into an arbitrary weighted scalar.

### 33.10 Candidate Feasibility

A sampling candidate is feasible only if:

$$
x\in R,
$$

and:

$$
x\in\Omega.
$$

Additionally, it must satisfy numerical validity conditions.

A candidate should be rejected if:

- it contains NaN or infinite coordinates;
- it lies outside the region;
- it lies outside the global domain;
- it duplicates an existing evaluation within the configured tolerance;
- its generation violates an explicit numerical constraint.

### 33.11 Deterministic Candidate Generation

ARRGO remains deterministic.

Therefore, candidate generation must not rely on an uncontrolled random process.

For identical:

$$
(R,D_R,G_t,\Theta),
$$

candidate generation must produce the same candidate set.

Deterministic candidate generation may use:

- region center;
- coordinate-wise interior points;
- deterministic geometric patterns;
- deterministic farthest-point constructions;
- deterministic behavior-driven locations;
- certified-potential-driven locations.

The selected construction must be specified explicitly by the implementation.

### 33.12 Avoiding Full Grid Enumeration

A naive multidimensional extension could generate:

$$
m^d
$$

grid points.

This grows exponentially with dimension.

Such exhaustive enumeration is not consistent with the intended adaptive character of ARRGO.

Therefore, candidate generation should use a finite adaptive candidate set rather than constructing a complete Cartesian grid.

The number of candidates should depend on the regional information state and computational budget rather than automatically growing exponentially with \(d\).

### 33.13 Coordinate-Wise Candidate Generation

A candidate can be constructed by modifying selected coordinates of the region center.

For example:

$$
x
=
c(R)+\delta,
$$

where:

$$
\delta\in\mathbb{R}^d
$$

is a deterministic displacement satisfying:

$$
x\in R.
$$

However, coordinate-wise perturbations must not imply that the objective is separable.

They are only candidate-generation mechanisms.

### 33.14 Joint Candidate Generation

Because:

$$
f=f(x_1,\ldots,x_d)
$$

may contain interactions between coordinates, ARRGO should also support candidates that modify multiple coordinates simultaneously.

Thus, a sampling candidate may satisfy:

$$
x_j\neq c_j(R)
$$

for several or all coordinates.

This allows the evaluation process to observe joint objective behavior.

### 33.15 Sampling Candidate Diversity

A candidate set should avoid unnecessary clustering.

For candidates:

$$
C=
\{x^{(1)},\ldots,x^{(m)}\},
$$

a deterministic diversity criterion can be expressed through pairwise distances:

$$
\|x^{(i)}-x^{(k)}\|_2.
$$

Candidates that are numerically indistinguishable or excessively redundant may be removed.

This improves the information value of the candidate set without introducing randomness.

### 33.16 Sampling vs Structural Refinement

Sampling does not change the region geometry.

If:

$$
x\in R
$$

is evaluated, then the region boundaries remain:

$$
(l_1,\ldots,l_d),
\qquad
(r_1,\ldots,r_d).
$$

Therefore:

$$
\boxed{
\text{Sampling}
\Rightarrow
\text{Information Update}
}
$$

while:

$$
\boxed{
\text{Sampling}
\not\Rightarrow
\text{Geometric Split}.
}
$$

A later action-selection step may decide that additional structural refinement is justified.

### 33.17 Sampling and Global Incumbent

After evaluating:

$$
x_{\mathrm{new}},
$$

the objective value:

$$
f(x_{\mathrm{new}})
$$

is compared with the current incumbent.

The updated incumbent is:

$$
f_{\mathrm{best}}^{(t+1)}
=
\max
\left(
f_{\mathrm{best}}^{(t)},
f(x_{\mathrm{new}})
\right).
$$

Therefore:

$$
f_{\mathrm{best}}^{(t)}
\leq
f_{\mathrm{best}}^{(t+1)}.
$$

Sampling can therefore directly improve the global optimization state.

### 33.18 Sampling and Certified Bounds

In Certified Mode, adding a valid evaluation:

$$
(x_{\mathrm{new}},f(x_{\mathrm{new}}))
$$

adds another Lipschitz cone to the regional bound construction.

The upper envelope becomes:

$$
U_R^{(t+1)}(x)
=
\min
\left(
U_R^{(t)}(x),
f(x_{\mathrm{new}})
+
L\|x-x_{\mathrm{new}}\|_2
\right).
$$

Therefore:

$$
U_R^{(t+1)}(x)
\leq
U_R^{(t)}(x).
$$

Similarly:

$$
L_R^{(t+1)}(x)
\geq
L_R^{(t)}(x).
$$

Thus, valid sampling cannot widen the certified pointwise interval.

### 33.19 Empty Region Sampling

If:

$$
D_R=\varnothing,
$$

there is no existing regional information.

The first sampling action should therefore prioritize acquiring a valid evaluation inside the region.

The implementation must not attempt to calculate distance-based coverage measures that require a non-empty evaluation set without explicitly handling the empty case.

### 33.20 Sampling Progress

A successful sampling action should create measurable information progress.

Possible progress indicators include:

- a new unique evaluation;
- increased spatial coverage;
- reduced local geometric uncertainty;
- improved behavior resolution;
- reduced certified uncertainty;
- improved incumbent.

If a generated candidate repeatedly fails to provide a valid new evaluation, candidate generation must move to another candidate or terminate the action selection process.

### 33.21 Sampling Under a Limited Budget

The sampling process must respect:

$$
N_t<N_{\max}.
$$

If:

$$
N_t=N_{\max},
$$

no additional objective evaluation may be performed.

The algorithm must then terminate or select a non-evaluation structural action if such an action is valid and meaningful under the execution policy.

### 33.22 One-Dimensional Reduction

When:

$$
d=1,
$$

the sampling point becomes:

$$
x\in[l,r].
$$

The center becomes:

$$
c=\frac{l+r}{2},
$$

and Euclidean distance reduces to:

$$
\|x-x_i\|_2=|x-x_i|.
$$

Therefore, the generalized sampling framework naturally contains the original one-dimensional formulation.

### 33.23 Sampling Generalization Principle

The multidimensional sampling mechanism can be summarized as:

$$
\boxed{
\begin{aligned}
&\text{Represent Region as Hyper-Rectangle}
\\
&\Downarrow
\\
&\text{Analyze Existing Evaluations}
\\
&\Downarrow
\\
&\text{Identify Unresolved Spatial/Behavioral Information}
\\
&\Downarrow
\\
&\text{Generate Deterministic Feasible Candidates}
\\
&\Downarrow
\\
&\text{Evaluate Candidate Information Profiles}
\\
&\Downarrow
\\
&\text{Remove Redundant Candidates}
\\
&\Downarrow
\\
&\text{Select Candidate}
\\
&\Downarrow
\\
&\text{Acquire New Objective Information}
\\
&\Downarrow
\\
&\text{Update Regional and Global State}
\end{aligned}
}
$$

The purpose of multidimensional sampling is therefore not exhaustive exploration.

It is the controlled acquisition of new objective information in spatially and behaviorally unresolved parts of the search domain.

This preserves ARRGO's information-driven character while allowing the same framework to operate in arbitrary finite dimensions.

## 34. Multidimensional Global Region Selection

In a multidimensional search space, ARRGO may maintain a large collection of regions at different spatial scales and with different information states.

The algorithm must therefore determine which region should receive the next refinement action.

This decision is distinct from local action selection.

The global stage answers:

$$
\boxed{
\text{Which region should be examined next?}
}
$$

while the local stage answers:

$$
\boxed{
\text{What should be done inside that region?}
}
$$

This separation is essential for preserving the global optimization character of ARRGO.

### 34.1 Region Selection Domain

Let the current region hierarchy be:

$$
\mathcal{H}_t.
$$

The set of regions currently eligible for refinement is:

$$
\mathcal{R}_{\mathrm{eligible}}^{(t)}
\subseteq
\mathcal{H}_t.
$$

Eligibility depends on:

- region validity;
- numerical resolution;
- lifecycle state;
- availability of feasible actions;
- global execution state.

A region may remain stored in the hierarchy without being currently eligible for refinement.

### 34.2 Global Relevance

A region is globally relevant when it may contain information that can affect the current optimization state.

Relevant information may include:

- high observed objective values;
- unresolved spatial coverage;
- unresolved local behavior;
- high certified potential;
- high certified uncertainty;
- insufficient geometric resolution.

Global relevance therefore cannot be represented solely by distance to the current incumbent.

### 34.3 Observed Objective Relevance

Let:

$$
f_{\mathrm{best}}
$$

be the current incumbent value.

For a region \(R\), define its observed regional best as:

$$
f_{\mathrm{best}}(R)
=
\max_{e_i\in D_R}f(x_i),
$$

when:

$$
D_R\neq\varnothing.
$$

A region containing high-valued evaluations may deserve further attention because it contains evidence of promising objective behavior.

However, observed value alone cannot determine global relevance because an unexplored region may contain the global optimum.

### 34.4 Certified Potential Relevance

In Certified Mode, a region may have a valid upper potential:

$$
P(R)
=
\sup_{x\in R}U_R(x).
$$

A region with a high potential remains globally relevant because its unknown objective values may still exceed the incumbent.

In particular, if:

$$
P(R)>f_{\mathrm{best}},
$$

then the current certificate does not rule out improvement inside the region.

If:

$$
P(R)\leq f_{\mathrm{best}},
$$

the region is currently dominated by the incumbent under the certified assumptions.

Nevertheless, ARRGO does not delete the region.

### 34.5 Information Relevance

A region may remain important even when its current objective values are not high.

For example, insufficient spatial coverage may indicate that the region is poorly observed.

Define a regional information state:

$$
Q_R
=
\left(
Q_{\mathrm{coverage}},
Q_{\mathrm{behavior}},
Q_{\mathrm{uncertainty}},
Q_{\mathrm{potential}}
\right).
$$

Global selection may therefore consider whether a region contains unresolved information that is important for maintaining the algorithm's global refinement guarantees.

### 34.6 Geometric Resolution

For:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j],
$$

define:

$$
w_j(R)=r_j-l_j.
$$

The maximum side width is:

$$
w_{\max}(R)
=
\max_j w_j(R).
$$

The diameter is:

$$
\operatorname{diam}(R)
=
\sqrt{
\sum_{j=1}^{d}w_j(R)^2
}.
$$

Large unresolved dimensions may increase the structural relevance of a region.

However, geometric size alone does not determine optimization relevance.

### 34.7 Region Priority Profile

The global region state can therefore be represented by a structured priority profile:

$$
\Pi(R)
=
\left(
Q_{\mathrm{information}}(R),
Q_{\mathrm{global}}(R),
Q_{\mathrm{geometry}}(R)
\right).
$$

The information component may contain:

$$
Q_{\mathrm{information}}(R)
=
\left(
Q_{\mathrm{coverage}},
Q_{\mathrm{behavior}},
Q_{\mathrm{uncertainty}}
\right).
$$

The global component may contain observed objective relevance and, in Certified Mode, optimization potential.

The geometric component may contain:

$$
Q_{\mathrm{geometry}}(R)
=
\left(
w_{\max}(R),
\operatorname{diam}(R)
\right).
$$

These are structured criteria rather than an arbitrary weighted sum.

### 34.8 No Arbitrary Global Score

ARRGO should not require a fixed scalar priority such as:

$$
S(R)
=
\alpha_1Q_1(R)
+
\alpha_2Q_2(R)
+
\cdots
$$

unless those weights are mathematically justified and explicitly part of the algorithm definition.

Such arbitrary weighting could make the behavior dependent on undocumented parameter choices.

Instead, ARRGO can use:

- lexicographic rules;
- Pareto/non-dominance;
- deterministic hierarchical comparisons;
- explicit certified conditions.

### 34.9 Global Candidate Set

The set of candidate regions is:

$$
\mathcal{C}_{R}^{(t)}
=
\mathcal{R}_{\mathrm{eligible}}^{(t)}.
$$

The selected region is:

$$
R^\ast
=
\operatorname{SelectRegion}
\left(
\mathcal{C}_{R}^{(t)}
\right).
$$

The selection procedure must be deterministic.

### 34.10 Certified Priority

In Certified Mode, a particularly important ordering signal is the certified potential.

For regions with valid potentials:

$$
P(R_1)>P(R_2),
$$

region \(R_1\) has stronger current evidence of possible global relevance than \(R_2\).

A potential-based ordering is therefore naturally connected to the global upper-bound framework.

However, potential ordering alone may neglect regions requiring information refinement.

Therefore, certified potential should be combined with the unresolved-information policy rather than automatically replacing it.

### 34.11 Empirical Priority

In Empirical Mode, no rigorous upper bound exists.

The algorithm therefore relies on observable information such as:

- regional objective values;
- spatial coverage;
- behavioral variation;
- geometric resolution;
- candidate information profiles.

The resulting priority is an empirical refinement policy.

It must not be interpreted as a mathematical probability that the region contains the global optimum.

### 34.12 Global Fairness

A global selection rule must preserve the fairness assumption required by the convergence argument.

For every globally relevant region that remains unresolved, the selection process must prevent permanent starvation.

Conceptually:

$$
R\text{ relevant and unresolved}
\Rightarrow
R\text{ is selected eventually},
$$

subject to the defined fairness assumptions.

This does not require every region to receive equal computational effort.

It requires that relevant unresolved regions cannot be ignored forever.

### 34.13 Interaction With Directional Fairness

Global fairness and directional fairness are different.

Global fairness concerns:

$$
\text{Which region receives attention?}
$$

Directional fairness concerns:

$$
\text{Which coordinate direction receives structural refinement?}
$$

The complete multidimensional convergence mechanism therefore requires both:

$$
\boxed{
\text{Global Fairness}
+
\text{Directional Fairness}
}
$$

for the corresponding unresolved regions.

### 34.14 Region Selection and No-Pruning

Suppose:

$$
P(R)\leq f_{\mathrm{best}}.
$$

The region may currently have no potential to improve the incumbent under the certified model.

ARRGO nevertheless retains:

$$
R\in\mathcal{H}_t.
$$

The region may simply become temporarily ineligible for active refinement.

Therefore:

$$
\boxed{
\text{Low Priority}
\neq
\text{Deleted}.
}
$$

### 34.15 Parent and Child Regions

Because parents remain in the hierarchy, multiple structural scales may coexist:

$$
R_{\mathrm{parent}},
R_{\mathrm{child}},
R_{\mathrm{grandchild}},
\ldots
$$

Global selection must therefore distinguish structural history from the current refinement frontier.

A parent region may remain stored for provenance while its descendants represent the current geometric refinement.

This distinction becomes particularly important for certified global potential calculations.

### 34.16 Certification Cover

For certification, the region collection used to construct:

$$
P_{\mathrm{global}}
$$

must provide valid coverage of the complete domain.

The structural hierarchy itself may contain overlapping historical regions.

Therefore, the implementation should distinguish:

$$
\boxed{
\text{Structural Hierarchy}
}
$$

from:

$$
\boxed{
\text{Certification Cover}.
}
$$

The hierarchy preserves history and provenance.

The certification cover is the collection of regions whose union covers the domain for the purpose of establishing:

$$
f^\ast\leq P_{\mathrm{global}}.
$$

This prevents parent retention from accidentally invalidating or unnecessarily weakening the interpretation of the global certificate.

### 34.17 Frontier Representation

A natural certification cover is a collection of current frontier regions:

$$
\mathcal{F}_t
\subseteq
\mathcal{H}_t,
$$

such that:

$$
\Omega
\subseteq
\bigcup_{R\in\mathcal{F}_t}R.
$$

The frontier should contain regions that collectively represent the current unresolved spatial decomposition.

Historical parent regions remain in:

$$
\mathcal{H}_t
$$

but need not independently participate in the certification cover when their descendants already provide the required coverage.

This preserves both:

- no-pruning of structural history;
- a clean global certification representation.

### 34.18 Global Potential Over the Frontier

When a valid certification frontier exists, define:

$$
P_{\mathrm{global}}
=
\sup_{R\in\mathcal{F}_t}P(R).
$$

Since:

$$
\Omega
\subseteq
\bigcup_{R\in\mathcal{F}_t}R,
$$

we obtain:

$$
f^\ast
\leq
P_{\mathrm{global}}.
$$

Therefore:

$$
\boxed{
f_{\mathrm{best}}
\leq
f^\ast
\leq
P_{\mathrm{global}}.
}
$$

This formulation separates the structural hierarchy from the certification cover.

### 34.19 Global Selection Pipeline

The global selection mechanism can therefore be summarized as:

$$
\boxed{
\begin{aligned}
&\text{Construct Eligible Region Set}
\\
&\Downarrow
\\
&\text{Evaluate Global Relevance}
\\
&\Downarrow
\\
&\text{Evaluate Unresolved Information}
\\
&\Downarrow
\\
&\text{Apply Certified Potential When Available}
\\
&\Downarrow
\\
&\text{Preserve Global Fairness}
\\
&\Downarrow
\\
&\text{Apply Deterministic Tie-Breaking}
\\
&\Downarrow
\\
&\text{Select Region}
\end{aligned}
}
$$

### 34.20 Separation of Decisions

The complete decision process is therefore:

$$
\boxed{
\text{Global Region Selection}
\rightarrow
\text{Local Action Selection}
\rightarrow
\text{Direction Selection}
\rightarrow
\text{Split Location Selection}
}
$$

or, for sampling:

$$
\boxed{
\text{Global Region Selection}
\rightarrow
\text{Local Action Selection}
\rightarrow
\text{Sampling Candidate Selection}.
}
$$

Each level solves a different problem.

### 34.21 One-Dimensional Reduction

For:

$$
d=1,
$$

regions become intervals:

$$
R=[l,r].
$$

There is only one possible structural direction.

All geometric quantities reduce to their one-dimensional forms.

Therefore, the generalized global-selection framework naturally contains the original one-dimensional region-selection process.

### 34.22 Generalized Global Selection Principle

The multidimensional global selection mechanism can be summarized as:

$$
\boxed{
\text{Global Relevance}
+
\text{Unresolved Information}
+
\text{Geometric Resolution}
+
\text{Certified Potential When Available}
}
$$

subject to:

$$
\boxed{
\text{Determinism}
+
\text{Global Fairness}
+
\text{No Pruning}.
}
$$

This provides the global decision layer required for extending ARRGO from one-dimensional intervals to arbitrary finite collections of multidimensional hyper-rectangular regions.

## 35. Multidimensional Region Analysis

Region analysis converts the evaluations associated with a region into a structured representation of the information currently available inside that region.

In one dimension, regional analysis can rely heavily on ordered points along a line.

In multiple dimensions, the same principle must be extended to spatial, directional, and geometric relationships.

The purpose of regional analysis is not to estimate the objective perfectly.

Its purpose is to identify which aspects of the region remain unresolved and therefore which refinement actions may be justified.

### 35.1 Regional Information State

For a region:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j],
$$

define its regional evaluation set:

$$
D_R
=
\{e_i\in D_t:x_i\in R\}.
$$

The regional information state is represented as:

$$
Q_R
=
\left(
Q_{\mathrm{coverage}},
Q_{\mathrm{behavior}},
Q_{\mathrm{uncertainty}},
Q_{\mathrm{potential}}
\right).
$$

The components are maintained separately.

No arbitrary weighted combination is required.

### 35.2 Empty Region

If:

$$
D_R=\varnothing,
$$

the region contains no direct objective information.

Such a region is therefore unresolved.

The analysis must explicitly recognize this state rather than assigning artificial behavioral or uncertainty values.

An empty region may require an initial sampling action.

### 35.3 Regional Sample Count

The number of evaluations associated with the region is:

$$
n_R
=
|D_R|.
$$

This quantity describes the amount of acquired information.

However:

$$
\boxed{
n_R\text{ large}
\not\Rightarrow
R\text{ well resolved}.
}
$$

Multiple evaluations may be highly clustered.

Therefore, sample count must be interpreted together with spatial coverage.

### 35.4 Geometric Resolution

For each direction:

$$
w_j(R)
=
r_j-l_j.
$$

The width vector is:

$$
W(R)
=
(w_1,\ldots,w_d).
$$

Two useful scalar summaries are:

$$
w_{\max}(R)
=
\max_j w_j(R),
$$

and:

$$
\operatorname{diam}(R)
=
\sqrt{
\sum_{j=1}^{d}w_j(R)^2
}.
$$

These quantities describe the current geometric scale of the region.

They do not directly measure objective resolution.

### 35.5 Directional Coverage

For coordinate \(j\), define the observed coordinate set:

$$
X_{R,j}
=
\{x_{i,j}:x_i\in R\}.
$$

After ordering these values and including the regional boundaries, define:

$$
g_j(R)
=
\max_k
\left(
z_{k+1,j}-z_{k,j}
\right).
$$

The normalized directional gap is:

$$
\gamma_j(R)
=
\frac{g_j(R)}{w_j(R)},
$$

for:

$$
w_j(R)>0.
$$

The coverage profile is:

$$
\Gamma(R)
=
(\gamma_1,\ldots,\gamma_d).
$$

A large value of \(\gamma_j\) indicates weak coverage along direction \(j\).

### 35.6 Geometric Covering Radius

Directional gaps alone cannot completely characterize multidimensional coverage.

A complementary measure is the geometric covering radius:

$$
h(R)
=
\sup_{x\in R}
\min_{x_i\in D_R}
\|x-x_i\|_2.
$$

A large \(h(R)\) indicates that substantial portions of the region remain far from evaluated points.

This is a geometric information measure.

It should not be interpreted as an objective uncertainty bound unless combined with an appropriate mathematical assumption such as Lipschitz continuity.

### 35.7 Regional Objective Range

If:

$$
D_R\neq\varnothing,
$$

the observed objective range can be defined as:

$$
\operatorname{Range}_f(R)
=
\max_{e_i\in D_R}f(x_i)
-
\min_{e_i\in D_R}f(x_i).
$$

A larger observed range indicates stronger variation among the sampled points.

However, it does not establish that the unsampled objective behaves similarly.

### 35.8 Local Behavioral Variation

Observed objective differences can be related to spatial distances.

For two evaluated points:

$$
x_i,x_k\in R,
$$

define:

$$
\Delta x_{ik}
=
\|x_i-x_k\|_2,
$$

and:

$$
\Delta f_{ik}
=
|f(x_i)-f(x_k)|.
$$

A finite-difference variation measure may then be represented as:

$$
v_{ik}
=
\frac{\Delta f_{ik}}{\Delta x_{ik}},
$$

when:

$$
\Delta x_{ik}>0.
$$

This is an observed variation ratio.

It should not automatically be called a derivative.

### 35.9 Directional Behavioral Variation

To identify possible directional structure, point pairs can additionally be examined according to their displacement vector:

$$
\delta_{ik}
=
x_k-x_i.
$$

The contribution of the displacement to coordinate \(j\) is:

$$
|\delta_{ik,j}|.
$$

This allows the analysis to distinguish observations that differ primarily along particular coordinate directions.

A directional behavior signal can therefore be represented as:

$$
B_R^{(j)}
=
\operatorname{BehaviorVariation}(D_R,j).
$$

The exact implementation must define how pairs are selected and how the resulting measurements are aggregated.

### 35.10 Joint Behavioral Structure

Coordinate-wise analysis alone may miss interactions between variables.

For example, an objective may remain nearly constant when changing \(x_1\) alone or \(x_2\) alone while changing substantially when both coordinates change together.

Therefore, ARRGO must preserve the possibility of joint behavioral analysis.

Conceptually:

$$
B_R
=
\left(
B_R^{(1)},\ldots,B_R^{(d)},
B_R^{(\mathrm{joint})}
\right).
$$

The joint component represents information that cannot be attributed reliably to a single coordinate.

### 35.11 Avoiding Separability Assumptions

The multidimensional formulation must not assume:

$$
f(x_1,\ldots,x_d)
=
\sum_{j=1}^{d}f_j(x_j).
$$

Such an assumption would convert the problem into independent one-dimensional subproblems.

ARRGO instead treats the objective as a single joint black-box function:

$$
f:\mathbb{R}^d\rightarrow\mathbb{R}.
$$

Coordinate-wise analysis is therefore an information mechanism, not a decomposition of the optimization problem.

### 35.12 Certified Uncertainty

In Certified Mode, valid regional bounds provide:

$$
L_R(x)
\leq
f(x)
\leq
U_R(x).
$$

The pointwise uncertainty is:

$$
u_R(x)
=
U_R(x)-L_R(x).
$$

The regional uncertainty is:

$$
u_{\max}(R)
=
\sup_{x\in R}u_R(x).
$$

This quantity has a stronger mathematical interpretation than empirical behavioral variation because it follows from the Lipschitz assumption.

### 35.13 Empirical Uncertainty

In Empirical Mode, no rigorous upper and lower envelopes are available solely from finite samples.

Therefore, quantities such as:

- local sample sparsity;
- observed variation;
- distance to nearest evaluation;
- interpolation residuals;

must be interpreted as empirical information signals.

They must not be presented as guaranteed bounds on unseen objective values.

### 35.14 Optimization Potential

In Certified Mode, the regional potential is:

$$
P(R)
=
\sup_{x\in R}U_R(x).
$$

It represents the largest objective value that remains mathematically possible under the certified model.

In Empirical Mode, a corresponding empirical potential may be used as a heuristic information signal, but it does not have the same mathematical guarantee.

The two concepts must therefore remain explicitly distinguished.

### 35.15 Unresolved Information Profile

The complete unresolved profile can be written as:

$$
Q_R
=
\left(
Q_{\mathrm{coverage}},
Q_{\mathrm{behavior}},
Q_{\mathrm{uncertainty}},
Q_{\mathrm{potential}}
\right).
$$

A region is not considered resolved merely because one component becomes small.

For example:

$$
Q_{\mathrm{coverage}}\text{ small}
$$

does not imply:

$$
Q_{\mathrm{behavior}}\text{ small}.
$$

Similarly:

$$
Q_{\mathrm{behavior}}\text{ small}
$$

does not automatically imply that the certified uncertainty is small.

### 35.16 Resolution State

The regional analysis determines whether the available information is sufficiently resolved for the current decision policy.

Conceptually:

$$
\operatorname{State}(R)
\in
\{
\mathrm{Unresolved},
\mathrm{Resolved},
\mathrm{NumericallyLimited}
\}.
$$

This information state is distinct from the structural lifecycle state:

$$
\{
\mathrm{ACTIVE},
\mathrm{STABLE},
\mathrm{REFINED}
\}.
$$

The structural state describes the region's position in the hierarchy.

The resolution state describes the quality of the current information.

### 35.17 Information Acquisition Need

A region has an information-acquisition need when one or more important information dimensions remain unresolved.

For example:

$$
Q_{\mathrm{coverage}}
\text{ unresolved}
$$

may justify sampling.

Similarly:

$$
Q_{\mathrm{behavior}}
\text{ unresolved}
$$

may justify additional sampling or structural refinement.

A high structural scale combined with unresolved behavior may justify splitting.

### 35.18 Analysis and Action Selection

Region analysis does not directly execute an action.

Instead:

$$
\boxed{
\text{Region Analysis}
\rightarrow
\text{Information State}
\rightarrow
\text{Action Selection}
}
$$

This separation prevents the analysis layer from becoming tightly coupled to execution logic.

### 35.19 Analysis and Global Selection

Regional analysis is also distinct from global region selection.

The analysis answers:

$$
\text{What is unresolved inside }R?
$$

Global selection answers:

$$
\text{Which }R\text{ should receive attention next?}
$$

Therefore:

$$
\boxed{
\text{Local Analysis}
\neq
\text{Global Selection}.
}
$$

### 35.20 Deterministic Analysis

For identical:

$$
(D_R,R,\Theta),
$$

regional analysis must return identical information.

Any ordering, aggregation, candidate construction, or comparison operation must therefore use deterministic rules.

Numerical tolerances must be applied consistently.

### 35.21 Information Sharing

Because evaluations are globally stored, the same evaluation may contribute to the analysis of multiple regions when the region representation requires it.

However, the objective evaluation itself is still stored exactly once.

Thus:

$$
\boxed{
\text{Information Reuse}
\neq
\text{Evaluation Duplication}.
}
$$

### 35.22 Parent and Child Analysis

When a region is split:

$$
R
\rightarrow
\{R_1,R_2\},
$$

the children inherit access to evaluations located inside them.

The parent retains its historical information.

The child analyses are therefore derived from the global evaluation history and their geometric membership rather than from duplicated evaluation ownership.

### 35.23 Analysis After Sampling

After a new evaluation:

$$
e_{t+1}
=
(x_{t+1},f(x_{t+1}),\operatorname{Id}_{t+1}),
$$

the affected regional information states are recomputed.

Possible changes include:

- improved directional coverage;
- changed observed behavior;
- reduced empirical sparsity;
- reduced certified uncertainty;
- changed regional potential;
- improved global incumbent.

### 35.24 Analysis After Splitting

Splitting changes the geometric representation:

$$
R
\rightarrow
\{R_1,R_2\}.
$$

No new objective value is created by the split.

Therefore, the children initially derive their information from existing evaluations.

The primary immediate effect is improved spatial localization.

### 35.25 Numerical Robustness

Multidimensional analysis requires explicit numerical handling for:

- vector equality;
- near-boundary points;
- zero-width dimensions;
- very small region diameters;
- duplicate evaluations;
- near-zero pairwise distances;
- floating-point ordering;
- NaN and infinite objective values.

A denominator such as:

$$
\|x_i-x_k\|_2
$$

must never be used without handling the zero-distance case.

### 35.26 Complexity Considerations

A naive pairwise behavioral analysis over \(n_R\) evaluations requires:

$$
O(n_R^2)
$$

pair comparisons.

As dimensionality and evaluation count increase, this can become expensive.

Therefore, the future implementation may use deterministic bounded subsets, neighborhood structures, or incremental statistics.

Any approximation must be explicitly identified as an implementation approximation rather than an exact mathematical quantity.

### 35.27 One-Dimensional Reduction

For:

$$
d=1,
$$

the width vector contains one element:

$$
W(R)=(r-l),
$$

and:

$$
\operatorname{diam}(R)=r-l.
$$

The Euclidean distance becomes:

$$
\|x_i-x_k\|_2
=
|x_i-x_k|.
$$

Directional coverage becomes ordinary interval coverage.

Directional behavior becomes scalar local behavior.

Thus, the one-dimensional analysis is recovered as a special case.

### 35.28 Generalized Region-Analysis Principle

The complete multidimensional analysis process is:

$$
\boxed{
\begin{aligned}
&\text{Collect Regional Evaluations}
\\
&\Downarrow
\\
&\text{Measure Geometric Resolution}
\\
&\Downarrow
\\
&\text{Measure Spatial Coverage}
\\
&\Downarrow
\\
&\text{Analyze Observed Objective Behavior}
\\
&\Downarrow
\\
&\text{Analyze Directional and Joint Structure}
\\
&\Downarrow
\\
&\text{Compute Certified Quantities When Available}
\\
&\Downarrow
\\
&\text{Construct Unresolved Information Profile}
\\
&\Downarrow
\\
&\text{Determine Resolution State}
\\
&\Downarrow
\\
&\text{Pass Information to Action Selection}
\end{aligned}
}
$$

The region-analysis layer therefore acts as the information engine of multidimensional ARRGO.

It does not directly optimize the objective or execute refinement.

Instead, it converts the currently available evaluations into structured evidence that guides the subsequent global and local refinement decisions.

## 36. Multidimensional State Transition Model

ARRGO can be represented as a deterministic state-transition system.

At every iteration, the complete algorithm state is transformed into a new state through a valid refinement action.

Let the global state at iteration \(t\) be:

$$
G_t.
$$

The next state is:

$$
G_{t+1}
=
T(G_t,A_t),
$$

where:

$$
A_t
$$

is the selected valid action and \(T\) is the deterministic state-transition operator.

### 36.1 Global State

The generalized global state contains:

$$
G_t
=
\left(
D_t,
\mathcal{H}_t,
\mathcal{F}_t,
x_{\mathrm{best}}^{(t)},
f_{\mathrm{best}}^{(t)},
N_t,
C_t,
\mathcal{S}_t
\right),
$$

where:

- \(D_t\) is the global evaluation history;
- \(\mathcal{H}_t\) is the complete structural hierarchy;
- \(\mathcal{F}_t\) is the current certification/refinement frontier when applicable;
- \(x_{\mathrm{best}}^{(t)}\) is the incumbent location;
- \(f_{\mathrm{best}}^{(t)}\) is the incumbent value;
- \(N_t\) is the number of objective evaluations;
- \(C_t\) contains current regional information;
- \(\mathcal{S}_t\) contains execution and termination state.

The exact software representation may differ, but these conceptual components must remain represented.

### 36.2 State Transition Inputs

A valid transition depends on:

$$
(G_t,\Theta),
$$

where \(\Theta\) contains the fixed algorithm configuration.

The objective function is also part of the execution environment:

$$
f:\Omega\rightarrow\mathbb{R}.
$$

Thus, for deterministic execution:

$$
G_{t+1}
=
T(G_t,\Theta,f).
$$

### 36.3 Analysis Transition

The first stage of a refinement cycle is regional analysis.

For eligible regions:

$$
R\in\mathcal{R}_{\mathrm{eligible}}^{(t)},
$$

ARRGO computes:

$$
Q_R^{(t)}
=
\operatorname{Analyze}(R,D_t,\Theta).
$$

The global state is therefore enriched with current information about each relevant region.

### 36.4 Global Selection Transition

The algorithm then selects a region:

$$
R_t^\ast
=
\operatorname{SelectRegion}
\left(
\mathcal{R}_{\mathrm{eligible}}^{(t)},
Q^{(t)},
G_t
\right).
$$

This operation does not modify the evaluation history or hierarchy.

It only determines where the next local decision will occur.

### 36.5 Local Action Transition

For the selected region:

$$
R_t^\ast,
$$

ARRGO constructs feasible actions:

$$
\mathcal{A}_t
=
\operatorname{GenerateActions}
(R_t^\ast,G_t).
$$

The selected action is:

$$
A_t
=
\operatorname{SelectAction}
(\mathcal{A}_t,Q_{R_t^\ast}^{(t)},G_t).
$$

The action must satisfy:

$$
A_t\in\mathcal{A}_t.
$$

### 36.6 Sampling Transition

If:

$$
A_t
=
(\mathrm{Sample},x_t),
$$

then:

$$
x_t\in R_t^\ast.
$$

The objective is evaluated:

$$
y_t=f(x_t).
$$

The evaluation history becomes:

$$
D_{t+1}
=
D_t
\cup
\{(x_t,y_t,\operatorname{Id}_t)\}.
$$

The evaluation count changes:

$$
N_{t+1}=N_t+1.
$$

The incumbent becomes:

$$
f_{\mathrm{best}}^{(t+1)}
=
\max
\left(
f_{\mathrm{best}}^{(t)},
y_t
\right).
$$

### 36.7 Split Transition

If:

$$
A_t
=
(\mathrm{Split},j_t,s_{j_t}),
$$

then the selected region is partitioned along coordinate \(j_t\).

The children satisfy:

$$
R_1\cup R_2
=
R_t^\ast
$$

up to the explicitly defined boundary-ownership convention.

The hierarchy is updated:

$$
\mathcal{H}_{t+1}
=
\mathcal{H}_t
\cup
\{R_1,R_2\}.
$$

The parent remains stored.

No objective evaluation is automatically created:

$$
D_{t+1}=D_t,
$$

and therefore:

$$
N_{t+1}=N_t.
$$

### 36.8 Stable Transition

If:

$$
A_t=\mathrm{Stable},
$$

the region is marked as currently stable.

Its structural representation remains in the hierarchy:

$$
R_t^\ast\in\mathcal{H}_{t+1}.
$$

A stable transition must nevertheless produce a meaningful state change or lead to a valid termination decision.

### 36.9 Incumbent Transition

For every state transition:

$$
f_{\mathrm{best}}^{(t+1)}
\geq
f_{\mathrm{best}}^{(t)}.
$$

Therefore:

$$
\boxed{
\{f_{\mathrm{best}}^{(t)}\}_{t\geq0}
\text{ is monotone non-decreasing}.
}
$$

The incumbent location must always correspond to an evaluation in the global history.

### 36.10 Hierarchy Transition

The hierarchy evolves only through valid structural operations.

For a split:

$$
R
\rightarrow
\{R_1,R_2\}.
$$

For sampling:

$$
\mathcal{H}_{t+1}
=
\mathcal{H}_t.
$$

Therefore, sampling and splitting have explicitly different state effects.

### 36.11 Frontier Transition

The structural hierarchy and current refinement frontier are distinct.

The hierarchy preserves all historical regions:

$$
\mathcal{H}_t.
$$

The frontier contains the current spatial representation used for active refinement and, when required, certification:

$$
\mathcal{F}_t
\subseteq
\mathcal{H}_t.
$$

After a split:

$$
R\in\mathcal{F}_t
$$

may be replaced in the frontier by its children:

$$
R_1,R_2\in\mathcal{F}_{t+1}.
$$

The parent remains in:

$$
\mathcal{H}_{t+1}.
$$

This mechanism preserves no-pruning while preventing historical parents from being confused with the current spatial cover.

### 36.12 Certified Transition

In Certified Mode, every relevant state transition updates the valid regional bounds.

The current regional upper envelope is:

$$
U_R^{(t)}(x)
=
\min_{x_i\in D_R^{(t)}}
\left[
f(x_i)+L\|x-x_i\|_2
\right].
$$

The regional potential is:

$$
P_t(R)
=
\sup_{x\in R}U_R^{(t)}(x).
$$

The global potential is computed over a valid covering frontier:

$$
P_{\mathrm{global}}^{(t)}
=
\sup_{R\in\mathcal{F}_t}P_t(R).
$$

The certified gap is:

$$
\Delta_{\mathrm{global}}^{(t)}
=
P_{\mathrm{global}}^{(t)}
-
f_{\mathrm{best}}^{(t)}.
$$

### 36.13 Certified Monotonicity

Additional evaluations inside a fixed region can only tighten its valid envelopes:

$$
U_R^{(t+1)}(x)
\leq
U_R^{(t)}(x),
$$

and:

$$
L_R^{(t+1)}(x)
\geq
L_R^{(t)}(x).
$$

However, after structural splitting, the frontier changes.

Therefore, global potential monotonicity must be established from the validity of the new covering representation rather than assumed solely from the local envelope property.

This distinction is important for a mathematically correct implementation.

### 36.14 Termination Transition

After every state update, ARRGO evaluates termination conditions.

If Certified Mode is active:

$$
\Delta_{\mathrm{global}}^{(t)}
\leq
\varepsilon
$$

may trigger certified termination.

If:

$$
N_t\geq N_{\max},
$$

budget termination may occur.

Other terminal states include:

- numerical limit;
- no valid action;
- configuration error.

The termination reason becomes part of the global state.

### 36.15 Non-Terminal Transition

If no termination condition is satisfied:

$$
G_t
\rightarrow
G_{t+1}
\rightarrow
\text{next analysis cycle}.
$$

Thus, the complete iteration is:

$$
\boxed{
G_t
\rightarrow
\text{Analyze}
\rightarrow
\text{Select Region}
\rightarrow
\text{Generate Actions}
\rightarrow
\text{Select Action}
\rightarrow
\text{Execute}
\rightarrow
\text{Update}
\rightarrow
\text{Check Termination}
\rightarrow
G_{t+1}.
}
$$

### 36.16 Deterministic Transition Requirement

For fixed:

$$
G_t,\Theta,f,
$$

the transition must be deterministic:

$$
T(G_t,\Theta,f)
=
G_{t+1}.
$$

If multiple actions have identical information profiles, the implementation must resolve the tie deterministically.

No uncontrolled randomness may determine the next state.

### 36.17 Progress Requirement

A non-terminal transition must produce meaningful progress.

Let:

$$
\mathcal{M}(G_t)
$$

represent the measurable state of the algorithm.

A valid transition must satisfy at least one meaningful progress condition, such as:

$$
N_{t+1}>N_t,
$$

or:

$$
\operatorname{diam}(R_{t+1})
<
\operatorname{diam}(R_t),
$$

or a meaningful information-state improvement.

The exact progress criterion may depend on the selected action.

The purpose is to prevent infinite execution caused by repeated no-op decisions.

### 36.18 Global Fairness Across Transitions

The state-transition system must preserve the fairness requirement.

If a region remains globally relevant and unresolved, the sequence of transitions must not permanently exclude it from consideration.

Conceptually:

$$
R
\text{ relevant and unresolved}
\Rightarrow
R
\text{ receives attention eventually},
$$

subject to the formal fairness assumptions of the convergence theorem.

### 36.19 Directional Fairness Across Transitions

Similarly, for a relevant region, unresolved coordinate directions must not be permanently ignored.

If direction \(j\) remains unresolved and refinement is required, the transition sequence must eventually allow a split along that direction.

Thus:

$$
\text{Global Fairness}
+
\text{Directional Fairness}
$$

together support the spatial contraction requirement:

$$
\operatorname{diam}(R_t)\rightarrow0.
$$

### 36.20 State-Transition Invariants

Every valid transition must preserve:

$$
\boxed{
\begin{aligned}
&x_i\in\Omega
\\
&R\subseteq\Omega
\\
&D_t\text{ contains valid evaluations}
\\
&N_t\leq N_{\max}
\\
&f_{\mathrm{best}}^{(t+1)}
\geq
f_{\mathrm{best}}^{(t)}
\\
&\text{parent regions remain stored}
\\
&\text{valid splits preserve coverage}
\\
&\text{actions remain feasible}
\\
&\text{execution remains deterministic}
\end{aligned}
}
$$

In Certified Mode, additionally:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
\leq
f^\ast
\leq
P_{\mathrm{global}}^{(t)}
}
$$

must remain valid under the stated assumptions.

### 36.21 One-Dimensional Recovery

When:

$$
d=1,
$$

the state-transition model reduces naturally to the original one-dimensional process.

The point becomes scalar, the region becomes an interval, and there is only one possible structural direction.

Therefore:

$$
T_d
\big|_{d=1}
=
T_{1D}.
$$

### 36.22 State-Transition Principle

The complete generalized ARRGO execution can therefore be expressed as:

$$
\boxed{
G_{t+1}
=
T
\left(
G_t,
\operatorname{SelectAction}
\left(
\operatorname{SelectRegion}
\left(
\operatorname{Analyze}(G_t)
\right)
\right)
\right).
}
$$

The state-transition formulation establishes a clean boundary between:

- mathematical state;
- regional information;
- decision logic;
- structural refinement;
- objective evaluation;
- global state update;
- termination.

This boundary will be used directly when designing the dimension-parameterized implementation architecture.

## 37. Formal Multidimensional ARRGO Specification

The previous sections generalized the individual components of ARRGO from one-dimensional intervals to multidimensional hyper-rectangular regions.

This section combines those components into one implementation-independent specification.

The purpose is to define the complete multidimensional ARRGO process before introducing implementation-specific data structures or numerical procedures.

### 37.1 Problem Definition

Let:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j]
\subset
\mathbb{R}^d,
\qquad
d<\infty.
$$

Let the deterministic objective be:

$$
f:\Omega\rightarrow\mathbb{R}.
$$

The optimization problem is:

$$
\max_{x\in\Omega}f(x).
$$

Assume that \(f\) is continuous on the compact domain \(\Omega\).

Therefore, at least one global maximizer exists.

### 37.2 Initial Global State

ARRGO begins with the root region:

$$
R_0=\Omega.
$$

The initial hierarchy is:

$$
\mathcal{H}_0=\{R_0\}.
$$

The initial evaluation history is:

$$
D_0=\varnothing.
$$

The initial evaluation count is:

$$
N_0=0.
$$

No incumbent exists before the first valid evaluation.

### 37.3 Initial Sampling

ARRGO performs a deterministic initial sampling procedure.

Let:

$$
X_0
=
\{x_1,\ldots,x_m\}
\subseteq
R_0.
$$

Each candidate must satisfy:

$$
x_i\in\Omega.
$$

The corresponding objective evaluations are:

$$
e_i
=
(x_i,f(x_i),\operatorname{Id}_i).
$$

The initial history becomes:

$$
D_m
=
\{e_1,\ldots,e_m\}.
$$

The initial incumbent is:

$$
f_{\mathrm{best}}^{(m)}
=
\max_{1\leq i\leq m}f(x_i).
$$

The implementation must ensure that the configured evaluation budget is not exceeded during initialization.

### 37.4 Global Evaluation History

At iteration \(t\), the global evaluation history is:

$$
D_t
=
\{e_1,\ldots,e_{N_t}\}.
$$

Each evaluation is stored exactly once.

Regional information is derived from the global history through spatial membership.

Thus:

$$
D_R^{(t)}
=
\{e_i\in D_t:x_i\in R\}.
$$

### 37.5 Incumbent State

The global incumbent is:

$$
x_{\mathrm{best}}^{(t)}
\in
\operatorname*{arg\,max}_{x_i\in D_t}
f(x_i),
$$

with:

$$
f_{\mathrm{best}}^{(t)}
=
f(x_{\mathrm{best}}^{(t)}).
$$

For every valid new evaluation:

$$
f_{\mathrm{best}}^{(t+1)}
=
\max
\left(
f_{\mathrm{best}}^{(t)},
f(x_{t+1})
\right).
$$

Therefore:

$$
f_{\mathrm{best}}^{(t+1)}
\geq
f_{\mathrm{best}}^{(t)}.
$$

### 37.6 Region Representation

Every region is a hyper-rectangle:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j].
$$

The region must satisfy:

$$
a_j\leq l_j<r_j\leq b_j
$$

for every coordinate \(j\).

Its width vector is:

$$
W(R)
=
(r_1-l_1,\ldots,r_d-l_d).
$$

Its diameter is:

$$
\operatorname{diam}(R)
=
\sqrt{
\sum_{j=1}^{d}(r_j-l_j)^2
}.
$$

### 37.7 Regional Analysis

For each eligible region:

$$
R\in\mathcal{R}_{\mathrm{eligible}}^{(t)},
$$

ARRGO computes its current information state:

$$
Q_R^{(t)}
=
\operatorname{Analyze}
(R,D_t,\Theta).
$$

The analysis may contain:

$$
Q_R^{(t)}
=
\left(
Q_{\mathrm{coverage}},
Q_{\mathrm{behavior}},
Q_{\mathrm{uncertainty}},
Q_{\mathrm{potential}}
\right).
$$

The exact components available depend on the execution mode.

### 37.8 Global Region Selection

ARRGO constructs the set of eligible regions:

$$
\mathcal{R}_{\mathrm{eligible}}^{(t)}.
$$

The next region is selected deterministically:

$$
R_t^\ast
=
\operatorname{SelectRegion}
\left(
\mathcal{R}_{\mathrm{eligible}}^{(t)},
Q^{(t)},
G_t
\right).
$$

The selection must preserve global fairness under the convergence assumptions.

### 37.9 Local Action Selection

For the selected region:

$$
R_t^\ast,
$$

ARRGO constructs feasible actions:

$$
\mathcal{A}(R_t^\ast).
$$

The action set may contain:

$$
\mathrm{Sample},
\qquad
\mathrm{Split},
\qquad
\mathrm{Stable}.
$$

The selected action is:

$$
A_t
=
\operatorname{SelectAction}
\left(
R_t^\ast,
Q_{R_t^\ast}^{(t)},
\mathcal{A}(R_t^\ast),
G_t
\right).
$$

### 37.10 Sampling Action

If:

$$
A_t
=
(\mathrm{Sample},x_t),
$$

then:

$$
x_t\in R_t^\ast.
$$

The objective is evaluated exactly:

$$
y_t=f(x_t).
$$

The history becomes:

$$
D_{t+1}
=
D_t
\cup
\{(x_t,y_t,\operatorname{Id}_t)\}.
$$

The evaluation count becomes:

$$
N_{t+1}=N_t+1.
$$

The region geometry remains unchanged.

### 37.11 Structural Split Action

If:

$$
A_t
=
(\mathrm{Split},j_t,s_{j_t}),
$$

then:

$$
j_t\in\{1,\ldots,d\},
$$

and:

$$
l_{j_t}<s_{j_t}<r_{j_t}.
$$

The parent is divided along coordinate \(j_t\).

The resulting children are:

$$
R_1
=
[l_1,r_1]
\times\cdots\times
[l_{j_t},s_{j_t}]
\times\cdots\times
[l_d,r_d],
$$

and:

$$
R_2
=
[l_1,r_1]
\times\cdots\times
[s_{j_t},r_{j_t}]
\times\cdots\times
[l_d,r_d].
$$

The boundary ownership convention must be explicitly defined by the implementation.

### 37.12 Contraction Requirement

The split must satisfy:

$$
\max
\left(
\frac{s_{j_t}-l_{j_t}}
{r_{j_t}-l_{j_t}},
\frac{r_{j_t}-s_{j_t}}
{r_{j_t}-l_{j_t}}
\right)
\leq
\rho.
$$

Therefore, the selected coordinate is strictly contracted.

Global spatial contraction additionally requires directional fairness across repeated refinements.

### 37.13 Hierarchy Update

After splitting:

$$
\mathcal{H}_{t+1}
=
\mathcal{H}_t
\cup
\{R_1,R_2\}.
$$

The parent remains in the hierarchy.

Therefore:

$$
\boxed{
\text{Split}
\neq
\text{Delete Parent}.
}
$$

The current refinement frontier may replace the parent with its children.

### 37.14 Certification Frontier

When Certified Mode is active, the current certification frontier:

$$
\mathcal{F}_t
$$

must continue to cover the complete domain.

Thus:

$$
\Omega
\subseteq
\bigcup_{R\in\mathcal{F}_t}R.
$$

Historical regions remain in:

$$
\mathcal{H}_t,
$$

but do not need to be independently included in the frontier when their descendants provide the required coverage.

### 37.15 Certified Bounds

Under a valid global Lipschitz constant:

$$
|f(x)-f(y)|
\leq
L\|x-y\|_2,
$$

define:

$$
U_R(x)
=
\min_{x_i\in D_R}
\left[
f(x_i)+L\|x-x_i\|_2
\right],
$$

and:

$$
L_R(x)
=
\max_{x_i\in D_R}
\left[
f(x_i)-L\|x-x_i\|_2
\right].
$$

Then:

$$
L_R(x)
\leq
f(x)
\leq
U_R(x).
$$

The regional potential is:

$$
P(R)
=
\sup_{x\in R}U_R(x).
$$

### 37.16 Global Certified Potential

Using the certification frontier:

$$
P_{\mathrm{global}}^{(t)}
=
\sup_{R\in\mathcal{F}_t}P_t(R).
$$

Because the frontier covers the domain:

$$
f^\ast
\leq
P_{\mathrm{global}}^{(t)}.
$$

Together with the incumbent:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast
\leq
P_{\mathrm{global}}^{(t)}.
$$

The certified gap is:

$$
\Delta_{\mathrm{global}}^{(t)}
=
P_{\mathrm{global}}^{(t)}
-
f_{\mathrm{best}}^{(t)}.
$$

### 37.17 Certified Termination

If:

$$
\Delta_{\mathrm{global}}^{(t)}
\leq
\varepsilon,
$$

then:

$$
f^\ast-f_{\mathrm{best}}^{(t)}
\leq
\varepsilon.
$$

Therefore, the incumbent is epsilon-optimal under the certified assumptions.

### 37.18 Empirical Mode

In Empirical Mode, no mathematical optimality certificate is assumed.

The returned result is:

$$
x_{\mathrm{best}}^{(N_{\max})},
$$

with:

$$
f_{\mathrm{best}}^{(N_{\max})}
=
\max_{x_i\in D_{N_{\max}}}f(x_i).
$$

This is a best-found solution.

It must not be described as a guaranteed global optimum solely from empirical execution.

### 37.19 Termination Conditions

ARRGO may terminate because of:

1. Certified tolerance;
2. Evaluation budget exhaustion;
3. Numerical resolution limit;
4. No valid action;
5. Configuration error.

If:

$$
N_t=N_{\max},
$$

no additional objective evaluation may be performed.

A structural action may only be performed when it is valid, meaningful, and consistent with the execution policy.

### 37.20 Progress Requirement

Every non-terminal iteration must produce meaningful progress.

Possible progress includes:

$$
N_{t+1}>N_t,
$$

or a valid geometric refinement:

$$
\operatorname{diam}(R_{\mathrm{child}})
<
\operatorname{diam}(R_{\mathrm{parent}}),
$$

or a measurable improvement in the information state.

Repeated no-op transitions are not valid execution progress.

### 37.21 Global Fairness

The convergence argument requires that globally relevant unresolved regions are not permanently ignored.

Thus, subject to the formal assumptions:

$$
R
\text{ relevant and unresolved}
\Rightarrow
R
\text{ receives attention eventually}.
$$

This does not require equal computational effort across regions.

It requires the absence of permanent starvation.

### 37.22 Directional Fairness

Similarly, if a coordinate direction remains unresolved in a relevant region, it must eventually be eligible for structural refinement.

For each required coordinate \(j\):

$$
w_j(R_t)\rightarrow0.
$$

Consequently:

$$
\operatorname{diam}(R_t)\rightarrow0.
$$

### 37.23 No-Pruning Invariant

At every iteration:

$$
\mathcal{H}_t
\subseteq
\mathcal{H}_{t+1}.
$$

Structural history is therefore monotone.

Regions may become:

$$
\mathrm{ACTIVE},
\qquad
\mathrm{STABLE},
\qquad
\mathrm{REFINED},
$$

but they are not deleted from the hierarchy.

### 37.24 Determinism Invariant

For fixed:

$$
(f,\Theta,G_t),
$$

all candidate generation, analysis, selection, and state-transition procedures must be deterministic.

Therefore:

$$
G_{t+1}
=
T(G_t,\Theta,f)
$$

is uniquely determined by the current state and configuration.

### 37.25 Global Invariant Set

The complete execution must preserve:

$$
\boxed{
\begin{aligned}
&x_i\in\Omega
\\
&R\subseteq\Omega
\\
&D_t\text{ contains valid evaluations}
\\
&N_t\leq N_{\max}
\\
&f_{\mathrm{best}}^{(t+1)}
\geq
f_{\mathrm{best}}^{(t)}
\\
&\mathcal{H}_t
\subseteq
\mathcal{H}_{t+1}
\\
&\text{valid splits preserve spatial coverage}
\\
&\text{selected actions are feasible}
\\
&\text{execution is deterministic}
\end{aligned}
}
$$

Certified Mode additionally requires:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
\leq
f^\ast
\leq
P_{\mathrm{global}}^{(t)}.
}
$$

### 37.26 Complete Multidimensional ARRGO Cycle

The complete cycle is:

$$
\boxed{
\begin{aligned}
&\text{Initialize Domain}
\\
&\Downarrow
\\
&\text{Create Root Hyper-Rectangle}
\\
&\Downarrow
\\
&\text{Initial Deterministic Sampling}
\\
&\Downarrow
\\
&\text{Initialize Incumbent}
\\
&\Downarrow
\\
&\text{Analyze Eligible Regions}
\\
&\Downarrow
\\
&\text{Select Globally Relevant Region}
\\
&\Downarrow
\\
&\text{Generate Feasible Actions}
\\
&\Downarrow
\\
&\text{Select Local Action}
\\
&\Downarrow
\\
&\begin{cases}
\text{Sample Candidate},\\
\text{or}\\
\text{Split Direction and Location},\\
\text{or}\\
\text{Stable State}
\end{cases}
\\
&\Downarrow
\\
&\text{Update Evaluation History / Hierarchy}
\\
&\Downarrow
\\
&\text{Update Incumbent}
\\
&\Downarrow
\\
&\text{Update Certification Frontier When Applicable}
\\
&\Downarrow
\\
&\text{Compute Certified Gap When Applicable}
\\
&\Downarrow
\\
&\text{Check Termination}
\\
&\Downarrow
\\
&\text{Repeat if Non-Terminal}
\end{aligned}
}
$$

### 37.27 Dimension Independence

The algorithm does not require a separate mathematical algorithm for each dimension.

Instead:

$$
d
$$

is a parameter of the same generalized framework.

Therefore:

$$
\mathrm{ARRGO}_d
$$

uses the same conceptual process for:

$$
d=1,2,3,\ldots
$$

subject to computational feasibility and the stated assumptions.

### 37.28 One-Dimensional Recovery

For:

$$
d=1,
$$

the generalized objects reduce to their original forms:

$$
x\in[a,b],
$$

$$
R=[l,r],
$$

and:

$$
\operatorname{diam}(R)=r-l.
$$

There is only one structural direction.

The multidimensional specification therefore reduces to the previously defined one-dimensional ARRGO process.

### 37.29 Formal Specification Summary

The generalized ARRGO framework can be summarized as:

$$
\boxed{
\begin{aligned}
\Omega
&=
\prod_{j=1}^{d}[a_j,b_j]
\\
R
&=
\prod_{j=1}^{d}[l_j,r_j]
\\
A_R
&\in
\{\mathrm{Sample},\mathrm{Split},\mathrm{Stable}\}
\\
\mathrm{Split}
&=
(j,s_j)
\\
f_{\mathrm{best}}^{(t)}
&=
\max_{e_i\in D_t}f(x_i)
\\
P_{\mathrm{global}}^{(t)}
&=
\sup_{R\in\mathcal{F}_t}P_t(R)
\quad
\text{(Certified Mode)}
\end{aligned}
}
$$

subject to:

$$
\boxed{
\text{Feasibility}
+
\text{Contraction}
+
\text{Information Acquisition}
+
\text{Global Fairness}
+
\text{Directional Fairness}
+
\text{Determinism}
+
\text{No Pruning}.
}
$$

This specification establishes the dimension-independent mathematical and algorithmic structure of ARRGO.

The next stage is to translate these abstractions into a generalized implementation architecture without prematurely committing to the one-dimensional data structures used in the current implementation.

## 38. Multidimensional Implementation Architecture

The mathematical generalization established that ARRGO can be formulated for any finite dimension \(d\).

The next requirement is to ensure that the implementation reflects the generalized mathematical model.

The implementation must therefore be dimension-parameterized rather than based on one-dimensional special cases.

The central design principle is:

$$
\boxed{
\text{One ARRGO Framework}
+
\text{Dimension-Parameterized Geometry}
}
$$

rather than separate implementations for one, two, and three dimensions.

### 38.1 Implementation Layers

The generalized implementation is organized into the following layers:

1. Objective Evaluation
2. Numerical Utilities
3. Evaluation History
4. Domain Representation
5. Region Representation
6. Region Geometry
7. Region Analysis
8. Candidate Generation
9. Action Selection
10. Global Region Selection
11. Hierarchy Management
12. Certified Bounds
13. Termination
14. ARRGO Execution

The separation between these layers is important because dimensionality primarily affects geometry and spatial analysis, while the global optimization state remains conceptually unchanged.

### 38.2 Dimension Representation

The dimension is:

$$
d\in\mathbb{N},
\qquad
d\geq1.
$$

A domain is represented by two vectors:

$$
a=(a_1,\ldots,a_d),
$$

and:

$$
b=(b_1,\ldots,b_d).
$$

The domain is:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j].
$$

The implementation must verify:

$$
a_j<b_j
$$

for every coordinate.

The dimension must also satisfy:

$$
d=|a|=|b|.
$$

### 38.3 Point Representation

A point is represented as:

$$
x=(x_1,\ldots,x_d)
\in\mathbb{R}^d.
$$

The implementation must verify:

$$
|x|=d.
$$

Point feasibility requires:

$$
a_j\leq x_j\leq b_j
$$

for every \(j\).

Therefore:

$$
x\in\Omega.
$$

The implementation should use a consistent numerical representation for points across evaluation history, candidate generation, region membership, and distance calculations.

### 38.4 Objective Interface

The objective remains a black-box evaluator:

$$
f:\mathbb{R}^d\rightarrow\mathbb{R}.
$$

ARRGO provides a valid point:

$$
x\in\Omega,
$$

and receives:

$$
y=f(x).
$$

The algorithm does not require:

- analytical derivatives;
- symbolic expressions;
- gradients;
- Hessians;
- separability;
- probabilistic models.

Optional numerical behavior analysis may be derived from evaluated points.

### 38.5 Evaluation Representation

A generalized evaluation is:

$$
e_i
=
(x_i,y_i,\operatorname{Id}_i),
$$

where:

$$
x_i\in\mathbb{R}^d,
\qquad
y_i\in\mathbb{R}.
$$

The evaluation history is:

$$
D_t
=
\{e_1,\ldots,e_{N_t}\}.
$$

The evaluation structure therefore changes from a scalar coordinate to a vector coordinate, while its conceptual role remains unchanged.

### 38.6 Global Incumbent

The incumbent representation remains:

$$
x_{\mathrm{best}}^{(t)}
\in
\mathbb{R}^d,
$$

with:

$$
f_{\mathrm{best}}^{(t)}
=
\max_{e_i\in D_t}y_i.
$$

The incumbent must correspond to an existing evaluation.

No dimensional redesign of the incumbent logic is required beyond supporting vector locations.

### 38.7 Domain Abstraction

The domain abstraction should expose operations such as:

- dimension;
- lower bounds;
- upper bounds;
- feasibility checking;
- center computation;
- width computation;
- diameter computation.

Conceptually:

$$
\operatorname{center}(\Omega)
=
\frac{a+b}{2}.
$$

The width vector is:

$$
W(\Omega)
=
b-a.
$$

The diameter is:

$$
\operatorname{diam}(\Omega)
=
\|b-a\|_2.
$$

### 38.8 Region Abstraction

A generalized region contains:

$$
R=(l,r,\operatorname{ParentId},\operatorname{ChildrenIds},\operatorname{State}),
$$

where:

$$
l,r\in\mathbb{R}^d.
$$

The geometric region is:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j].
$$

The region must satisfy:

$$
l_j<r_j
$$

for every coordinate.

### 38.9 Region Geometry Operations

The region abstraction should provide deterministic geometric operations:

- `contains(x)`;
- `center()`;
- `widths()`;
- `max_width()`;
- `diameter()`;
- `split(direction, location)`;
- `is_numerically_resolved()`.

The containment relation is:

$$
x\in R
\iff
l_j\leq x_j\leq r_j
\quad
\forall j.
$$

### 38.10 Region Membership and Boundary Ownership

A numerical implementation must explicitly define how points on shared split boundaries are assigned.

Mathematically, two closed children may satisfy:

$$
R_1\cap R_2\neq\varnothing
$$

at the split boundary.

This is acceptable for geometric coverage but can create ambiguity when assigning an evaluation to exactly one child.

Therefore, implementation-level membership should use an explicit ownership convention.

Possible approaches include:

- half-open internal intervals;
- deterministic boundary assignment;
- canonical child ordering.

The chosen convention must preserve complete domain coverage.

### 38.11 Region Hierarchy

The hierarchy remains a graph of region relationships:

$$
\mathcal{H}
=
(\mathcal{R},\mathcal{E}).
$$

Each child stores:

$$
\operatorname{ParentId}.
$$

Each parent stores:

$$
\operatorname{ChildrenIds}.
$$

Parent regions remain stored after splitting.

Therefore:

$$
\mathcal{H}_t
\subseteq
\mathcal{H}_{t+1}.
$$

The hierarchy records structural history rather than only the current spatial partition.

### 38.12 Current Frontier

Because historical parents remain in the hierarchy, the implementation must distinguish:

$$
\mathcal{H}_t
$$

from the current frontier:

$$
\mathcal{F}_t.
$$

The frontier represents the current spatial decomposition.

It must satisfy:

$$
\Omega
\subseteq
\bigcup_{R\in\mathcal{F}_t}R.
$$

After splitting a frontier region:

$$
R
\rightarrow
\{R_1,R_2\},
$$

the parent remains in \(\mathcal{H}\), while the children become part of the current frontier.

### 38.13 Region Evaluation View

Regional evaluations are obtained from global history:

$$
D_R
=
\{e_i\in D_t:x_i\in R\}.
$$

The implementation should avoid creating independent copies of objective evaluations for every region.

Instead, evaluation identity remains globally unique.

This preserves:

$$
\boxed{
\text{Single Evaluation Ownership}
+
\text{Multiple Information Views}.
}
$$

### 38.14 Region Analysis Interface

Region analysis receives:

$$
(R,D_R,\Theta).
$$

It produces a structured information state:

$$
Q_R
=
\operatorname{Analyze}(R,D_R,\Theta).
$$

The result may contain:

- sample count;
- directional coverage;
- geometric coverage;
- observed objective variation;
- directional behavior;
- joint behavior;
- empirical resolution signals;
- certified uncertainty;
- certified potential;
- resolution state.

The analysis layer should not directly execute sampling or splitting.

### 38.15 Candidate Representation

A generalized sampling candidate is:

$$
C_{\mathrm{sample}}
=
(x,\Pi_{\mathrm{sample}}),
$$

where:

$$
x\in R
$$

and:

$$
\Pi_{\mathrm{sample}}
$$

is the candidate information profile.

A generalized split candidate is:

$$
C_{\mathrm{split}}
=
(j,s_j,\Pi_{\mathrm{split}}),
$$

where:

$$
j\in\{1,\ldots,d\},
$$

and:

$$
l_j<s_j<r_j.
$$

### 38.16 Action Representation

The action abstraction becomes:

$$
A
\in
\{
\mathrm{Sample}(x),
\mathrm{Split}(j,s_j),
\mathrm{Stable}
\}.
$$

Sampling contains a vector location.

Splitting contains both:

- a coordinate direction;
- a split location.

Stable contains no new geometric or objective operation.

### 38.17 Candidate Generation

Candidate generation must be dimension-aware.

Sampling candidates may be generated from:

- region center;
- coverage gaps;
- geometric distance;
- observed behavior;
- potential-aware information;
- deterministic diversity rules.

Split candidates require:

$$
(j,s_j).
$$

The implementation must never assume a single spatial direction.

### 38.18 Direction Selection

For each candidate direction:

$$
j\in\{1,\ldots,d\},
$$

the implementation may construct a directional information profile:

$$
P_{\mathrm{dir}}^{(j)}
=
\left(
\gamma_j,
B_R^{(j)},
U_R^{(j)},
W_j,
\ldots
\right).
$$

Direction selection should use the same non-arbitrary decision philosophy established in the algorithm design.

A direction must not be selected merely because its coordinate index is smaller.

### 38.19 Split Location Selection

Direction selection and split-location selection are separate decisions.

After selecting:

$$
j^\ast,
$$

a feasible location:

$$
s_{j^\ast}
$$

must satisfy:

$$
\max
\left(
\frac{s_{j^\ast}-l_{j^\ast}}
{r_{j^\ast}-l_{j^\ast}},
\frac{r_{j^\ast}-s_{j^\ast}}
{r_{j^\ast}-l_{j^\ast}}
\right)
\leq\rho.
$$

The implementation should generate only valid locations.

For binary splitting:

$$
\rho\geq\frac12
$$

is required for a non-empty feasible split interval.

### 38.20 Action Selection

The local action-selection layer receives:

$$
Q_R,
\mathcal{C}_{\mathrm{sample}},
\mathcal{C}_{\mathrm{split}},
\Theta.
$$

It determines:

$$
A_R^\ast
\in
\{
\mathrm{Sample},
\mathrm{Split},
\mathrm{Stable}
\}.
$$

The action-selection layer must preserve the distinction between:

$$
\boxed{
\text{Information Acquisition}
\neq
\text{Structural Refinement}.
}
$$

### 38.21 Global Region Selection

Global region selection operates over eligible regions:

$$
\mathcal{R}_{\mathrm{eligible}}.
$$

It receives regional information and global state.

Its purpose is to determine:

$$
R^\ast
=
\operatorname{SelectRegion}(\mathcal{R}_{\mathrm{eligible}},G_t).
$$

This decision is distinct from choosing the action inside the selected region.

### 38.22 Certified Bounds Layer

In Certified Mode, the bounds layer uses Euclidean distance:

$$
\|x-x_i\|_2.
$$

The upper envelope is:

$$
U_R(x)
=
\min_{x_i\in D_R}
\left[
f(x_i)+L\|x-x_i\|_2
\right].
$$

The lower envelope is:

$$
L_R(x)
=
\max_{x_i\in D_R}
\left[
f(x_i)-L\|x-x_i\|_2
\right].
$$

The implementation must distinguish certified quantities from empirical signals.

### 38.23 Certification Frontier

Certified global quantities should be computed over a valid current frontier:

$$
\mathcal{F}_t.
$$

Thus:

$$
P_{\mathrm{global}}
=
\sup_{R\in\mathcal{F}_t}P(R).
$$

Historical parent regions remain available for analysis and structural history but should not automatically replace the current frontier in certification calculations.

### 38.24 Termination Layer

Termination remains dimension-independent conceptually.

Possible conditions include:

- certified tolerance;
- evaluation budget;
- numerical resolution;
- no valid action;
- configuration error.

Spatial numerical resolution may depend on:

$$
w_{\max}(R),
$$

or:

$$
\operatorname{diam}(R).
$$

### 38.25 Numerical Utilities

The numerical layer must support vector operations including:

- Euclidean distance;
- vector comparison;
- tolerance-aware equality;
- region membership;
- finite-value validation;
- stable ordering;
- boundary handling.

For example:

$$
\operatorname{distance}(x,y)
=
\|x-y\|_2.
$$

Duplicate detection may use:

$$
\|x-y\|_2
\leq
\tau_x.
$$

### 38.26 Complexity Awareness

The generalized implementation must account for the fact that dimensionality increases computational cost.

In particular:

- pairwise behavioral analysis may scale as \(O(n_R^2)\);
- naive Cartesian candidate grids scale exponentially with \(d\);
- exact geometric covering calculations may become expensive;
- certified potential calculations may require numerical optimization over hyperrectangles.

Therefore, the generalized implementation should avoid unnecessary exponential constructions.

Approximations may be introduced later, but each approximation must be explicitly distinguished from an exact mathematical quantity.

### 38.27 Determinism

Every implementation layer must preserve deterministic behavior.

For fixed:

$$
(f,\Theta,G_t),
$$

the complete next-state transition must remain deterministic.

This includes:

- candidate generation;
- candidate ordering;
- dominance comparisons;
- direction selection;
- split-location selection;
- global region selection;
- tie-breaking.

### 38.28 Dimension-Independent Components

The following components retain essentially the same conceptual implementation across dimensions:

- objective evaluation;
- evaluation history;
- evaluation identifiers;
- incumbent management;
- region lifecycle;
- hierarchy relationships;
- global state;
- termination;
- no-pruning;
- determinism;
- empirical/certified mode distinction.

### 38.29 Dimension-Dependent Components

The following components require generalized implementations:

- domain geometry;
- point representation;
- region geometry;
- spatial membership;
- distance;
- coverage;
- behavioral analysis;
- split generation;
- directional selection;
- certified envelopes.

These should depend on the dimension parameter rather than duplicate algorithmic logic.

### 38.30 Proposed Package Structure

The implementation can evolve toward the following conceptual structure:

src/
    arrgo/
        core/
            config.py
            evaluation.py
            history.py
            state.py
            result.py
        geometry/
            domain.py
            point.py
            region.py
        analysis/
            coverage.py
            behavior.py
            uncertainty.py
            region_analysis.py
        candidates/
            sampling.py
            splitting.py
            profiles.py
        selection/
            action.py
            region.py
            dominance.py
        hierarchy/
            tree.py
            frontier.py
        certification/
            envelopes.py
            potential.py
        termination/
            criteria.py
        engine/
            arrgo.py

This structure is a conceptual target rather than an immediate requirement to rewrite the current implementation.

### 38.31 One-Dimensional Compatibility

The generalized architecture must preserve the existing one-dimensional behavior.

When:

$$
d=1,
$$

vectors contain one coordinate:

$$
x=(x_1),
$$

and regions contain one interval:

$$
R=[l_1,r_1].
$$

The generalized geometry operations therefore reduce to scalar operations.

This provides an important compatibility requirement:

$$
\boxed{
\mathrm{ARRGO}(1)
\text{ must remain a valid special case of }
\mathrm{ARRGO}(d).
}
$$

### 38.32 Implementation Principle

The implementation should therefore follow:

$$
\boxed{
\text{Generalize the Abstraction}
\rightarrow
\text{Generalize the Geometry}
\rightarrow
\text{Generalize the Analysis}
\rightarrow
\text{Generalize the Decisions}
\rightarrow
\text{Preserve the Global Engine}.
}
$$

The objective is not to create a collection of independent one-, two-, and three-dimensional algorithms.

Instead, ARRGO should become one dimension-parameterized optimization framework whose behavior is specialized only by the geometry and information available in the chosen dimension.

This architecture provides the implementation foundation required for subsequent multidimensional coding, validation, and experimental evaluation.

## 39. Multidimensional Data Model

The multidimensional implementation requires a precise data model that represents geometric objects, objective evaluations, regional information, candidate actions, and global state.

The purpose of this data model is to provide a stable interface between the mathematical formulation and the implementation.

The model must represent arbitrary finite dimensions without introducing separate data structures for each dimension.

### 39.1 Design Principle

The central design principle is:

$$
\boxed{
\text{Dimension is Data}
\neq
\text{Dimension is Algorithm Logic}
}
$$

The dimension \(d\) determines the size of geometric vectors.

It should not require separate implementations for:

$$
d=1,\quad d=2,\quad d=3,\ldots
$$

### 39.2 Point

A point is represented as:

$$
x=(x_1,\ldots,x_d).
$$

It satisfies:

$$
x\in\mathbb{R}^d.
$$

A point therefore contains exactly \(d\) coordinates.

The implementation should treat the point as one geometric object rather than exposing individual coordinate fields such as `x`, `y`, or `z`.

This allows the same representation to support arbitrary finite dimensions.

### 39.3 Domain

A domain is represented by two points:

$$
a=(a_1,\ldots,a_d),
$$

and:

$$
b=(b_1,\ldots,b_d).
$$

The domain is:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j].
$$

The domain invariant is:

$$
a_j<b_j
\qquad
\forall j.
$$

The dimension is therefore:

$$
d=|a|=|b|.
$$

### 39.4 Domain Properties

The domain should provide:

$$
\operatorname{dimension}(\Omega)=d,
$$

$$
\operatorname{lower}(\Omega)=a,
$$

$$
\operatorname{upper}(\Omega)=b.
$$

Its center is:

$$
c(\Omega)
=
\frac{a+b}{2}.
$$

Its width vector is:

$$
W(\Omega)
=
b-a.
$$

Its diameter is:

$$
\operatorname{diam}(\Omega)
=
\|b-a\|_2.
$$

### 39.5 Region

A region is represented by lower and upper vectors:

$$
l=(l_1,\ldots,l_d),
$$

and:

$$
r=(r_1,\ldots,r_d).
$$

The region is:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j].
$$

Its validity condition is:

$$
l_j<r_j
\qquad
\forall j.
$$

Additionally:

$$
a_j\leq l_j<r_j\leq b_j
\qquad
\forall j.
$$

Therefore:

$$
R\subseteq\Omega.
$$

### 39.6 Region Geometric Properties

The region provides:

$$
W(R)=r-l,
$$

where:

$$
W(R)
=
(w_1,\ldots,w_d).
$$

The maximum side width is:

$$
w_{\max}(R)
=
\max_j w_j.
$$

The center is:

$$
c(R)
=
\frac{l+r}{2}.
$$

The diameter is:

$$
\operatorname{diam}(R)
=
\|r-l\|_2.
$$

These quantities are derived from the region boundaries rather than stored as independent mutable values.

### 39.7 Region Identity

Every region has a unique identifier:

$$
\operatorname{Id}(R).
$$

The identifier is independent of its geometric coordinates.

This allows region geometry to change only through explicit structural operations while hierarchy relationships remain stable.

### 39.8 Region Hierarchy Metadata

A region may contain:

$$
\operatorname{ParentId},
$$

and:

$$
\operatorname{ChildrenIds}.
$$

Its structural state is:

$$
\operatorname{State}(R)
\in
\{
\mathrm{ACTIVE},
\mathrm{STABLE},
\mathrm{REFINED}
\}.
$$

The hierarchy metadata does not change the geometric definition of the region.

### 39.9 Evaluation

An objective evaluation is:

$$
e_i
=
(x_i,y_i,\operatorname{Id}_i),
$$

where:

$$
x_i\in\Omega,
$$

and:

$$
y_i=f(x_i).
$$

The evaluation identifier is unique within the global evaluation history.

The same evaluation must not be duplicated merely because it is visible from multiple regional views.

### 39.10 Evaluation History

The global evaluation history is:

$$
D_t
=
\{e_1,\ldots,e_{N_t}\}.
$$

It is the authoritative source of objective evaluations.

Regional evaluation sets are derived views:

$$
D_R
=
\{e_i\in D_t:x_i\in R\}.
$$

This provides:

$$
\boxed{
\text{One Global Evaluation History}
+
\text{Many Regional Views}.
}
$$

### 39.11 Incumbent

The incumbent consists of an evaluated point and its objective value:

$$
I_t
=
(x_{\mathrm{best}}^{(t)},
f_{\mathrm{best}}^{(t)}).
$$

It must satisfy:

$$
x_{\mathrm{best}}^{(t)}
\in
\{x_i:e_i\in D_t\}.
$$

The incumbent value is:

$$
f_{\mathrm{best}}^{(t)}
=
\max_{e_i\in D_t}y_i.
$$

### 39.12 Regional Coverage State

The regional coverage representation contains directional information:

$$
\Gamma(R)
=
(\gamma_1,\ldots,\gamma_d),
$$

where each component describes the relative unresolved spacing along one coordinate direction.

A geometric coverage measure may additionally be represented by:

$$
h(R)
=
\sup_{x\in R}
\min_{x_i\in D_R}
\|x-x_i\|_2.
$$

These quantities are information signals rather than objective-value guarantees in Empirical Mode.

### 39.13 Regional Behavior State

The regional behavior representation may contain:

$$
B_R
=
\left(
B_R^{(1)},\ldots,B_R^{(d)},
B_R^{(\mathrm{joint})}
\right).
$$

The directional components describe observed behavior associated with coordinate directions.

The joint component preserves information about interactions between coordinates.

No separability assumption is introduced.

### 39.14 Regional Certified State

In Certified Mode, the regional certified state may contain:

$$
U_R(x),
$$

$$
L_R(x),
$$

$$
u_{\max}(R),
$$

and:

$$
P(R).
$$

These quantities are derived from the supplied valid Lipschitz constant and regional evaluations.

They must remain distinct from empirical behavior signals.

### 39.15 Regional Information State

The complete regional information object is:

$$
Q_R
=
\left(
Q_{\mathrm{coverage}},
Q_{\mathrm{behavior}},
Q_{\mathrm{uncertainty}},
Q_{\mathrm{potential}}
\right).
$$

Additional metadata may include:

- evaluation count;
- maximum width;
- diameter;
- resolution state;
- directional profiles;
- numerical status.

The information object should be treated as derived state whenever possible.

### 39.16 Sampling Candidate

A sampling candidate is:

$$
C_{\mathrm{sample}}
=
(x,\Pi_{\mathrm{sample}}),
$$

where:

$$
x\in R.
$$

The candidate profile:

$$
\Pi_{\mathrm{sample}}
$$

contains deterministic information signals describing why the candidate may be useful.

The candidate itself does not contain an objective value before evaluation.

### 39.17 Split Candidate

A split candidate is:

$$
C_{\mathrm{split}}
=
(j,s_j,\Pi_{\mathrm{split}}),
$$

where:

$$
j\in\{1,\ldots,d\},
$$

and:

$$
l_j<s_j<r_j.
$$

The candidate represents a structural action and therefore does not require a new objective evaluation before execution.

### 39.18 Action

The action model is:

$$
A
\in
\{
\mathrm{Sample}(x),
\mathrm{Split}(j,s_j),
\mathrm{Stable}
\}.
$$

Sampling modifies objective information.

Splitting modifies the spatial hierarchy.

Stable modifies the resolution state.

These actions must remain semantically distinct.

### 39.19 Split Direction

The split direction is represented by:

$$
j\in\{1,\ldots,d\}.
$$

The corresponding directional width is:

$$
w_j=r_j-l_j.
$$

Only directions satisfying:

$$
w_j>0
$$

are structurally splittable.

### 39.20 Split Location

For selected direction \(j\), the split location is:

$$
s_j\in(l_j,r_j).
$$

The contraction condition is:

$$
\max
\left(
\frac{s_j-l_j}{r_j-l_j},
\frac{r_j-s_j}{r_j-l_j}
\right)
\leq
\rho.
$$

The split representation therefore contains both geometric decisions:

$$
\boxed{
\text{Direction}
+
\text{Location}.
}
$$

### 39.21 Global Frontier

The current frontier is:

$$
\mathcal{F}_t.
$$

It contains the current spatial regions used for refinement and, when applicable, certification.

The frontier must satisfy:

$$
\Omega
\subseteq
\bigcup_{R\in\mathcal{F}_t}R.
$$

The full hierarchy:

$$
\mathcal{H}_t
$$

contains historical parent regions as well.

Therefore:

$$
\mathcal{F}_t
\subseteq
\mathcal{H}_t.
$$

### 39.22 Global State

The generalized global state can be represented conceptually as:

$$
G_t
=
\left(
D_t,
\mathcal{H}_t,
\mathcal{F}_t,
I_t,
N_t,
Q_t,
S_t
\right).
$$

Here:

- \(D_t\) is evaluation history;
- \(\mathcal{H}_t\) is structural history;
- \(\mathcal{F}_t\) is the current frontier;
- \(I_t\) is the incumbent;
- \(N_t\) is evaluation count;
- \(Q_t\) contains current regional information;
- \(S_t\) contains execution status.

### 39.23 Configuration

The generalized configuration should contain:

$$
\Theta
=
\left(
a,b,
N_{\max},
\rho,
\tau_x,
\tau_{\mathrm{num}},
\mathrm{Mode},
L,
\varepsilon
\right).
$$

Not every parameter is required in every execution mode.

For example, \(L\) and \(\varepsilon\) are relevant to Certified Mode.

The configuration must validate dimensional consistency:

$$
|a|=|b|=d.
$$

### 39.24 Termination State

The execution status should explicitly represent:

- running;
- certified tolerance reached;
- budget exhausted;
- numerical limit;
- no valid action;
- configuration error.

A terminal state must prevent additional invalid iterations.

### 39.25 Derived Versus Stored Data

The implementation should distinguish between authoritative data and derived data.

Authoritative data includes:

- domain boundaries;
- region boundaries;
- evaluation history;
- region identifiers;
- hierarchy relationships;
- action history when recorded.

Derived data includes:

- region widths;
- region centers;
- distances;
- coverage metrics;
- behavior metrics;
- certified envelopes;
- regional potential;
- global potential.

Derived quantities should be recomputed or invalidated when their underlying information changes.

### 39.26 Data Ownership

Objective evaluations have global ownership:

$$
e_i\in D_t.
$$

Regions have structural ownership within the hierarchy.

Regional analysis does not own objective evaluations.

Therefore:

$$
\boxed{
\text{Data Ownership}
\neq
\text{Information Visibility}.
}
$$

A single evaluation may be visible to every regional analysis for which it is geometrically relevant.

### 39.27 Numerical Identity

Two points:

$$
x,y\in\mathbb{R}^d
$$

may be considered numerically identical under:

$$
\|x-y\|_2\leq\tau_x.
$$

The exact tolerance policy must be centralized.

It should not be independently redefined by different implementation layers.

### 39.28 Serialization Principle

The core geometric and evaluation objects should be representable without depending on runtime-specific objects.

For example, a region should be reconstructible from:

$$
(l,r,\operatorname{Id},\operatorname{ParentId},
\operatorname{ChildrenIds},\operatorname{State}).
$$

This supports reproducibility, debugging, testing, and later experiment logging.

### 39.29 One-Dimensional Compatibility

When:

$$
d=1,
$$

the generalized data structures reduce to the original scalar structures.

A point becomes:

$$
x=(x_1),
$$

a region becomes:

$$
R=[l_1,r_1],
$$

and the width vector contains one element.

The Euclidean distance becomes:

$$
\|x-y\|_2
=
|x_1-y_1|.
$$

Therefore, the multidimensional data model preserves the one-dimensional case as a special case.

### 39.30 Data Model Principle

The complete data-model hierarchy is:

$$
\boxed{
\begin{aligned}
&\text{Domain}
\\
&\Downarrow
\\
&\text{Point and Region Geometry}
\\
&\Downarrow
\\
&\text{Objective Evaluations}
\\
&\Downarrow
\\
&\text{Regional Information Views}
\\
&\Downarrow
\\
&\text{Candidates and Actions}
\\
&\Downarrow
\\
&\text{Hierarchy and Frontier}
\\
&\Downarrow
\\
&\text{Global State}
\end{aligned}
}
$$

This data model provides the conceptual contract required before modifying the current ARRGO implementation.

The implementation should first establish these generalized data abstractions and preserve the existing one-dimensional behavior as a compatibility case.

## 40. Multidimensional Data Model Contracts

The generalized data model defines the objects used by ARRGO.

The next requirement is to define the behavioral contracts that these objects must satisfy during execution.

A data structure is not sufficient by itself.

Each operation must preserve the mathematical and algorithmic invariants established by the ARRGO framework.

### 40.1 Contract Principle

Every ARRGO operation must satisfy:

$$
\boxed{
\text{Valid Input}
+
\text{Valid Transformation}
\Rightarrow
\text{Valid State}
}
$$

An operation must not introduce an invalid domain, invalid region, infeasible point, inconsistent hierarchy, duplicated evaluation, or invalid certified state.

### 40.2 Domain Contract

For:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j],
$$

the domain contract requires:

$$
d\geq1,
$$

and:

$$
a_j<b_j
\qquad
\forall j.
$$

The domain is immutable during an ARRGO execution.

Therefore:

$$
\Omega_{t+1}=\Omega_t.
$$

The optimization problem itself must not change during refinement.

### 40.3 Point Contract

Every point supplied to the objective must satisfy:

$$
x\in\Omega.
$$

Equivalently:

$$
a_j\leq x_j\leq b_j
\qquad
\forall j.
$$

An infeasible point must never be passed to the objective evaluator.

### 40.4 Region Contract

For every region:

$$
R=
\prod_{j=1}^{d}[l_j,r_j],
$$

the following must hold:

$$
a_j\leq l_j<r_j\leq b_j
\qquad
\forall j.
$$

Therefore:

$$
R\subseteq\Omega.
$$

A region operation must preserve these constraints.

### 40.5 Region Geometry Contract

For every valid region:

$$
w_j=r_j-l_j>0.
$$

Therefore:

$$
w_{\max}(R)>0,
$$

and:

$$
\operatorname{diam}(R)>0.
$$

Numerical resolution may eventually classify a region as resolved, but numerical resolution must not be confused with an invalid geometric region.

### 40.6 Region Containment Contract

If a region is created as a child of \(R\), then:

$$
R_{\mathrm{child}}\subseteq R_{\mathrm{parent}}.
$$

For every coordinate:

$$
l_j^{\mathrm{parent}}
\leq
l_j^{\mathrm{child}}
<
r_j^{\mathrm{child}}
\leq
r_j^{\mathrm{parent}}.
$$

This must hold independently of the selected split direction.

### 40.7 Split Contract

A split operation selects:

$$
j\in\{1,\ldots,d\},
$$

and:

$$
l_j<s_j<r_j.
$$

Only coordinate \(j\) is changed.

For all:

$$
k\neq j,
$$

the child boundaries remain unchanged:

$$
l_k^{(1)}=l_k^{(2)}=l_k,
$$

and:

$$
r_k^{(1)}=r_k^{(2)}=r_k.
$$

The selected coordinate is divided at \(s_j\).

### 40.8 Split Coverage Contract

The children must cover the parent:

$$
R_{\mathrm{parent}}
\subseteq
R_1\cup R_2.
$$

Ideally:

$$
R_{\mathrm{parent}}
=
R_1\cup R_2.
$$

If closed regions are used mathematically, the children may overlap on the split boundary.

This overlap is a geometric property and must not be confused with duplicate ownership of an evaluation.

### 40.9 Boundary Ownership Contract

The implementation must define a deterministic ownership rule for the split boundary.

For an evaluation located exactly at:

$$
x_j=s_j,
$$

the implementation must determine which child owns that evaluation.

The ownership rule must be:

- deterministic;
- consistent;
- independent of execution order;
- compatible with complete spatial coverage.

This prevents the same evaluation from being counted multiple times when regional information requires exclusive membership.

### 40.10 Contraction Contract

For the selected coordinate \(j\), valid splitting requires:

$$
\max
\left(
\frac{s_j-l_j}{r_j-l_j},
\frac{r_j-s_j}{r_j-l_j}
\right)
\leq\rho.
$$

Therefore each child width along the selected coordinate satisfies:

$$
w_j^{(\mathrm{child})}
\leq
\rho w_j^{(\mathrm{parent})}.
$$

For binary splitting:

$$
\rho\geq\frac12.
$$

The implementation must reject configurations that cannot produce a valid binary split.

### 40.11 Directional Progress Contract

A split modifies one coordinate direction.

Therefore, a single split does not guarantee that every coordinate width decreases.

Global spatial contraction requires a fairness condition ensuring that unresolved directions are eventually selected.

Thus:

$$
\boxed{
\text{Local Contraction}
\neq
\text{Global Multidimensional Contraction}.
}
$$

The second requires directional fairness.

### 40.12 Evaluation Contract

Every evaluation must contain:

$$
x_i\in\Omega,
$$

and a finite objective value:

$$
y_i=f(x_i).
$$

Each evaluation has a unique identifier:

$$
\operatorname{Id}_i
\neq
\operatorname{Id}_k
\qquad
\text{for }i\neq k.
$$

The objective evaluation itself is deterministic.

### 40.13 Duplicate Evaluation Contract

Before evaluating a candidate point \(x\), the implementation should check whether an equivalent evaluation already exists.

A numerical duplicate may be defined by:

$$
\|x-x_i\|_2\leq\tau_x.
$$

A duplicate candidate should not consume an unnecessary objective evaluation.

The tolerance must be centrally defined and consistently applied.

### 40.14 Evaluation History Contract

The global history must contain only valid evaluations:

$$
D_t
=
\{e_1,\ldots,e_{N_t}\}.
$$

The history must preserve evaluation identity and insertion order where execution order is relevant.

No regional operation may silently remove an evaluation from the global history.

Therefore:

$$
D_t
\subseteq
D_{t+1}.
$$

### 40.15 Incumbent Contract

The incumbent must always correspond to an evaluation in the global history.

Therefore:

$$
x_{\mathrm{best}}^{(t)}
=
x_i
$$

for some:

$$
e_i\in D_t.
$$

Its value must satisfy:

$$
f_{\mathrm{best}}^{(t)}
=
\max_{e_i\in D_t}y_i.
$$

### 40.16 Incumbent Monotonicity Contract

Because evaluations are never removed:

$$
f_{\mathrm{best}}^{(t+1)}
\geq
f_{\mathrm{best}}^{(t)}.
$$

Thus the global best observed objective value is monotone non-decreasing.

This is an invariant of the evaluation history, independent of the dimensionality.

### 40.17 Region Evaluation View Contract

Regional evaluation information must be derived from the global history.

For a region \(R\):

$$
D_R
=
\{e_i\in D_t:x_i\in R\}.
$$

The regional view must not create a second logical identity for the same evaluation.

### 40.18 Regional Analysis Contract

Region analysis must be deterministic:

$$
Q_R
=
\operatorname{Analyze}(R,D_R,\Theta).
$$

For identical inputs:

$$
(R,D_R,\Theta)
$$

the output must be identical.

Analysis must not modify:

- objective evaluations;
- domain boundaries;
- region identity;
- hierarchy relationships.

### 40.19 Candidate Contract

Every sampling candidate must satisfy:

$$
x\in R.
$$

Every split candidate must satisfy:

$$
j\in\{1,\ldots,d\},
$$

and:

$$
l_j<s_j<r_j.
$$

Invalid candidates must be rejected before action execution.

### 40.20 Action Contract

Every action must be one of:

$$
\mathrm{Sample}(x),
\qquad
\mathrm{Split}(j,s_j),
\qquad
\mathrm{Stable}.
$$

The action must be feasible for the selected region.

An action must not silently change its semantic meaning during execution.

### 40.21 Sampling Contract

A sampling action:

$$
\mathrm{Sample}(x)
$$

must:

1. verify \(x\in R\);
2. verify budget availability;
3. verify duplicate policy;
4. evaluate \(f(x)\);
5. add the evaluation to global history;
6. update the incumbent if necessary;
7. invalidate or update affected derived regional information.

Sampling does not create child regions.

### 40.22 Splitting Contract

A splitting action:

$$
\mathrm{Split}(j,s_j)
$$

must:

1. verify that direction \(j\) is feasible;
2. verify that \(s_j\) satisfies contraction;
3. create valid child regions;
4. preserve the parent;
5. update hierarchy relationships;
6. update the current frontier;
7. preserve global evaluation history.

A split does not automatically consume an objective-evaluation budget.

### 40.23 Stable Contract

A Stable action means that the current region does not require immediate refinement under the configured resolution criteria.

Stable must not mean:

$$
\text{``Region can never become relevant again.''}
$$

Global information may change after later evaluations.

Therefore, stable regions remain represented in the hierarchy and may become eligible again if the implementation permits reactivation.

### 40.24 Hierarchy Contract

The hierarchy must remain structurally consistent.

For every child:

$$
\operatorname{ParentId}(R_{\mathrm{child}})
=
\operatorname{Id}(R_{\mathrm{parent}}).
$$

For every parent:

$$
\operatorname{Id}(R_{\mathrm{child}})
\in
\operatorname{ChildrenIds}(R_{\mathrm{parent}}).
$$

These relationships must remain mutually consistent.

### 40.25 No-Pruning Contract

ARRGO does not delete historical regions.

Therefore:

$$
\mathcal{H}_t
\subseteq
\mathcal{H}_{t+1}.
$$

A region may become:

$$
\mathrm{STABLE}
\quad\text{or}\quad
\mathrm{REFINED},
$$

but it remains part of the structural history.

This is distinct from the current frontier.

### 40.26 Frontier Contract

The current frontier must represent the current spatial decomposition.

It must satisfy:

$$
\Omega
\subseteq
\bigcup_{R\in\mathcal{F}_t}R.
$$

When a frontier region is split:

$$
\mathcal{F}_{t+1}
=
(\mathcal{F}_t\setminus\{R\})
\cup
\{R_1,R_2\}.
$$

The parent remains in the full hierarchy:

$$
R\in\mathcal{H}_{t+1}.
$$

### 40.27 Certified Frontier Contract

In Certified Mode, global certification must be based on a valid covering collection.

Therefore:

$$
P_{\mathrm{global}}
=
\sup_{R\in\mathcal{F}_t}P(R).
$$

This prevents historical parent regions from automatically dominating the certification calculation merely because they remain stored.

### 40.28 Certified Bound Contract

Under a valid global Lipschitz constant:

$$
|f(x)-f(y)|
\leq
L\|x-y\|_2,
$$

the regional envelopes must satisfy:

$$
L_R(x)
\leq
f(x)
\leq
U_R(x).
$$

Consequently:

$$
f_{\mathrm{best}}
\leq
f^\ast
\leq
P_{\mathrm{global}}.
$$

If:

$$
\Delta_{\mathrm{global}}
=
P_{\mathrm{global}}
-
f_{\mathrm{best}}
\leq
\varepsilon,
$$

the incumbent is epsilon-optimal under the stated assumptions.

### 40.29 Budget Contract

Only actual objective evaluations consume the evaluation budget.

Therefore:

$$
N_t\leq N_{\max}.
$$

Sampling may increase:

$$
N_{t+1}=N_t+1.
$$

Splitting does not automatically increase:

$$
N_{t+1}=N_t.
$$

The implementation must nevertheless prevent unlimited structural operations when no meaningful progress remains.

### 40.30 Progress Contract

Every non-terminal iteration must produce meaningful state change.

Possible progress includes:

- a new objective evaluation;
- a valid region split;
- a valid resolution-state transition;
- a certified-gap improvement;
- another explicitly defined state transition.

A sequence of iterations that repeatedly selects an action without changing the relevant state must be detected as stagnation.

### 40.31 Determinism Contract

For fixed:

$$
(f,\Theta,G_t),
$$

the next state must be deterministic.

Therefore:

$$
G_{t+1}
=
T(G_t,A_t;\Theta,f)
$$

must produce the same result for identical inputs.

Tie-breaking must therefore be explicit and deterministic.

### 40.32 Information-Driven Refinement Contract

Refinement decisions must be based on available information.

A region should not be refined solely because of its coordinate index, creation order, or arbitrary fixed spatial preference.

The decision should depend on quantities such as:

- unresolved coverage;
- observed behavior;
- geometric resolution;
- uncertainty;
- optimization potential;
- global relevance.

### 40.33 Joint-Objective Contract

Multidimensional ARRGO must optimize:

$$
f(x_1,\ldots,x_d)
$$

as a joint objective.

The implementation must not implicitly replace the problem with independent one-dimensional optimizations:

$$
f(x_1,\ldots,x_d)
\not\equiv
\sum_j f_j(x_j)
$$

unless separability is explicitly part of a particular benchmark.

Coordinate-wise analysis is therefore an information mechanism, not a separability assumption.

### 40.34 Empirical Mode Contract

In Empirical Mode, ARRGO may return:

$$
x_{\mathrm{best}},
f_{\mathrm{best}},
$$

as the best observed solution.

No mathematical global-optimality certificate is implied merely by observing a high-quality result.

### 40.35 Certified Mode Contract

Certified Mode requires:

1. a valid global Lipschitz constant;
2. valid regional evaluations;
3. valid frontier coverage;
4. mathematically valid upper and lower envelopes;
5. a correctly computed global potential;
6. a valid global gap.

The certificate is therefore conditional on its assumptions.

### 40.36 One-Dimensional Compatibility Contract

When:

$$
d=1,
$$

all multidimensional contracts must reduce to the existing one-dimensional contracts.

In particular:

$$
\|x-y\|_2
=
|x-y|,
$$

and:

$$
R=[l,r].
$$

The split direction becomes unique:

$$
j=1.
$$

Thus the multidimensional contract must preserve the existing ARRGO behavior as its one-dimensional special case.

### 40.37 Contract Summary

The generalized ARRGO implementation must preserve the following chain:

$$
\boxed{
\begin{aligned}
&\text{Valid Domain}
\\
&\Downarrow
\\
&\text{Valid Points and Regions}
\\
&\Downarrow
\\
&\text{Valid Evaluations}
\\
&\Downarrow
\\
&\text{Consistent Regional Information}
\\
&\Downarrow
\\
&\text{Valid Candidates and Actions}
\\
&\Downarrow
\\
&\text{Consistent Hierarchy and Frontier}
\\
&\Downarrow
\\
&\text{Valid Global State}
\\
&\Downarrow
\\
&\text{Valid Empirical or Certified Result}
\end{aligned}
}
$$

These contracts define the behavioral requirements that the multidimensional implementation must satisfy.

They will later serve as the basis for implementation-level invariants, unit tests, integration tests, and end-to-end multidimensional validation.

## 41. Formal Proof of Multidimensional Region Splitting

We now formally establish that the binary region-splitting operation used by ARRGO remains valid in arbitrary finite dimensions.

The proof establishes three properties:

1. validity of the generated child regions;
2. containment of the children inside the parent;
3. preservation of spatial coverage.

### 41.1 Parent Region

Let the parent region be:

$$
R
=
\prod_{k=1}^{d}[l_k,r_k],
$$

where:

$$
l_k<r_k
\qquad
\forall k\in\{1,\ldots,d\}.
$$

Let a split direction be selected:

$$
j\in\{1,\ldots,d\}.
$$

Choose a split location:

$$
s_j\in(l_j,r_j).
$$

Only coordinate \(j\) is divided.

### 41.2 Child Regions

The first child is:

$$
R_1
=
\left(
\prod_{k\neq j}[l_k,r_k]
\right)
\times
[l_j,s_j].
$$

The second child is:

$$
R_2
=
\left(
\prod_{k\neq j}[l_k,r_k]
\right)
\times
[s_j,r_j].
$$

The two children differ only in the selected coordinate.

### 41.3 Validity of Child Regions

Because:

$$
l_j<s_j<r_j,
$$

we have:

$$
l_j<s_j,
$$

and:

$$
s_j<r_j.
$$

Therefore the interval:

$$
[l_j,s_j]
$$

is valid, and:

$$
[s_j,r_j]
$$

is also valid.

For every other coordinate:

$$
l_k<r_k
\qquad
(k\neq j).
$$

Therefore both child regions are valid \(d\)-dimensional hyperrectangles.

Hence:

$$
R_1
\quad\text{and}\quad
R_2
$$

are valid regions.

### 41.4 Containment of Child Regions

Consider any point:

$$
x\in R_1.
$$

For every coordinate \(k\neq j\):

$$
l_k\leq x_k\leq r_k.
$$

For the split coordinate:

$$
l_j\leq x_j\leq s_j.
$$

Since:

$$
s_j<r_j,
$$

we also have:

$$
l_j\leq x_j\leq r_j.
$$

Therefore:

$$
x\in R.
$$

Thus:

$$
R_1\subseteq R.
$$

The same argument applies to \(R_2\).

Therefore:

$$
R_2\subseteq R.
$$

Consequently:

$$
\boxed{
R_1\cup R_2\subseteq R.
}
$$

### 41.5 Coverage Preservation

Now consider an arbitrary point:

$$
x\in R.
$$

For every coordinate \(k\neq j\):

$$
l_k\leq x_k\leq r_k.
$$

For the selected coordinate \(j\), exactly one of the following relations holds:

$$
x_j\leq s_j,
$$

or:

$$
x_j\geq s_j.
$$

If:

$$
x_j\leq s_j,
$$

then:

$$
x\in R_1.
$$

If:

$$
x_j\geq s_j,
$$

then:

$$
x\in R_2.
$$

Therefore every point in \(R\) belongs to at least one child.

Hence:

$$
R\subseteq R_1\cup R_2.
$$

Combining this with the containment result:

$$
R_1\cup R_2\subseteq R,
$$

we obtain:

$$
\boxed{
R=R_1\cup R_2.
}
$$

Thus the split preserves complete spatial coverage.

### 41.6 Boundary Intersection

Because both children include the split boundary:

$$
x_j=s_j,
$$

their intersection is:

$$
R_1\cap R_2
=
\left(
\prod_{k\neq j}[l_k,r_k]
\right)
\times
\{s_j\}.
$$

Therefore the children may overlap on a \((d-1)\)-dimensional boundary.

This does not violate spatial coverage.

However, implementation-level evaluation ownership must still define how an evaluation located exactly on this boundary is assigned to a regional information view.

### 41.7 Parent-Child Structural Consistency

The geometric result gives:

$$
R_1,R_2\subseteq R.
$$

Therefore a valid hierarchy may establish:

$$
\operatorname{Parent}(R_1)
=
\operatorname{Parent}(R_2)
=
R.
$$

The parent remains part of the structural hierarchy.

Thus splitting changes the hierarchy by adding children without deleting the parent.

### 41.8 Preservation of Domain Feasibility

Assume:

$$
R\subseteq\Omega.
$$

Since:

$$
R_1\subseteq R
$$

and:

$$
R_2\subseteq R,
$$

it follows that:

$$
R_1\subseteq\Omega,
$$

and:

$$
R_2\subseteq\Omega.
$$

Therefore a valid split cannot generate a child outside the original optimization domain.

### 41.9 Contraction of the Split Direction

The parent width in direction \(j\) is:

$$
w_j=r_j-l_j.
$$

The child widths are:

$$
w_j^{(1)}
=
s_j-l_j,
$$

and:

$$
w_j^{(2)}
=
r_j-s_j.
$$

Under the contraction condition:

$$
\max
\left(
\frac{s_j-l_j}{r_j-l_j},
\frac{r_j-s_j}{r_j-l_j}
\right)
\leq\rho,
$$

we obtain:

$$
w_j^{(1)}
\leq
\rho w_j,
$$

and:

$$
w_j^{(2)}
\leq
\rho w_j.
$$

For:

$$
0<\rho<1,
$$

both child widths in the selected direction are strictly smaller than the parent width.

### 41.10 Effect on Region Diameter

The parent diameter is:

$$
\operatorname{diam}(R)
=
\sqrt{
\sum_{k=1}^{d}w_k^2
}.
$$

Only the selected width \(w_j\) changes.

For either child:

$$
\operatorname{diam}(R_m)^2
=
\sum_{k\neq j}w_k^2
+
\left(w_j^{(m)}\right)^2,
\qquad
m\in\{1,2\}.
$$

Because:

$$
w_j^{(m)}<w_j,
$$

we obtain:

$$
\operatorname{diam}(R_m)
<
\operatorname{diam}(R).
$$

Therefore every valid split strictly decreases the diameter of each child.

### 41.11 Proposition

The preceding arguments establish the following proposition.

**Proposition — Multidimensional Region Split Validity**

Let:

$$
R
=
\prod_{k=1}^{d}[l_k,r_k]
$$

be a valid hyperrectangular region.

Let:

$$
j\in\{1,\ldots,d\}
$$

and:

$$
s_j\in(l_j,r_j).
$$

Construct \(R_1\) and \(R_2\) by splitting \(R\) along coordinate \(j\) at \(s_j\).

Then:

1. \(R_1\) and \(R_2\) are valid hyperrectangles;
2. \(R_1\subseteq R\);
3. \(R_2\subseteq R\);
4. \(R=R_1\cup R_2\);
5. the children remain inside the original optimization domain;
6. the split boundary is the only geometric overlap when closed intervals are used;
7. if the contraction condition is satisfied, both child widths in the selected direction are reduced;
8. both child diameters are strictly smaller than the parent diameter.

Therefore the binary structural refinement operation is geometrically valid in every finite dimension:

$$
\boxed{
d<\infty
\Rightarrow
\text{Valid Hyperrectangular Splitting}.
}
$$

### 41.12 One-Dimensional Reduction

For:

$$
d=1,
$$

the parent becomes:

$$
R=[l_1,r_1],
$$

and the two children become:

$$
R_1=[l_1,s_1],
$$

and:

$$
R_2=[s_1,r_1].
$$

Thus the multidimensional proposition reduces exactly to the original one-dimensional interval-splitting operation.

### 41.13 Consequence for ARRGO

The proposition establishes that the structural refinement mechanism itself does not depend on one-dimensional geometry.

The same binary splitting principle can therefore be used in:

$$
\mathbb{R}^1,
\mathbb{R}^2,
\ldots,
\mathbb{R}^d.
$$

However, this proposition alone does not establish global convergence.

It establishes only the geometric validity of the structural refinement operation.

Global convergence additionally requires:

- sufficient refinement of relevant regions;
- directional fairness;
- sufficient information acquisition;
- valid global region selection;
- continuity of the objective;
- and, for certification, valid regularity assumptions such as a global Lipschitz bound.

The distinction is therefore:

$$
\boxed{
\text{Valid Splitting}
\neq
\text{Global Convergence}.
}
$$

The next proof must establish how repeated directional refinement leads to global spatial contraction when the refinement policy satisfies an appropriate fairness condition.

## 42. Formal Proof of Multidimensional Spatial Contraction

The previous proposition established that a single multidimensional split is geometrically valid and strictly reduces the diameter of the resulting child regions.

However, a single split reduces only one coordinate width.

Therefore, global spatial contraction requires an additional condition on the sequence of selected split directions.

This section formalizes that condition and proves that directional fairness combined with local contraction leads to vanishing region diameter.

### 42.1 Coordinate Width Vector

For a region:

$$
R_t
=
\prod_{j=1}^{d}[l_j^{(t)},r_j^{(t)}],
$$

define the coordinate-width vector:

$$
W_t
=
\left(
w_1^{(t)},\ldots,w_d^{(t)}
\right),
$$

where:

$$
w_j^{(t)}
=
r_j^{(t)}-l_j^{(t)}.
$$

The maximum coordinate width is:

$$
w_{\max}^{(t)}
=
\max_{1\leq j\leq d}
w_j^{(t)}.
$$

The Euclidean diameter is:

$$
D_t
=
\operatorname{diam}(R_t)
=
\sqrt{
\sum_{j=1}^{d}
\left(w_j^{(t)}\right)^2
}.
$$

### 42.2 Relationship Between Maximum Width and Diameter

Because:

$$
w_j^{(t)}
\leq
w_{\max}^{(t)}
$$

for every \(j\), we have:

$$
D_t^2
=
\sum_{j=1}^{d}
\left(w_j^{(t)}\right)^2
\leq
d\left(w_{\max}^{(t)}\right)^2.
$$

Therefore:

$$
D_t
\leq
\sqrt d\,w_{\max}^{(t)}.
$$

Conversely, because at least one coordinate has width \(w_{\max}^{(t)}\):

$$
D_t
\geq
w_{\max}^{(t)}.
$$

Hence:

$$
\boxed{
w_{\max}^{(t)}
\leq
D_t
\leq
\sqrt d\,w_{\max}^{(t)}.
}
$$

Consequently:

$$
w_{\max}^{(t)}\rightarrow0
\iff
D_t\rightarrow0.
$$

Thus global contraction can be established by proving that every coordinate width converges to zero.

### 42.3 Local Contraction

Suppose a split is performed in direction \(j\).

Under the contraction condition:

$$
\max
\left(
\frac{w_j^{(1)}}{w_j},
\frac{w_j^{(2)}}{w_j}
\right)
\leq\rho,
$$

where:

$$
0<\rho<1,
$$

the selected coordinate satisfies:

$$
w_j^{(t+1)}
\leq
\rho w_j^{(t)}.
$$

For every other coordinate:

$$
w_k^{(t+1)}
=
w_k^{(t)},
\qquad
k\neq j.
$$

Therefore a split contracts exactly one coordinate direction.

### 42.4 Why Local Contraction Is Not Sufficient

Consider:

$$
d=2.
$$

Suppose every split selects direction \(1\).

Then:

$$
w_1^{(t)}\rightarrow0,
$$

while:

$$
w_2^{(t)}=w_2^{(0)}.
$$

Therefore:

$$
D_t
\not\rightarrow0.
$$

This demonstrates:

$$
\boxed{
\text{Local Contraction}
\not\Rightarrow
\text{Global Contraction}.
}
$$

A condition controlling the selection of coordinate directions is therefore necessary.

### 42.5 Directional Fairness

We define directional fairness as follows.

**Definition — Directional Fairness**

A refinement sequence is directionally fair if every coordinate direction that remains spatially unresolved in a persistently refined relevant region is selected for refinement infinitely often, unless that direction has already reached the configured numerical resolution limit.

Equivalently, for every unresolved direction \(j\):

$$
\#\{t:\text{direction }j
\text{ is selected at iteration }t\}
\rightarrow\infty.
$$

This condition prevents any unresolved coordinate direction from being permanently ignored.

### 42.6 Repeated Contraction in One Direction

Consider a fixed coordinate direction \(j\).

Suppose it is selected \(n_j(t)\) times by iteration \(t\).

Each such refinement multiplies its width by at most \(\rho\).

Therefore:

$$
w_j^{(t)}
\leq
\rho^{\,n_j(t)}
w_j^{(0)}.
$$

Because:

$$
0<\rho<1,
$$

we have:

$$
\rho^n\rightarrow0
\qquad
\text{as }n\rightarrow\infty.
$$

Directional fairness gives:

$$
n_j(t)\rightarrow\infty.
$$

Therefore:

$$
\boxed{
w_j^{(t)}\rightarrow0.
}
$$

This holds for every coordinate:

$$
j=1,\ldots,d.
$$

### 42.7 Global Contraction

Since:

$$
w_j^{(t)}\rightarrow0
\qquad
\forall j,
$$

we obtain:

$$
w_{\max}^{(t)}
=
\max_jw_j^{(t)}
\rightarrow0.
$$

Using:

$$
w_{\max}^{(t)}
\leq
D_t
\leq
\sqrt d\,w_{\max}^{(t)},
$$

we conclude:

$$
\boxed{
\operatorname{diam}(R_t)
\rightarrow0.
}
$$

Thus directional fairness combined with local contraction establishes global spatial contraction.

### 42.8 Contraction Proposition

The previous arguments establish the following proposition.

**Proposition — Multidimensional Spatial Contraction**

Let:

$$
R_t
=
\prod_{j=1}^{d}
[l_j^{(t)},r_j^{(t)}]
$$

be a sequence of regions generated by valid binary refinements.

Assume:

1. \(d<\infty\);
2. every selected split satisfies a contraction factor \(0<\rho<1\);
3. each split contracts the selected coordinate by at most \(\rho\);
4. every unresolved coordinate direction is selected infinitely often.

Then:

$$
\lim_{t\rightarrow\infty}
w_j^{(t)}
=
0
\qquad
\forall j,
$$

and therefore:

$$
\boxed{
\lim_{t\rightarrow\infty}
\operatorname{diam}(R_t)
=
0.
}
$$

### 42.9 Nested Region Interpretation

For a fixed refinement path containing a point \(x^\ast\), the regions satisfy:

$$
R_{t+1}\subseteq R_t.
$$

If:

$$
\operatorname{diam}(R_t)\rightarrow0,
$$

then the nested regions become arbitrarily small around the points they contain.

In particular, if:

$$
x^\ast\in R_t
\qquad
\forall t,
$$

then:

$$
\sup_{x\in R_t}
\|x-x^\ast\|_2
\leq
\operatorname{diam}(R_t)
\rightarrow0.
$$

Hence every point in the region approaches \(x^\ast\) in the sense of shrinking geometric containment.

### 42.10 Important Qualification

Directional fairness must be interpreted relative to a relevant refinement path.

It is not sufficient that every coordinate direction is occasionally selected somewhere in the global hierarchy.

For global optimization, the relevant regions containing globally important points must receive sufficient refinement.

Therefore two fairness conditions are conceptually distinct:

$$
\boxed{
\text{Directional Fairness}
+
\text{Global Region Fairness}.
}
$$

Directional fairness controls which coordinate directions are refined.

Global region fairness controls which spatial regions continue to receive refinement.

Both are required for the convergence argument.

### 42.11 Interaction With Global Selection

Suppose \(R_t^\ast\) denotes a sequence of globally relevant regions.

The global selection mechanism must ensure that regions containing unresolved globally relevant areas cannot be permanently ignored.

If:

$$
x^\ast\in R_t^\ast
$$

for the global maximizer \(x^\ast\), and the corresponding refinement path satisfies directional fairness, then:

$$
\operatorname{diam}(R_t^\ast)
\rightarrow0.
$$

This provides the spatial component required by the multidimensional convergence argument.

### 42.12 Interaction With Sampling

Spatial contraction alone does not produce a high-quality objective value.

The algorithm must also acquire objective information inside or sufficiently near the shrinking relevant regions.

Therefore the convergence chain becomes:

$$
\boxed{
\text{Global Region Fairness}
+
\text{Directional Fairness}
+
\text{Local Contraction}
}
$$

leading to:

$$
\boxed{
\operatorname{diam}(R_t^\ast)\rightarrow0.
}
$$

Sampling and information acquisition then connect this spatial result to objective-value convergence.

### 42.13 One-Dimensional Reduction

When:

$$
d=1,
$$

there is only one coordinate direction.

Therefore directional fairness is automatically satisfied whenever refinement continues.

The width becomes:

$$
w_1^{(t)}
=
r^{(t)}-l^{(t)},
$$

and:

$$
\operatorname{diam}(R_t)
=
w_1^{(t)}.
$$

Thus the multidimensional contraction proposition reduces to the original one-dimensional contraction argument.

### 42.14 Consequence for the ARRGO Framework

The proposition establishes an important extension of the one-dimensional theory.

ARRGO does not require every coordinate to be refined simultaneously.

Instead, it is sufficient that:

1. each valid split contracts its selected direction;
2. unresolved directions are eventually selected;
3. globally relevant regions are not permanently ignored.

Under these conditions:

$$
\boxed{
\text{Local Directional Refinement}
\Rightarrow
\text{Global Spatial Contraction}.
}
$$

This result is the geometric bridge between the one-dimensional refinement mechanism and the multidimensional convergence framework.

It does not yet establish objective-value convergence by itself.

The next step is to formalize the information-acquisition condition required to connect shrinking regions to evaluations approaching the global optimum.

## 43. Formal Information Acquisition Condition in Multidimensional ARRGO

The previous section established that directional fairness and local contraction can force the diameter of a relevant refinement path to converge to zero.

Spatial contraction alone, however, is not sufficient for global optimization.

ARRGO also requires objective information to be acquired sufficiently close to globally relevant points.

This section formalizes the information-acquisition condition required to connect geometric refinement with objective-value convergence.

### 43.1 Global Maximizer

Let:

$$
x^\ast\in\Omega
$$

be a global maximizer of the objective:

$$
f(x^\ast)
=
f^\ast
=
\max_{x\in\Omega}f(x).
$$

The objective is assumed to be continuous on the compact domain \(\Omega\).

Therefore such an \(x^\ast\) exists.

### 43.2 Relevant Region Sequence

Consider a sequence of regions:

$$
R_t^\ast
$$

such that:

$$
x^\ast\in R_t^\ast
\qquad
\forall t.
$$

Assume the regions form a nested refinement path:

$$
R_{t+1}^\ast
\subseteq
R_t^\ast.
$$

From the previous contraction result:

$$
\operatorname{diam}(R_t^\ast)
\rightarrow0.
$$

Thus:

$$
R_t^\ast
\rightarrow
x^\ast
$$

in the sense of geometric containment.

### 43.3 Evaluation Distance to the Optimum

For an evaluation point \(x_i\), define its distance from the global maximizer:

$$
d_i^\ast
=
\|x_i-x^\ast\|_2.
$$

The objective information becomes increasingly relevant to the optimum as:

$$
d_i^\ast\rightarrow0.
$$

However, ARRGO does not know \(x^\ast\) in advance.

Therefore the algorithm cannot explicitly minimize \(d_i^\ast\).

The convergence condition must instead be expressed through the selected region and its information-acquisition policy.

### 43.4 Information Acquisition Condition

We define the following condition.

**Definition — Multidimensional Information Acquisition**

A refinement and sampling process satisfies the information-acquisition condition at \(x^\ast\) if, for every:

$$
\epsilon>0,
$$

there exists an evaluation point \(x_i\) obtained after a sufficiently large iteration such that:

$$
\|x_i-x^\ast\|_2<\delta_\epsilon,
$$

where continuity of \(f\) guarantees:

$$
|f(x_i)-f(x^\ast)|<\epsilon.
$$

Equivalently, the process generates evaluations arbitrarily close to \(x^\ast\):

$$
\inf_{e_i\in D_t}
\|x_i-x^\ast\|_2
\rightarrow0.
$$

This is the key information condition required for objective-value convergence.

### 43.5 Why Region Contraction Helps

Suppose:

$$
x^\ast\in R_t^\ast
$$

and:

$$
\operatorname{diam}(R_t^\ast)\rightarrow0.
$$

If the sampling mechanism continues to acquire evaluations inside \(R_t^\ast\), then any sufficiently refined sample is geometrically close to \(x^\ast\).

For:

$$
x_i\in R_t^\ast,
$$

we have:

$$
\|x_i-x^\ast\|_2
\leq
\operatorname{diam}(R_t^\ast).
$$

Therefore:

$$
\operatorname{diam}(R_t^\ast)\rightarrow0
\quad\Rightarrow\quad
\|x_i-x^\ast\|_2\rightarrow0
$$

for evaluations acquired from sufficiently refined relevant regions.

### 43.6 Continuity of the Objective

Because \(f\) is continuous at \(x^\ast\), for every:

$$
\epsilon>0,
$$

there exists:

$$
\delta>0
$$

such that:

$$
\|x-x^\ast\|_2<\delta
$$

implies:

$$
|f(x)-f(x^\ast)|<\epsilon.
$$

Therefore, once ARRGO obtains an evaluation satisfying:

$$
\|x_i-x^\ast\|_2<\delta,
$$

we obtain:

$$
f(x_i)>f^\ast-\epsilon.
$$

Since:

$$
f_{\mathrm{best}}^{(t)}
\geq
f(x_i),
$$

it follows that:

$$
f_{\mathrm{best}}^{(t)}
>
f^\ast-\epsilon.
$$

### 43.7 Objective-Value Convergence

The incumbent is always bounded above by the global optimum:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast.
$$

From the information-acquisition argument, for every:

$$
\epsilon>0,
$$

there exists a sufficiently large iteration such that:

$$
f_{\mathrm{best}}^{(t)}
>
f^\ast-\epsilon.
$$

Therefore:

$$
\boxed{
\lim_{t\rightarrow\infty}
f_{\mathrm{best}}^{(t)}
=
f^\ast.
}
$$

This establishes the objective-value component of global convergence.

### 43.8 Information Acquisition Does Not Require Knowing the Optimum

The condition does not require ARRGO to know:

$$
x^\ast.
$$

It is a property of the refinement and sampling process.

The algorithm uses observable information such as:

- regional coverage;
- observed objective behavior;
- geometric resolution;
- certified uncertainty when available;
- optimization potential.

The theoretical condition states what must eventually happen near globally relevant regions.

It does not require the algorithm to explicitly calculate the unknown optimum.

### 43.9 Sufficient Sampling Condition

A practical sufficient condition is:

> Every globally relevant region whose diameter continues to decrease must receive infinitely many valid sampling opportunities, unless the region has already reached a valid terminal resolution.

Under this condition, together with:

$$
\operatorname{diam}(R_t^\ast)\rightarrow0,
$$

the evaluation sequence contains points approaching \(x^\ast\).

Thus:

$$
\boxed{
\text{Relevant Region Refinement}
+
\text{Continued Information Acquisition}
}
$$

implies:

$$
\boxed{
\text{Evaluations Approach the Global Optimum}.
}
$$

### 43.10 Sampling Is Not the Same as Splitting

A critical distinction must be preserved.

Splitting changes:

$$
R_t
\rightarrow
\{R_{t+1}^{(1)},R_{t+1}^{(2)}\}.
$$

Sampling changes:

$$
D_t
\rightarrow
D_{t+1}.
$$

Therefore:

$$
\boxed{
\text{Splitting}
\rightarrow
\text{Spatial Resolution},
}
$$

while:

$$
\boxed{
\text{Sampling}
\rightarrow
\text{Objective Information}.
}
$$

Both mechanisms are required for the complete convergence argument.

### 43.11 Coverage and Information Acquisition

Multidimensional coverage provides a mechanism for identifying regions where spatial information remains incomplete.

For example, the geometric covering radius:

$$
h(R)
=
\sup_{x\in R}
\min_{x_i\in D_R}
\|x-x_i\|_2
$$

measures the largest distance from a point in the region to the nearest observed evaluation.

A decreasing covering radius provides increasingly dense geometric information.

However:

$$
h(R)\rightarrow0
$$

is stronger than merely reducing coordinate-wise gaps.

Therefore the theoretical information condition should be understood as a geometric requirement, while coordinate-wise coverage can serve as an implementation-level approximation or diagnostic.

### 43.12 Joint Multidimensional Information

The objective remains:

$$
f(x_1,\ldots,x_d).
$$

Therefore information acquisition must preserve possible interactions between coordinates.

A collection of coordinate-wise samples does not necessarily provide sufficient information about joint behavior.

The sampling mechanism must therefore be capable of producing genuine multidimensional points:

$$
x=(x_1,\ldots,x_d),
$$

rather than treating the optimization problem as \(d\) independent one-dimensional problems.

### 43.13 Information Acquisition Proposition

The preceding arguments establish the following proposition.

**Proposition — Multidimensional Information Acquisition**

Let \(f\) be continuous on:

$$
\Omega\subset\mathbb{R}^d,
$$

and let \(x^\ast\) be a global maximizer.

Suppose there exists a sequence of relevant regions \(R_t^\ast\) satisfying:

$$
x^\ast\in R_t^\ast,
$$

and:

$$
\operatorname{diam}(R_t^\ast)\rightarrow0.
$$

If the sampling mechanism acquires evaluations from these regions sufficiently often so that:

$$
\inf_{e_i\in D_t}
\|x_i-x^\ast\|_2
\rightarrow0,
$$

then:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast.
}
$$

### 43.14 Proof

Let:

$$
\epsilon>0.
$$

By continuity of \(f\) at \(x^\ast\), there exists:

$$
\delta>0
$$

such that:

$$
\|x-x^\ast\|_2<\delta
\Rightarrow
|f(x)-f^\ast|<\epsilon.
$$

Because:

$$
\inf_{e_i\in D_t}
\|x_i-x^\ast\|_2
\rightarrow0,
$$

there exists \(T\) such that for some evaluation \(x_i\), with \(i\geq T\),

$$
\|x_i-x^\ast\|_2<\delta.
$$

Therefore:

$$
f(x_i)>f^\ast-\epsilon.
$$

Since the incumbent is the best observed value:

$$
f_{\mathrm{best}}^{(i)}
\geq
f(x_i),
$$

so:

$$
f_{\mathrm{best}}^{(i)}
>
f^\ast-\epsilon.
$$

At the same time:

$$
f_{\mathrm{best}}^{(i)}
\leq
f^\ast.
$$

Hence:

$$
f^\ast-\epsilon
<
f_{\mathrm{best}}^{(i)}
\leq
f^\ast.
$$

Because \(\epsilon\) is arbitrary:

$$
\boxed{
\lim_{t\rightarrow\infty}
f_{\mathrm{best}}^{(t)}
=
f^\ast.
}
$$

### 43.15 Combined Spatial-Information Principle

The previous contraction and information results can now be combined:

$$
\boxed{
\begin{aligned}
&\text{Valid Local Contraction}
\\
&+
\text{Directional Fairness}
\\
&+
\text{Global Region Fairness}
\\
&+
\text{Information Acquisition}
\\
&+
\text{Continuity}
\\
&\Downarrow
\\
&\text{Global Objective-Value Convergence}.
\end{aligned}
}
$$

This is the central mathematical bridge from the geometry of multidimensional regions to global optimization behavior.

### 43.16 One-Dimensional Reduction

For:

$$
d=1,
$$

the Euclidean distance becomes:

$$
\|x-x^\ast\|_2
=
|x-x^\ast|.
$$

The region becomes:

$$
R_t=[l_t,r_t],
$$

and:

$$
\operatorname{diam}(R_t)
=
r_t-l_t.
$$

Thus the information-acquisition condition reduces to the original one-dimensional requirement that evaluations approach the relevant optimum region.

### 43.17 Scope of the Result

This result establishes a sufficient convergence condition.

It does not claim that every possible implementation of ARRGO automatically satisfies the information-acquisition condition.

Instead, it identifies the property that the implementation and refinement policy must satisfy for the convergence theorem to apply.

Therefore:

$$
\boxed{
\text{Implementation}
+
\text{Fair Refinement}
+
\text{Sufficient Information Acquisition}
}
$$

is required to realize the theoretical convergence result.

The next step is to formalize global region fairness and show how the global selection mechanism can satisfy the relevance condition required by the convergence argument.

## 44. Formal Proof of Global Region Fairness

The previous sections established two essential components of the multidimensional convergence argument:

1. valid directional refinement can reduce the diameter of a relevant region;
2. sufficient information acquisition inside such a region allows the incumbent objective value to approach the global optimum.

A remaining requirement is to ensure that a globally relevant region is not permanently ignored by the global selection mechanism.

This section formalizes the concept of Global Region Fairness and establishes its role in the convergence argument.

### 44.1 Motivation for Global Region Fairness

ARRGO maintains multiple regions simultaneously.

At any iteration, the algorithm must decide which region receives the next refinement action.

Therefore, even if a region containing a global maximizer is geometrically refinable, convergence may fail if the global selection mechanism permanently ignores that region.

This creates a distinction between two different fairness requirements:

$$
\boxed{
\text{Global Region Fairness}
}
$$

determines whether a relevant region continues to receive attention, while:

$$
\boxed{
\text{Directional Fairness}
}
$$

determines whether all unresolved spatial directions inside that region are eventually refined.

These two conditions operate at different levels of the hierarchy.

### 44.2 Eligible Region Set

Let the hierarchy at iteration \(t\) be:

$$
\mathcal H_t.
$$

Define the set of regions eligible for further refinement as:

$$
\mathcal E_t
\subseteq
\mathcal H_t.
$$

An eligible region must satisfy the structural and numerical requirements necessary for further processing.

For example, a region may be ineligible if:

- its numerical width has reached the resolution limit;
- no valid refinement action exists;
- the configuration prevents further refinement;
- the region has reached a valid terminal state.

Eligibility is therefore distinct from global relevance.

### 44.3 Globally Relevant Regions

A region is globally relevant when information from that region can still affect the global optimization state.

In empirical mode, relevance may be associated with unresolved objective information and the possibility of improving the incumbent.

In Certified Mode, a stronger criterion can be obtained from the regional potential:

$$
P(R)
=
\sup_{x\in R}U_R(x).
$$

A region with:

$$
P(R)>f_{\mathrm{best}}
$$

cannot yet be ruled out as a source of an improvement under the certified model.

Thus define the certified relevant set:

$$
\mathcal G_t
=
\left\{
R\in\mathcal F_t:
P(R)>f_{\mathrm{best}}^{(t)}
\right\},
$$

where \(\mathcal F_t\) is the certification frontier.

This definition is intentionally different from pruning.

A region may be considered globally irrelevant for the current certified decision without being deleted from the hierarchy.

### 44.4 Global Region Fairness

We now define the central condition.

**Definition — Global Region Fairness**

A global region-selection mechanism is globally fair if every region that remains globally relevant and eligible for refinement cannot be ignored indefinitely.

Formally, let:

$$
R^\ast_t
$$

denote a region on a persistent relevant refinement path containing a global maximizer \(x^\ast\).

If \(R^\ast_t\) remains eligible and globally relevant over infinitely many iterations, then the selection mechanism must select that refinement path infinitely often.

Equivalently:

$$
\boxed{
R^\ast_t
\text{ remains relevant and eligible}
\Rightarrow
\text{it receives infinitely many refinement opportunities}.
}
$$

This is a fairness condition rather than a particular scoring formula.

### 44.5 Why Greedy Selection Alone Is Not Sufficient

Suppose two regions exist:

$$
R_1
\quad\text{and}\quad
R_2.
$$

Assume \(R_1\) currently has a slightly larger observed value than \(R_2\).

A purely greedy strategy may repeatedly select \(R_1\):

$$
R_1,R_1,R_1,\ldots
$$

even though \(R_2\) contains the global maximizer.

Such a policy may improve the incumbent temporarily while permanently neglecting the globally optimal region.

Therefore:

$$
\boxed{
\text{Current Best Value}
\neq
\text{Sufficient Global Selection Criterion}.
}
$$

A globally convergent strategy must preserve access to regions whose relevance has not been eliminated.

### 44.6 Fairness Does Not Mean Uniform Selection

Global fairness does not require:

$$
\Pr(R_i\text{ selected})
=
\Pr(R_j\text{ selected}).
$$

In fact, ARRGO is deterministic and does not require probabilistic selection.

Fairness means only that a relevant eligible region cannot be permanently starved.

Therefore a deterministic priority policy can still be globally fair.

The selection sequence may strongly prefer one region for many iterations while still eventually returning to other relevant regions.

### 44.7 Deterministic Fairness

Because ARRGO is deterministic, the selection mechanism can be represented as:

$$
R_t
=
S(G_t),
$$

where \(S\) is a deterministic region-selection function and \(G_t\) is the global state.

The sequence:

$$
R_1,R_2,\ldots
$$

is therefore uniquely determined by:

$$
\Theta,\ f,\ G_0.
$$

Global fairness requires that this deterministic sequence does not permanently starve any persistent relevant refinement path.

Thus:

$$
\boxed{
\text{Deterministic}
\not\Rightarrow
\text{Unfair}.
}
$$

Determinism and fairness are separate properties.

### 44.8 Fairness Along a Refinement Path

Consider a sequence of regions:

$$
R_0^\ast
\supseteq
R_1^\ast
\supseteq
R_2^\ast
\supseteq
\cdots
$$

such that:

$$
x^\ast\in R_t^\ast.
$$

For convergence, it is not enough for the root region \(R_0^\ast\) to be selected.

The descendants containing \(x^\ast\) must continue receiving refinement opportunities.

Therefore global fairness must apply recursively along the relevant refinement path.

If:

$$
R_t^\ast
$$

is replaced by a child containing \(x^\ast\), then that child becomes the next relevant region.

This gives the chain:

$$
R_0^\ast
\rightarrow
R_1^\ast
\rightarrow
R_2^\ast
\rightarrow
\cdots
$$

and global fairness requires continued access to this path.

### 44.9 Interaction with Directional Fairness

Global fairness and directional fairness are complementary.

Global fairness guarantees:

$$
\text{Which region receives attention?}
$$

Directional fairness guarantees:

$$
\text{Which spatial directions receive refinement?}
$$

Therefore the complete structural condition is:

$$
\boxed{
\text{Global Region Fairness}
+
\text{Directional Fairness}.
}
$$

Without global fairness, the relevant region may never be selected.

Without directional fairness, the selected region may repeatedly refine only one coordinate direction while leaving another unresolved.

Both levels are therefore required.

### 44.10 Interaction with Information Acquisition

Suppose global fairness guarantees that:

$$
R_t^\ast
$$

continues receiving refinement opportunities.

Suppose directional fairness gives:

$$
\operatorname{diam}(R_t^\ast)
\rightarrow0.
$$

If the sampling mechanism also satisfies the information-acquisition condition, then evaluations approach \(x^\ast\).

Thus the complete chain becomes:

$$
\boxed{
\begin{aligned}
&\text{Global Region Fairness}
\\
&\Downarrow
\\
&\text{Relevant Region Receives Refinement}
\\
&\Downarrow
\\
&\text{Directional Fairness}
\\
&\Downarrow
\\
&\operatorname{diam}(R_t^\ast)\rightarrow0
\\
&\Downarrow
\\
&\text{Information Acquisition}
\\
&\Downarrow
\\
&f_{\mathrm{best}}^{(t)}\rightarrow f^\ast.
\end{aligned}
}
$$

### 44.11 Certified Relevance

In Certified Mode, the potential provides a mathematically meaningful relevance signal.

For a frontier region \(R\):

$$
P(R)
=
\sup_{x\in R}U_R(x).
$$

If:

$$
P(R)
\leq
f_{\mathrm{best}},
$$

then that region cannot contain an objective value larger than the current incumbent under the valid certified model.

Such a region need not receive further refinement for certification progress.

However, ARRGO does not delete it.

Therefore:

$$
\boxed{
\text{Certification Irrelevance}
\neq
\text{Deletion}.
}
$$

The region remains represented in the structural hierarchy.

### 44.12 Frontier-Based Global Fairness

Because ARRGO retains parents after splitting, the structural hierarchy contains historical regions.

Certification and global refinement should therefore operate on an appropriate current frontier.

Let:

$$
\mathcal F_t
$$

be a frontier whose regions cover the domain:

$$
\Omega
\subseteq
\bigcup_{R\in\mathcal F_t}R.
$$

The globally relevant certification set is then:

$$
\mathcal G_t
=
\left\{
R\in\mathcal F_t:
P(R)>f_{\mathrm{best}}^{(t)}
\right\}.
$$

Global fairness should be interpreted primarily over eligible frontier regions and persistent refinement paths rather than over every historical parent in \(\mathcal H_t\).

This preserves both:

- the no-pruning structural hierarchy;
- the correctness of current global reasoning.

### 44.13 Fairness and Stable Regions

A region may become:

$$
\mathrm{Stable}.
$$

Stability means that no unresolved local refinement objective currently requires additional action.

A stable region does not necessarily need to be selected repeatedly.

Therefore fairness applies to regions that remain:

$$
\text{relevant}
+
\text{eligible}
+
\text{unresolved}.
$$

If a region legitimately becomes stable, temporary non-selection does not violate fairness.

However, if new global information makes the region relevant again, the selection mechanism must allow it to become eligible again.

Thus stability is not permanent deletion.

### 44.14 Fairness and Global Incumbent Improvement

The global incumbent satisfies:

$$
f_{\mathrm{best}}^{(t+1)}
\geq
f_{\mathrm{best}}^{(t)}.
$$

An improvement in another region can change the relative relevance of all regions.

Therefore global fairness must be evaluated dynamically.

A region that is relevant at iteration \(t\) may become irrelevant at iteration \(t+1\).

Conversely, information acquired elsewhere may change the global state and cause a previously stable region to become relevant again.

The selection mechanism must therefore operate on the current global state rather than on a permanently fixed ranking.

### 44.15 Global Fairness Proposition

**Proposition — Global Region Fairness**

Assume that:

1. the domain is represented by a valid finite-dimensional region hierarchy;
2. the certification frontier remains a valid cover when certification is used;
3. globally relevant regions are identifiable from the current global state;
4. the region-selection mechanism is deterministic;
5. every persistent relevant and eligible refinement path receives infinitely many refinement opportunities.

Then no persistent globally relevant refinement path is permanently starved.

In particular, if a global maximizer \(x^\ast\) remains contained in a relevant and eligible refinement path, that path receives infinitely many refinement opportunities.

### 44.16 Proof

Consider a persistent refinement path:

$$
R_0^\ast
\supseteq
R_1^\ast
\supseteq
R_2^\ast
\supseteq
\cdots
$$

such that:

$$
x^\ast\in R_t^\ast
$$

for every stage of the path.

Assume that the path remains globally relevant and eligible.

By the global fairness assumption, the path receives infinitely many refinement opportunities.

Therefore the algorithm cannot permanently ignore the path.

Whenever the path is selected, a valid refinement action can either:

1. acquire additional objective information through sampling; or
2. reduce the region geometrically through structural splitting.

If directional fairness also holds, repeated refinement along the path reduces all unresolved coordinate widths.

Therefore:

$$
\operatorname{diam}(R_t^\ast)
\rightarrow0.
$$

If the information-acquisition condition also holds, evaluations approach \(x^\ast\).

By continuity of \(f\):

$$
f(x_i)
\rightarrow
f^\ast
$$

along the corresponding informative evaluation sequence.

Since the incumbent is the maximum observed value:

$$
f_{\mathrm{best}}^{(t)}
\geq
f(x_i),
$$

while always:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast.
$$

Therefore:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast.
}
$$

### 44.17 Fairness Is a Sufficient Condition

Global fairness is a sufficient condition for the convergence argument.

It is not claimed to be the only possible mechanism capable of producing convergence.

Other selection policies may satisfy an equivalent property under different formalizations.

The essential requirement is:

$$
\boxed{
\text{No Persistent Relevant Region Is Permanently Starved}.
}
$$

The exact implementation of this principle may vary.

### 44.18 Relation to the Global Selection Mechanism

The previous algorithmic sections defined global region selection using information relevance, optimization potential, spatial resolution, and deterministic priority.

The present proof identifies the theoretical property that such a mechanism must satisfy.

Therefore the relationship is:

$$
\boxed{
\text{Global Selection Policy}
\Longrightarrow
\text{Fairness Property}.
}
$$

The policy itself is an algorithmic design choice.

Fairness is the mathematical condition required by the convergence proof.

This distinction prevents a specific heuristic ranking rule from being incorrectly presented as a theorem.

### 44.19 One-Dimensional Reduction

For:

$$
d=1,
$$

there is only one spatial direction.

Therefore directional fairness becomes automatic whenever valid refinement continues.

The remaining fairness requirement is the global region-selection condition:

$$
\text{relevant interval}
\rightarrow
\text{continued refinement opportunities}.
$$

Thus the multidimensional fairness framework reduces naturally to the original one-dimensional setting.

### 44.20 Combined Convergence Requirements

The results established so far can now be combined into a stronger sufficient condition:

$$
\boxed{
\begin{aligned}
&\text{Compact Domain}
\\
&+
\text{Continuity}
\\
&+
\text{Valid Region Splitting}
\\
&+
\text{Spatial Contraction}
\\
&+
\text{Global Region Fairness}
\\
&+
\text{Directional Fairness}
\\
&+
\text{Information Acquisition}
\\
&\Longrightarrow
\\
&\text{Global Objective-Value Convergence}.
\end{aligned}
}
$$

This result still does not establish location convergence.

If the global maximizer is not unique, the sequence of incumbent locations may converge to any point in the maximizer set.

Even when the maximizer is unique, additional conditions are required to prove convergence of the locations themselves.

### 44.21 Central Interpretation

The multidimensional ARRGO convergence mechanism can therefore be viewed as a three-level process:

$$
\boxed{
\begin{aligned}
&\text{Global Level:}
&&\text{Do not permanently ignore relevant regions}
\\
&\text{Geometric Level:}
&&\text{Do not permanently ignore unresolved directions}
\\
&\text{Information Level:}
&&\text{Acquire objective information near relevant points}.
\end{aligned}
}
$$

These three levels connect global selection, spatial refinement, and objective evaluation into a single convergence argument.

The next step is to combine the spatial contraction, information acquisition, and global fairness results into one complete multidimensional convergence theorem.

## 45. Complete Multidimensional Global Convergence Theorem

The previous sections established three separate sufficient mechanisms:

1. spatial contraction of relevant regions;
2. information acquisition near globally optimal points;
3. global fairness of the region-selection mechanism.

This section combines these mechanisms into a single multidimensional convergence theorem.

The purpose is to establish the mathematical conditions under which ARRGO converges in objective value to the global optimum in any finite dimension.

### 45.1 Theorem Statement

Let:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j]
\subset\mathbb R^d,
\qquad
d<\infty,
$$

be a compact hyper-rectangular domain.

Let:

$$
f:\Omega\rightarrow\mathbb R
$$

be deterministic and continuous.

Consider an ARRGO process that maintains a hierarchy of valid regions and a global evaluation history.

Assume the following conditions.

#### A1 — Valid Domain

The domain is bounded and closed:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j].
$$

Therefore \(\Omega\) is compact.

#### A2 — Deterministic Objective

Every valid evaluation of a point \(x\in\Omega\) returns the deterministic value:

$$
f(x).
$$

No stochastic objective noise is considered.

#### A3 — Valid Region Representation

Every active region is a valid hyper-rectangle:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j]
\subseteq
\Omega.
$$

Every structural split produces valid child regions contained in the parent.

#### A4 — Valid Spatial Contraction

Whenever a region is structurally refined in direction \(j\), the corresponding side length satisfies:

$$
w_j^{(t+1)}
\leq
\rho w_j^{(t)},
$$

where:

$$
0<\rho<1.
$$

#### A5 — Global Region Fairness

Every persistent region path that remains globally relevant and eligible for refinement receives infinitely many refinement opportunities.

#### A6 — Directional Fairness

For every persistently refined relevant region, every unresolved spatial direction is selected infinitely often unless that direction has reached a valid numerical resolution limit.

#### A7 — Information Acquisition

Along every relevant refinement path containing a global maximizer \(x^\ast\), the evaluation process acquires points satisfying:

$$
\inf_{e_i\in D_t}
\|x_i-x^\ast\|_2
\rightarrow0.
$$

#### A8 — Exact Objective Evaluation

Evaluations used by the incumbent are exact evaluations of the deterministic objective:

$$
y_i=f(x_i).
$$

#### A9 — Valid Global Incumbent

At every iteration:

$$
f_{\mathrm{best}}^{(t)}
=
\max_{e_i\in D_t}f(x_i).
$$

Then:

$$
\boxed{
\lim_{t\rightarrow\infty}
f_{\mathrm{best}}^{(t)}
=
f^\ast
}
$$

where:

$$
f^\ast
=
\max_{x\in\Omega}f(x).
$$

This is objective-value global convergence.

### 45.2 Existence of a Global Optimum

Because \(\Omega\) is compact and \(f\) is continuous, the extreme value theorem guarantees the existence of at least one:

$$
x^\ast\in\Omega
$$

such that:

$$
f(x^\ast)
=
f^\ast
=
\max_{x\in\Omega}f(x).
$$

Therefore the target value \(f^\ast\) is well-defined.

### 45.3 Incumbent Monotonicity

The evaluation history is cumulative:

$$
D_t
\subseteq
D_{t+1}.
$$

Therefore:

$$
f_{\mathrm{best}}^{(t+1)}
=
\max_{e_i\in D_{t+1}}f(x_i)
\geq
\max_{e_i\in D_t}f(x_i).
$$

Hence:

$$
\boxed{
f_{\mathrm{best}}^{(t+1)}
\geq
f_{\mathrm{best}}^{(t)}.
}
$$

The best observed objective value is monotonically nondecreasing.

### 45.4 Upper Bound by the Global Optimum

Every evaluated point belongs to the domain:

$$
x_i\in\Omega.
$$

Therefore:

$$
f(x_i)
\leq
f^\ast.
$$

Taking the maximum over all observed evaluations gives:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
\leq
f^\ast.
}
$$

Thus the incumbent is bounded above by the global optimum.

### 45.5 Existence of a Relevant Refinement Path

Let \(x^\ast\) be a global maximizer.

The hierarchy initially contains the root region:

$$
R_0=\Omega.
$$

Because every valid split preserves domain coverage, there exists a descendant region containing \(x^\ast\) whenever the region containing \(x^\ast\) is split.

Thus there exists a nested sequence:

$$
R_0^\ast
\supseteq
R_1^\ast
\supseteq
R_2^\ast
\supseteq
\cdots
$$

such that:

$$
x^\ast\in R_t^\ast.
$$

This sequence represents the refinement path relevant to the global maximizer.

### 45.6 Global Fairness Guarantees Continued Access

By A5, if the relevant path remains globally relevant and eligible, it cannot be permanently ignored.

Therefore the path receives infinitely many refinement opportunities.

This prevents the algorithm from permanently focusing on unrelated regions.

Hence:

$$
\boxed{
\text{Global Relevance}
+
\text{Global Fairness}
\Rightarrow
\text{Continued Access to the Relevant Path}.
}
$$

### 45.7 Directional Fairness Produces Spatial Contraction

By A6, every unresolved coordinate direction on the relevant refinement path is eventually refined.

Let:

$$
w_j^{(0)}
$$

be the initial width in direction \(j\).

If direction \(j\) is selected \(n_j(t)\) times, then A4 gives:

$$
w_j^{(t)}
\leq
\rho^{n_j(t)}
w_j^{(0)}.
$$

Because A6 requires:

$$
n_j(t)\rightarrow\infty,
$$

and:

$$
0<\rho<1,
$$

we obtain:

$$
\rho^{n_j(t)}
\rightarrow0.
$$

Therefore:

$$
w_j^{(t)}
\rightarrow0
\qquad
\forall j.
$$

Consequently:

$$
\operatorname{diam}(R_t^\ast)
=
\sqrt{
\sum_{j=1}^{d}
\left(w_j^{(t)}\right)^2
}
\rightarrow0.
$$

Hence:

$$
\boxed{
\operatorname{diam}(R_t^\ast)
\rightarrow0.
}
$$

### 45.8 Information Acquisition Near the Optimum

By A7:

$$
\inf_{e_i\in D_t}
\|x_i-x^\ast\|_2
\rightarrow0.
$$

Therefore, for every:

$$
\delta>0,
$$

there exists a sufficiently large iteration \(T\) and an evaluated point \(x_i\), with \(i\geq T\), such that:

$$
\|x_i-x^\ast\|_2<\delta.
$$

Because \(f\) is continuous at \(x^\ast\), for every:

$$
\epsilon>0,
$$

there exists \(\delta>0\) satisfying:

$$
\|x-x^\ast\|_2<\delta
\Rightarrow
|f(x)-f^\ast|<\epsilon.
$$

Therefore sufficiently informative evaluations satisfy:

$$
f(x_i)
>
f^\ast-\epsilon.
$$

### 45.9 Incumbent Approaches the Global Optimum

Since:

$$
f_{\mathrm{best}}^{(i)}
\geq
f(x_i),
$$

we obtain:

$$
f_{\mathrm{best}}^{(i)}
>
f^\ast-\epsilon.
$$

At the same time, from Section 45.4:

$$
f_{\mathrm{best}}^{(i)}
\leq
f^\ast.
$$

Therefore:

$$
f^\ast-\epsilon
<
f_{\mathrm{best}}^{(i)}
\leq
f^\ast.
$$

Since \(\epsilon>0\) is arbitrary:

$$
\boxed{
\lim_{t\rightarrow\infty}
f_{\mathrm{best}}^{(t)}
=
f^\ast.
}
$$

This proves global convergence in objective value.

### 45.10 Main Convergence Chain

The complete logical dependency can be expressed as:

$$
\boxed{
\begin{aligned}
&\text{Compact Domain}
+
\text{Continuity}
\\
&\Downarrow
\\
&\text{Global Optimum Exists}
\\
&\Downarrow
\\
&\text{Global Fairness}
\\
&+
\text{Directional Fairness}
\\
&\Downarrow
\\
&\text{Relevant Regions Continue to Refine}
\\
&\Downarrow
\\
&\operatorname{diam}(R_t^\ast)\rightarrow0
\\
&\Downarrow
\\
&\text{Information Acquisition Near }x^\ast
\\
&\Downarrow
\\
&f(x_i)\rightarrow f^\ast
\\
&\Downarrow
\\
&f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast.
\end{aligned}
}
$$

This is the central global convergence argument for finite-dimensional ARRGO.

### 45.11 What Has Actually Been Proved

The theorem proves:

$$
\boxed{
\text{Objective-Value Global Convergence}.
}
$$

That is:

$$
f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast.
$$

It does not automatically prove:

$$
x_{\mathrm{best}}^{(t)}
\rightarrow
x^\ast.
$$

The distinction is important when multiple global maximizers exist.

### 45.12 Location Convergence

Suppose the global maximizer is unique:

$$
\operatorname*{arg\,max}_{x\in\Omega}f(x)
=
\{x^\ast\}.
$$

Under suitable additional assumptions ensuring that near-optimal objective values can only occur close to \(x^\ast\), objective-value convergence can imply:

$$
\|x_{\mathrm{best}}^{(t)}-x^\ast\|_2
\rightarrow0.
$$

However, uniqueness alone should not be used as an automatic substitute for this additional isolation property.

Therefore the primary theorem remains the objective-value result.

### 45.13 Non-Unique Global Optima

If the maximizer set is:

$$
\mathcal X^\ast
=
\left\{
x\in\Omega:
f(x)=f^\ast
\right\},
$$

then convergence of the objective value only requires the incumbent to approach this set in objective value.

Different executions or refinement paths may approach different points in:

$$
\mathcal X^\ast.
$$

This is consistent with the flat-peak behavior observed in the one-dimensional experiments.

### 45.14 No Probabilistic Assumption

The theorem does not require:

$$
\Pr(x_t\in R)>0
$$

or any stochastic sampling assumption.

All conditions are deterministic.

The information-acquisition and fairness requirements describe deterministic properties of the refinement process.

Therefore the theorem is consistent with the deterministic design of ARRGO.

### 45.15 No Pruning Requirement

The proof does not require deletion of regions.

Parents may remain in the structural hierarchy:

$$
R_{\mathrm{parent}}
\in
\mathcal H_t
$$

after splitting.

The convergence argument operates on the relevant refinement path and current frontier rather than requiring historical regions to be removed.

Therefore:

$$
\boxed{
\text{Parent Retention}
\not\Rightarrow
\text{Failure of Convergence}.
}
$$

The hierarchy may preserve complete structural history while the refinement process operates on currently relevant regions.

### 45.16 Empirical Mode

In Empirical Mode, the theorem applies only if the stated structural and information-acquisition assumptions are satisfied.

The implementation does not obtain a mathematical certificate merely from observing convergence in experiments.

Therefore an empirical result should be reported as:

$$
\boxed{
\text{Best Found Solution}.
}
$$

not automatically as:

$$
\boxed{
\text{Certified Global Optimum}.
}
$$

### 45.17 Certified Mode

If, in addition, the objective satisfies a valid global Lipschitz condition:

$$
|f(x)-f(y)|_2
\leq
L\|x-y\|_2,
$$

then the multidimensional upper and lower envelopes provide certified bounds.

For every relevant region:

$$
L_R(x)
\leq
f(x)
\leq
U_R(x).
$$

Therefore:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast
\leq
P_{\mathrm{global}}^{(t)}.
$$

Define:

$$
\Delta_{\mathrm{global}}^{(t)}
=
P_{\mathrm{global}}^{(t)}
-
f_{\mathrm{best}}^{(t)}.
$$

If:

$$
\Delta_{\mathrm{global}}^{(t)}
\leq
\epsilon,
$$

then:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
\geq
f^\ast-\epsilon.
}
$$

Thus Certified Mode provides an explicit epsilon-optimality statement under the validity of the supplied Lipschitz constant and the correctness of the certification frontier.

### 45.18 Finite-Dimensional Requirement

The theorem assumes:

$$
d<\infty.
$$

This is important because the diameter relation:

$$
w_{\max}(R)
\leq
\operatorname{diam}(R)
\leq
\sqrt d\,w_{\max}(R)
$$

depends on finite \(d\).

The theorem therefore establishes a generalization to arbitrary finite-dimensional spaces:

$$
\boxed{
d=1,2,3,\ldots
}
$$

without requiring a separate convergence theorem for each dimension.

### 45.19 One-Dimensional Recovery

For:

$$
d=1,
$$

the domain becomes:

$$
\Omega=[a_1,b_1],
$$

the region becomes:

$$
R_t=[l_t,r_t],
$$

and:

$$
\operatorname{diam}(R_t)
=
r_t-l_t.
$$

The Euclidean distance becomes:

$$
\|x-x^\ast\|_2
=
|x-x^\ast|.
$$

Directional fairness becomes trivial because only one spatial direction exists.

Therefore the multidimensional theorem reduces to the original one-dimensional convergence structure.

### 45.20 Final Theoretical Result

The complete result can be summarized as:

$$
\boxed{
\begin{aligned}
&
\text{Compactness}
+
\text{Continuity}
+
\text{Valid Splitting}
\\
&+
\text{Contraction}
+
\text{Global Region Fairness}
+
\text{Directional Fairness}
\\
&+
\text{Information Acquisition}
+
\text{Exact Evaluation}
\\
&\Longrightarrow
\\
&
\boxed{
f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast
}
\end{aligned}
}
$$

Therefore ARRGO has a mathematically consistent global-convergence formulation for arbitrary finite-dimensional bounded optimization problems, provided the stated assumptions are satisfied.

This result establishes the theoretical validity of extending the ARRGO framework from one-dimensional intervals to finite-dimensional hyper-rectangular domains.

It does not claim that every concrete implementation automatically satisfies all assumptions.

The next step is to establish the corresponding certified convergence theorem separately, because certified convergence requires stronger assumptions than empirical objective-value convergence.

## 46. Formal Proof of Multidimensional Certified Convergence

The previous theorem established objective-value global convergence under compactness, continuity, valid refinement, fairness, and sufficient information acquisition.

Certified convergence requires a stronger argument.

In Certified Mode, ARRGO must maintain mathematically valid upper and lower bounds on the objective and use these bounds to quantify the remaining uncertainty about the global optimum.

This section formally establishes the correctness of the multidimensional certification mechanism.

### 46.1 Additional Certified Assumption

Assume that the objective satisfies a global Lipschitz condition on \(\Omega\):

$$
|f(x)-f(y)|
\leq
L\|x-y\|_2
\qquad
\forall x,y\in\Omega,
$$

where:

$$
L\geq0
$$

is a valid global Lipschitz constant.

The validity of this constant is an assumption of Certified Mode.

If the supplied value of \(L\) is not valid, the resulting quantities cannot be interpreted as rigorous certificates.

### 46.2 Evaluation-Based Upper Bound

Let a region \(R\) contain evaluations:

$$
D_R
=
\{(x_i,f(x_i))\}_{i=1}^{n_R}.
$$

For any:

$$
x\in R,
$$

the Lipschitz condition gives, for every evaluated point \(x_i\),

$$
f(x)
\leq
f(x_i)
+
L\|x-x_i\|_2.
$$

Therefore \(f(x)\) is bounded above by every such expression.

Taking their minimum gives:

$$
U_R(x)
=
\min_{x_i\in D_R}
\left[
f(x_i)
+
L\|x-x_i\|_2
\right].
$$

Hence:

$$
\boxed{
f(x)\leq U_R(x)
}
$$

for every \(x\in R\).

### 46.3 Evaluation-Based Lower Bound

Similarly, the Lipschitz condition gives:

$$
f(x)
\geq
f(x_i)
-
L\|x-x_i\|_2.
$$

Taking the maximum over all available evaluations gives:

$$
L_R(x)
=
\max_{x_i\in D_R}
\left[
f(x_i)
-
L\|x-x_i\|_2
\right].
$$

Therefore:

$$
\boxed{
L_R(x)\leq f(x)
}
$$

for every \(x\in R\).

### 46.4 Regional Envelope Validity

Combining the two inequalities gives:

$$
\boxed{
L_R(x)
\leq
f(x)
\leq
U_R(x)
}
$$

for every:

$$
x\in R.
$$

Therefore the interval:

$$
[L_R(x),U_R(x)]
$$

contains the true objective value at every point in the region.

This is a mathematical bound rather than an empirical estimate.

### 46.5 Regional Uncertainty

Define the pointwise certified uncertainty:

$$
u_R(x)
=
U_R(x)-L_R(x).
$$

Because:

$$
L_R(x)\leq f(x)\leq U_R(x),
$$

we have:

$$
u_R(x)\geq0.
$$

Define the worst-case regional uncertainty:

$$
u_{\max}(R)
=
\sup_{x\in R}
u_R(x).
$$

Thus:

$$
u_{\max}(R)
$$

measures the maximum remaining certified interval width inside the region.

### 46.6 Regional Optimization Potential

Define the regional potential:

$$
P(R)
=
\sup_{x\in R}U_R(x).
$$

Because:

$$
f(x)\leq U_R(x),
$$

we obtain:

$$
\sup_{x\in R}f(x)
\leq
P(R).
$$

Therefore \(P(R)\) is a valid upper bound on the best objective value that can still occur inside \(R\).

### 46.7 Certification Frontier

ARRGO retains historical parent regions after splitting.

Therefore the complete structural hierarchy:

$$
\mathcal H_t
$$

contains both current regions and historical ancestors.

Certification should instead operate on a current frontier:

$$
\mathcal F_t\subseteq\mathcal H_t
$$

such that:

$$
\Omega
\subseteq
\bigcup_{R\in\mathcal F_t}R.
$$

The frontier represents the current structural cover used for certification.

### 46.8 Global Certified Potential

Define:

$$
P_{\mathrm{global}}^{(t)}
=
\sup_{R\in\mathcal F_t}P(R).
$$

Because the frontier covers the entire domain:

$$
\Omega
\subseteq
\bigcup_{R\in\mathcal F_t}R,
$$

every possible global maximizer belongs to at least one frontier region.

Therefore:

$$
f^\ast
\leq
P_{\mathrm{global}}^{(t)}.
$$

### 46.9 Lower Bound from the Incumbent

Every evaluated point belongs to the domain.

Therefore:

$$
f(x_i)
\leq
f^\ast.
$$

Since the incumbent is the maximum observed value:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast.
$$

Thus the incumbent is a valid lower bound on the unknown global optimum.

### 46.10 Global Certified Sandwich

Combining the previous results gives:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
\leq
f^\ast
\leq
P_{\mathrm{global}}^{(t)}.
}
$$

This is the fundamental certification inequality.

The true optimum is mathematically trapped between an observed lower bound and a certified upper bound.

### 46.11 Global Certification Gap

Define:

$$
\Delta_{\mathrm{global}}^{(t)}
=
P_{\mathrm{global}}^{(t)}
-
f_{\mathrm{best}}^{(t)}.
$$

Because:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast
\leq
P_{\mathrm{global}}^{(t)},
$$

we obtain:

$$
0
\leq
f^\ast-f_{\mathrm{best}}^{(t)}
\leq
\Delta_{\mathrm{global}}^{(t)}.
$$

Therefore:

$$
\boxed{
\Delta_{\mathrm{global}}^{(t)}
}
$$

is a rigorous upper bound on the objective-value error.

### 46.12 Epsilon-Optimality

Suppose:

$$
\Delta_{\mathrm{global}}^{(t)}
\leq
\epsilon.
$$

Then:

$$
f^\ast-f_{\mathrm{best}}^{(t)}
\leq
\Delta_{\mathrm{global}}^{(t)}
\leq
\epsilon.
$$

Therefore:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
\geq
f^\ast-\epsilon.
}
$$

The incumbent is consequently epsilon-optimal in objective value.

This is a genuine mathematical certificate under the stated assumptions.

### 46.13 Why the Lipschitz Constant Must Be Valid

The entire certification chain depends on:

$$
|f(x)-f(y)|
\leq
L\|x-y\|_2.
$$

If the supplied \(L\) is too small, the upper and lower envelopes may fail to contain the true function.

In that case:

$$
L_R(x)\leq f(x)\leq U_R(x)
$$

is no longer guaranteed.

Therefore the resulting gap cannot be called a rigorous certificate.

A conservative overestimate of \(L\) preserves validity, although it may produce looser bounds.

Thus:

$$
\boxed{
\text{Valid }L
\Rightarrow
\text{Valid Certificate}.
}
$$

### 46.14 Effect of Additional Sampling

Suppose a new evaluation is added to \(D_R\).

The upper envelope becomes:

$$
U_R^{\mathrm{new}}(x)
=
\min
\left(
U_R^{\mathrm{old}}(x),
f(x_{\mathrm{new}})
+
L\|x-x_{\mathrm{new}}\|_2
\right).
$$

Therefore:

$$
U_R^{\mathrm{new}}(x)
\leq
U_R^{\mathrm{old}}(x).
$$

Similarly:

$$
L_R^{\mathrm{new}}(x)
\geq
L_R^{\mathrm{old}}(x).
$$

Hence:

$$
u_R^{\mathrm{new}}(x)
\leq
u_R^{\mathrm{old}}(x).
$$

Additional valid evaluations cannot increase the pointwise certified uncertainty.

### 46.15 Effect of Structural Splitting

Suppose:

$$
R
\rightarrow
R_1\cup R_2.
$$

Splitting does not itself create a new objective evaluation.

Therefore:

$$
D_{R_1}\cup D_{R_2}
$$

contains the information assigned to the child regions according to the implementation's explicit boundary-ownership convention.

The structural domain is partitioned into smaller regions while preserving the objective information.

Because:

$$
R_1,R_2\subseteq R,
$$

their geometric diameters are smaller under valid contraction.

Thus splitting improves the spatial resolution of the certification problem without itself consuming an objective evaluation.

### 46.16 No-Pruning Compatibility

Certification does not require historical regions to be deleted.

The structural hierarchy may preserve:

$$
R_{\mathrm{parent}}
\rightarrow
\{R_1,R_2\}
$$

while the certification frontier contains only the current cover.

Therefore:

$$
\boxed{
\text{Structural History}
\neq
\text{Certification Frontier}.
}
$$

This separation preserves the no-pruning principle while maintaining a mathematically valid global certificate.

### 46.17 Certified Convergence

Assume that the structural and information conditions of the multidimensional convergence theorem hold.

Then:

$$
f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast.
$$

In addition, under valid Lipschitz certification and sufficiently refined frontier regions, the certified upper bound can be made arbitrarily close to the optimum:

$$
P_{\mathrm{global}}^{(t)}
\rightarrow
f^\ast.
$$

Consequently:

$$
\Delta_{\mathrm{global}}^{(t)}
=
P_{\mathrm{global}}^{(t)}
-
f_{\mathrm{best}}^{(t)}
\rightarrow0.
$$

Therefore:

$$
\boxed{
\lim_{t\rightarrow\infty}
\Delta_{\mathrm{global}}^{(t)}
=
0.
}
$$

This establishes certified convergence in objective value.

### 46.18 Interpretation of the Certified Gap

The certified gap has two important interpretations.

First:

$$
\Delta_{\mathrm{global}}^{(t)}
$$

is an upper bound on the unknown objective error:

$$
f^\ast-f_{\mathrm{best}}^{(t)}
\leq
\Delta_{\mathrm{global}}^{(t)}.
$$

Second, it measures the remaining worst-case uncertainty allowed by the current certified model.

Therefore:

$$
\boxed{
\Delta_{\mathrm{global}}
\downarrow
\quad\Rightarrow\quad
\text{Certified Knowledge Increases}.
}
$$

A small gap means that the algorithm has mathematically narrowed the possible location of the unknown optimum in objective-value terms.

### 46.19 Certified Termination

Certified termination occurs when:

$$
\Delta_{\mathrm{global}}^{(t)}
\leq
\epsilon.
$$

At that point:

$$
f_{\mathrm{best}}^{(t)}
\geq
f^\ast-\epsilon.
$$

Therefore the algorithm can terminate with a mathematically justified epsilon-optimality statement.

This differs fundamentally from budget termination.

Budget termination only establishes:

$$
\boxed{
\text{Best solution found within the available evaluation budget}.
}
$$

Certified termination establishes:

$$
\boxed{
\text{Epsilon-optimal solution under the certification assumptions}.
}
$$

### 46.20 Certified Mode and Global Fairness

Global fairness remains important even in Certified Mode.

If a frontier region has:

$$
P(R)>f_{\mathrm{best}},
$$

then it remains potentially capable of containing a better solution.

A fair refinement mechanism must therefore continue to provide refinement opportunities to globally relevant frontier regions.

Regions whose certified potential has become sufficiently low may cease to require refinement for certification purposes.

Again, this does not require deleting them from the structural hierarchy.

### 46.21 Certified Convergence Proposition

**Proposition — Multidimensional Certified Convergence**

Let \(f\) be continuous on a compact bounded domain:

$$
\Omega\subset\mathbb R^d,
$$

and suppose:

$$
|f(x)-f(y)|
\leq
L\|x-y\|_2
$$

for a valid global Lipschitz constant \(L\).

Assume that ARRGO maintains:

1. a valid current certification frontier;
2. valid regional upper and lower envelopes;
3. a correct global incumbent;
4. valid region contraction;
5. global region fairness;
6. directional fairness;
7. sufficient information acquisition.

Then:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast
\leq
P_{\mathrm{global}}^{(t)}
$$

for every certified iteration.

Furthermore, if the refinement and information process drives the certification frontier to arbitrarily fine resolution, then:

$$
\boxed{
\Delta_{\mathrm{global}}^{(t)}
\rightarrow0.
}
$$

Consequently, for every:

$$
\epsilon>0,
$$

there exists a sufficiently large iteration \(T\) such that:

$$
\Delta_{\mathrm{global}}^{(T)}
\leq
\epsilon,
$$

and therefore:

$$
\boxed{
f_{\mathrm{best}}^{(T)}
\geq
f^\ast-\epsilon.
}
$$

### 46.22 Proof

The Lipschitz assumption establishes valid regional envelopes:

$$
L_R(x)
\leq
f(x)
\leq
U_R(x).
$$

Because the certification frontier covers \(\Omega\), every global maximizer is contained in at least one frontier region.

Therefore:

$$
f^\ast
\leq
P_{\mathrm{global}}.
$$

The incumbent is an observed feasible value, so:

$$
f_{\mathrm{best}}
\leq
f^\ast.
$$

Hence:

$$
f_{\mathrm{best}}
\leq
f^\ast
\leq
P_{\mathrm{global}}.
$$

Under global fairness and directional fairness, relevant frontier regions receive continued refinement opportunities.

Under valid contraction, their diameters can be reduced arbitrarily.

Under sufficient information acquisition and continuity, the incumbent approaches the global optimum:

$$
f_{\mathrm{best}}
\rightarrow
f^\ast.
$$

As the frontier becomes arbitrarily fine and valid evaluations are acquired, the remaining certified upper uncertainty can also be reduced arbitrarily.

Therefore:

$$
P_{\mathrm{global}}
\rightarrow
f^\ast.
$$

Combining:

$$
f_{\mathrm{best}}
\rightarrow
f^\ast
$$

and:

$$
P_{\mathrm{global}}
\rightarrow
f^\ast,
$$

we obtain:

$$
\boxed{
\Delta_{\mathrm{global}}
=
P_{\mathrm{global}}
-
f_{\mathrm{best}}
\rightarrow0.
}
$$

Therefore, for every \(\epsilon>0\), a sufficiently small certified gap eventually satisfies:

$$
\Delta_{\mathrm{global}}\leq\epsilon.
$$

The epsilon-optimality result follows:

$$
f_{\mathrm{best}}
\geq
f^\ast-\epsilon.
$$

### 46.23 One-Dimensional Reduction

For:

$$
d=1,
$$

the Euclidean norm becomes:

$$
\|x-y\|_2
=
|x-y|.
$$

Therefore the multidimensional certified envelopes reduce to:

$$
U_R(x)
=
\min_i
\left[
f(x_i)+L|x-x_i|
\right],
$$

and:

$$
L_R(x)
=
\max_i
\left[
f(x_i)-L|x-x_i|
\right].
$$

Thus the multidimensional certification framework recovers the original one-dimensional formulation exactly.

### 46.24 Final Certified Result

The complete certified reasoning can therefore be summarized as:

$$
\boxed{
\begin{aligned}
&\text{Valid Lipschitz Assumption}
\\
&+
\text{Valid Certification Frontier}
\\
&+
\text{Valid Upper/Lower Envelopes}
\\
&+
\text{Fair Refinement}
\\
&+
\text{Information Acquisition}
\\
&\Longrightarrow
\\
&f_{\mathrm{best}}
\leq
f^\ast
\leq
P_{\mathrm{global}}
\\
&\Longrightarrow
\\
&0
\leq
f^\ast-f_{\mathrm{best}}
\leq
\Delta_{\mathrm{global}}
\\
&\Longrightarrow
\\
&\Delta_{\mathrm{global}}\rightarrow0
\\
&\Longrightarrow
\\
&\text{Certified Epsilon-Optimality}.
\end{aligned}
}
$$

Therefore Certified Mode provides a mathematically meaningful convergence and stopping mechanism for the finite-dimensional ARRGO formulation, provided that the Lipschitz constant, certification frontier, refinement process, and information-acquisition conditions are valid.

The next step is to prove the relationship between the multidimensional convergence theorem and the original one-dimensional ARRGO theorem in a single formal equivalence result.

## 47. Formal One-Dimensional Equivalence of the Generalized ARRGO Framework

The multidimensional formulation developed in the previous sections was constructed as a dimension-parameterized extension of the original one-dimensional ARRGO framework.

This section formally establishes that the one-dimensional formulation is recovered when:

$$
d=1.
$$

This result is important because a valid generalization must preserve the original framework as a special case rather than defining an unrelated optimization method.

### 47.1 Generalized Domain Reduction

The multidimensional domain is:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j].
$$

Set:

$$
d=1.
$$

Then:

$$
\Omega
=
[a_1,b_1].
$$

Renaming:

$$
a_1=a,
\qquad
b_1=b,
$$

gives:

$$
\boxed{
\Omega=[a,b].
}
$$

This is exactly the original one-dimensional bounded optimization domain.

### 47.2 Decision Variable Reduction

In the generalized formulation:

$$
x=(x_1,\ldots,x_d)
\in\mathbb R^d.
$$

For:

$$
d=1,
$$

we obtain:

$$
x=(x_1).
$$

The vector therefore reduces to the scalar:

$$
\boxed{
x=x_1.
}
$$

Consequently:

$$
f:\Omega\rightarrow\mathbb R
$$

becomes the original scalar-input objective:

$$
f:[a,b]\rightarrow\mathbb R.
$$

### 47.3 Region Reduction

The generalized region is:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j].
$$

For:

$$
d=1,
$$

this becomes:

$$
R=[l_1,r_1].
$$

Renaming:

$$
l_1=l,
\qquad
r_1=r,
$$

gives:

$$
\boxed{
R=[l,r].
}
$$

Thus the multidimensional hyper-rectangle becomes exactly the original interval.

### 47.4 Geometry Reduction

The generalized side-width vector is:

$$
W(R)
=
(w_1,\ldots,w_d),
$$

where:

$$
w_j=r_j-l_j.
$$

For:

$$
d=1,
$$

we obtain:

$$
W(R)=(r_1-l_1).
$$

Therefore the scalar region width is:

$$
\boxed{
w(R)=r-l.
}
$$

The generalized diameter is:

$$
\operatorname{diam}(R)
=
\sqrt{
\sum_{j=1}^{d}w_j^2
}.
$$

For \(d=1\):

$$
\operatorname{diam}(R)
=
\sqrt{w_1^2}
=
w_1
=
r-l.
$$

Hence:

$$
\boxed{
\operatorname{diam}(R)=r-l.
}
$$

The multidimensional diameter therefore becomes exactly the original interval width.

### 47.5 Center Reduction

The generalized region center is:

$$
c(R)
=
\left(
\frac{l_1+r_1}{2},
\ldots,
\frac{l_d+r_d}{2}
\right).
$$

For:

$$
d=1,
$$

this becomes:

$$
c(R)
=
\frac{l_1+r_1}{2}.
$$

Therefore:

$$
\boxed{
c(R)=\frac{l+r}{2}.
}
$$

This recovers the original midpoint.

### 47.6 Distance Reduction

The generalized Euclidean distance is:

$$
\|x-y\|_2
=
\sqrt{
\sum_{j=1}^{d}(x_j-y_j)^2
}.
$$

For:

$$
d=1,
$$

we obtain:

$$
\|x-y\|_2
=
\sqrt{(x-y)^2}
=
|x-y|.
$$

Thus:

$$
\boxed{
\|x-y\|_2=|x-y|.
}
$$

This is exactly the distance used by the one-dimensional framework.

### 47.7 Evaluation History Reduction

The generalized evaluation is:

$$
e_i
=
(x_i,f(x_i),\operatorname{Id}_i),
$$

with:

$$
x_i\in\mathbb R^d.
$$

For:

$$
d=1,
$$

the point becomes scalar:

$$
x_i\in\mathbb R.
$$

Therefore:

$$
\boxed{
e_i
=
(x_i,f(x_i),\operatorname{Id}_i).
}
$$

The evaluation-history structure is unchanged.

### 47.8 Regional Evaluation Reduction

The generalized regional evaluation set is:

$$
D_R
=
\left\{
e_i\in D_t:
x_i\in R
\right\}.
$$

For:

$$
R=[l,r],
$$

membership becomes:

$$
l\leq x_i\leq r.
$$

Therefore:

$$
\boxed{
D_R
=
\left\{
e_i\in D_t:
l\leq x_i\leq r
\right\}.
}
$$

This is exactly the original one-dimensional regional data definition.

### 47.9 Sampling Reduction

The generalized sampling action selects:

$$
x\in R
\subset\mathbb R^d.
$$

For:

$$
d=1,
$$

this becomes:

$$
x\in[l,r].
$$

The action:

$$
\operatorname{Sample}(x)
$$

therefore becomes the original scalar sampling operation.

Duplicate detection also reduces from:

$$
\|x-x_i\|_2\leq\tau_x
$$

to:

$$
|x-x_i|\leq\tau_x.
$$

Thus the sampling mechanism is recovered exactly.

### 47.10 Spatial Coverage Reduction

The generalized coordinate-wise evaluation set is:

$$
X_{R,j}
=
\{x_{i,j}:x_i\in R\}.
$$

For:

$$
d=1,
$$

there is only one coordinate:

$$
X_{R,1}
=
\{x_i:x_i\in R\}.
$$

The ordered gaps between evaluations and region boundaries therefore reduce to the original one-dimensional spatial coverage gaps.

The geometric covering radius becomes:

$$
h(R)
=
\sup_{x\in R}
\min_{x_i\in D_R}
|x-x_i|.
$$

Thus the multidimensional coverage formulation reduces naturally to interval coverage.

### 47.11 Structural Splitting Reduction

The generalized split action is:

$$
\operatorname{Split}(j,s_j).
$$

For:

$$
d=1,
$$

the only possible direction is:

$$
j=1.
$$

Therefore the action becomes:

$$
\operatorname{Split}(1,s_1).
$$

Renaming:

$$
s_1=s,
$$

gives:

$$
\boxed{
\operatorname{Split}(s).
}
$$

The two children become:

$$
R_1=[l,s],
$$

and:

$$
R_2=[s,r].
$$

This is exactly the original binary interval split.

### 47.12 Directional Fairness Reduction

In the multidimensional formulation, directional fairness requires every unresolved coordinate direction to receive refinement opportunities.

When:

$$
d=1,
$$

there is only one coordinate direction.

Therefore there is no directional-selection ambiguity.

Directional fairness reduces to the requirement that the single spatial direction continues to receive refinement when refinement remains necessary.

Thus the multidimensional directional-fairness condition becomes the original one-dimensional refinement condition.

### 47.13 Contraction Reduction

The generalized contraction condition is:

$$
\max
\left(
\frac{s_j-l_j}{r_j-l_j},
\frac{r_j-s_j}{r_j-l_j}
\right)
\leq
\rho.
$$

For:

$$
d=1,
\qquad
j=1,
$$

this becomes:

$$
\boxed{
\max
\left(
\frac{s-l}{r-l},
\frac{r-s}{r-l}
\right)
\leq
\rho.
}
$$

Therefore the generalized contraction condition exactly recovers the one-dimensional binary-splitting contraction condition.

### 47.14 Certified Envelope Reduction

The generalized upper envelope is:

$$
U_R(x)
=
\min_{x_i\in D_R}
\left[
f(x_i)
+
L\|x-x_i\|_2
\right].
$$

For:

$$
d=1,
$$

the norm becomes absolute distance:

$$
\|x-x_i\|_2
=
|x-x_i|.
$$

Therefore:

$$
\boxed{
U_R(x)
=
\min_i
\left[
f(x_i)
+
L|x-x_i|
\right].
}
$$

Similarly:

$$
\boxed{
L_R(x)
=
\max_i
\left[
f(x_i)
-
L|x-x_i|
\right].
}
$$

These are precisely the original one-dimensional Lipschitz envelopes.

### 47.15 Regional Potential Reduction

The generalized regional potential is:

$$
P(R)
=
\sup_{x\in R}U_R(x).
$$

For:

$$
R=[l,r],
$$

this becomes:

$$
\boxed{
P(R)
=
\sup_{x\in[l,r]}U_R(x).
}
$$

Thus the optimization-potential definition is dimension-independent and remains unchanged in one dimension.

### 47.16 Global Certified Bound Reduction

The generalized certification frontier satisfies:

$$
\Omega
\subseteq
\bigcup_{R\in\mathcal F_t}R.
$$

For \(d=1\):

$$
[a,b]
\subseteq
\bigcup_{R\in\mathcal F_t}R.
$$

Therefore:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast
\leq
P_{\mathrm{global}}^{(t)}.
$$

The global certification gap remains:

$$
\boxed{
\Delta_{\mathrm{global}}^{(t)}
=
P_{\mathrm{global}}^{(t)}
-
f_{\mathrm{best}}^{(t)}.
}
$$

Thus the complete certification logic is preserved.

### 47.17 Incumbent Reduction

The generalized incumbent is:

$$
x_{\mathrm{best}}^{(t)}
\in
\arg\max_{x_i\in D_t}f(x_i).
$$

For:

$$
d=1,
$$

the same definition applies to scalar evaluation points.

Therefore:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
=
\max_{x_i\in D_t}f(x_i).
}
$$

No change in the global incumbent logic is required.

### 47.18 Hierarchy Reduction

The generalized hierarchy consists of hyper-rectangular parent-child relationships.

For:

$$
d=1,
$$

the hierarchy becomes a tree of intervals:

$$
R_0
\rightarrow
\{R_1,R_2\}
\rightarrow
\{R_{1,1},R_{1,2},R_{2,1},R_{2,2}\}
\rightarrow\cdots
$$

Parents remain represented after splitting.

Therefore the original no-pruning hierarchy is recovered.

### 47.19 Global Selection Reduction

The generalized global-selection mechanism evaluates candidate regions according to:

- global relevance;
- unresolved information;
- spatial resolution;
- optimization potential;
- deterministic tie-breaking;
- fairness.

When:

$$
d=1,
$$

all multidimensional geometric quantities reduce to scalar interval quantities.

Therefore the same global-selection principle applies without introducing a separate one-dimensional theory.

### 47.20 Convergence Reduction

The multidimensional convergence theorem establishes:

$$
f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast
$$

under the stated assumptions.

For:

$$
d=1,
$$

these assumptions reduce to:

- compact bounded interval;
- continuity;
- valid interval splitting;
- contraction;
- global region fairness;
- information acquisition.

Therefore:

$$
\boxed{
\mathrm{ARRGO}_{d=1}
=
\mathrm{ARRGO}_{1D}.
}
$$

The generalized theorem therefore contains the original one-dimensional convergence result as a special case.

### 47.21 Certified Convergence Reduction

Under the valid Lipschitz assumption:

$$
|f(x)-f(y)|
\leq
L|x-y|,
$$

the generalized certified result becomes:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast
\leq
P_{\mathrm{global}}^{(t)},
$$

and:

$$
\Delta_{\mathrm{global}}^{(t)}
\rightarrow0.
$$

Thus the original one-dimensional Certified Mode is recovered exactly.

### 47.22 Equivalence Proposition

**Proposition — One-Dimensional Equivalence**

Let the generalized ARRGO framework be defined for:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j].
$$

When:

$$
d=1,
$$

the generalized formulation reduces to the original one-dimensional ARRGO formulation with respect to:

1. domain;
2. decision variable;
3. region representation;

6. coverage;

8. contraction;
9. hierarchy;
10. global selection;

13. certification gap;
14. convergence conditions.

Therefore:

$$
\boxed{
\mathrm{ARRGO}(1)
\equiv
\mathrm{ARRGO}_{1D}.
}
$$

### 47.23 Proof

Each generalized component was reduced explicitly in Sections 47.1–47.21.

The domain becomes an interval.

The vector decision variable becomes a scalar.

The hyper-rectangular region becomes an interval.

The Euclidean norm becomes absolute distance.

The multidimensional split action becomes a scalar binary split.

The multidimensional contraction condition becomes the original interval contraction condition.

The directional-selection problem disappears because only one direction exists.

The generalized certified envelopes become the original one-dimensional Lipschitz envelopes.

The hierarchy remains a parent-preserving interval tree.

The incumbent and global optimization state remain unchanged.

Finally, the multidimensional convergence assumptions reduce to the corresponding one-dimensional assumptions.

Therefore no essential component of the generalized framework remains inconsistent with the original formulation when:

$$
d=1.
$$

Hence:

$$
\boxed{
\mathrm{ARRGO}(1)
\equiv
\mathrm{ARRGO}_{1D}.
}
$$

### 47.24 Significance of the Equivalence

This equivalence establishes that the multidimensional formulation is a true extension rather than a separate algorithm.

The relationship can be represented as:

$$
\boxed{
\mathrm{ARRGO}(d)
\quad
\text{for finite }d\geq1,
}
$$

with:

$$
\boxed{
\mathrm{ARRGO}(1)
=
\text{original ARRGO}.
}
$$

Therefore the existing one-dimensional implementation can be interpreted as the \(d=1\) realization of a more general mathematical framework.

### 47.25 Important Scope Distinction

The equivalence theorem is mathematical and structural.

It does not claim that the current one-dimensional software implementation automatically supports arbitrary dimensions.

The current implementation remains a one-dimensional implementation.

The result established here is:

$$
\boxed{
\text{The ARRGO Theory Generalizes to Finite Dimensions}.
}
$$

A separate multidimensional implementation would require the engineering changes described in the implementation-architecture sections.

Therefore:

$$
\text{Theoretical Generalization}
\neq
\text{Implemented Generalization}.
$$

This distinction preserves the scientific validity of the project.

### 47.26 Final Equivalence Result

The complete relationship can be summarized as:

$$
\boxed{
\begin{aligned}
\mathrm{ARRGO}(d)
&=
\text{Dimension-Parameterized General Framework},
\\
\mathrm{ARRGO}(1)
&=
\text{Original One-Dimensional ARRGO},

d&<\infty.

$$

Thus the one-dimensional algorithm is formally recovered as a special case of the finite-dimensional ARRGO framework.

The next step is to examine the limits of this generalization and identify exactly which claims can and cannot be made about scalability, high-dimensional behavior, and practical performance.

## 48. Limits of the Multidimensional Generalization

The previous sections established that the mathematical structure of ARRGO can be generalized from one-dimensional intervals to arbitrary finite-dimensional hyper-rectangular domains.

However, mathematical generality does not automatically imply computational scalability or practical efficiency.

This section identifies the precise limits of the generalization and separates theoretical validity from implementation complexity and empirical performance.

### 48.1 Mathematical Generality

The generalized formulation applies to:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j],
\qquad
d<\infty.
$$

The convergence argument remains valid under the stated assumptions:

$$
\boxed{
\text{Finite Dimension}
+
\text{Valid Refinement}
+
\text{Fairness}
+
\text{Information Acquisition}
+
\text{Continuity}
\Rightarrow
\text{Objective-Value Convergence}.
}
$$

This establishes mathematical generality.

It does not establish that the computational cost remains constant as \(d\) increases.

### 48.2 Dimensionality and Computational Cost

As the dimension increases, the geometric complexity of the search domain increases rapidly.

A hyper-rectangle has:

$$
2d
$$

boundary faces and:

$$
2^d
$$

vertices.

More importantly, representing and exploring sufficiently fine spatial structure may require an increasing number of evaluations.

Therefore:

$$
\boxed{
\text{Theoretical Validity}
\neq
\text{Dimension-Independent Computational Cost}.
}
$$

ARRGO is mathematically defined for arbitrary finite \(d\), but this does not imply constant computational complexity with respect to dimension.

### 48.3 The Curse of Dimensionality

High-dimensional optimization can suffer from the curse of dimensionality.

For example, suppose each coordinate is discretized into \(m\) levels.

A full Cartesian grid would contain:

$$
m^d
$$

points.

This grows exponentially with \(d\).

ARRGO does not require construction of such a full Cartesian grid.

Instead, it performs adaptive region refinement and information-driven sampling.

Therefore its objective is to avoid unnecessary global enumeration.

Nevertheless:

$$
\boxed{
\text{Adaptive Refinement}
\neq
\text{Elimination of the Curse of Dimensionality}.
}
$$

No claim should be made that ARRGO completely removes the fundamental difficulty of high-dimensional global optimization.

### 48.4 Adaptive Refinement as a Response

The central response of ARRGO to high-dimensional complexity is to avoid refining every region and every direction uniformly.

Instead, the framework attempts to identify:

- globally relevant regions;
- unresolved spatial directions;
- informative sampling locations;
- regions with meaningful optimization potential;
- regions with large certified uncertainty when certification is available.

Thus the intended computational principle is:

$$
\boxed{
\text{Refine According to Information}
}
$$

rather than:

$$
\boxed{
\text{Refine Uniformly Everywhere}.
}
$$

This is a design principle rather than a proven complexity bound.

### 48.5 Directional Complexity

In \(d\) dimensions, a structural split must specify both:

$$
(j,s_j),
$$

where:

- \(j\) identifies the coordinate direction;
- \(s_j\) identifies the split location.

Therefore the number of possible structural decisions increases with the number of available dimensions and candidate split locations.

The generalized algorithm must therefore distinguish:

$$
\text{Direction Selection}
$$

from:

$$
\text{Split-Location Selection}.
$$

This separation is essential for maintaining a deterministic and interpretable refinement process.

### 48.6 Joint Objective Interactions

The objective remains a joint function:

$$
f(x_1,\ldots,x_d).
$$

Therefore the optimization cannot generally be decomposed as:

$$
f(x_1,\ldots,x_d)
=
\sum_{j=1}^{d}f_j(x_j).
$$

Without a separability assumption, changing one coordinate can alter the effect of another coordinate.

Consequently, coordinate-wise refinement alone cannot characterize the complete objective landscape.

ARRGO therefore requires multidimensional sampling capable of observing joint configurations.

### 48.7 Sampling Complexity

The multidimensional sampling mechanism may require evaluating points that explore combinations of coordinates.

This can become expensive as \(d\) increases.

The framework therefore does not claim:

$$
N_{\mathrm{evaluations}}
=
O(d)
$$

or any other universal dimension-independent evaluation bound.

Instead, the number of evaluations remains dependent on:

- objective structure;
- dimension;
- domain geometry;
- refinement policy;
- information resolution requirements;
- numerical tolerances;
- desired optimization accuracy.

### 48.8 Region Count

Binary splitting creates two children from one parent.

Because parents are retained structurally, after \(s\) binary splits the hierarchy contains:

$$
1+2s
$$

region nodes, assuming every split replaces one current frontier region by two children.

This is a structural-history count.

The number of current frontier regions is instead:

$$
1+s.
$$

Therefore:

$$
\boxed{
|\mathcal H_t|
=
1+s,
}
$$

only if \(\mathcal H_t\) denotes the current frontier.

If \(\mathcal H_t\) denotes the complete parent-preserving hierarchy, then:

$$
\boxed{
|\mathcal H_t|
=
1+2s.
}
$$

This distinction is important for both implementation and complexity analysis.

### 48.9 No-Pruning and Memory Growth

The no-pruning principle has an explicit computational consequence.

Because historical parents remain represented, the structural hierarchy grows monotonically.

Therefore memory usage associated with hierarchy metadata generally increases with the number of structural splits.

This is intentional because the hierarchy stores structural history.

Thus:

$$
\boxed{
\text{No Pruning}
\Rightarrow
\text{Increasing Structural History}.
}
$$

This is a design trade-off rather than a correctness problem.

The current certification frontier can remain separate from the historical hierarchy so that certification calculations do not need to treat every historical parent as a current region.

### 48.10 Certified Computation Complexity

In Certified Mode, regional envelopes require operations involving:

$$
f(x_i)
\pm
L\|x-x_i\|_2.
$$

Each distance evaluation in \(d\) dimensions requires, in general:

$$
O(d)
$$

coordinate operations.

If a region contains \(n_R\) evaluations, evaluating an envelope at a single point can therefore require approximately:

$$
O(d\,n_R)
$$

distance-related work.

This is not a universal total runtime bound because the number of envelope queries and regional evaluations depends on the implementation.

The important conclusion is:

$$
\boxed{
\text{Certified Computation Has Dimension-Dependent Cost}.
}
$$

### 48.11 Certified Bounds and the Lipschitz Assumption

The certified convergence result requires a valid global Lipschitz constant:

$$
|f(x)-f(y)|
\leq
L\|x-y\|_2.
$$

Such a constant may not always be known.

If \(L\) is unavailable, ARRGO can still operate in Empirical Mode.

Therefore:

$$
\boxed{
\text{Lipschitz Constant Required for Certification}
}
$$

but:

$$
\boxed{
\text{Lipschitz Constant Not Required for Empirical Search}.
}
$$

This distinction is fundamental to the framework.

### 48.12 Effect of an Overly Conservative Lipschitz Constant

A valid but very large value of \(L\) preserves certification validity.

However, the resulting envelopes may be loose:

$$
U_R(x)-L_R(x)
\quad\text{large}.
$$

Consequently:

$$
\Delta_{\mathrm{global}}
$$

may decrease slowly.

Thus:

$$
\boxed{
\text{Validity}
\neq
\text{Tightness}.
}
$$

A conservative Lipschitz constant protects correctness but may reduce the practical efficiency of certified convergence.

### 48.13 Effect of an Invalid Lipschitz Constant

If:

$$
L
<
L_{\mathrm{true}}
$$

for the required global Lipschitz bound, then the inequality:

$$
|f(x)-f(y)|
\leq
L\|x-y\|_2
$$

may fail.

The regional envelopes can then cease to be valid.

Therefore:

$$
P_{\mathrm{global}}
$$

cannot be interpreted as a rigorous upper bound.

Consequently:

$$
\Delta_{\mathrm{global}}
$$

cannot be interpreted as a rigorous certification gap.

Thus:

$$
\boxed{
\text{Invalid }L
\Rightarrow
\text{No Valid Certification}.
}
$$

### 48.14 Finite-Dimensional Scope

The generalized theorem assumes:

$$
d<\infty.
$$

It does not establish convergence in infinite-dimensional spaces.

Infinite-dimensional optimization may require substantially different mathematical tools concerning:

- compactness;
- topology;
- geometry;
- approximation;
- information acquisition.

Therefore the present generalization should be explicitly described as:

$$
\boxed{
\text{Finite-Dimensional Generalization}.
}
$$

### 48.15 Dependence on Fairness

The convergence theorem depends on fairness conditions.

In particular:

$$
\text{Global Region Fairness}
$$

prevents permanent starvation of relevant regions, while:

$$
\text{Directional Fairness}
$$

prevents permanent starvation of unresolved coordinate directions.

If either property fails, the convergence theorem may no longer apply.

Therefore the theorem is conditional on the refinement policy satisfying these properties.

### 48.16 Dependence on Information Acquisition

Spatial contraction alone does not establish convergence.

A region may become arbitrarily small while objective information remains insufficient.

Therefore:

$$
\boxed{
\text{Spatial Contraction}
\neq
\text{Objective Convergence}.
}
$$

The information-acquisition condition is a separate requirement.

This is why ARRGO distinguishes:

$$
\text{Structural Refinement}
$$

from:

$$
\text{Information Refinement}.
$$

### 48.17 Dependence on Objective Regularity

Continuity is sufficient for the objective-value convergence argument presented here.

If continuity is removed, geometric proximity to a maximizer does not necessarily imply objective-value proximity.

For example, discontinuous objectives may contain arbitrarily sharp changes near a point.

Therefore the convergence theorem cannot simply be extended to arbitrary black-box functions without additional assumptions.

The theoretical statement is:

$$
\boxed{
\text{Continuity}
+
\text{Information Acquisition}
\Rightarrow
\text{Objective-Value Convergence}.
}
$$

### 48.18 Numerical Precision Limits

The mathematical model assumes exact geometric reasoning.

Real implementations operate with finite-precision arithmetic.

Therefore sufficiently small regions may eventually encounter:

- floating-point equality issues;
- duplicate evaluation detection limits;
- indistinguishable split locations;
- underflow or overflow in objective calculations;
- numerical instability in distance calculations.

Consequently the practical implementation requires numerical tolerances and explicit resolution limits.

The theoretical interpretation then becomes:

$$
\boxed{
\text{Convergence Until Numerical Resolution Limits}
}
$$

rather than an assertion of infinite-precision computation.

### 48.19 Budget-Limited Execution

The theoretical convergence result concerns an asymptotic refinement process.

A practical run has a finite evaluation budget:

$$
N_{\max}<\infty.
$$

Therefore the algorithm may terminate before the asymptotic conditions are fully realized.

Budget termination therefore means:

$$
\boxed{
\text{Best Found Within Budget}.
}
$$

It does not mean:

$$
\boxed{
\text{Global Optimum Proven}.
}
$$

Certified termination is different because it can provide:

$$
f^\ast-f_{\mathrm{best}}
\leq
\epsilon
$$

when the certified gap condition is satisfied.

### 48.20 No Universal Evaluation Complexity Claim

The current theory establishes convergence conditions, not a universal evaluation-complexity theorem.

Therefore ARRGO should not currently claim a bound such as:

$$
N(\epsilon,d)
\leq
C d^p\epsilon^{-q}
$$

without a separate complexity analysis and explicit assumptions on the objective class.

The appropriate current claim is:

$$
\boxed{
\text{Convergence Validity}
}
$$

rather than:

$$
\boxed{
\text{Universal Complexity Optimality}.
}
$$

### 48.21 Practical Scalability Claim

The multidimensional framework is designed to scale conceptually through dimension-parameterized abstractions.

However, practical scalability must be established experimentally.

A scientifically valid evaluation would compare behavior across increasing dimensions, for example:

$$
d=1,\ 2,\ 3,\ 5,\ 10,\ldots
$$

using controlled benchmark families.

Relevant measurements could include:

- evaluation count;
- objective error;
- region count;
- hierarchy depth;
- refinement distribution;
- computation time;
- certified gap where applicable.

Such experiments would provide empirical evidence about scalability but would not replace the mathematical convergence theorem.

### 48.22 Current Scope of the ARRGO Project

The present project intentionally separates theoretical generalization from multidimensional implementation.

The current software implementation is one-dimensional.

The mathematical framework, however, has been generalized to:

$$
\boxed{
\Omega\subset\mathbb R^d,
\qquad
d<\infty.
}
$$

Therefore the correct project statement is:

> ARRGO has a mathematically generalized formulation for finite-dimensional bounded optimization, while the current implementation and experimental validation remain one-dimensional.

This statement is stronger scientifically than claiming unsupported multidimensional software performance.

### 48.23 What the Generalization Establishes

The completed theoretical development establishes that:

1. the domain representation generalizes;
2. point representation generalizes;
3. region representation generalizes;
4. sampling generalizes;
5. coverage generalizes;
6. behavioral analysis generalizes;
7. structural splitting generalizes;
8. directional fairness can be formalized;
9. global fairness can be formalized;
10. spatial contraction can be proved;
11. information acquisition can be formalized;
12. objective-value convergence can be proved conditionally;
13. certified bounds generalize under a valid Lipschitz assumption;
14. the one-dimensional formulation is recovered when \(d=1\).

Therefore:

$$
\boxed{
\text{Theoretical Generalization Is Established}.
}
$$

### 48.24 What the Generalization Does Not Establish

The theoretical development does not establish:

- universal polynomial complexity;
- immunity to the curse of dimensionality;
- optimal evaluation complexity;
- automatic efficiency in high dimensions;
- empirical superiority over other optimizers;
- automatic validity of an estimated Lipschitz constant;
- correctness of a future multidimensional implementation without validation;
- convergence for infinite-dimensional optimization.

Therefore:

$$
\boxed{
\text{General Mathematical Validity}
\neq
\text{Universal Practical Superiority}.
}
$$

### 48.25 Final Boundary of the Claim

The scientifically defensible claim is:

$$
\boxed{
\begin{aligned}
&\text{ARRGO admits a mathematically consistent}
\\
&\text{generalization to arbitrary finite-dimensional}
\\
&\text{bounded optimization domains,}
\\
&\text{under explicitly stated assumptions.}
\end{aligned}
}
$$

The one-dimensional implementation provides a concrete realization of this framework.

The multidimensional formulation establishes the theoretical structure required for future implementations and validation.

This separation between:

$$
\boxed{
\text{Theory}
\quad\text{and}\quad
\text{Implementation}
}
$$

is intentional and should be preserved throughout the final scientific presentation of the project.

## 48. Limits of the Multidimensional Generalization

The previous sections established that the mathematical structure of ARRGO can be generalized from one-dimensional intervals to arbitrary finite-dimensional hyper-rectangular domains.

However, mathematical generality does not automatically imply computational scalability or practical efficiency.

This section identifies the precise limits of the generalization and separates theoretical validity from implementation complexity and empirical performance.

### 48.1 Mathematical Generality

The generalized formulation applies to:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j],
\qquad
d<\infty.
$$

The convergence argument remains valid under the stated assumptions:

$$
\boxed{
\text{Finite Dimension}
+
\text{Valid Refinement}
+
\text{Fairness}
+
\text{Information Acquisition}
+
\text{Continuity}
\Rightarrow
\text{Objective-Value Convergence}.
}
$$

This establishes mathematical generality.

It does not establish that the computational cost remains constant as \(d\) increases.

### 48.2 Dimensionality and Computational Cost

As the dimension increases, the geometric complexity of the search domain increases rapidly.

A hyper-rectangle has:

$$
2d
$$

boundary faces and:

$$
2^d
$$

vertices.

More importantly, representing and exploring sufficiently fine spatial structure may require an increasing number of evaluations.

Therefore:

$$
\boxed{
\text{Theoretical Validity}
\neq
\text{Dimension-Independent Computational Cost}.
}
$$

ARRGO is mathematically defined for arbitrary finite \(d\), but this does not imply constant computational complexity with respect to dimension.

### 48.3 The Curse of Dimensionality

High-dimensional optimization can suffer from the curse of dimensionality.

For example, suppose each coordinate is discretized into \(m\) levels.

A full Cartesian grid would contain:

$$
m^d
$$

points.

This grows exponentially with \(d\).

ARRGO does not require construction of such a full Cartesian grid.

Instead, it performs adaptive region refinement and information-driven sampling.

Therefore its objective is to avoid unnecessary global enumeration.

Nevertheless:

$$
\boxed{
\text{Adaptive Refinement}
\neq
\text{Elimination of the Curse of Dimensionality}.
}
$$

No claim should be made that ARRGO completely removes the fundamental difficulty of high-dimensional global optimization.

### 48.4 Adaptive Refinement as a Response

The central response of ARRGO to high-dimensional complexity is to avoid refining every region and every direction uniformly.

Instead, the framework attempts to identify:

- globally relevant regions;
- unresolved spatial directions;
- informative sampling locations;
- regions with meaningful optimization potential;
- regions with large certified uncertainty when certification is available.

Thus the intended computational principle is:

$$
\boxed{
\text{Refine According to Information}
}
$$

rather than:

$$
\boxed{
\text{Refine Uniformly Everywhere}.
}
$$

This is a design principle rather than a proven complexity bound.

### 48.5 Directional Complexity

In \(d\) dimensions, a structural split must specify both:

$$
(j,s_j),
$$

where:

- \(j\) identifies the coordinate direction;
- \(s_j\) identifies the split location.

Therefore the number of possible structural decisions increases with the number of available dimensions and candidate split locations.

The generalized algorithm must therefore distinguish:

$$
\text{Direction Selection}
$$

from:

$$
\text{Split-Location Selection}.
$$

This separation is essential for maintaining a deterministic and interpretable refinement process.

### 48.6 Joint Objective Interactions

The objective remains a joint function:

$$
f(x_1,\ldots,x_d).
$$

Therefore the optimization cannot generally be decomposed as:

$$
f(x_1,\ldots,x_d)
=
\sum_{j=1}^{d}f_j(x_j).
$$

Without a separability assumption, changing one coordinate can alter the effect of another coordinate.

Consequently, coordinate-wise refinement alone cannot characterize the complete objective landscape.

ARRGO therefore requires multidimensional sampling capable of observing joint configurations.

### 48.7 Sampling Complexity

The multidimensional sampling mechanism may require evaluating points that explore combinations of coordinates.

This can become expensive as \(d\) increases.

The framework therefore does not claim:

$$
N_{\mathrm{evaluations}}
=
O(d)
$$

or any other universal dimension-independent evaluation bound.

Instead, the number of evaluations remains dependent on:

- objective structure;
- dimension;
- domain geometry;
- refinement policy;
- information resolution requirements;
- numerical tolerances;
- desired optimization accuracy.

### 48.8 Region Count

Binary splitting creates two children from one parent.

Because parents are retained structurally, after \(s\) binary splits the hierarchy contains:

$$
1+2s
$$

region nodes, assuming every split replaces one current frontier region by two children.

This is a structural-history count.

The number of current frontier regions is instead:

$$
1+s.
$$

1+s,

## 50. Final Multidimensional Theoretical Conclusion

The preceding sections developed a finite-dimensional mathematical generalization of the ARRGO framework.

The generalization was constructed by preserving the central optimization principles of the original one-dimensional framework while replacing scalar geometry with dimension-parameterized hyper-rectangular geometry.

This section states the final theoretical conclusions and defines the exact scientific claims supported by the development.

### 50.1 Generalized Optimization Domain

ARRGO can be formulated over any bounded finite-dimensional hyper-rectangular domain:

$$
\boxed{
\Omega
=
\prod_{j=1}^{d}[a_j,b_j]
\subset\mathbb R^d,
\qquad
d<\infty.
}
$$

The one-dimensional interval is therefore not a separate mathematical model.

It is the special case:

$$
d=1.
$$

### 50.2 Generalized Region Structure

A region is represented as:

$$
\boxed{
R
=
\prod_{j=1}^{d}[l_j,r_j].
}
$$

The generalized region preserves the central ARRGO idea that optimization is performed through adaptive refinement of spatial regions.

The geometry is described through:

$$
W(R)
=
(w_1,\ldots,w_d),
$$

where:

$$
w_j=r_j-l_j,
$$

and through the region diameter:

$$
\operatorname{diam}(R)
=
\sqrt{
\sum_{j=1}^{d}w_j^2
}.
$$

Thus the geometric resolution of ARRGO remains explicitly measurable in any finite dimension.

### 50.3 Generalized Information Model

Objective evaluations remain:

$$
e_i
=
(x_i,f(x_i),\operatorname{Id}_i),
$$

with:

$$
x_i\in\Omega.
$$

The global evaluation history remains:

$$
D_t
=
\{e_1,\ldots,e_t\}.
$$

Therefore the fundamental information model does not depend on dimension.

Only the representation of the evaluation point changes from:

$$
x\in\mathbb R
$$

to:

$$
x\in\mathbb R^d.
$$

### 50.4 Generalized Refinement Principle

The fundamental ARRGO refinement principle is preserved:

$$
\boxed{
\text{Identify Unresolved Information}
\rightarrow
\text{Refine}
\rightarrow
\text{Acquire Information}
\rightarrow
\text{Reanalyze}.
}
$$

In multiple dimensions, refinement consists of two complementary mechanisms:

$$
\boxed{
\text{Sampling}
}
$$

and:

$$
\boxed{
\text{Structural Splitting}.
}
$$

Sampling improves objective information.

Splitting improves spatial resolution.

The two operations therefore remain conceptually distinct.

### 50.5 Generalized Structural Splitting

A multidimensional split is represented by:

$$
(j,s_j),
$$

where \(j\) identifies the selected coordinate direction and \(s_j\) identifies the split location.

The resulting children preserve domain coverage:

$$
R=R_1\cup R_2.
$$

Under the contraction condition:

$$
\max
\left(
\frac{s_j-l_j}{r_j-l_j},
\frac{r_j-s_j}{r_j-l_j}
\right)
\leq
\rho,
$$

with:

$$
0<\rho<1,
$$

the selected spatial direction contracts.

Repeated fair refinement can therefore force:

$$
\boxed{
\operatorname{diam}(R_t)\rightarrow0.
}
$$

### 50.6 Global and Directional Fairness

Two distinct fairness requirements are necessary in the multidimensional formulation.

Global region fairness ensures that:

$$
\text{relevant regions are not permanently ignored}.
$$

Directional fairness ensures that:

$$
\text{unresolved spatial directions are not permanently ignored}.
$$

Together:

$$
\boxed{
\text{Global Fairness}
+
\text{Directional Fairness}
}
$$

provide the structural access required by the convergence argument.

### 50.7 Information Acquisition

Spatial contraction alone is not sufficient.

The framework must also acquire objective information increasingly close to globally relevant points.

For a global maximizer \(x^\ast\):

$$
\inf_{e_i\in D_t}
\|x_i-x^\ast\|_2
\rightarrow0.
$$

Combined with continuity:

$$
f(x_i)
\rightarrow
f(x^\ast)
=
f^\ast.
$$

Therefore:

$$
\boxed{
\text{Information Acquisition}
+
\text{Continuity}
\Rightarrow
\text{Objective-Value Convergence}.
}
$$

### 50.8 Global Objective-Value Convergence

Under the complete set of sufficient assumptions:

$$

This is the primary theoretical guarantee of the finite-dimensional ARRGO framework.

### 50.9 Certified Optimization

When a valid global Lipschitz constant exists:

$$
|f(x)-f(y)|
\leq
L\|x-y\|_2,
$$

the framework additionally provides certified regional bounds:

$$
L_R(x)
\leq
f(x)
\leq
U_R(x).
$$

A valid certification frontier then gives:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast

0
\leq

\geq
f^\ast-\epsilon.

Thus Certified Mode provides a rigorous epsilon-optimality statement under the stated assumptions.

### 50.10 Certification Gap Convergence

\boxed{

\rightarrow0.

The certification gap consequently represents a mathematically meaningful measure of remaining objective uncertainty.

## 51. Theoretical Consistency Audit

The multidimensional generalization has now been developed through definitions, propositions, convergence conditions, certification results, and the one-dimensional recovery result.

Before presenting the final theoretical conclusions, a consistency audit is required.

The purpose of this audit is to verify that:

1. every major guarantee has explicit assumptions;
2. structural properties are not confused with convergence guarantees;
3. empirical observations are not presented as mathematical proofs;
4. Certified Mode is not confused with Empirical Mode;
5. the no-pruning hierarchy is consistent with certification;
6. multidimensional refinement does not implicitly assume separability;
7. finite-dimensional generality is not confused with computational scalability.

### 51.1 Domain Consistency

The generalized domain is:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j],
\qquad
d<\infty.
$$

Each interval is closed and bounded.

Therefore:

$$
\Omega
\text{ is compact}.
$$

For a continuous objective:

$$
f\in C(\Omega),
$$

the extreme value theorem guarantees the existence of at least one global maximizer:

$$
\boxed{
\exists x^\ast\in\Omega:
f(x^\ast)=f^\ast.
}
$$

Therefore the convergence target is mathematically well-defined.

### 51.2 Region Consistency

Every ARRGO region has the form:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j]
\subseteq
\Omega.
$$

A valid split in direction \(j\) produces:

$$
R_1
=
\left(
\prod_{k\neq j}[l_k,r_k]
\right)
\times
[l_j,s_j],
$$

and:

$$
R_2
=
\left(
\prod_{k\neq j}[l_k,r_k]
\right)
\times
[s_j,r_j].
$$

Therefore:

$$
R_1,R_2\subseteq R
$$

and:

$$
R=R_1\cup R_2.
$$

Hence structural refinement preserves domain coverage.

### 51.3 Boundary Consistency

Because the mathematical regions are closed:

$$
R_1\cap R_2
$$

contains the split boundary:

$$
\left\{
x\in R:
x_j=s_j
\right\}.
$$

This overlap does not invalidate the mathematical union:

$$
R_1\cup R_2=R.
$$

However, an implementation must define how evaluation and region membership are assigned when a point lies exactly on a shared boundary.

A deterministic ownership convention is therefore required.

This is an implementation-level requirement rather than a mathematical contradiction.

### 51.4 Contraction Consistency

For a valid split:

$$
\max
\left(
\frac{s_j-l_j}{r_j-l_j},
\frac{r_j-s_j}{r_j-l_j}
\right)
\leq
\rho,
$$

with:

$$
0<\rho<1.
$$

The selected directional width therefore contracts:

$$
w_j'
\leq
\rho w_j.
$$

However, binary splitting requires:

$$
\rho\geq\frac12
$$

for the admissible split interval to be nonempty.

Therefore a practical binary-splitting configuration must satisfy:

$$
\boxed{
\frac12\leq\rho<1.
}
$$

This condition should be distinguished from the abstract contraction condition:

$$
0<\rho<1.
$$

The former is required specifically for the binary split mechanism used by ARRGO.

### 51.5 Multidimensional Contraction Consistency

A split contracts only one coordinate direction.

Therefore a single split does not imply:

$$
\operatorname{diam}(R')<\operatorname{diam}(R)
$$

by itself unless the selected direction has positive width.

For a non-degenerate region and a valid interior split:

$$
0<s_j-l_j<r_j-l_j,
$$

and:

$$
0<r_j-s_j<r_j-l_j.
$$

Consequently the selected width decreases.

Since all other widths remain unchanged:

$$
\operatorname{diam}(R')
<
\operatorname{diam}(R).
$$

Repeated contraction of every unresolved direction is obtained only under directional fairness.

Thus:

$$
\boxed{
\text{Single Split}
\neq
\text{Global Spatial Contraction}.
}
$$

Rather:

$$
\boxed{
\text{Repeated Fair Directional Splitting}
\Rightarrow
\text{Global Spatial Contraction}.
}
$$

### 51.6 Fairness Consistency

Two different fairness properties are required.

Global region fairness concerns:

$$
\text{which region is refined}.
$$

Directional fairness concerns:

$$
\text{which coordinate direction is refined inside a region}.
$$

These properties solve different starvation problems.

Therefore they must not be collapsed into a single informal statement.

The convergence argument requires that globally relevant refinement paths remain accessible and that unresolved directions on such paths are not permanently ignored.

Hence:

$$
\boxed{
\text{Global Fairness}
+
\text{Directional Fairness}
}
$$

is the appropriate combined condition.

### 51.7 Information-Acquisition Consistency

Spatial contraction does not automatically imply that objective values are observed near the global maximizer.

The convergence argument therefore requires an information-acquisition condition.

For a global maximizer \(x^\ast\), a sufficient condition is:

$$
\operatorname{dist}(x^\ast,D_t)
=
\inf_{x_i\in D_t}
\|x_i-x^\ast\|_2
\rightarrow0.
$$

By continuity:

$$
f(x_i)\rightarrow f(x^\ast).
$$

Since:

$$
f_{\mathrm{best}}^{(t)}
\geq
f(x_i),
$$

and:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast,
$$

we obtain:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast.
}
$$

Therefore the information-acquisition assumption is explicitly connected to the final objective-value convergence result.

### 51.8 Continuity Consistency

The convergence argument uses continuity to convert:

$$
x_i\rightarrow x^\ast
$$

into:

$$
f(x_i)\rightarrow f(x^\ast).
$$

Without continuity, this implication is not generally valid.

Therefore the theorem should not be stated for arbitrary deterministic black-box objectives without additional regularity assumptions.

The correct theoretical scope is:

$$
\boxed{
f\in C(\Omega).
}
$$

This does not mean that ARRGO cannot be executed on discontinuous objectives.

It means that the stated global convergence theorem does not automatically apply to them.

### 51.9 Incumbent Consistency

The incumbent is defined by:

$$
f_{\mathrm{best}}^{(t)}
=
\max_{x_i\in D_t}f(x_i).
$$

Therefore:

$$
f_{\mathrm{best}}^{(t+1)}
\geq
f_{\mathrm{best}}^{(t)}.
$$

Thus the incumbent objective value is monotone nondecreasing.

Because every evaluated point belongs to the domain:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast.
$$

Hence:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
\text{ is monotone nondecreasing and bounded above by }f^\ast.
}
$$

This establishes consistency of the global incumbent reasoning.

### 51.10 Empirical Mode Consistency

In Empirical Mode, ARRGO uses observed objective values without requiring a valid global upper bound.

Therefore the result after finite budget termination is:

$$
\boxed{
x_{\mathrm{best}}
=
\text{best evaluated solution found within the budget}.
}
$$

No universal claim of exact global optimality follows from empirical execution alone.

Thus:

$$
\boxed{
\text{Empirical Result}
=
\text{Best Found},
}
$$

not:

$$
\boxed{
\text{Empirical Result}
=
\text{Proven Global Optimum}.
}
$$

### 51.11 Certified Mode Consistency

Certified Mode additionally assumes:

$$
|f(x)-f(y)|
\leq
L\|x-y\|_2
\qquad
\forall x,y\in\Omega.
$$

This gives:

$$
f(x_i)-L\|x-x_i\|_2
\leq
f(x)
\leq
f(x_i)+L\|x-x_i\|_2.
$$

Taking the maximum of valid lower bounds and minimum of valid upper bounds gives:

$$
L_R(x)
\leq
f(x)
\leq
U_R(x).
$$

Therefore the certification mechanism is logically separate from the empirical optimization mechanism.

### 51.12 Certification-Frontier Consistency

The complete hierarchy contains historical parent regions.

Therefore defining certification over every historical region would unnecessarily mix:

$$
\text{structural history}
$$

with:

$$
\text{current spatial cover}.
$$

Instead, certification uses a frontier:

$$
\mathcal F_t
\subseteq
\mathcal H_t
$$

such that:

$$
\Omega
\subseteq
\bigcup_{R\in\mathcal F_t}R.
$$

The global potential is then:

$$
P_{\mathrm{global}}^{(t)}
=
\sup_{R\in\mathcal F_t}P(R).
$$

This preserves both principles:

$$
\boxed{
\text{No Pruning of Structural History}
}
$$

and:

$$
\boxed{
\text{Complete Current Certification Cover}.
}
$$

Therefore no-pruning and certified reasoning are compatible.

### 51.13 Certification-Gap Consistency

Because:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast
\leq
P_{\mathrm{global}}^{(t)},
$$

we obtain:

$$
0
\leq
f^\ast-f_{\mathrm{best}}^{(t)}
\leq
P_{\mathrm{global}}^{(t)}
-
f_{\mathrm{best}}^{(t)}.
$$

Define:

$$
\Delta_{\mathrm{global}}^{(t)}
=
P_{\mathrm{global}}^{(t)}
-
f_{\mathrm{best}}^{(t)}.
$$

Therefore:

$$
\boxed{
f^\ast-f_{\mathrm{best}}^{(t)}
\leq
\Delta_{\mathrm{global}}^{(t)}.
}
$$

This confirms that the certification gap is a valid upper bound on the incumbent's objective error under the certification assumptions.

### 51.14 Certification Versus Convergence

The certification gap provides a rigorous finite-time error bound when the certification assumptions hold.

The convergence theorem describes asymptotic behavior under the refinement and information assumptions.

These are related but distinct statements.

Therefore:

$$
\boxed{
\text{Certification}
\neq
\text{Convergence Proof}.
}
$$

Certification provides a bound at a particular state.

Convergence establishes what happens under continued valid refinement and information acquisition.

### 51.15 Location-Convergence Consistency

Objective-value convergence does not imply convergence to a unique location.

If:

$$
\operatorname*{arg\,max}_{x\in\Omega}f(x)
$$

contains multiple points, different evaluated sequences may approach different global maximizers.

Therefore the primary convergence statement is:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast.
}
$$

Location convergence requires additional assumptions such as uniqueness or isolation of the global maximizer.

### 51.16 No Separability Assumption

The multidimensional theory does not require:

$$
f(x_1,\ldots,x_d)
=
\sum_j f_j(x_j).
$$

The objective may contain arbitrary coordinate interactions subject to the assumed regularity.

Therefore multidimensional ARRGO remains a joint optimization framework.

Coordinate-wise structural refinement is only a geometric operation.

It does not imply coordinate-wise decomposition of the objective.

### 51.17 Determinism Consistency

For deterministic objective evaluation and deterministic candidate-generation, region-selection, tie-breaking, and action-selection rules, the transition from one algorithmic state to the next is deterministic.

Therefore:

$$
G_{t+1}
=
T(G_t,A_t)
$$

is uniquely determined once:

$$
G_t
$$

and the selected action are fixed.

Consequently identical initial states and configurations produce identical algorithmic trajectories, subject to deterministic numerical behavior.

This establishes reproducibility as a design property.

### 51.18 Budget Consistency

Only objective evaluations consume the evaluation budget.

Structural operations such as:

- region analysis;
- candidate generation;
- split evaluation;
- hierarchy updates;

do not themselves evaluate the objective.

Therefore:

$$
N_t
\leq
N_{\max}
$$

must hold throughout execution.

A split-only sequence cannot be allowed to continue indefinitely without either:

- reaching a valid numerical resolution limit;
- producing a terminal state;
- or acquiring objective information.

This ensures that structural refinement cannot bypass termination requirements.

### 51.19 Progress Consistency

Every non-terminal iteration must produce meaningful state progress.

A valid action may change:

- objective information;
- spatial resolution;
- hierarchy structure;
- certification state;
- or another explicitly tracked unresolved quantity.

Therefore an action that leaves the complete algorithmic state unchanged must not be repeatedly executed.

This prevents infinite loops caused by repeatedly selecting a formally valid but ineffective action.

### 51.20 Numerical Consistency

The mathematical theory assumes exact real arithmetic.

The implementation uses finite-precision arithmetic.

Therefore numerical tolerances are required for:

- duplicate evaluation detection;
- point equality;
- region membership;
- split-location comparisons;
- minimum region width;
- distance calculations.

These tolerances define the practical numerical resolution of the implementation.

They do not change the mathematical convergence theorem, but they limit its literal computational realization.

### 51.21 Complexity Consistency

The theoretical framework is defined for:

$$
1\leq d<\infty.
$$

This establishes finite-dimensional mathematical generality.

It does not establish:

$$
O(1)
$$

or polynomial computational complexity with respect to dimension.

The framework may still experience increasing evaluation and computational cost as \(d\) grows.

Therefore:

$$
\boxed{
\text{Finite-Dimensional Generality}
\neq
\text{Guaranteed High-Dimensional Efficiency}.
}
$$

### 51.22 Experimental Consistency

The one-dimensional experiments provide empirical observations about the implemented ARRGO system.

They can support claims concerning:

- observed optimization behavior;
- reproducibility;
- numerical stability;
- benchmark performance;
- refinement patterns.

They cannot prove:

- universal global optimality;
- universal high-dimensional scalability;
- universal superiority over other methods.

Therefore:

$$
\boxed{
\text{Experimental Evidence}
\text{ supports but does not replace }
\text{Mathematical Proof}.
}
$$

### 51.23 One-Dimensional Consistency

Setting:

$$
d=1
$$

reduces:

$$
\Omega
=
\prod_{j=1}^{1}[a_j,b_j]
$$

to:

$$
\Omega=[a,b].
$$

The generalized geometry, splitting, distance, certification, and convergence conditions all reduce to their one-dimensional counterparts.

Therefore:

$$
\boxed{
\mathrm{ARRGO}(1)
\equiv
\mathrm{ARRGO}_{1D}.
}
$$

This confirms that the generalized theory does not replace the original theory; it contains it as a special case.

## 52. Master Theorem of the Finite-Dimensional ARRGO Framework

This section combines the structural, spatial, informational, optimization, and certification results developed throughout the multidimensional generalization.

The purpose is to state the complete mathematical result in one formal theorem.

### 52.1 Theorem Statement

Let:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j]
\subset\mathbb R^d,
\qquad
1\leq d<\infty,
$$

be a bounded finite-dimensional hyper-rectangular domain.

Let:

$$
f:\Omega\rightarrow\mathbb R
$$

be deterministic and continuous.

Consider an ARRGO process that satisfies the following conditions.

#### C1 — Valid Regions

Every region is a hyper-rectangle:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j]
\subseteq\Omega.
$$

#### C2 — Valid Structural Splitting

Every split produces valid children whose union equals the parent:

$$
R=R_1\cup R_2,
\qquad
R_1,R_2\subseteq R.
$$

#### C3 — Spatial Contraction

Whenever a region is split in direction \(j\):

$$
w_j'
\leq
\rho w_j,
$$

where:

$$
\frac12\leq\rho<1.
$$

The lower bound \(\rho\geq1/2\) corresponds to the binary splitting mechanism.

#### C4 — Global Region Fairness

Every persistent globally relevant and eligible refinement path receives refinement opportunities and cannot be permanently ignored.

#### C5 — Directional Fairness

Every unresolved coordinate direction on a persistently refined relevant path is selected for refinement infinitely often, unless a valid numerical resolution limit is reached.

#### C6 — Information Acquisition

For every global maximizer \(x^\ast\), the evaluation history satisfies:

$$
\operatorname{dist}(x^\ast,D_t)
=
\inf_{x_i\in D_t}
\|x_i-x^\ast\|_2
\rightarrow0.
$$

#### C7 — Exact Objective Evaluation

Every stored objective value is an exact deterministic evaluation:

$$
y_i=f(x_i).
$$

Under C1–C7, the ARRGO incumbent satisfies:

$$
\boxed{
\lim_{t\rightarrow\infty}
f_{\mathrm{best}}^{(t)}
=
f^\ast,
}
$$

where:

$$
f^\ast
=
\max_{x\in\Omega}f(x).
$$

### 52.2 Proof

Because:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j]
$$

is closed and bounded in finite-dimensional Euclidean space, it is compact.

Since:

$$
f\in C(\Omega),
$$

the extreme value theorem gives:

$$
\exists x^\ast\in\Omega
\quad
\text{such that}
\quad
f(x^\ast)=f^\ast.
$$

Therefore a global maximizer exists.

The evaluation history is cumulative:

$$
D_t
\subseteq
D_{t+1}.
$$

The incumbent is defined by:

$$
f_{\mathrm{best}}^{(t)}
=
\max_{x_i\in D_t}f(x_i).
$$

Therefore:

$$
f_{\mathrm{best}}^{(t+1)}
\geq
f_{\mathrm{best}}^{(t)}.
$$

Since every evaluated point belongs to \(\Omega\):

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast.
$$

Thus the incumbent sequence is monotone nondecreasing and bounded above.

Now consider any global maximizer \(x^\ast\).

By C4, the refinement process cannot permanently ignore the relevant region containing \(x^\ast\).

By C5 and C3, every unresolved coordinate direction on that relevant path is repeatedly contracted.

For coordinate \(j\), if it is selected \(n_j(t)\) times:

$$
w_j^{(t)}
\leq
\rho^{n_j(t)}w_j^{(0)}.
$$

Since:

$$
n_j(t)\rightarrow\infty
$$

and:

$$
0<\rho<1,
$$

we obtain:

$$
w_j^{(t)}\rightarrow0.
$$

Because this holds for every:

$$
j=1,\ldots,d,
$$

the region diameter satisfies:

$$
\operatorname{diam}(R_t^\ast)
=
\sqrt{
\sum_{j=1}^{d}
(w_j^{(t)})^2
}
\rightarrow0.
$$

C6 additionally guarantees that objective evaluations approach \(x^\ast\):

$$
\operatorname{dist}(x^\ast,D_t)\rightarrow0.
$$

Therefore there exists a sequence of evaluated points:

$$
x_{i_t}\in D_t
$$

such that:

$$
x_{i_t}\rightarrow x^\ast.
$$

By continuity of \(f\):

$$
f(x_{i_t})
\rightarrow
f(x^\ast)
=
f^\ast.
$$

Since the incumbent is the maximum observed value:

$$
f_{\mathrm{best}}^{(t)}
\geq
f(x_{i_t}),
$$

while:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast,
$$

we obtain:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast.
}
$$

This proves global objective-value convergence.

### 52.3 Certified Extension

Assume additionally that \(f\) satisfies a valid global Lipschitz condition:

$$
|f(x)-f(y)|
\leq
L\|x-y\|_2
\qquad
\forall x,y\in\Omega.
$$

For each region \(R\), define:

$$
U_R(x)
=
\min_{x_i\in D_R}
\left[
f(x_i)
+
L\|x-x_i\|_2
\right],
$$

and:

$$
L_R(x)
=
\max_{x_i\in D_R}
\left[
f(x_i)
-
L\|x-x_i\|_2
\right].
$$

Then:

$$
L_R(x)
\leq
f(x)
\leq
U_R(x).
$$

For a certification frontier \(\mathcal F_t\) satisfying:

$$
\Omega
\subseteq
\bigcup_{R\in\mathcal F_t}R,
$$

define:

$$
P_{\mathrm{global}}^{(t)}
=
\sup_{R\in\mathcal F_t}
\sup_{x\in R}
U_R(x).
$$

Then:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast
\leq
P_{\mathrm{global}}^{(t)}.
$$

Therefore:

$$
0
\leq
f^\ast-f_{\mathrm{best}}^{(t)}
\leq
\Delta_{\mathrm{global}}^{(t)},
$$

where:

$$
\Delta_{\mathrm{global}}^{(t)}
=
P_{\mathrm{global}}^{(t)}
-
f_{\mathrm{best}}^{(t)}.
$$

If:

$$
\Delta_{\mathrm{global}}^{(t)}
\leq
\epsilon,
$$

then:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
\geq
f^\ast-\epsilon.
}
$$

Hence the current incumbent is mathematically certified as epsilon-optimal.

### 52.4 Certified Gap Convergence

Under the additional condition that the certification frontier becomes spatially resolving and each frontier region has valid regional information, the certified upper bound becomes asymptotically tight.

Let:

$$
\delta_t
=
\max_{R\in\mathcal F_t}
\operatorname{diam}(R).
$$

Suppose:

$$
\delta_t\rightarrow0.
$$

For every frontier region \(R\), assume at least one valid evaluation \(x_i\in D_R\).

For every \(x\in R\):

$$
\|x-x_i\|_2
\leq
\operatorname{diam}(R).
$$

Therefore:

$$
U_R(x)
\leq
f(x_i)
+
L\operatorname{diam}(R).
$$

Since:

$$
f(x_i)\leq f^\ast,
$$

we obtain:

$$
U_R(x)
\leq
f^\ast
+
L\operatorname{diam}(R).
$$

Taking the supremum over \(x\in R\):

$$
P(R)
\leq
f^\ast
+
L\operatorname{diam}(R).
$$

Taking the supremum over the certification frontier:

$$
P_{\mathrm{global}}^{(t)}
\leq
f^\ast
+
L\delta_t.
$$

At the same time, because the frontier covers \(x^\ast\):

$$
P_{\mathrm{global}}^{(t)}
\geq
f^\ast.
$$

Therefore:

$$
f^\ast
\leq
P_{\mathrm{global}}^{(t)}
\leq
f^\ast+L\delta_t.
$$

Since:

$$
\delta_t\rightarrow0,
$$

the squeeze theorem gives:

$$
\boxed{
P_{\mathrm{global}}^{(t)}
\rightarrow
f^\ast.
}
$$

Combined with:

$$
f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast,
$$

we obtain:

$$
\boxed{
\Delta_{\mathrm{global}}^{(t)}
\rightarrow0.
}
$$

This establishes the asymptotic validity of the certified convergence gap under the stated frontier-resolution condition.

### 52.5 Special Case \(d=1\)

For:

$$
d=1,
$$

the generalized domain becomes:

$$
\Omega=[a,b].
$$

The hyper-rectangle becomes an interval:

$$
R=[l,r].
$$

The Euclidean distance becomes:

$$
\|x-y\|_2=|x-y|.
$$

The directional split becomes a scalar split.

The multidimensional contraction condition becomes:

$$
\max
\left(
\frac{s-l}{r-l},
\frac{r-s}{r-l}
\right)
\leq
\rho.
$$

The certified envelopes become:

$$
U_R(x)
=
\min_i
\left[
f(x_i)+L|x-x_i|
\right],
$$

and:

$$
L_R(x)
=
\max_i
\left[
f(x_i)-L|x-x_i|
\right].
$$

Thus the generalized theorem reduces to the original one-dimensional theoretical framework.

### 52.6 What the Master Theorem Establishes

The theorem establishes a conditional mathematical result for arbitrary finite dimensions:

$$
\boxed{
1\leq d<\infty.
}
$$

It establishes:

- existence of a global optimum under compactness and continuity;
- validity of hyper-rectangular region refinement;
- spatial contraction under fair directional refinement;
- accessibility of relevant regions under global fairness;
- objective-value convergence under sufficient information acquisition;
- certified upper and lower bounds under a valid Lipschitz condition;
- certified epsilon-optimality under a sufficiently small global gap;
- asymptotic certification-gap convergence under a resolving certification frontier;
- exact recovery of the one-dimensional framework when \(d=1\).

### 52.7 What the Master Theorem Does Not Establish

The theorem does not establish:

- polynomial complexity in dimension;
- immunity to the curse of dimensionality;
- practical efficiency for large \(d\);
- superiority over other global optimization algorithms;
- automatic satisfaction of fairness or information-acquisition conditions by every possible implementation;
- convergence for arbitrary discontinuous objectives;
- correctness of an invalid Lipschitz constant;
- infinite-dimensional convergence;
- empirical performance in dimensions that have not been implemented and tested.

Therefore the theorem is a conditional mathematical guarantee rather than an unrestricted performance claim.

### 52.8 Final Theoretical Statement

The finite-dimensional ARRGO framework can therefore be stated as:

$$
\boxed{
\begin{aligned}
&\text{For every finite }d\geq1,
\\
&\text{ARRGO admits a mathematically consistent}
\\
&\text{region-based formulation over }
\Omega\subset\mathbb R^d.
\end{aligned}
}
$$

Under the stated structural, fairness, contraction, information, and regularity assumptions:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast.
}
$$

Under the additional valid Lipschitz and certification-frontier assumptions:

$$
\boxed{
\Delta_{\mathrm{global}}^{(t)}
\rightarrow0,
}
$$

and finite-time termination with:

$$
\Delta_{\mathrm{global}}^{(t)}
\leq
\epsilon
$$

provides:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
\geq
f^\ast-\epsilon.
}
$$

The one-dimensional ARRGO implementation is therefore the concrete \(d=1\) realization of a broader finite-dimensional mathematical framework.

The generalized theory does not claim that the current software already implements higher-dimensional optimization.

It establishes that the underlying ARRGO principles admit a coherent and conditionally convergent mathematical formulation beyond one dimension.

## 53. Theoretical Scope and Claim Boundary

The mathematical development of ARRGO establishes a finite-dimensional theoretical framework under explicit assumptions.

However, mathematical generality, convergence guarantees, certification guarantees, and computational performance represent different types of claims.

This section defines the exact scope of each claim.

### 53.1 Mathematical Generality

The ARRGO formulation is defined for:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j]
\subset\mathbb R^d,
\qquad
1\leq d<\infty.
$$

Therefore the mathematical formulation is not restricted to:

$$
d=1.
$$

The one-dimensional formulation is the special case:

$$
d=1.
$$

Hence:

$$
\boxed{
\text{Theoretical ARRGO is finite-dimensional}.
}
$$

### 53.2 Structural Validity

The structural part of the framework establishes that valid hyper-rectangular regions can be recursively refined while preserving domain coverage.

For every valid split:

$$
R=R_1\cup R_2
$$

with:

$$
R_1,R_2\subseteq R\subseteq\Omega.
$$

Therefore structural refinement does not inherently leave the feasible domain.

This establishes:

$$
\boxed{
\text{Structural Validity}.
}
$$

Structural validity alone does not imply optimization convergence.

### 53.3 Spatial Resolution

Under the contraction condition and directional fairness:

$$
\operatorname{diam}(R_t)\rightarrow0
$$

along a persistently refined relevant path.

Therefore the framework has a mathematically defined mechanism for increasing spatial resolution.

However, spatial resolution alone is insufficient.

The algorithm must also acquire objective information near globally relevant points.

Thus:

$$
\boxed{
\text{Spatial Resolution}
\neq
\text{Objective Convergence}.
}
$$

### 53.4 Information Resolution

The convergence argument additionally requires:

$$
\operatorname{dist}(x^\ast,D_t)
\rightarrow0.
$$

This condition connects spatial refinement to actual objective information.

Under continuity:

$$
x_i\rightarrow x^\ast
\quad\Longrightarrow\quad
f(x_i)\rightarrow f^\ast.
$$

Consequently:

$$
f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast.
$$

Thus the convergence result depends on both:

$$
\boxed{
\text{Spatial Refinement}
+
\text{Information Acquisition}.
}
$$

### 53.5 Conditional Nature of Convergence

The global convergence theorem is conditional.

It does not state:

> Every implementation of ARRGO automatically converges for every deterministic objective.

Instead, it states that if the specified assumptions and sufficient refinement conditions hold, then:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast.
}
$$

The distinction is essential because fairness and information acquisition are behavioral properties of an algorithmic policy.

They cannot be inferred merely from the existence of region splitting.

### 53.6 Certification Scope

Certified Mode requires an additional global regularity condition:

$$
|f(x)-f(y)|
\leq
L\|x-y\|_2.
$$

The value \(L\) must be valid over the complete domain.

Under this assumption:

$$
L_R(x)
\leq
f(x)
\leq
U_R(x).
$$

Consequently:

$$
f_{\mathrm{best}}
\leq
f^\ast
\leq
P_{\mathrm{global}}.
$$

Therefore the certification gap:

$$
\Delta_{\mathrm{global}}
=
P_{\mathrm{global}}
-
f_{\mathrm{best}}
$$

is a valid upper bound on the objective error.

### 53.7 Importance of a Valid Lipschitz Constant

A numerically estimated or guessed value of \(L\) is not automatically a valid certification constant.

If:

$$
L_{\mathrm{used}}<L_{\mathrm{true}},
$$

the resulting envelopes may fail to contain the objective.

Therefore:

$$
\boxed{
\text{Invalid }L
\Rightarrow
\text{No Valid Certification}.
}
$$

A conservative value:

$$
L_{\mathrm{used}}\geq L_{\mathrm{true}}
$$

preserves validity, although it may produce looser bounds.

Thus certification correctness depends on the validity of the supplied regularity constant.

### 53.8 Certification Frontier Requirement

Because ARRGO retains historical parent regions, the structural hierarchy and the current certification cover must remain conceptually separate.

The certification frontier must satisfy:

$$
\Omega
\subseteq
\bigcup_{R\in\mathcal F_t}R.
$$

Without complete coverage:

$$
f^\ast
\leq
P_{\mathrm{global}}
$$

cannot be guaranteed globally.

Therefore:

$$
\boxed{
\text{Complete Frontier Coverage}
}
$$

is a required component of global certification.

### 53.9 Finite-Time Versus Asymptotic Guarantees

Two different statements must be distinguished.

#### Finite-Time Certification

If:

$$
\Delta_{\mathrm{global}}^{(t)}
\leq
\epsilon,
$$

then:

$$
f_{\mathrm{best}}^{(t)}
\geq
f^\ast-\epsilon.
$$

This is a finite-time mathematical guarantee.

#### Asymptotic Convergence

Under the convergence assumptions:

$$
f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast.
$$

This is an asymptotic statement.

Therefore:

$$
\boxed{
\text{Finite-Time Certification}
\neq
\text{Asymptotic Convergence}.
}
$$

### 53.10 Budget-Limited Execution

In practical execution:

$$
N_t\leq N_{\max}.
$$

If the budget is exhausted before the certified gap reaches the required tolerance, the algorithm returns its best observed solution.

Therefore:

$$
\boxed{
\text{Budget Exhaustion}
\Rightarrow
\text{Best Found},
}
$$

unless a separate valid certificate has already been established.

This distinction applies directly to the current one-dimensional experiments.

### 53.11 Experimental Claim Boundary

The current experimental validation concerns the one-dimensional implementation.

Therefore the experiments establish evidence about:

- one-dimensional optimization behavior;
- benchmark performance;
- reproducibility;
- numerical stability;
- refinement behavior;
- empirical and certified execution where tested.

They do not establish:

- high-dimensional empirical performance;
- scalability to arbitrary \(d\);
- computational superiority;
- universal convergence of every implementation;
- universal optimality after a finite budget.

Thus:

$$
\boxed{
\text{Current Experiments}
\Rightarrow
\text{1D Empirical Evidence}.
}
$$

### 53.12 Theory-to-Implementation Boundary

The project currently contains two distinct levels:

$$
\boxed{
\text{Generalized Mathematical Framework}
}
$$

and:

$$
\boxed{
\text{Concrete 1D Implementation}.
}
$$

The generalized theory defines:

$$
\mathrm{ARRGO}(d),
\qquad
1\leq d<\infty.
$$

The implementation realizes:

$$
\mathrm{ARRGO}(1).
$$

Therefore the project should not state that the current software is already a multidimensional optimizer.

Instead:

> The current implementation realizes the one-dimensional case of a finite-dimensional mathematical ARRGO framework.

### 53.13 Performance Claim Boundary

The theoretical framework contains no universal complexity theorem of the form:

$$
T(d,N)
\leq
\text{polynomial bound}
$$

or:

$$
N(d,\epsilon)
\leq
\text{universal finite bound}.
$$

Therefore no such claim should be made without additional analysis.

The framework is mathematically general in finite dimensions, but its practical computational behavior may still depend strongly on:

- dimension;
- objective geometry;
- number of relevant regions;
- refinement policy;
- sampling policy;
- certification cost;
- numerical precision.

Hence:

$$
\boxed{
\text{Generality}
\neq
\text{Efficiency}.
}
$$

### 53.14 Curse of Dimensionality

Increasing dimension may increase the number of spatial degrees of freedom that must be resolved.

Even though ARRGO does not require a full Cartesian grid, multidimensional refinement may still become increasingly expensive.

Therefore the theoretical generalization does not eliminate the curse of dimensionality.

The appropriate claim is:

$$
\boxed{
\text{ARRGO Avoids Mandatory Full-Grid Enumeration}
}
$$

rather than:

$$
\boxed{
\text{ARRGO Eliminates the Curse of Dimensionality}.
}
$$

### 53.15 No Separability Requirement

The theoretical framework does not require:

$$
f(x_1,\ldots,x_d)
=
\sum_{j=1}^{d}f_j(x_j).
$$

Therefore the framework allows joint coordinate interactions.

This is important because selecting one coordinate direction for structural splitting is a geometric operation rather than an assumption that the objective is separable.

Thus:

$$
\boxed{
\text{Directional Refinement}
\neq
\text{Coordinate-Wise Optimization}.
}
$$

### 53.16 Determinism Claim

Under deterministic:

- objective evaluation;
- candidate generation;
- region selection;
- action selection;
- tie-breaking;

the ARRGO state transition is deterministic.

Therefore identical initial conditions produce identical trajectories, subject to deterministic numerical behavior.

This supports reproducibility.

However, determinism alone does not imply convergence.

Thus:

$$
\boxed{
\text{Determinism}
\neq
\text{Global Optimality}.
}
$$

### 53.17 No-Pruning Claim

ARRGO retains historical parent regions.

Therefore structural history is preserved:

$$
R_{\mathrm{parent}}
\in
\mathcal H_t
$$

even after refinement.

This does not require historical regions to remain part of the current certification frontier.

Hence:

$$
\boxed{
\text{No Structural Pruning}
+
\text{Current Frontier Certification}
}
$$

can coexist without contradiction.

### 53.18 Non-Unique Optima

The theoretical convergence result concerns objective value:

$$
f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast.
$$

It does not require a unique maximizer.

If:

$$
\operatorname*{arg\,max}_{x\in\Omega}f(x)
$$

contains multiple points, different valid trajectories may approach different maximizers.

Therefore location convergence requires additional assumptions.

This is consistent with the flat-peak benchmark.

### 53.19 Numerical Realization Boundary

The mathematical theory is formulated in exact real arithmetic.

The implementation uses floating-point arithmetic.

Therefore practical execution introduces tolerances for:

- duplicate detection;
- region membership;
- geometric comparisons;
- convergence thresholds;
- numerical resolution.

These tolerances create a practical resolution limit.

They should therefore be reported as implementation parameters rather than hidden assumptions.

### 53.20 Final Claim Classification

The claims of the ARRGO project can be classified as follows.

#### Mathematically Established

- finite-dimensional formulation;
- valid hyper-rectangular region model;
- valid multidimensional splitting;
- contraction mechanism;
- one-dimensional recovery;
- certified bound construction under a valid Lipschitz constant.

#### Conditionally Established

- global objective-value convergence;
- asymptotic certified-gap convergence;
- epsilon-optimality under a valid certification gap;
- convergence under fairness and information-acquisition conditions.

#### Empirically Established for the Current Implementation

- deterministic reproducibility;
- benchmark behavior;
- observed convergence patterns;
- numerical robustness for tested cases;
- one-dimensional empirical/certified execution.

#### Not Established

- universal high-dimensional efficiency;
- universal complexity bounds;
- universal superiority over competing methods;
- empirical high-dimensional performance;
- convergence for arbitrary discontinuous objectives;
- infinite-dimensional convergence.

### 53.21 Scientific Integrity Principle

The final interpretation of ARRGO should follow:

$$
\boxed{
\text{Claim Strength}
\leq
\text{Evidence Strength}.
}
$$

A theoretical result should be stated only with its required assumptions.

An empirical result should be stated only for the tested configuration and domain.

A computational claim should be supported by complexity or scalability analysis.

This principle prevents the multidimensional generalization from being overstated.

### 53.22 Final Scope Statement

The scientifically precise scope of the project is:

$$
\boxed{
\begin{aligned}
&\text{ARRGO is formulated as a deterministic,}
\\
&\text{region-based, information-driven framework}
\\
&\text{for bounded finite-dimensional global optimization.}
\\
&\\
&\text{Under explicit refinement, fairness,}
\\
&\text{information, and regularity assumptions,}
\\
&\text{its objective-value convergence can be established}
\\
&\text{conditionally for finite dimensions.}
\\
&\\
&\text{Under an additional valid Lipschitz assumption,}
\\
&\text{rigorous certification bounds can be constructed.}
\\
&\\
&\text{The current implementation and experiments}
\\
&\text{remain one-dimensional.}
\end{aligned}
}
$$

This statement defines the boundary between the theoretical contribution, the implemented system, and the empirical evidence of the ARRGO project.

## 54. Formal Research Contribution

The preceding theoretical development defines the mathematical scope and guarantees of the finite-dimensional ARRGO framework.

This section identifies the contribution of the present work in a form suitable for scientific reporting.

### 54.1 Contribution at the Framework Level

The primary contribution is the formulation of ARRGO as a unified deterministic, region-based, information-driven global optimization framework.

The framework is organized around:

$$
\boxed{
\text{Region Representation}
+
\text{Information Analysis}
+
\text{Adaptive Refinement}
+
\text{Global Selection}.
}
$$

Rather than treating sampling and spatial subdivision as interchangeable operations, ARRGO distinguishes:

$$
\boxed{
\text{Sampling}
=
\text{Information Refinement}
}
$$

from:

$$
\boxed{
\text{Splitting}
=
\text{Structural Refinement}.
}
$$

This distinction is maintained throughout the theoretical formulation.

### 54.2 Contribution at the Geometric Level

The original one-dimensional interval-based representation is generalized to:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j]
\subset\mathbb R^d.
$$

Regions become hyper-rectangles:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j].
$$

This provides a dimension-parameterized representation:

$$
\boxed{
\mathrm{ARRGO}(d),
\qquad
1\leq d<\infty.
}
$$

The geometry is therefore expressed independently of a particular fixed dimension.

### 54.3 Contribution at the Refinement Level

Multidimensional structural refinement is represented by:

$$
(j,s_j),
$$

where:

- \(j\) identifies the selected spatial direction;
- \(s_j\) identifies the split location.

This separates two decisions:

$$
\boxed{
\text{Direction Selection}
}
$$

and:

$$
\boxed{
\text{Split-Location Selection}.
}
$$

The distinction allows the refinement mechanism to reason about directional resolution separately from the position of the structural split.

### 54.4 Contribution at the Information Level

ARRGO represents unresolved information through multiple dimensions rather than collapsing them into an arbitrary weighted scalar score.

A regional information state may be represented as:

$$
Q_R
=
(
Q_{\mathrm{coverage}},
Q_{\mathrm{behavior}},
Q_{\mathrm{uncertainty}},
Q_{\mathrm{potential}}
).
$$

The framework can therefore distinguish different reasons why a region may remain unresolved.

Candidate actions can be compared using non-dominance and Pareto-style reasoning rather than requiring arbitrary fixed weights.

This preserves the semantic distinction between different information objectives.

### 54.5 Contribution at the Global-Selection Level

ARRGO separates:

$$
\boxed{
\text{Global Region Selection}
}
$$

from:

$$
\boxed{
\text{Local Action Selection}.
}
$$

The first determines:

> Which region should receive attention?

The second determines:

> What should be done inside that region?

Possible actions are:

$$
A_R
\in
\{
\mathrm{Sample},
\mathrm{Split},
\mathrm{Stable}
\}.
$$

This separation allows global relevance and local unresolved information to be modeled independently.

### 54.6 Contribution at the Hierarchy Level

ARRGO preserves structural history.

After a split:

$$
R\rightarrow R_1\cup R_2,
$$

the parent remains part of the structural hierarchy:

$$
R\in\mathcal H_t.
$$

Therefore:

$$
\boxed{
\text{No Structural Pruning}.
}
$$

This preserves the complete refinement history.

At the same time, certification uses a current frontier:

$$
\mathcal F_t
\subseteq
\mathcal H_t,
$$

with:

$$
\Omega
\subseteq
\bigcup_{R\in\mathcal F_t}R.
$$

This separates:

$$
\text{Historical Structure}
$$

from:

$$
\text{Current Certification Cover}.
$$

### 54.7 Contribution at the Convergence Level

The convergence argument makes explicit that several properties must work together.

The core chain is:

$$
\boxed{
\begin{aligned}
&\text{Valid Refinement}
\\
&+
\text{Spatial Contraction}

&+
\text{Global Fairness}

This identifies the exact conditions connecting local region refinement to global objective-value convergence.

The result is conditional and does not claim automatic convergence for every possible refinement policy.

### 54.8 Contribution at the Certification Level

Under the additional assumption:

|f(x)-f(y)|

ARRGO constructs regional envelopes:

L_R(x)

### 54.9 Contribution at the Theoretical-Generalization Level

The one-dimensional framework is shown to be a special case of the finite-dimensional formulation:

$$
\boxed{
\mathrm{ARRGO}(1)

# Final Summary

Notebook 07 established the mathematical generalization of ARRGO from one-dimensional bounded optimization to arbitrary finite-dimensional bounded optimization.

The development proceeded from the generalized optimization domain and objective evaluation model to multidimensional regions, hierarchy, sampling, coverage, structural splitting, information-driven refinement, global selection, convergence, certification, and theoretical claim boundaries.

## 1. Generalized Problem Formulation

The generalized optimization problem is defined over:

$$
\Omega
=
\prod_{j=1}^{d}[a_j,b_j]
\subset\mathbb R^d,
\qquad
1\leq d<\infty.
$$

The objective is:

$$
\max_{x\in\Omega}f(x).
$$

For continuous \(f\), compactness of the bounded closed domain guarantees the existence of a global maximizer.

## 2. Generalized Region Model

One-dimensional intervals are generalized to hyper-rectangular regions:

$$
R
=
\prod_{j=1}^{d}[l_j,r_j].
$$

Each region has:

- coordinate-wise widths;
- maximum width;
- center;
- Euclidean diameter;
- regional evaluation information;
- hierarchy metadata;
- refinement state.

This provides a dimension-parameterized spatial representation.

## 3. Generalized Information Model

Objective evaluations remain dimension-independent in structure:

$$
e_i
=
(x_i,f(x_i),\operatorname{Id}_i),
\qquad
x_i\in\mathbb R^d.
$$

The global evaluation history remains:

$$
D_t
=
\{e_1,\ldots,e_t\}.
$$

The incumbent remains:

$$
f_{\mathrm{best}}^{(t)}
=
\max_{x_i\in D_t}f(x_i).
$$

Therefore the fundamental optimization information model is preserved.

## 4. Sampling and Structural Refinement

The framework explicitly distinguishes two types of refinement:

$$
\boxed{
\text{Sampling}
=
\text{Information Refinement}
}
$$

and:

$$
\boxed{
\text{Splitting}
=
\text{Structural Refinement}.
}
$$

Sampling acquires new objective information.

Splitting changes the spatial representation of the search domain.

This distinction remains central in the multidimensional formulation.

## 5. Multidimensional Structural Splitting

A multidimensional split is represented by:

$$
(j,s_j),
$$

where \(j\) is the selected coordinate direction and \(s_j\) is the split location.

The resulting children satisfy:

$$
R=R_1\cup R_2,
$$

with:

$$
R_1,R_2\subseteq R.
$$

Thus structural refinement preserves domain coverage.

Under the contraction condition:

$$
\max
\left(
\frac{s_j-l_j}{r_j-l_j},
\frac{r_j-s_j}{r_j-l_j}
\right)
\leq
\rho,
$$

with:

$$
\frac12\leq\rho<1,
$$

the selected directional width contracts.

## 6. Spatial Convergence Mechanism

A single directional split does not guarantee complete multidimensional contraction.

Instead, directional fairness requires that unresolved directions on persistently refined relevant paths are not permanently ignored.

Therefore:

$$
\boxed{
\text{Directional Fairness}
+
\text{Contraction}
\Rightarrow
\operatorname{diam}(R_t)\rightarrow0.
}
$$

This provides the spatial component of the convergence argument.

## 7. Global Region Fairness

Spatial contraction must occur on globally relevant paths.

Global region fairness therefore requires that persistent relevant and eligible regions cannot be permanently starved.

Together:

$$
\boxed{
\text{Global Fairness}
+
\text{Directional Fairness}
}
$$

provide the structural access required by the global convergence argument.

## 8. Information Acquisition

Spatial refinement alone is insufficient.

The objective must also be evaluated increasingly close to a global maximizer \(x^\ast\):

$$
\operatorname{dist}(x^\ast,D_t)
\rightarrow0.
$$

By continuity:

$$
f(x_i)
\rightarrow
f(x^\ast)
=
f^\ast.
$$

Since the incumbent is the maximum observed value:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast.
}
$$

Thus the global convergence mechanism requires both spatial refinement and objective information.

## 9. Global Convergence Result

The complete conditional convergence chain is:

$$
\boxed{
\begin{aligned}
&\text{Compact Domain}
\\
&+
\text{Continuity}
\\
&+
\text{Valid Region Representation}
\\
&+
\text{Valid Splitting}
\\
&+
\text{Spatial Contraction}
\\
&+
\text{Global Fairness}
\\
&+
\text{Directional Fairness}
\\
&+
\text{Information Acquisition}
\\
&+
\text{Exact Objective Evaluation}
\\
&\Longrightarrow
\\
&\boxed{
f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast.
}
\end{aligned}
}
$$

This is a conditional mathematical guarantee.

It is not a claim that every implementation automatically satisfies all of these conditions.

## 10. Certified Optimization

For Certified Mode, an additional valid global Lipschitz condition is required:

$$
|f(x)-f(y)|
\leq
L\|x-y\|_2.
$$

This produces valid regional envelopes:

$$
L_R(x)
\leq
f(x)
\leq
U_R(x).
$$

For a complete certification frontier:

$$
\Omega
\subseteq
\bigcup_{R\in\mathcal F_t}R,
$$

the global potential satisfies:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^\ast
\leq
P_{\mathrm{global}}^{(t)}.
$$

Therefore:

$$
0
\leq
f^\ast-f_{\mathrm{best}}^{(t)}
\leq
\Delta_{\mathrm{global}}^{(t)}.
$$

If:

$$
\Delta_{\mathrm{global}}^{(t)}
\leq
\epsilon,
$$

then:

$$
\boxed{
f_{\mathrm{best}}^{(t)}
\geq
f^\ast-\epsilon.
}
$$

Thus Certified Mode provides a rigorous epsilon-optimality statement under the stated assumptions.

## 11. Certified Gap Convergence

If the certification frontier becomes spatially resolving and each frontier region has valid regional information, then:

$$
P_{\mathrm{global}}^{(t)}
\rightarrow
f^\ast.
$$

Together with:

$$
f_{\mathrm{best}}^{(t)}
\rightarrow
f^\ast,
$$

this gives:

$$
\boxed{
\Delta_{\mathrm{global}}^{(t)}
\rightarrow0.
}
$$

The certification gap therefore becomes asymptotically tight under the stated conditions.

## 12. No-Pruning and Certification

ARRGO retains historical parent regions:

$$
R_{\mathrm{parent}}\in\mathcal H_t.
$$

The certification frontier is maintained separately:

$$
\mathcal F_t\subseteq\mathcal H_t.
$$

Therefore:

$$
\boxed{
\text{Structural History}
\neq
\text{Certification Frontier}.
}
$$

This allows the framework to preserve complete structural history without incorrectly treating every historical region as part of the current certification cover.

## 13. One-Dimensional Recovery

Setting:

$$
d=1
$$

recovers:

$$
\Omega=[a,b],
$$

$$
R=[l,r],
$$

and:

$$
\operatorname{diam}(R)=r-l.
$$

The Euclidean distance becomes:

$$
\|x-y\|_2=|x-y|.
$$

The multidimensional split becomes the scalar split.

The certified envelopes become:

$$
U_R(x)
=
\min_i
\left[
f(x_i)+L|x-x_i|
\right],
$$

and:

$$
L_R(x)
=
\max_i
\left[
f(x_i)-L|x-x_i|
\right].
$$

Therefore:

$$
\boxed{
\mathrm{ARRGO}(1)
\equiv
\mathrm{ARRGO}_{1D}.
}
$$

The one-dimensional framework is consequently recovered exactly as a special case of the generalized formulation.

The mathematical development establishes a framework for:

$$

\mathrm{ARRGO}(d),
\qquad
1\leq d<\infty.
}

It establishes conditional objective-value convergence under explicit assumptions and certified epsilon-optimality under additional valid Lipschitz and certification-frontier conditions.

It does not establish universal computational efficiency or performance superiority.

The multidimensional formulation developed in this notebook is theoretical.

Therefore:

$$
\boxed{

}
$$
