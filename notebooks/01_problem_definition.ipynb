# ARRGO: Adaptive Region Refinement for Global Optimization

## Problem Definition

### Objective

We consider the global optimization problem

$$
\max_{x \in \Omega} f(x)
$$

where $\Omega$ is a bounded search domain and $f$ is an objective function
that can be evaluated at selected points.

The algorithm does not require the analytical expression of $f$ and does
not assume that derivatives are directly available.

## Information Available to ARRGO

ARRGO does not require the analytical expression of the objective function.

Instead, the algorithm interacts with the objective function through
function evaluations.

For a selected point $x$, the objective function returns its value:

$$
y = f(x)
$$

Therefore, the fundamental observation available to the algorithm is:

$$
(x, f(x))
$$

For a region $R$, the collected observations are represented as:

$$
\mathcal{D}_R =
\left\{
(x_1,f(x_1)),
(x_2,f(x_2)),
\ldots,
(x_n,f(x_n))
\right\}
$$

## What ARRGO Does Not Assume

ARRGO does not require:

- The analytical expression of $f$
- Direct access to derivatives
- A predefined location for sampling
- A fixed partition of the domain
- A predefined local model for every region

If derivative-like information is useful, it may be estimated from
function evaluations. Such estimates are treated as numerical
information rather than exact derivatives.

## Search Domain

The optimization is performed over a bounded search domain:

$$
\Omega = [a,b]
$$

where:

- $a$ is the lower bound of the search domain.
- $b$ is the upper bound of the search domain.
- $a < b$.

The domain $\Omega$ represents the complete set of points in which
the global optimum is searched.

## Region Representation

ARRGO recursively analyzes the search domain through regions.

A region is a subdomain of the original search domain:

$$
R = [l,r]
$$

where:

$$
a \leq l < r \leq b
$$

Each region contains the function evaluations collected within its
boundaries.

The observations associated with a region are represented by:

$$
\mathcal{D}_R =
\left\{
(x_i,f(x_i)) \mid x_i \in R
\right\}
$$

Thus, a region is not defined only by its geometric boundaries.
It also contains the information that has been collected about the
objective function inside that region.

## Function Evaluation and Samples

A function evaluation is performed by selecting a point $x$ inside the
search domain and obtaining the corresponding objective value:

$$
y = f(x)
$$

The resulting observation is called a sample:

$$
q = (x,y) = (x,f(x))
$$

A sample therefore contains two pieces of information:

- the location at which the objective function was evaluated;
- the objective value observed at that location.

For a region $R$, its available samples are represented by:

$$
\mathcal{D}_R =
\{q_1,q_2,\ldots,q_n\}
$$

where:

$$
q_i=(x_i,f(x_i)), \qquad x_i\in R
$$

## Sample Distribution

The samples inside a region provide information about the observed
behavior of the objective function.

Given the ordered sample locations

$$
x_1 < x_2 < \cdots < x_n,
$$

the distance between two consecutive samples is:

$$
d_i = x_{i+1}-x_i.
$$

These distances describe how densely the region has been explored.

A large sampling gap indicates that a relatively large portion of the
region has not yet been directly observed, while smaller gaps indicate
denser sampling.

## Observed Local Behavior

ARRGO does not assume that derivatives of the objective function are
available.

Instead, local behavior can initially be inferred from consecutive
function evaluations.

For two consecutive samples

$$
(x_i,f(x_i)), \qquad (x_{i+1},f(x_{i+1})),
$$

the observed secant slope is defined as:

$$
s_i =
\frac{f(x_{i+1})-f(x_i)}
{x_{i+1}-x_i}
$$

Since $x_{i+1}>x_i$, the sign of $s_i$ describes the observed direction
of change between the two samples:

$$
s_i > 0
\quad\Rightarrow\quad
\text{the observed function values increase}
$$

$$
s_i < 0
\quad\Rightarrow\quad
\text{the observed function values decrease}
$$

$$
s_i = 0
\quad\Rightarrow\quad
\text{the observed function values are equal}
$$

The secant slope is an observation derived from sampled function
values and is not treated as an exact derivative.

## Candidate Extrema from Observed Slopes

Consider three consecutive samples:

$$
(x_{i-1},f(x_{i-1})),\quad
(x_i,f(x_i)),\quad
(x_{i+1},f(x_{i+1}))
$$

with the corresponding observed secant slopes:

$$
s_{i-1}
=
\frac{f(x_i)-f(x_{i-1})}
{x_i-x_{i-1}}
$$

and

$$
s_i
=
\frac{f(x_{i+1})-f(x_i)}
{x_{i+1}-x_i}.
$$

A change in the sign of the observed slopes can indicate a candidate
extremum.

For a candidate maximum, the observed pattern is:

$$
s_{i-1}>0
\quad\text{and}\quad
s_i<0.
$$

This indicates that the observed function values increase before
$x_i$ and decrease after $x_i$.

Similarly, a candidate minimum is indicated by:

$$
s_{i-1}<0
\quad\text{and}\quad
s_i>0.
$$

These patterns are treated as behavioral evidence rather than proof
of an exact local extremum.

## Variation of Observed Behavior

The sequence of observed secant slopes provides information about how
the local behavior of the objective function changes across a region.

For consecutive secant slopes $s_i$ and $s_{i+1}$, define the observed
slope variation as:

$$
\Delta s_i = s_{i+1}-s_i
$$

The magnitude

$$
|\Delta s_i|
$$

measures the amount of change between two consecutive observed slopes.

Small changes in consecutive slopes indicate relatively consistent
observed behavior, whereas large changes may indicate that the local
behavior of the objective function is changing more rapidly.

For a region containing multiple samples, the sequence

$$
\{s_1,s_2,\ldots,s_{n-1}\}
$$

and its variations

$$
\{\Delta s_1,\Delta s_2,\ldots,\Delta s_{n-2}\}
$$

form part of the behavioral information used by ARRGO.

## Behavioral Complexity

The observed behavior of a region is not determined only by whether
the objective function is increasing or decreasing.

Changes in the sequence of observed secant slopes provide additional
information about the local structure of the objective function.

A region may exhibit:

- approximately consistent behavior;
- gradual changes in behavior;
- frequent changes in the direction of variation;
- strongly varying observed slopes.

Therefore, behavioral complexity is considered a property of the
observed structure within a region rather than a fixed global threshold.

ARRGO aims to characterize complexity from the available observations
and use this information to determine whether additional information
should be obtained through sampling or structural refinement.

No universal fixed threshold is assumed at this stage.

## Region Behavior Profile

For each region $R$, ARRGO maintains a behavior profile derived from
the function evaluations available inside that region.

The behavior profile summarizes the observed characteristics of the
region and is not intended to represent the exact behavior of the
objective function.

Conceptually, the behavior profile contains:

$$
B_R =
(V_R,\;M_R,\;E_R,\;C_R,\;G_R)
$$

where:

- $V_R$ represents the observed variation of function values.
- $M_R$ represents the observed directional behavior, such as increasing,
  decreasing, or changing direction.
- $E_R$ represents candidate extrema detected from changes in observed
  slopes.
- $C_R$ represents the observed behavioral complexity.
- $G_R$ represents the spatial distribution of samples and unresolved
  sampling gaps.

The behavior profile is updated whenever new function evaluations are
added to the region.

## Observed Function Variation

Let the available samples in a region be ordered according to their
locations:

$$
x_1 < x_2 < \cdots < x_n.
$$

The corresponding observed function values are:

$$
y_i=f(x_i).
$$

The change in the observed objective value between two consecutive
samples is defined as:

$$
\Delta y_i = y_{i+1}-y_i.
$$

The magnitude of this change is:

$$
|\Delta y_i| = |y_{i+1}-y_i|.
$$

The sequence

$$
\{|\Delta y_1|,|\Delta y_2|,\ldots,|\Delta y_{n-1}|\}
$$

describes how much the observed objective values change across the
region.

This quantity is a descriptive property of the available samples.
It does not by itself provide a bound on the behavior of the objective
function between sampled points.

## Observed Directional Behavior

The sign of each observed secant slope provides information about the
direction of change between consecutive samples.

For each interval $[x_i,x_{i+1}]$, define its observed direction as:

$$
d_i = \operatorname{sign}(s_i)
$$

where:

$$
d_i =
\begin{cases}
+1, & s_i > 0,\\
0, & s_i = 0,\\
-1, & s_i < 0.
\end{cases}
$$

Therefore, the directional behavior of a region can be represented by
the sequence:

$$
D_R = \{d_1,d_2,\ldots,d_{n-1}\}.
$$

For example,

$$
D_R=(+1,+1,+1)
$$

indicates consistently increasing observed behavior, while

$$
D_R=(-1,-1,-1)
$$

indicates consistently decreasing observed behavior.

A sequence such as

$$
D_R=(+1,+1,-1,-1)
$$

indicates an observed change from increasing to decreasing behavior.

The directional sequence is derived only from available function
evaluations and does not constitute a proof of the exact behavior
between sampled points.

## Candidate Extrema

Candidate extrema are identified from changes in the observed
directional behavior.

For an interior sample $x_i$, consider the directions of the two
neighboring intervals:

$$
d_{i-1}=\operatorname{sign}(s_{i-1})
$$

and

$$
d_i=\operatorname{sign}(s_i).
$$

A candidate maximum is identified when the observed direction changes
from increasing to decreasing:

$$
d_{i-1}=+1
\quad\text{and}\quad
d_i=-1.
$$

A candidate minimum is identified when the observed direction changes
from decreasing to increasing:

$$
d_{i-1}=-1
\quad\text{and}\quad
d_i=+1.
$$

The corresponding sample location $x_i$ is recorded as a candidate
extremum.

These candidates represent hypotheses generated from the currently
available observations. They are not treated as exact local extrema,
because unsampled behavior between consecutive observations may
contain additional extrema.

## Sampling Gaps

Let the sample locations inside a region be ordered as:

$$
x_1 < x_2 < \cdots < x_n.
$$

The spatial gap between two consecutive samples is defined as:

$$
g_i = x_{i+1}-x_i.
$$

The set of observed sampling gaps is therefore:

$$
G_R =
\{g_1,g_2,\ldots,g_{n-1}\}.
$$

A larger value of $g_i$ indicates a larger unsampled interval between
two consecutive observations.

The maximum observed sampling gap is defined as:

$$
g_{\max}(R)=\max_i g_i.
$$

Sampling gaps provide spatial information about where the region has
been explored densely or sparsely.

A sampling gap does not imply that the objective function is
uncertain by a specific amount. Its interpretation depends on the
available information about the behavior of the objective function
and the assumptions used by the optimization framework.

## Information Sufficiency

The amount of information required to analyze a region depends on the
observed behavior of the objective function rather than on a fixed
number of samples.

A region is considered sufficiently informed for a particular decision
when additional observations are unlikely to change that decision
meaningfully.

Let the current information state of a region be represented by:

$$
S_R = (\mathcal{D}_R,B_R)
$$

where $\mathcal{D}_R$ is the set of available observations and $B_R$
is the corresponding behavior profile.

Suppose a new observation is obtained at a point $x\in R$. The updated
state is:

$$
S_R^{+x}.
$$

Information sufficiency can then be viewed as the stability of the
decision under additional information:

$$
\operatorname{Decision}(S_R)
\approx
\operatorname{Decision}(S_R^{+x}).
$$

Thus, information sufficiency is decision-dependent and is not
defined by a universal fixed sample count.

## Refinement of a Region

ARRGO refines a region when the currently available information is not
sufficient for the optimization process.

Two fundamental refinement operations are considered:

1. Information refinement through additional function evaluations.
2. Structural refinement through partitioning the region.

These operations address two different forms of insufficient information.

Sampling improves the knowledge of the objective function within the
current geometric region, while splitting improves the spatial
resolution of the search domain.

## Sampling as Information Refinement

Sampling refines the information available about a region without
changing its geometric boundaries.

For a region

$$
R=[l,r],
$$

suppose its current observations are:

$$
\mathcal{D}_R.
$$

Selecting a new point $x_{\mathrm{new}}\in R$ and evaluating the
objective function produces:

$$
y_{\mathrm{new}}=f(x_{\mathrm{new}}).
$$

The updated observation set becomes:

$$
\mathcal{D}_R^{+}
=
\mathcal{D}_R
\cup
\{(x_{\mathrm{new}},y_{\mathrm{new}})\}.
$$

Therefore, sampling can be represented as:

$$
R(\mathcal{D}_R)
\longrightarrow
R(\mathcal{D}_R^{+}).
$$

The geometric boundaries of the region remain unchanged.
Only the information available about the objective function is
increased.

## Splitting as Structural Refinement

Splitting refines a region by increasing the spatial resolution of the
search domain.

Consider a region:

$$
R=[l,r].
$$

A valid split point $s$ is selected such that:

$$
l<s<r.
$$

The region is then partitioned into two child regions:

$$
R_L=[l,s]
$$

and

$$
R_R=[s,r].
$$

Therefore:

$$
R
\longrightarrow
\{R_L,R_R\}.
$$

The union of the child regions covers the original region:

$$
R_L\cup R_R=R.
$$

Their intersection contains only the split boundary:

$$
R_L\cap R_R=\{s\}.
$$

Splitting does not discard the original region or its observations.
Existing function evaluations can be reused by the resulting child
regions whenever their locations belong to those regions.

## Sampling versus Splitting

When a region requires further refinement, ARRGO considers two
fundamentally different actions:

### Information Refinement

A new function evaluation is performed inside the current region:

$$
R(\mathcal{D}_R)
\longrightarrow
R(\mathcal{D}_R^{+}).
$$

The geometric boundaries remain unchanged, while information about the
objective function increases.

### Structural Refinement

The region is partitioned into smaller regions:

$$
R
\longrightarrow
\{R_L,R_R\}.
$$

The spatial resolution of the search domain increases.

Therefore, sampling and splitting are not competing operations with the
same immediate effect. They provide different types of refinement.

ARRGO selects between them according to their expected contribution
to reducing the unresolved information relevant to global optimization.

## Refinement Benefit

Let the unresolved optimization information of a region be represented
conceptually by:

$$
\mathcal{U}(R).
$$

For a possible refinement action $A$, define its refinement benefit as
the reduction in unresolved information:

$$
\operatorname{Benefit}(A\mid R)
=
\mathcal{U}(R)
-
\mathcal{U}(R\mid A).
$$

The two principal actions are therefore compared through:

$$
\operatorname{Benefit}(\text{Sampling}\mid R)
$$

and

$$
\operatorname{Benefit}(\text{Splitting}\mid R).
$$

The preferred action is the one that provides greater justified
refinement benefit under the information currently available.

The exact definition of $\mathcal{U}(R)$ and the corresponding
estimation of refinement benefit are problem-dependent and are
specified by the uncertainty and analysis components of ARRGO.

## Unresolved Optimization Information

Function evaluations provide information only at selected points.

For a region

$$
R=[l,r],
$$

the available observations are:

$$
\mathcal{D}_R =
\{(x_i,f(x_i))\}_{i=1}^{n}.
$$

Between and around these observations, the exact behavior of the
objective function may remain unknown.

This motivates the concept of unresolved optimization information.

Let

$$
\mathcal{U}(R)
$$

denote a measure of the information about the objective function that
remains unresolved within region $R$.

The quantity $\mathcal{U}(R)$ is not assumed to have a universal form.
Its definition depends on the mathematical assumptions available about
the objective function and the uncertainty model used by ARRGO.

In particular, finite function evaluations alone cannot provide a
non-trivial universal bound on unseen function values without additional
assumptions about the objective function.

## Certified Uncertainty under a Lipschitz Assumption

Suppose the objective function satisfies a Lipschitz continuity
condition on a region $R$:

$$
|f(x)-f(z)| \leq L|x-z|,
\qquad \forall x,z\in R,
$$

where $L$ is a valid Lipschitz constant.

For an observed sample

$$
q_i=(x_i,y_i),
\qquad y_i=f(x_i),
$$

the Lipschitz condition implies:

$$
y_i-L|x-x_i|
\leq f(x)
\leq
y_i+L|x-x_i|.
$$

Therefore, every observed sample provides a lower and upper bound on
the unknown function value at any point $x\in R$.

## Lipschitz-Based Envelopes

Given all observations in a region, the lower and upper bounds can be
combined.

The lower envelope is defined as:

$$
L_R(x)
=
\max_i
\left[
y_i-L|x-x_i|
\right].
$$

The upper envelope is defined as:

$$
U_R(x)
=
\min_i
\left[
y_i+L|x-x_i|
\right].
$$

Under the Lipschitz assumption, the true objective function satisfies:

$$
L_R(x)
\leq
f(x)
\leq
U_R(x),
\qquad \forall x\in R.
$$

Thus, the interval

$$
[L_R(x),U_R(x)]
$$

represents a mathematically justified range of possible objective
values at $x$.

## Pointwise Uncertainty

Under the Lipschitz assumption, the uncertainty associated with a point
$x\in R$ is defined as the width of the corresponding valid interval:

$$
u_R(x)
=
U_R(x)-L_R(x).
$$

Therefore,

$$
u_R(x)\geq 0.
$$

A small value of $u_R(x)$ indicates that the available observations
provide a relatively narrow range of possible objective values at $x$.

A large value of $u_R(x)$ indicates that the available observations
allow a wider range of possible objective values at $x$.

The pointwise uncertainty is therefore a measure of unresolved
function-value information at a specific location.

## Region-Level Uncertainty

To characterize the unresolved information across an entire region,
ARRGO considers the maximum pointwise uncertainty:

$$
\mathcal{U}(R)
=
\max_{x\in R}u_R(x).
$$

Equivalently,

$$
\mathcal{U}(R)
=
\max_{x\in R}
\left[
U_R(x)-L_R(x)
\right].
$$

A region with a larger value of $\mathcal{U}(R)$ contains locations at
which the currently available information permits a wider range of
possible objective values.

Therefore, $\mathcal{U}(R)$ provides a region-level measure of unresolved
information under the selected uncertainty model.

## Information-Driven Sampling

Under the Lipschitz-based uncertainty model, the pointwise uncertainty
is given by:

$$
u_R(x)=U_R(x)-L_R(x).
$$

A natural information-driven sampling rule is to select a point where
the current uncertainty is largest:

$$
x_{\mathrm{new}}
\in
\operatorname*{arg\,max}_{x\in R} u_R(x).
$$

The selected point represents a location at which an additional
function evaluation can provide information about one of the most
unresolved parts of the region.

After evaluating the objective function at this point,

$$
y_{\mathrm{new}}=f(x_{\mathrm{new}}),
$$

the new observation is added to the region:

$$
\mathcal{D}_R^{+}
=
\mathcal{D}_R
\cup
\{(x_{\mathrm{new}},y_{\mathrm{new}})\}.
$$

The uncertainty envelopes are then recomputed using the expanded
observation set.

## Optimization Potential

Let the best observed objective value currently available to ARRGO be:

$$
f_{\mathrm{best}}
=
\max_{x_i\in\mathcal{D}} f(x_i).
$$

Under the Lipschitz-based uncertainty model, every point $x\in R$
has an upper bound:

$$
f(x)\leq U_R(x).
$$

Therefore, the largest possible objective value that is consistent
with the current information in region $R$ is bounded by:

$$
P(R)
=
\max_{x\in R} U_R(x).
$$

The quantity $P(R)$ is called the optimization potential of the
region.

It represents the largest objective value that the current information
allows the region to potentially contain.

If

$$
P(R)>f_{\mathrm{best}},
$$

the region is still capable, under the uncertainty model, of containing
a solution better than the best solution observed so far.

If

$$
P(R)\leq f_{\mathrm{best}},
$$

the current information does not allow the region to contain a value
greater than the best observed solution.

## Region Priority

ARRGO must decide which region should receive the next refinement
operation.

Two different types of information are relevant to this decision:

1. **Information uncertainty**

   This describes how much unknown behavior remains inside the region.

2. **Optimization potential**

   This describes how much the region could still improve upon the
   best solution observed so far.

These two concepts represent different properties of a region.

A region can have high uncertainty but low optimization potential.
Similarly, a region can have high optimization potential while its
current uncertainty is relatively small.

Therefore, ARRGO does not identify uncertainty with optimization
potential.

For a region $R$, define:

$$
\mathcal{U}(R)
=
\max_{x\in R}
\left(
U_R(x)-L_R(x)
\right)
$$

as its region-level uncertainty, and:

$$
P(R)
=
\max_{x\in R} U_R(x)
$$

as its optimization potential.

Let the current best observed objective value be:

$$
f_{\mathrm{best}}.
$$

The potential improvement allowed by the current information is:

$$
\Delta_R
=
P(R)-f_{\mathrm{best}}.
$$

Thus:

- $\mathcal{U}(R)$ measures unresolved information.
- $P(R)$ measures the maximum value still allowed by the model.
- $\Delta_R$ measures the region's potential to improve the incumbent.

These quantities should remain conceptually separate.

The priority mechanism of ARRGO will use these properties to determine
where additional information or structural refinement is most valuable.

## Region Priority Criterion

The purpose of region priority is not to estimate which region
contains the global optimum with certainty.

Instead, it determines which region should receive attention next,
given the information currently available.

ARRGO considers two fundamental questions:

1. How much unresolved information remains in the region?
2. Can the region still contain an improvement over the current best
   solution?

For region $R$, these are represented by:

$$
\mathcal{U}(R)
=
\max_{x\in R}
\left(
U_R(x)-L_R(x)
\right)
$$

and

$$
\Delta_R
=
P(R)-f_{\mathrm{best}}.
$$

The first quantity measures uncertainty, while the second measures
optimization potential.

Rather than combining these quantities using arbitrary fixed weights,
ARRGO treats them as separate decision dimensions.

A region is therefore evaluated in the following conceptual space:

$$
R
\longrightarrow
\left(
\mathcal{U}(R),
\Delta_R
\right).
$$

This leads to three important cases.

### Case 1: High uncertainty and high potential

The region is both poorly understood and capable of containing a
better solution.

Such a region is a strong candidate for refinement.

### Case 2: High uncertainty and low potential

The region contains substantial unresolved information, but the current
upper bound does not indicate a strong possibility of improvement.

The region may still be refined because ARRGO does not prune regions.

### Case 3: Low uncertainty and high potential

The region is relatively well understood, but its remaining feasible
upper bound is still competitive with the current best solution.

Further refinement may therefore be useful for resolving the remaining
optimization uncertainty.

Consequently, region priority is treated as a decision problem rather
than as a single arbitrary numerical score.

The exact priority rule will be defined after the refinement actions
and their expected information gain have been formally specified.

## Refinement Decision: Sample or Split?

Once a region has been selected for refinement, ARRGO must determine
which refinement operation is more appropriate.

The two available operations are:

- **Sampling:** acquire new function information inside the current region.
- **Splitting:** introduce new spatial structure by dividing the region
  into smaller subregions.

These operations solve different problems.

Sampling is appropriate when the existing spatial structure is useful,
but the available function information is insufficient.

Splitting is appropriate when the region is too large or its internal
behavior cannot be adequately represented as a single region.

Therefore, the refinement decision should depend on the information
contained in the region rather than on a fixed rule such as:

$$
\text{if } n<n_0 \text{ then SAMPLE else SPLIT}.
$$

Let $A$ denote a possible refinement action.

The value of an action can be viewed through its expected reduction
in unresolved information:

$$
\operatorname{Benefit}(A\mid R)
=
\mathcal{U}(R)
-
\mathcal{U}(R\mid A).
$$

For the two refinement operators:

$$
\operatorname{Benefit}_{\mathrm{sample}}(R)
=
\mathcal{U}(R)
-
\mathcal{U}(R\mid\mathrm{SAMPLE}),
$$

and

$$
\operatorname{Benefit}_{\mathrm{split}}(R)
=
\mathcal{U}(R)
-
\mathcal{U}(R\mid\mathrm{SPLIT}).
$$

ARRGO therefore treats sampling and splitting as competing refinement
actions.

The selected action should provide the most meaningful improvement in
the region's representation, while respecting the global evaluation
budget and the structural requirements of the search.

The exact action-selection criterion will be developed after defining
how ARRGO estimates the effect of a refinement action before actually
performing it.

## Expected Refinement Effect

A refinement action is performed before its exact outcome is known.

For sampling, the value of the objective function at the new point is
unknown before evaluation:

$$
y_{\mathrm{new}}=f(x_{\mathrm{new}}).
$$

Therefore, the exact post-sampling state cannot be computed in advance.

Similarly, the effect of splitting depends on how the available
information and behavioral structure are distributed across the
resulting subregions.

ARRGO therefore distinguishes between:

1. **Observed refinement effect:** the change measured after an action
   has been performed.

2. **Expected refinement effect:** the estimated usefulness of an action
   before it is performed.

For an action $A$ applied to region $R$, define its observed information
gain as:

$$
G_{\mathrm{obs}}(A\mid R)
=
\mathcal{U}(R)
-
\mathcal{U}(R_A),
$$

where $R_A$ denotes the state resulting from performing action $A$.

Before performing the action, ARRGO instead uses an estimate:

Thus:

$$
\widehat{G}_{\mathrm{sample}}(R)
$$

represents the estimated value of acquiring a new function evaluation,

while:

$$
\widehat{G}_{\mathrm{split}}(R)
$$

represents the estimated value of introducing additional spatial
structure.

These estimates are not treated as mathematical guarantees.

They are decision signals used by ARRGO to choose the next refinement
operation.

After the action is performed, the resulting information is measured
again and the region is re-analyzed.

Therefore, ARRGO follows an adaptive feedback cycle:

$$
\boxed{
\text{Analyze}
\rightarrow
\text{Estimate}
\rightarrow
\text{Refine}
\rightarrow
\text{Observe}
\rightarrow
\text{Re-analyze}
}
$$

This feedback mechanism allows refinement decisions to change as new
function evaluations and new spatial structures become available.

## Informative Sampling

When ARRGO decides to perform a sampling operation, it must select a
point inside the current region.

Let

$$
R=[l,r]
$$

and let the currently observed samples be

$$
\mathcal{D}_R
=
\{(x_i,f(x_i))\}_{i=1}^{n}.
$$

The sampling point should not be selected solely according to its
geometric position.

Instead, ARRGO seeks a point whose evaluation is expected to provide
useful information about the unresolved behavior of the region.

Potentially informative locations include:

- locations inside large unsampled gaps,
- locations where the observed behavior changes rapidly,
- locations near candidate extrema,
- locations where the uncertainty model indicates large uncertainty,
- and locations where additional information may affect the optimization
  decision.

Therefore, the sampling problem is defined as:

$$
x_{\mathrm{sample}}
\in
\arg\max_{x\in R}
\widehat{I}(x\mid R),
$$

where

$$
\widehat{I}(x\mid R)
$$

is an estimate of the information value of evaluating $f(x)$.

The information value is not defined only by geometric distance.

It depends on the current state of the region:

$$
\widehat{I}(x\mid R)
=
\operatorname{InformationValue}
\left(
x,\mathcal{D}_R,B_R,\mathcal{U}(R),P(R)
\right).
$$

After selecting the point, ARRGO evaluates the objective:

$$
y_{\mathrm{new}}=f(x_{\mathrm{sample}})
$$

and adds the new observation to the region:

$$
\mathcal{D}_R
\leftarrow
\mathcal{D}_R
\cup
\{(x_{\mathrm{sample}},y_{\mathrm{new}})\}.
$$

The region is then analyzed again because the new observation can
change its behavior profile, uncertainty, candidate extrema, and
optimization potential.

Thus, sampling is an information-acquisition operation rather than
simply a geometric subdivision operation.

## Candidate Sampling Locations

For a region

$$
R=[l,r],
$$

ARRGO constructs a set of candidate sampling locations from the
information currently available in the region.

Let the ordered observations be:

$$
x_1<x_2<\cdots<x_n.
$$

The candidate set is generated from regions where additional function
evaluations may provide useful information.

### Gap-based candidates

For each consecutive pair of observations,

$$
I_i=[x_i,x_{i+1}],
$$

ARRGO identifies the interior of the gap as a potential source of new
information.

A candidate location may therefore be generated inside each
sufficiently informative gap.

### Behavior-based candidates

Observed changes in directional behavior or secant slopes can indicate
locations where the function may contain previously unresolved local
structure.

Such locations can be included in the candidate set.

### Uncertainty-based candidates

When an uncertainty model is available, locations with large estimated
uncertainty can be added as candidates.

### Potential-based candidates

Locations associated with large upper bounds can also be considered,
because evaluating them may help determine whether a potentially
promising objective value actually exists.

The resulting candidate set is:

$$
\mathcal{C}_R
=
\mathcal{C}_{\mathrm{gap}}
\cup
\mathcal{C}_{\mathrm{behavior}}
\cup
\mathcal{C}_{\mathrm{uncertainty}}
\cup
\mathcal{C}_{\mathrm{potential}}.
$$

Duplicate candidates are removed before the selection step.

The candidate set is only a collection of possible evaluation points.
It does not determine which point will actually be sampled.

ARRGO subsequently evaluates the information value of each candidate
and selects the most informative feasible location:

$$
x_{\mathrm{sample}}
\in
\arg\max_{x\in\mathcal{C}_R}
\widehat{I}(x\mid R).
$$

Therefore, ARRGO separates candidate generation from candidate
selection:

$$
\boxed{
\text{Candidate Generation}
\neq
\text{Candidate Selection}
}
$$

## Information Value of a Sampling Point

Let $x$ be a candidate sampling location inside region $R$.

Before evaluating the objective at $x$, the value $f(x)$ is unknown.
Therefore, ARRGO cannot directly compute the actual information gain
that would result from evaluating $f(x)$.

Instead, ARRGO estimates the information value of a candidate from the
current state of the region.

We define:

$$
\widehat{I}(x\mid R)
$$

as the estimated information value of evaluating the objective at $x$.

This quantity is based only on information already available to ARRGO.

An informative point should have the potential to resolve an important
part of the current uncertainty or ambiguity of the region.

In particular, the information value may depend on:

- the spatial distance from existing observations,
- the uncertainty associated with the location,
- the local variation of observed behavior,
- the proximity to candidate extrema,
- and the potential of the surrounding region to improve the current
  best solution.

The exact form of $\widehat{I}(x\mid R)$ is not assumed in advance.

Instead, ARRGO defines the information value through measurable
properties of the current region and evaluates candidate points
according to these properties.

The sampling decision is therefore:

$$
x_{\mathrm{sample}}
\in
\arg\max_{x\in\mathcal{C}_R}
\widehat{I}(x\mid R).
$$

After the selected point is evaluated, its actual function value becomes
available:

$$
y_{\mathrm{new}}=f(x_{\mathrm{sample}}).
$$

The newly acquired information is then used to recompute the region
analysis.

Thus, the estimated information value is a decision mechanism, not a
claim about the unknown value of the objective function.

## Spatial Information and Coverage

A sampling point can provide information simply because it is located in
a part of the region that is poorly covered by existing observations.

Let the ordered observations in region $R$ be:

$$
x_1<x_2<\cdots<x_n.
$$

For each consecutive pair of observations, define the sampling gap:

$$
g_i=x_{i+1}-x_i.
$$

A large gap indicates that ARRGO has little spatial information about
the behavior of the objective function inside that interval.

For a candidate point $x\in[x_i,x_{i+1}]$, define its distance from the
nearest observed point as:

$$
d_{\mathrm{near}}(x)
=
\min
\left(
|x-x_i|,
|x-x_{i+1}|
\right).
$$

A larger value of $d_{\mathrm{near}}(x)$ means that the candidate is
located farther from the currently available observations.

Therefore, spatial information value can be represented by:

$$
I_{\mathrm{spatial}}(x\mid R)
=
d_{\mathrm{near}}(x).
$$

This quantity is purely geometric and does not require knowledge of the
objective function.

It measures how much the candidate can improve the spatial coverage of
the region.

However, spatial coverage alone is not sufficient for global
optimization.

A point may be far from existing observations while being located in a
region that has little optimization potential.

Therefore, spatial information is treated as one component of the
overall information value rather than as the complete sampling
criterion.

## Uncertainty Information

Spatial coverage measures how far a candidate point is from existing
observations.

However, spatial distance does not directly describe how uncertain the
objective value is at that point.

Under the Lipschitz assumption, ARRGO has lower and upper bounds for
the objective function:

$$
L_R(x)\leq f(x)\leq U_R(x).
$$

The width of this interval is:

$$
u_R(x)
=
U_R(x)-L_R(x).
$$

A large value of $u_R(x)$ indicates that the currently available
information allows a wide range of possible objective values at $x$.

Therefore, define the uncertainty information of a candidate point as:

$$
I_{\mathrm{uncertainty}}(x\mid R)
=
u_R(x)
=
U_R(x)-L_R(x).
$$

A candidate with a larger uncertainty value has greater potential to
reduce unresolved information about the objective function.

This criterion is different from spatial coverage.

A point can be spatially close to an existing observation while still
having substantial uncertainty under the current model.

Conversely, a point can be spatially distant while having relatively
small uncertainty.

Therefore:

$$
I_{\mathrm{spatial}}(x\mid R)
\neq
I_{\mathrm{uncertainty}}(x\mid R).
$$

The two quantities describe different aspects of the value of a
sampling location.

ARRGO keeps these aspects separate until the final candidate-selection
mechanism is defined.

## Optimization Relevance of a Sampling Point

Not every uncertain point is equally relevant to the optimization
problem.

Let the current best observed objective value be:

$$
f_{\mathrm{best}}.
$$

Under the uncertainty model, the objective value at a candidate point
$x$ is bounded by:

$$
f(x)\leq U_R(x).
$$

Therefore, the quantity

$$
r_R(x)
=
U_R(x)-f_{\mathrm{best}}
$$

measures how far the current upper bound at $x$ is above the best
observed objective value.

If

$$
r_R(x)>0,
$$

the current information still allows $x$ to contain an objective value
better than the incumbent.

If

$$
r_R(x)\leq0,
$$

the current upper bound at $x$ does not exceed the best observed value.

To focus on the possibility of improvement, define the non-negative
optimization relevance:

$$
I_{\mathrm{optimization}}(x\mid R)
=
\max
\left(
0,
U_R(x)-f_{\mathrm{best}}
\right).
$$

A larger value indicates that evaluating the candidate may be relevant
to determining whether the region contains a solution better than the
current best observation.

This quantity is different from uncertainty.

For example, a point may have:

$$
u_R(x)\gg0
$$

while:

$$
I_{\mathrm{optimization}}(x\mid R)=0.
$$

In this case, the objective value is uncertain, but the current upper
bound does not indicate a possible improvement over the incumbent.

Therefore, ARRGO distinguishes between:

$$
\boxed{\text{Uncertainty}}
$$

and

$$
\boxed{\text{Optimization Relevance}}.
$$

This distinction allows the sampling mechanism to consider both
information acquisition and the actual objective of global
optimization.

## Behavioral Information

The observations inside a region contain information about how the
objective function behaves between sampled points.

Let the ordered observations be:

$$
x_1<x_2<\cdots<x_n
$$

with corresponding function values:

$$
y_i=f(x_i).
$$

The observed secant slopes are:

$$
s_i
=
\frac{y_{i+1}-y_i}{x_{i+1}-x_i}.
$$

Changes in consecutive slopes are given by:

$$
\Delta s_i
=
s_{i+1}-s_i.
$$

A large change in observed slope indicates that the local behavior
between neighboring observations may be changing.

For a candidate point $x$, ARRGO can therefore consider the behavioral
structure of the interval containing $x$.

The purpose is not to assume that the exact derivative or curvature of
the objective function is known.

Instead, ARRGO uses only observable changes in the sampled data.

Behavioral information may indicate:

- a change in the observed direction of the function,
- a rapid change in consecutive secant slopes,
- proximity to a candidate extremum,
- or a transition between different observed local behaviors.

Therefore, define the behavioral information conceptually as:

$$
I_{\mathrm{behavior}}(x\mid R)
=
\operatorname{BehavioralEvidence}(x,\mathcal{D}_R).
$$

This quantity represents the amount of evidence in the current samples
that the candidate lies near an unresolved behavioral transition.

Behavioral information is observational rather than exact.

In particular, a change in secant slopes does not by itself prove the
existence or location of a true local extremum.

Its purpose is to guide adaptive sampling toward locations where the
observed behavior suggests that additional information may be useful.

Thus, ARRGO distinguishes between:

$$
\boxed{\text{Observed Behavior}}
$$

and

$$
\boxed{\text{Analytical Derivative Information}}.
$$

The former is available from function evaluations, while the latter is
not assumed by the algorithm.

## Multi-Criteria Candidate Evaluation

A candidate sampling point may be informative for different reasons.

For a candidate $x\in\mathcal{C}_R$, ARRGO can observe several
independent properties:

$$
I_{\mathrm{spatial}}(x\mid R),
$$

$$
I_{\mathrm{uncertainty}}(x\mid R),
$$

$$
I_{\mathrm{optimization}}(x\mid R),
$$

and

$$
I_{\mathrm{behavior}}(x\mid R).
$$

These quantities describe different dimensions of the candidate's
value.

They therefore form an information profile:

$$
\mathbf{I}(x\mid R)
=
\left[
I_{\mathrm{spatial}},
I_{\mathrm{uncertainty}},
I_{\mathrm{optimization}},
I_{\mathrm{behavior}}
\right].
$$

ARRGO does not assume that these dimensions should be combined using
arbitrary fixed weights.

Instead, candidate locations can first be compared according to their
information profiles.

A candidate is potentially preferable when it provides stronger
evidence across one or more important dimensions without being
dominated by another candidate.

For two candidates $x$ and $z$, candidate $x$ dominates candidate $z$
if:

$$
I_k(x\mid R)\geq I_k(z\mid R)
$$

for every considered information dimension $k$, and:

$$
I_j(x\mid R)>I_j(z\mid R)
$$

for at least one dimension $j$.

The set of candidates that are not dominated by another candidate forms
the current non-dominated candidate set:

$$
\mathcal{C}_R^{\mathrm{ND}}
\subseteq
\mathcal{C}_R.
$$

This allows ARRGO to preserve multiple meaningful trade-offs instead
of collapsing all information dimensions into a single arbitrary
score.

A final selection mechanism can then choose among the non-dominated
candidates using the current optimization context and refinement
objective.

Thus, candidate evaluation is treated as a multi-criteria decision
problem:

$$
\boxed{
\text{Candidate}
\rightarrow
\text{Information Profile}
\rightarrow
\text{Non-Dominated Set}
\rightarrow
\text{Final Selection}
}
$$

## Final Candidate Selection

After constructing the non-dominated candidate set

$$
\mathcal{C}_R^{\mathrm{ND}},
$$

ARRGO must select one candidate for actual function evaluation.

The final selection should consider the purpose of the current
refinement.

If the primary purpose is to reduce unresolved uncertainty, candidates
with greater uncertainty information should be preferred.

If the primary purpose is to investigate a potentially improving
solution, optimization relevance becomes more important.

If the region contains strong evidence of unresolved behavioral changes,
behavioral information can guide the selection.

Therefore, the final selection is conditioned on the current refinement
objective:

$$
x_{\mathrm{sample}}
=
\operatorname{Select}
\left(
\mathcal{C}_R^{\mathrm{ND}},
\text{RefinementObjective}(R)
\right).
$$

The refinement objective is determined from the current state of the
region.

For example, ARRGO may identify the current refinement objective as:

$$
\text{Information Resolution},
$$

when the region is poorly understood, or:

$$
\text{Optimization Investigation},
$$

when the region has strong potential to improve the incumbent.

The selected candidate is then evaluated:

$$
y_{\mathrm{new}}
=
f(x_{\mathrm{sample}}).
$$

The new observation is added to the region:

$$
\mathcal{D}_R
\leftarrow
\mathcal{D}_R
\cup
\{(x_{\mathrm{sample}},y_{\mathrm{new}})\}.
$$

The region is subsequently re-analyzed.

This creates a closed adaptive loop:

$$
\boxed{
\text{Analyze}
\rightarrow
\text{Generate Candidates}
\rightarrow
\text{Evaluate Candidates}
\rightarrow
\text{Select}
\rightarrow
\text{Sample}
\rightarrow
\text{Re-analyze}
}
$$

The selection mechanism therefore adapts to the current state of the
region instead of following a fixed sampling schedule.

## Splitting as Structural Refinement

Sampling increases the amount of information available inside a region,
while splitting changes the spatial structure of the search.

Let the current region be:

$$
R=[l,r].
$$

A split point $s$ is selected such that:

$$
l<s<r.
$$

The region is then divided into two child regions:

$$
R_L=[l,s]
$$

and

$$
R_R=[s,r].
$$

The original region is retained as part of the region hierarchy and is
not discarded.

The child regions inherit the observations that belong to their
respective domains.

Thus:

$$
\mathcal{D}_{R_L}
=
\{(x_i,f(x_i))\in\mathcal{D}_R\mid x_i\in R_L\},
$$

and:

$$
\mathcal{D}_{R_R}
=
\{(x_i,f(x_i))\in\mathcal{D}_R\mid x_i\in R_R\}.
$$

The split creates two smaller spatial units whose behavior can be
analyzed independently.

The purpose of splitting is therefore not to discard information.

Instead, it increases the structural resolution of the search:

$$
\boxed{
\text{One Region}
\rightarrow
\text{Multiple Spatial Regions}
}
$$

This allows ARRGO to distinguish between different behaviors that may
coexist inside a larger region.

A valid split must also contract the region.

Let:

$$
\rho<1
$$

be a structural contraction factor.

The split must satisfy:

$$
\max(s-l,r-s)
\leq
\rho(r-l).
$$

This prevents the algorithm from repeatedly producing nearly unchanged
regions and ensures that continued structural refinement can reduce the
size of the search regions.

The choice of the split point itself is adaptive and will be determined
from the information available inside the region.

## Adaptive Split Point Selection

A split should not necessarily occur at the midpoint of a region.

Let:

$$
R=[l,r].
$$

The purpose of selecting a split point is to create subregions that
provide a more useful structural representation of the current search
space.

A split point may be considered informative when it separates regions
with different observed characteristics.

For example, a useful split may occur near:

- a transition in observed directional behavior,
- a strong change in secant slopes,
- a candidate extremum,
- a large sampling gap,
- or a boundary between areas with different optimization potential.

Let the set of feasible split points be:

$$
\mathcal{S}_R
=
\{s\mid l<s<r\}.
$$

ARRGO seeks an adaptive split point:

$$
s^*
\in
\arg\max_{s\in\mathcal{S}_R}
\widehat{V}_{\mathrm{split}}(s\mid R),
$$

where

$$
\widehat{V}_{\mathrm{split}}(s\mid R)
$$

represents the estimated structural value of splitting the region at
$s$.

The structural value should reflect how effectively the split can
separate different observed behaviors or unresolved structures.

However, the split must also satisfy the contraction requirement:

$$
\max(s-l,r-s)
\leq
\rho(r-l),
\qquad
\rho<1.
$$

Therefore, the split-point problem can be expressed as:

$$
s^*
\in
\arg\max_{s\in\mathcal{S}_R^{\mathrm{feasible}}}
\widehat{V}_{\mathrm{split}}(s\mid R),
$$

where

$$
\mathcal{S}_R^{\mathrm{feasible}}
=
\left\{
s\in(l,r):
\max(s-l,r-s)\leq\rho(r-l)
\right\}.
$$

The contraction condition guarantees structural refinement, while the
adaptive criterion determines where that refinement should occur.

Thus, ARRGO separates two responsibilities:

$$
\boxed{
\text{Contraction}
=
\text{Guarantee}
}
$$

and

$$
\boxed{
\text{Split Value}
=
\text{Adaptation}
}
$$

## Structural Value of a Split

A split is useful when dividing a region allows ARRGO to represent its
internal structure more precisely.

Let:

$$
R=[l,r]
$$

and let a feasible split point be $s$.

The resulting child regions are:

$$
R_L=[l,s],
\qquad
R_R=[s,r].
$$

Before performing the split, ARRGO estimates whether the two child
regions are expected to provide a more informative representation than
the current parent region.

Define the estimated structural value of a split as:

$$
\widehat{V}_{\mathrm{split}}(s\mid R).
$$

This quantity represents the expected improvement in the structural
representation of the search space resulting from splitting $R$ at
$s$.

The structural value can depend on observable properties of the
current region, including:

- differences in observed behavior across the proposed split,
- changes in directional behavior,
- changes in secant slopes,
- distribution of observations,
- sampling gaps,
- differences in uncertainty,
- and differences in optimization potential between the resulting
  subregions.

Conceptually:

$$
\widehat{V}_{\mathrm{split}}(s\mid R)
=
\operatorname{StructuralValue}
\left(
s,\mathcal{D}_R,B_R,\mathcal{U}(R),P(R)
\right).
$$

The split point is then selected from the feasible set according to:

$$
s^*
\in
\arg\max_{s\in\mathcal{S}_R^{\mathrm{feasible}}}
\widehat{V}_{\mathrm{split}}(s\mid R).
$$

The structural value is an adaptive decision signal.

It is not a guarantee that the selected split contains the global
optimum.

Its purpose is to determine where additional spatial resolution is
most useful given the information currently available.

After splitting, the resulting child regions are independently
analyzed, and their behavior, uncertainty, and optimization potential
are recomputed.

Thus, splitting follows the same adaptive principle as sampling:

$$
\boxed{
\text{Current Information}
\rightarrow
\text{Refinement Decision}
\rightarrow
\text{New Structure}
\rightarrow
\text{Re-analysis}
}
$$

## Candidate Split Locations

For a region

$$
R=[l,r],
$$

ARRGO does not consider every possible point in the continuous
interval $(l,r)$ as a split location.

Instead, it constructs a finite set of structurally meaningful
candidate split points from the information currently available.

Let:

$$
\mathcal{S}_R
$$

denote the candidate split set.

Potential split locations can be generated from several sources.

### 1. Behavioral transition candidates

Locations near observed changes in directional behavior or secant slopes
can be considered as possible structural boundaries.

### 2. Sampling-gap candidates

Large gaps between consecutive observations may indicate that the
region contains unresolved structure.

A point inside such a gap can therefore become a candidate split
location.

### 3. Candidate-extremum candidates

A location associated with an observed candidate extremum can be
considered when separating different local behaviors may improve the
regional representation.

### 4. Information-boundary candidates

If neighboring parts of a region have substantially different
uncertainty or optimization potential, a point between them can be
considered as a structural boundary.

The resulting candidate set is:

$$
\mathcal{S}_R
=
\mathcal{S}_{\mathrm{behavior}}
\cup
\mathcal{S}_{\mathrm{gap}}
\cup
\mathcal{S}_{\mathrm{extremum}}
\cup
\mathcal{S}_{\mathrm{boundary}}.
$$

Candidates that violate the structural contraction requirement are
removed:

$$
\mathcal{S}_R^{\mathrm{feasible}}
=
\left\{
s\in\mathcal{S}_R:
\max(s-l,r-s)\leq\rho(r-l)
\right\}.
$$

The final split point is then selected from the feasible candidate set:

$$
s^*
\in
\arg\max_{s\in\mathcal{S}_R^{\mathrm{feasible}}}
\widehat{V}_{\mathrm{split}}(s\mid R).
$$

Thus, ARRGO separates three stages:

$$
\boxed{
\text{Candidate Generation}
\rightarrow
\text{Feasibility Filtering}
\rightarrow
\text{Split Selection}
}
$$

Candidate generation identifies structurally meaningful locations,
feasibility filtering preserves the required contraction property, and
split selection determines which feasible location provides the greatest
estimated structural value.

## Structural Difference Across a Split

A split is useful when the two resulting subregions exhibit different
observed characteristics.

Let a candidate split point $s$ divide region $R$ into:

$$
R_L=[l,s],
\qquad
R_R=[s,r].
$$

Let the behavior profiles of the two child regions be:

$$
B_{R_L}
$$

and

$$
B_{R_R}.
$$

The structural difference between the two regions can be represented
conceptually as:

$$
D_{\mathrm{struct}}(R_L,R_R).
$$

This quantity measures how different the currently observed structures
of the two subregions are.

Possible sources of structural difference include:

- different directional behavior,
- different patterns of secant slopes,
- different observed function variation,
- different sampling density,
- different uncertainty,
- and different optimization potential.

For example, if one side of the proposed split predominantly shows
increasing behavior while the other side shows decreasing behavior,
the split may separate two meaningfully different local structures.

However, structural difference must be based only on information that
has actually been observed.

Therefore, ARRGO does not assume that a structural difference proves
that the underlying function has a true discontinuity, extremum, or
change in analytical regime.

Instead:

$$
D_{\mathrm{struct}}(R_L,R_R)
$$

represents an observed difference in the current information available
to the algorithm.

A useful split should therefore provide both:

$$
\boxed{\text{Structural Difference}}
$$

and

$$
\boxed{\text{Sufficient Spatial Contraction}}.
$$

The first makes the split informative, while the second ensures that
the resulting regions become progressively smaller.

The precise mathematical definition of structural difference will be
developed from the observable region features rather than from
assumptions about the analytical form of the objective function.

## Directional Difference Across a Split

The observed directional behavior of a region can be represented using
the signs of its secant slopes.

For consecutive observations:

$$
s_i
=
\frac{y_{i+1}-y_i}{x_{i+1}-x_i}.
$$

Define the observed direction as:

$$
d_i=\operatorname{sign}(s_i).
$$

For a proposed split point $s$, the observations on the left and right
sides generate two directional profiles:

$$
D_{R_L}
$$

and

$$
D_{R_R}.
$$

A directional difference exists when the observed directional behavior
on the two sides is substantially different.

For example, consider:

$$
D_{R_L}=\{+1,+1,+1\}
$$

and:

$$
D_{R_R}=\{-1,-1,-1\}.
$$

The two sides exhibit opposite observed directions.

This provides evidence that the proposed split separates different local
behaviors.

A simple directional disagreement measure can be defined by comparing
the dominant observed directions of the two subregions.

Let:

$$
\bar d_{R_L}
$$

and

$$
\bar d_{R_R}
$$

represent their dominant observed directions.

Then:

$$
D_{\mathrm{direction}}
(R_L,R_R)
=
\mathbf{1}
\left[
\bar d_{R_L}
\neq
\bar d_{R_R}
\right].
$$

Here, $\mathbf{1}[\cdot]$ is an indicator function.

A value of $1$ indicates different dominant observed directions, while
a value of $0$ indicates that the dominant directions are the same.

This measure is intentionally based only on observed function
evaluations.

It does not claim that the underlying function has a true monotonic
structure on either side of the split.

Directional difference is therefore one source of structural evidence,
rather than a proof of analytical behavior.

More refined measures can later consider the strength, consistency, and
frequency of directional changes instead of using only the dominant
direction.

## Slope Variation Difference Across a Split

Directional behavior only describes whether the observed function values
increase or decrease.

However, two subregions may have the same dominant direction while having
very different rates of change.

For consecutive secant slopes:

$$
s_i
=
\frac{y_{i+1}-y_i}{x_{i+1}-x_i},
$$

the change between neighboring slopes is:

$$
\Delta s_i=s_{i+1}-s_i.
$$

The magnitude

$$
|\Delta s_i|
$$

represents the observed change in the local rate of variation.

For a region $R$, the observed slope-variation profile can therefore be
represented as:

$$
V_R^{(s)}
=
\{
|\Delta s_i|
\}.
$$

For a proposed split

$$
R\rightarrow R_L\cup R_R,
$$

we obtain two slope-variation profiles:

$$
V_{R_L}^{(s)}
$$

and

$$
V_{R_R}^{(s)}.
$$

A structural difference may exist when the magnitude and pattern of
slope variation differ between the two subregions.

One simple summary is the average observed slope variation:

$$
\mu_R^{(s)}
=
\frac{1}{|V_R^{(s)}|}
\sum_{v\in V_R^{(s)}}v,
$$

provided that the region contains enough observations to compute
slope variations.

The difference between the two subregions can then be represented as:

$$
D_{\mathrm{slope}}
(R_L,R_R)
=
\left|
\mu_{R_L}^{(s)}
-
\mu_{R_R}^{(s)}
\right|.
$$

A large value indicates that the observed rate of change varies
differently across the two subregions.

For example, both subregions may have positive secant slopes, but the
left side may show nearly constant slopes while the right side may show
rapidly changing slopes.

In this case, the directional difference may be small or zero, while
the slope-variation difference can still provide evidence that the split
separates different observed behaviors.

This measure is based entirely on function evaluations and secant
slopes.

It does not represent the exact derivative or curvature of the
underlying function.

If a subregion does not contain enough observations to compute a slope
variation, its slope-variation evidence is considered unavailable
rather than being assigned an artificial value of zero.

## Sampling Density and Coverage Difference Across a Split

The structural quality of a region depends not only on the observed
function values, but also on how those observations are distributed
inside the region.

For ordered observations

$$
x_1<x_2<\cdots<x_n,
$$

define the sampling gaps as:

$$
g_i=x_{i+1}-x_i.
$$

These gaps describe the spatial coverage of the available information.

For a region $R$, define its maximum sampling gap as:

$$
g_{\max}(R)=\max_i g_i.
$$

A large maximum gap indicates that a relatively large part of the region
has not been directly explored.

For a proposed split

$$
R\rightarrow R_L\cup R_R,
$$

we can compare the spatial coverage of the two subregions through their
sampling gaps.

For example, define:

$$
C_R=g_{\max}(R).
$$

Then the coverage difference between the two subregions can be
represented as:

$$
D_{\mathrm{coverage}}(R_L,R_R)
=
\left|
C_{R_L}-C_{R_R}
\right|.
$$

A large value indicates that the two subregions have different levels
of spatial coverage.

However, coverage difference alone does not determine whether a split
is useful.

A split can produce two regions with similar sampling densities while
still separating important behavioral structures.

Therefore, coverage information should be interpreted together with
directional and slope-variation information.

The spatial distribution of observations can also be considered through
the normalized maximum gap:

$$
\widetilde C_R
=
\frac{g_{\max}(R)}
{r-l},
$$

where

$$
R=[l,r].
$$

This makes the coverage measure relative to the size of the region.

For a proposed split, the normalized coverage profiles of the two
subregions can be compared as:

$$
D_{\mathrm{normalized\_coverage}}
=
\left|
\widetilde C_{R_L}
-
\widetilde C_{R_R}
\right|.
$$

These measures describe the distribution of available information and
do not make any assumption about the unobserved values of the objective
function.

## Uncertainty Difference Across a Split

The available observations provide only partial information about the
objective function inside a region.

Under a valid Lipschitz constant $L$, the function can be bounded using
the observed samples.

For a region $R$:

$$
L_R(x)
=
\max_i
\left[
y_i-L|x-x_i|
\right],
$$

and

$$
U_R(x)
=
\min_i
\left[
y_i+L|x-x_i|
\right].
$$

The pointwise uncertainty is:

$$
u_R(x)=U_R(x)-L_R(x).
$$

The region-level uncertainty is:

$$
\mathcal U(R)
=
\max_{x\in R}u_R(x).
$$

Now consider a proposed split:

$$
R\rightarrow R_L\cup R_R.
$$

The two subregions have their own uncertainty measures:

$$
\mathcal U(R_L)
$$

and

$$
\mathcal U(R_R).
$$

A simple measure of uncertainty difference is:

$$
D_{\mathrm{uncertainty}}
(R_L,R_R)
=
\left|
\mathcal U(R_L)
-
\mathcal U(R_R)
\right|.
$$

A large value indicates that the available information is distributed
unevenly between the two subregions.

However, uncertainty difference alone does not determine whether a split
is useful.

A more important quantity for refinement is the reduction in uncertainty
caused by the split:

$$
\Delta\mathcal U_{\mathrm{split}}
=
\mathcal U(R)
-
\max
\left(
\mathcal U(R_L),
\mathcal U(R_R)
\right).
$$

This quantity measures whether the worst remaining uncertainty becomes
smaller after the structural refinement.

The uncertainty of the parent region and the uncertainty of its children
must therefore be considered together.

A useful split should ideally preserve spatial contraction while
producing a meaningful reduction in unresolved uncertainty.

The uncertainty measures above are valid only when the Lipschitz
assumption and its constant $L$ are valid.

Without such an assumption, finite function evaluations alone do not
provide a universal numerical uncertainty bound for unseen points.

## Optimization Potential Difference Across a Split

In global optimization, observed function values alone do not determine
whether an unexplored part of a region can contain a better solution.

Under a valid Lipschitz assumption, however, an upper envelope can provide
an upper bound on the objective function.

For a region $R$:

$$
U_R(x)
=
\min_i
\left[
y_i+L|x-x_i|
\right].
$$

The optimization potential of the region is defined as:

$$
P(R)
=
\max_{x\in R}U_R(x).
$$

Let the current best observed objective value be:

$$
f_{\mathrm{best}}
=
\max_i f(x_i).
$$

The potential improvement of a region is then:

$$
\Delta_R
=
P(R)-f_{\mathrm{best}}.
$$

If

$$
\Delta_R>0,
$$

the current information does not rule out the possibility that the
region contains a solution better than the current best observed value.

If

$$
\Delta_R\leq0,
$$

the current certified information indicates that the region cannot
contain a value greater than the current best observed value.

This does not mean that such a region is removed from the algorithm.
ARRGO does not prune regions.

Now consider a proposed split:

$$
R\rightarrow R_L\cup R_R.
$$

The children have their own optimization potentials:

$$
P(R_L)
$$

and

$$
P(R_R).
$$

The difference in optimization potential can be represented as:

$$
D_{\mathrm{potential}}
(R_L,R_R)
=
\left|
P(R_L)-P(R_R)
\right|.
$$

A large value indicates that the two subregions have different levels
of remaining optimization potential.

More importantly, the split can be evaluated by the change in the
largest remaining potential:

$$
\Delta P_{\mathrm{split}}
=
P(R)
-
\max
\left(
P(R_L),P(R_R)
\right).
$$

This measures whether the structural refinement reduces the largest
remaining potential according to the current certified information.

Optimization potential is therefore different from uncertainty.

Uncertainty asks:

$$
\text{How much do we still not know?}
$$

Optimization potential asks:

$$
\text{How good could the best unexplored point still be?}
$$

Both quantities can influence refinement decisions, but they represent
different types of information.

The potential-based quantities above are valid only when the Lipschitz
assumption and its constant $L$ are valid.

## Structural Difference Profile

A proposed split can reveal several different types of structural
information.

For

$$
R\rightarrow R_L\cup R_R,
$$

the observed structural differences may include:

### Directional Difference

$$
D_{\mathrm{direction}}(R_L,R_R)
$$

which measures differences in the observed directional behavior of the
two subregions.

### Slope-Variation Difference

$$
D_{\mathrm{slope}}(R_L,R_R)
$$

which measures differences in the observed variation of secant slopes.

### Coverage Difference

$$
D_{\mathrm{coverage}}(R_L,R_R)
$$

which measures differences in the spatial distribution of available
observations.

### Uncertainty Difference

$$
D_{\mathrm{uncertainty}}(R_L,R_R)
$$

which measures differences in the remaining certified uncertainty.

### Optimization-Potential Difference

$$
D_{\mathrm{potential}}(R_L,R_R)
$$

which measures differences in the remaining optimization potential.

These quantities form a structural difference profile:

$$
\mathbf D_{\mathrm{struct}}(s\mid R)
=
\left[
D_{\mathrm{direction}},
D_{\mathrm{slope}},
D_{\mathrm{coverage}},
D_{\mathrm{uncertainty}},
D_{\mathrm{potential}}
\right].
$$

This vector should not initially be reduced to a single weighted score.

Different components represent different types of information and may
not have the same numerical meaning or scale.

Therefore, the structural value of a split is first treated as a
multi-criteria object:

$$
\widehat V_{\mathrm{split}}(s\mid R)
\equiv
\mathbf D_{\mathrm{struct}}(s\mid R).
$$

A split is considered structurally more informative when it provides
stronger evidence in one or more relevant dimensions without being
dominated by another feasible split.

This allows ARRGO to compare candidate splits without introducing
arbitrary fixed weights.

The final split-selection mechanism can therefore use the same
dominance-based principle previously used for sampling candidates.

The important distinction is:

$$
\text{Structural Difference}
\neq
\text{Single Numerical Score}.
$$

Structural difference is initially represented as a profile of
different kinds of evidence.

Only after the meaning and compatibility of these dimensions have been
established should a final scalar selection rule, if necessary, be
introduced.

## Dominance Between Candidate Splits

Suppose a region $R$ has several feasible candidate split points:

$$
\mathcal S_R^{\mathrm{feasible}}
=
\{s_1,s_2,\ldots,s_m\}.
$$

Each candidate split has a structural difference profile:

$$
\mathbf D_{\mathrm{struct}}(s\mid R)
=
\left[
D_{\mathrm{direction}},
D_{\mathrm{slope}},
D_{\mathrm{coverage}},
D_{\mathrm{uncertainty}},
D_{\mathrm{potential}}
\right].
$$

Because these dimensions represent different types of information, ARRGO
does not initially combine them using arbitrary fixed weights.

Instead, candidate splits can be compared using Pareto dominance.

A feasible split $s_1$ dominates another feasible split $s_2$ if:

$$
D_k(s_1\mid R)
\geq
D_k(s_2\mid R)
$$

for every structural dimension $k$, and:

$$
D_j(s_1\mid R)
>
D_j(s_2\mid R)
$$

for at least one dimension $j$.

Formally:

$$
s_1\succ s_2
$$

if and only if

$$
\forall k:
D_k(s_1\mid R)\geq D_k(s_2\mid R)
$$

and

$$
\exists j:
D_j(s_1\mid R)>D_j(s_2\mid R).
$$

The set of non-dominated candidate splits is then:

$$
\mathcal S_R^{\mathrm{ND}}
=
\left\{
s\in\mathcal S_R^{\mathrm{feasible}}
\mid
\nexists z\in\mathcal S_R^{\mathrm{feasible}}
\text{ such that }z\succ s
\right\}.
$$

A dominated split is not necessarily invalid.

It is simply inferior to another feasible split across all considered
structural dimensions.

Therefore, ARRGO can focus its final selection mechanism on the
non-dominated set:

$$
\mathcal S_R^{\mathrm{ND}}
\subseteq
\mathcal S_R^{\mathrm{feasible}}.
$$

This preserves the multi-dimensional nature of structural information
without introducing arbitrary weights.

The dominance relation is used only for comparing candidate split
locations.

It does not remove or prune regions from the global search structure.

The overall split-selection process is therefore:

$$
\boxed{
\text{Generate}
\rightarrow
\text{Check Feasibility}
\rightarrow
\text{Build Structural Profiles}
\rightarrow
\text{Find Non-Dominated Splits}
\rightarrow
\text{Select Split}
}
$$

The final selection among non-dominated splits requires an additional
refinement objective, because multiple candidates may remain
non-dominated.

## Final Selection Among Non-Dominated Splits

After Pareto dominance, multiple candidate splits may remain
non-dominated.

Therefore, dominance alone is not sufficient to determine the final
split point.

Let:

$$
\mathcal S_R^{\mathrm{ND}}
$$

denote the set of non-dominated feasible split candidates.

The final selection should depend on the current refinement objective
of the region.

For example, if the region is primarily limited by unresolved
uncertainty, the selection mechanism should prioritize candidates that
are expected to improve uncertainty resolution.

If the main objective is to distinguish different observed behaviors,
the selection should prioritize structural separation.

If the main objective is to investigate optimization potential, the
selection should prioritize the separation of regions with different
remaining potential.

Therefore, ARRGO separates:

$$
\boxed{
\text{Candidate Evaluation}
}
$$

from:

$$
\boxed{
\text{Refinement Objective}
}
$$

The candidate evaluation produces the structural profile:

$$
\mathbf D_{\mathrm{struct}}(s\mid R).
$$

The refinement objective determines which aspect of this profile is
currently most relevant.

Let the current refinement objective be:

$$
O_R.
$$

Then the final split is selected as:

$$
s^*
=
\operatorname{Select}
\left(
\mathcal S_R^{\mathrm{ND}},
O_R
\right).
$$

The refinement objective may be determined from the current state of
the region.

For example:

$$
O_R=
\begin{cases}
\text{uncertainty resolution},
&
\text{if uncertainty is unresolved},\\
\text{behavioral separation},
&
\text{if observed behavior is heterogeneous},\\
\text{potential separation},
&
\text{if optimization potential is unresolved},\\
\text{coverage improvement},
&
\text{if spatial information is insufficient}.
\end{cases}
$$

The objective is therefore state-dependent rather than fixed for every
region.

This prevents ARRGO from treating every region in the same way.

The final selection process becomes:

$$
\boxed{
\text{Feasible Splits}
\rightarrow
\text{Structural Profiles}
\rightarrow
\text{Pareto Dominance}
\rightarrow
\text{Region-Specific Refinement Objective}
\rightarrow
\text{Final Split}
}
$$

No arbitrary fixed weighting is required at this stage.

The exact definition of the region-specific refinement objective is
intentionally left open until the corresponding region-state measures
have been formally defined.

## Region Refinement Objective

A region may require refinement for different reasons.

A region may contain insufficient spatial information, unresolved
uncertainty, heterogeneous observed behavior, or significant remaining
optimization potential.

Therefore, refinement should not be governed by a single fixed
objective.

Let the state of a region be represented by:

$$
S_R=(\mathcal D_R,B_R,\mathcal U_R,P_R).
$$

From this state, ARRGO identifies the currently unresolved aspects of
the region.

Define the refinement-objective profile as:

$$
\mathbf O_R
=
\left[
O_{\mathrm{coverage}},
O_{\mathrm{behavior}},
O_{\mathrm{uncertainty}},
O_{\mathrm{potential}}
\right].
$$

Each component represents the degree to which a particular type of
information remains unresolved.

For example:

$$
O_{\mathrm{coverage}}
$$

represents unresolved spatial coverage,

$$
O_{\mathrm{behavior}}
$$

represents unresolved differences or transitions in observed behavior,

$$
O_{\mathrm{uncertainty}}
$$

represents unresolved uncertainty under a valid uncertainty model, and

$$
O_{\mathrm{potential}}
$$

represents unresolved optimization potential.

The refinement objective is therefore not an externally fixed label.
It is derived from the current information state of the region.

The dominant unresolved objective can be represented as:

$$
O_R^*
=
\operatorname{ArgMax}
\left(
\mathbf O_R
\right).
$$

This does not imply that all other objectives are ignored.

Instead, $O_R^*$ identifies the aspect that currently provides the
strongest reason for further refinement.

The refinement process can therefore be viewed as:

$$
\boxed{
\text{Region Information}
\rightarrow
\text{Unresolved Information}
\rightarrow
\text{Refinement Objective}
\rightarrow
\text{Refinement Action}
}
$$

This creates a feedback mechanism in which the objective of refinement
can change as new function evaluations and structural refinements
provide additional information.

Consequently, a region that initially requires coverage refinement may
later become behaviorally unresolved, while another region may become
primarily uncertainty- or potential-driven.

The refinement objective is therefore dynamic and region-specific.

## Coverage Resolution Objective

A region may be insufficiently explored when its available observations
do not provide adequate spatial coverage.

For a region

$$
R=[l,r]
$$

with ordered observations

$$
x_1<x_2<\cdots<x_n,
$$

define the sampling gaps as:

$$
g_i=x_{i+1}-x_i.
$$

The largest unresolved spatial gap is:

$$
g_{\max}(R)
=
\max_i g_i.
$$

To make this quantity independent of the absolute size of the region,
define the normalized maximum gap:

$$
\widetilde G_R
=
\frac{g_{\max}(R)}{r-l}.
$$

A large value of $\widetilde G_R$ indicates that a relatively large
portion of the region is represented by a sparse set of observations.

However, a large gap by itself does not necessarily imply that the
region requires splitting.

The importance of a gap also depends on whether the available
observations suggest unresolved behavior or optimization relevance near
that gap.

Therefore, coverage should be treated as one component of the
region's unresolved-information profile rather than as an isolated
splitting rule.

Define the coverage-resolution objective conceptually as:

$$
O_{\mathrm{coverage}}(R)
=
\operatorname{CoverageUnresolved}(R).
$$

Its value should increase when the spatial distribution of observations
leaves important portions of the region insufficiently characterized.

The objective is therefore based on the relationship between:

$$
\text{Region Geometry}
$$

and

$$
\text{Observation Distribution}.
$$

In particular, coverage resolution does not depend on the total number
of samples alone.

Two regions may contain the same number of observations while having
very different spatial coverage.

For example, observations concentrated in one part of a region may
leave a large unexplored interval elsewhere.

Therefore:

$$
\boxed{
\text{Coverage Quality}
\neq
\text{Number of Samples}
}
$$

The coverage-resolution objective provides evidence for determining
whether additional spatial refinement may be useful.

It does not by itself determine whether the next action should be
sampling or splitting.

## Behavior Resolution Objective

Spatial coverage alone does not determine whether a region has been
sufficiently characterized.

A region may contain many observations while its observed behavior
remains complex or ambiguous.

For ordered observations:

$$
x_1<x_2<\cdots<x_n,
$$

the secant slopes are:

$$
s_i
=
\frac{y_{i+1}-y_i}{x_{i+1}-x_i}.
$$

The corresponding directional sequence is:

$$
D_R
=
\{
\operatorname{sign}(s_1),
\ldots,
\operatorname{sign}(s_{n-1})
\}.
$$

Changes in direction provide evidence of possible behavioral
transitions.

For consecutive slopes, define:

$$
\Delta s_i=s_{i+1}-s_i.
$$

Large variations in consecutive slopes provide evidence that the
observed rate of change is not uniform.

Therefore, unresolved behavior can be associated with:

$$
\text{Directional Changes}
$$

and

$$
\text{Slope Variations}.
$$

Define the behavior-resolution objective conceptually as:

$$
O_{\mathrm{behavior}}(R)
=
\operatorname{BehaviorUnresolved}(R).
$$

Its value should increase when the currently observed data contain
significant or ambiguous changes in directional behavior or slope
variation.

The objective is not intended to identify the exact analytical
behavior of the underlying function.

Instead, it measures how much additional spatial resolution may be
needed to distinguish the observed local structures.

A region with a consistent observed pattern such as

$$
D_R=\{+1,+1,+1,+1\}
$$

may provide stronger behavioral evidence than a region with a pattern
such as

$$
D_R=\{+1,+1,-1,+1,-1\}.
$$

The second pattern contains several observed directional transitions
and may therefore require additional investigation.

However, directional changes alone are not sufficient evidence of
complexity.

They may also result from sparse sampling or numerical noise.

Therefore, the behavior-resolution objective should be interpreted
together with sampling coverage and the reliability of the available
observations.

In particular:

$$
\boxed{
\text{Observed Behavioral Complexity}
\neq
\text{True Function Complexity}
}
$$

The behavior-resolution objective provides evidence that further
refinement may be useful, but it does not by itself determine the
refinement action. 

## Uncertainty Resolution Objective

Under a valid Lipschitz assumption, the available function evaluations
provide upper and lower envelopes over a region.

For

$$
R=[l,r],
$$

define:

$$
L_R(x)
=
\max_i
\left[
y_i-L|x-x_i|
\right]
$$

and

$$
U_R(x)
=
\min_i
\left[
y_i+L|x-x_i|
\right].
$$

The pointwise uncertainty is:

$$
u_R(x)
=
U_R(x)-L_R(x).
$$

The region-level uncertainty is:

$$
\mathcal U(R)
=
\max_{x\in R}u_R(x).
$$

The value of $\mathcal U(R)$ represents the largest remaining
uncertainty permitted by the current observations and the assumed
Lipschitz bound.

Therefore, define the uncertainty-resolution objective as:

$$
O_{\mathrm{uncertainty}}(R)
=
\mathcal U(R).
$$

A larger value indicates that there exists at least one location in the
region where the current certified information remains relatively
uncertain.

This objective is fundamentally different from the behavioral
objective.

Behavioral analysis asks:

$$
\text{What structure do the observed samples suggest?}
$$

Uncertainty analysis asks:

$$
\text{What values are still mathematically possible under the
assumption?}
$$

Consequently, a region may have:

- simple observed behavior but high uncertainty,
- complex observed behavior but low uncertainty,
- both high behavioral complexity and high uncertainty,
- or neither.

The uncertainty-resolution objective therefore provides a separate
source of information for refinement decisions.

When a new sample is evaluated, the envelopes can be recomputed:

$$
\mathcal U(R)
\rightarrow
\mathcal U(R\mid \text{new observation}).
$$

The observed reduction is:

$$
\Delta\mathcal U
=
\mathcal U(R)
-
\mathcal U(R\mid \text{new observation}).
$$

This quantity measures the actual effect of the new function evaluation
on the certified uncertainty.

The uncertainty-resolution objective is valid only when the underlying
uncertainty model is valid.

In particular, the Lipschitz-based formulation requires a valid
Lipschitz constant $L$.

Without such an assumption, finite function evaluations alone do not
provide the same type of certified uncertainty bound.

## Optimization-Potential Resolution Objective

In global optimization, a region is important not only because its
behavior or uncertainty is unresolved, but also because it may still
contain a solution better than the current best observed solution.

Under a valid Lipschitz assumption, the upper envelope provides an upper
bound on the objective function:

$$
U_R(x)
=
\min_i
\left[
y_i+L|x-x_i|
\right].
$$

The optimization potential of a region is:

$$
P(R)
=
\max_{x\in R}U_R(x).
$$

Let the current best observed objective value be:

$$
f_{\mathrm{best}}.
$$

The remaining potential improvement is:

$$
\Delta_R
=
P(R)-f_{\mathrm{best}}.
$$

If:

$$
\Delta_R>0,
$$

the current certified information does not rule out the existence of a
better solution inside the region.

If:

$$
\Delta_R\leq0,
$$

the current certified information indicates that the region cannot
improve upon the current best observed value.

Define the optimization-potential resolution objective as:

$$
O_{\mathrm{potential}}(R)
=
\max(0,\Delta_R).
$$

This objective measures how much room remains, according to the current
certified information, for discovering an improvement over the
incumbent solution.

The optimization-potential objective is different from uncertainty.

A region may have high uncertainty while having little potential for
improvement.

Conversely, a region may have moderate uncertainty but still have
substantial potential to contain a better solution.

Therefore:

$$
\boxed{
\text{Uncertainty}
\neq
\text{Optimization Potential}
}
$$

When new information becomes available, the potential can change:

$$
P(R)
\rightarrow
P(R\mid\text{new information}).
$$

Consequently:

$$
O_{\mathrm{potential}}(R)
\rightarrow
O_{\mathrm{potential}}(R\mid\text{new information}).
$$

This allows the refinement process to continuously update the
importance of a region as the global search progresses.

A region whose potential decreases is not removed from the search
structure.

It remains part of the region hierarchy because ARRGO does not use
region pruning as a refinement mechanism.

The optimization-potential objective is valid only when the upper
envelope is based on a valid uncertainty assumption, such as a valid
Lipschitz constant.

## Unresolved Information State

The refinement objectives describe different types of unresolved
information inside a region.

For a region $R$, define the objective profile as:

$$
\mathbf O_R
=
\left[
O_{\mathrm{coverage}},
O_{\mathrm{behavior}},
O_{\mathrm{uncertainty}},
O_{\mathrm{potential}}
\right].
$$

These objectives do not necessarily have the same numerical scale.

Therefore, ARRGO does not compare their raw numerical values directly.

Instead, each objective is interpreted relative to its own meaning and
the information already available in the region.

For each objective, ARRGO determines whether the corresponding aspect is
sufficiently resolved.

Define the unresolved-information state as:

$$
\mathcal Q_R
=
\left[
Q_{\mathrm{coverage}},
Q_{\mathrm{behavior}},
Q_{\mathrm{uncertainty}},
Q_{\mathrm{potential}}
\right],
$$

where each component represents whether that aspect remains unresolved.

Conceptually:

$$
Q_k(R)
=
\operatorname{Unresolved}_k(R).
$$

The resulting state describes the current information requirements of
the region.

For example, a region may be:

$$
\mathcal Q_R
=
[
\text{unresolved},
\text{resolved},
\text{unresolved},
\text{unresolved}
].
$$

This means that spatial coverage, uncertainty, and optimization
potential still require investigation, while the observed behavior is
currently sufficiently characterized.

The unresolved state should be derived from the available information
rather than from a fixed global rule.

In particular, the interpretation of an objective should consider the
resolution already achieved inside the region.

Therefore:

$$
\boxed{
\text{Unresolved State}
=
\text{Current Information}
+
\text{Remaining Information Requirement}
}
$$

The purpose of $\mathcal Q_R$ is not to assign a permanent label to the
region.

As new samples are evaluated or the region is structurally refined,
the unresolved state can change:

$$
\mathcal Q_R
\rightarrow
\mathcal Q_R'.
$$

This allows ARRGO to adapt its refinement objective throughout the
optimization process.

The unresolved-information state is therefore a dynamic representation
of what the algorithm still needs to learn about a region.

## Resolution Criteria

An unresolved-information objective describes an aspect of a region that
may still require refinement.

However, ARRGO needs a criterion for determining when an aspect has
become sufficiently resolved.

A resolution criterion should not be based only on a fixed number of
samples or an arbitrary numerical threshold.

Instead, resolution should be connected to the effect of additional
information on the decision that ARRGO would make.

Let the current information state of a region be:

$$
S_R.
$$

Suppose a potential additional observation produces the updated state:

$$
S_R^{+x}.
$$

Let:

$$
\operatorname{Decision}(S_R)
$$

denote the refinement decision produced from the current information.

An aspect of the region can be considered sufficiently resolved when
additional information is unlikely to change the relevant decision.

Conceptually:

$$
\operatorname{Decision}(S_R)
\approx
\operatorname{Decision}(S_R^{+x}).
$$

Therefore, resolution is defined in terms of decision stability rather
than sample count.

For an objective $O_k$, define its resolution condition as:

$$
\operatorname{Resolution}_k(R)
=
\operatorname{DecisionStability}_k(R).
$$

The corresponding unresolved state is:

$$
Q_k(R)
=
1-\operatorname{Resolution}_k(R).
$$

This formulation emphasizes that the purpose of refinement is not to
collect observations indefinitely.

The purpose is to collect enough information to make the relevant
decision stable.

For example, additional sampling may be unnecessary when several
potential sampling locations would all lead to the same refinement
decision.

Conversely, a region may remain unresolved even with many observations
if a small amount of additional information could still change the
decision.

Therefore:

$$
\boxed{
\text{Resolution}
\neq
\text{Number of Samples}
}
$$

and:

$$
\boxed{
\text{Resolution}
=
\text{Stability of the Relevant Decision}
}
$$

This principle allows the resolution criteria to remain adaptive to the
current state of each region.

The exact operational test for decision stability must be defined for
each refinement objective.

## Decision Stability for Coverage

Coverage resolution concerns whether the current spatial distribution of
observations is sufficient for the next refinement decision.

For a region

$$
R=[l,r],
$$

let the current observation set be:

$$
\mathcal D_R.
$$

The current coverage state is determined from the spatial distribution
of the observations, including the sampling gaps:

$$
g_i=x_{i+1}-x_i.
$$

Let the current coverage-based decision be:

$$
\operatorname{Decision}_{\mathrm{coverage}}(\mathcal D_R).
$$

Now consider a hypothetical additional observation at a candidate
location $x$:

$$
\mathcal D_R^{+x}
=
\mathcal D_R
\cup
\{(x,f(x))\}.
$$

The corresponding coverage decision becomes:

$$
\operatorname{Decision}_{\mathrm{coverage}}
(\mathcal D_R^{+x}).
$$

Coverage is considered decision-stable when additional observations in
the relevant unexplored portions of the region do not change the
coverage-based refinement decision:

$$
\operatorname{Decision}_{\mathrm{coverage}}
(\mathcal D_R)
\approx
\operatorname{Decision}_{\mathrm{coverage}}
(\mathcal D_R^{+x}).
$$

This means that the current spatial information is sufficiently
consistent for the purpose of deciding whether further coverage
refinement is necessary.

The stability test should consider candidate locations in the largest
or most relevant sampling gaps rather than arbitrary points.

Therefore, coverage resolution is related to the stability of the
spatial refinement decision under additional observations.

Conceptually:

$$
\boxed{
\text{Coverage Resolved}
\iff
\text{Additional Spatial Information Does Not Change the Decision}
}
$$

This criterion does not require a fixed number of observations.

It also does not require a universal absolute gap threshold.

Instead, the required spatial resolution depends on the current
structure and information state of the region.

If additional observations can still substantially alter the
coverage-based refinement decision, the coverage objective remains
unresolved.

Otherwise, the region is considered sufficiently resolved with respect
to coverage.

The same decision-stability principle can later be adapted to the
behavior, uncertainty, and optimization-potential objectives.

## Coverage Stability Estimation

Coverage depends only on the spatial distribution of observations.

Therefore, unlike behavioral or optimization-based objectives, coverage
stability can be evaluated without knowing the function value at a
candidate point.

Let the current ordered observations of a region be:

$$
x_1<x_2<\cdots<x_n.
$$

The corresponding sampling gaps are:

$$
g_i=x_{i+1}-x_i.
$$

For each gap, consider a hypothetical observation inserted inside the
gap.

If:

$$
x_i<x<x_{i+1},
$$

the original gap

$$
g_i=x_{i+1}-x_i
$$

is replaced by two gaps:

$$
g_i^{(L)}=x-x_i
$$

and

$$
g_i^{(R)}=x_{i+1}-x.
$$

The resulting maximum gap is therefore:

$$
g_{\max}^{+x}
=
\max
\left(
\max_{j\neq i}g_j,
x-x_i,
x_{i+1}-x
\right).
$$

The best possible reduction of the selected gap is obtained by placing
the hypothetical observation at its midpoint:

$$
x^*
=
\frac{x_i+x_{i+1}}{2}.
$$

In that case:

$$
g_i^{(L)}
=
g_i^{(R)}
=
\frac{g_i}{2}.
$$

Therefore, the best achievable maximum-gap reduction inside gap $i$ is:

$$
\Delta G_i
=
g_{\max}(R)
-
\max
\left(
\max_{j\neq i}g_j,
\frac{g_i}{2}
\right).
$$

This quantity describes how much the worst spatial gap could be
reduced by obtaining one additional observation inside gap $i$.

The same reasoning can be applied to every relevant sampling gap.

Thus, ARRGO can estimate the potential coverage improvement of a new
observation without evaluating the objective function.

For a candidate location $x$ inside gap $[x_i,x_{i+1}]$, define the
coverage-improvement estimate as:

$$
\widehat I_{\mathrm{coverage}}(x\mid R)
=
g_{\max}(R)-g_{\max}^{+x}.
$$

A larger value indicates that the candidate observation is expected to
provide greater improvement in the spatial coverage of the region.

This produces a purely geometric information measure:

$$
\boxed{
\text{Coverage Improvement}
=
\text{Reduction in Maximum Sampling Gap}
}
$$

Coverage stability can therefore be estimated by examining whether
available candidate observations can still produce a meaningful change
in the current spatial coverage structure.

No function evaluation is required for this estimation.

## Behavior Stability Estimation

Behavioral information depends on the observed function values and their
spatial arrangement.

Unlike coverage, the effect of a new observation on behavioral
information cannot be determined exactly before evaluating the objective
function.

Let the current observations be:

$$
\mathcal D_R
=
\{(x_i,y_i)\}_{i=1}^{n},
$$

with:

$$
x_1<x_2<\cdots<x_n.
$$

The current secant slopes are:

$$
s_i
=
\frac{y_{i+1}-y_i}{x_{i+1}-x_i}.
$$

Suppose a candidate point $x$ lies inside the interval

$$
[x_i,x_{i+1}].
$$

The function value

$$
f(x)
$$

is unknown before evaluation.

Therefore, the new slopes

$$
\frac{f(x)-y_i}{x-x_i}
$$

and

$$
\frac{y_{i+1}-f(x)}{x_{i+1}-x}
$$

cannot be computed exactly in advance.

Consequently, ARRGO does not claim to know the exact behavioral effect
of the new observation.

Instead, it estimates the potential behavioral information of the
candidate from the currently observed structure.

Candidate locations become more behaviorally informative when they are
associated with regions where:

- the current secant slopes change rapidly,
- directional behavior changes,
- large sampling gaps contain potentially unresolved transitions,
- or the observed local behavior is inconsistent with neighboring
  observations.

Define the estimated behavioral information of a candidate point as:

$$
\widehat I_{\mathrm{behavior}}(x\mid R)
=
\operatorname{BehavioralEvidence}
(x,\mathcal D_R,B_R).
$$

This quantity represents the expected ability of the candidate
observation to clarify currently unresolved behavioral structure.

It is an estimate rather than a guaranteed reduction in behavioral
uncertainty.

After evaluating the candidate:

$$
y=f(x),
$$

the behavioral profile can be recomputed:

$$
B_R
\rightarrow
B_R^{+x}.
$$

The actual behavioral change can then be observed from the updated
secant slopes and directional sequence.

Thus, ARRGO separates:

$$
\boxed{
\text{Predicted Behavioral Information}
}
$$

from:

$$
\boxed{
\text{Observed Behavioral Effect}
}
$$

Before evaluation:

$$
\widehat I_{\mathrm{behavior}}(x\mid R)
$$

is used for candidate selection.

After evaluation:

$$
B_R^{+x}
$$

is used to update the region state.

This creates the same feedback mechanism used throughout ARRGO:

$$
\text{Estimate}
\rightarrow
\text{Evaluate}
\rightarrow
\text{Observe}
\rightarrow
\text{Update}.
$$

No unobserved function value is assumed during candidate selection.

## Optimization-Potential Stability Estimation

Optimization potential describes how strongly the current information
supports the possibility of finding a solution better than the current
incumbent.

Under a valid Lipschitz assumption, the upper envelope is:

$$
U_R(x)
=
\min_i
\left[
y_i+L|x-x_i|
\right].
$$

The current optimization potential is:

$$
P(R)
=
\max_{x\in R}U_R(x).
$$

Let:

$$
f_{\mathrm{best}}
=
\max_i y_i.
$$

The remaining potential improvement is:

$$
\Delta_R
=
P(R)-f_{\mathrm{best}}.
$$

Before evaluating a new point, its exact function value is unknown.

Therefore, the exact post-sampling potential cannot be determined in
advance.

However, the current upper envelope provides information about where
additional observations may affect the potential most strongly.

For a candidate point $x$, define the estimated potential information as:

$$
\widehat I_{\mathrm{potential}}(x\mid R)
=
\operatorname{PotentialEvidence}
(x,\mathcal D_R,U_R,P(R)).
$$

A candidate is potentially informative when it lies in a location whose
current certified upper bound contributes substantially to the
remaining optimization potential.

In particular, locations near regions with high values of

$$
U_R(x)
$$

are relevant because observations there may determine whether the
remaining potential is supported by an actual high objective value or
is reduced by additional information.

After evaluating the candidate:

$$
y=f(x),
$$

the upper envelope is recomputed:

$$
U_R
\rightarrow
U_R^{+x}.
$$

The updated optimization potential becomes:

$$
P(R)
\rightarrow
P(R\mid x,y).
$$

The actual change in potential is therefore:

$$
\Delta P
=
P(R)-P(R\mid x,y).
$$

This quantity can only be known after the function evaluation.

Therefore, ARRGO distinguishes between:

$$
\boxed{
\text{Estimated Potential Information}
}
$$

and:

$$
\boxed{
\text{Observed Potential Reduction}
}
$$

Before evaluation:

$$
\widehat I_{\mathrm{potential}}(x\mid R)
$$

guides candidate selection.

After evaluation:

$$
P(R\mid x,y)
$$

updates the optimization state.

This creates a feedback mechanism in which the algorithm can learn which
types of candidate locations are effective at resolving optimization
potential.

The potential-stability mechanism therefore follows:

$$
\text{Estimate}
\rightarrow
\text{Evaluate}
\rightarrow
\text{Update Potential}
\rightarrow
\text{Re-analyze}.
$$

The resulting objective remains certified only under the validity of the
underlying Lipschitz assumption.

## Refinement Action Selection: Sampling vs Splitting

ARRGO has two fundamental refinement actions:

$$
\text{Sampling}
$$

and

$$
\text{Splitting}.
$$

Sampling adds new function information inside the current region:

$$
R(\mathcal D_R)
\rightarrow
R(\mathcal D_R^+).
$$

The geometry of the region remains unchanged.

Splitting introduces a finer spatial structure:

$$
R
\rightarrow
R_L\cup R_R.
$$

The parent region remains part of the region hierarchy.

The choice between these two actions should depend on the current
information state of the region.

Let:

$$
\mathbf O_R
=
[
O_{\mathrm{coverage}},
O_{\mathrm{behavior}},
O_{\mathrm{uncertainty}},
O_{\mathrm{potential}}
]
$$

represent the unresolved-information profile.

The action-selection problem can therefore be expressed as:

$$
A_R^*
=
\operatorname{SelectAction}
\left(
R,\mathbf O_R
\right),
$$

where:

$$
A_R^*
\in
\{
\text{SAMPLE},
\text{SPLIT},
\text{STABLE}
\}.
$$

Sampling is appropriate when the main limitation is insufficient
information about the current region and an additional observation can
directly resolve that information.

Splitting is appropriate when the main limitation is insufficient
spatial resolution and separating the region into smaller subregions
can expose distinct local structures.

Conceptually:

$$
\boxed{
\text{Information Deficiency}
\rightarrow
\text{Sampling}
}
$$

while:

$$
\boxed{
\text{Structural Deficiency}
\rightarrow
\text{Splitting}
}
$$

However, these actions are not mutually exclusive in terms of their
possible benefits.

A region may simultaneously have unresolved behavioral structure,
uncertainty, and poor spatial coverage.

Therefore, ARRGO should compare the expected effect of both actions
rather than assigning a permanently fixed action to each type of
region.

For sampling:

$$
\widehat G_{\mathrm{sample}}(R)
=
\text{Estimated information gain from one additional observation}.
$$

For splitting:

$$
\widehat G_{\mathrm{split}}(R)
=
\text{Estimated structural gain from spatial refinement}.
$$

The final action can then be selected according to the current
refinement objective:

$$
A_R^*
=
\operatorname{Select}
\left(
\widehat G_{\mathrm{sample}},
\widehat G_{\mathrm{split}},
O_R^*
\right).
$$

The exact comparison rule is intentionally left unspecified until the
information-gain measures for sampling and splitting are placed on a
common and mathematically justified basis.

This prevents ARRGO from introducing arbitrary weights or thresholds
before the underlying quantities have been formally defined.

The complete decision cycle is therefore:

$$
\boxed{
\text{Analyze}
\rightarrow
\text{Identify Unresolved Information}
\rightarrow
\text{Estimate Action Effects}
\rightarrow
\text{Choose Sampling or Splitting}
\rightarrow
\text{Refine}
\rightarrow
\text{Re-analyze}
}
$$

A region is marked stable only when the relevant unresolved information
has become sufficiently decision-stable.

Stability does not remove the region from the global region hierarchy.

## Estimated Gain of Sampling

Sampling adds a new function evaluation to the current information
state of a region.

For a candidate point $x\in R$, the exact effect of the observation
cannot be known before evaluating:

$$
y=f(x).
$$

Therefore, ARRGO uses an estimated sampling gain before performing the
function evaluation.

Let the current state of the region be:

$$
S_R=(\mathcal D_R,B_R,\mathcal U_R,P_R).
$$

For a candidate point $x$, define the estimated sampling gain as:

$$
\widehat G_{\mathrm{sample}}(x\mid R)
=
\operatorname{ExpectedInformationGain}
(x,S_R).
$$

This quantity represents the expected ability of the new observation
to reduce unresolved information in the region.

The gain may contain several sources of information, including:

$$
\widehat G_{\mathrm{sample}}
=
\{
G_{\mathrm{coverage}},
G_{\mathrm{behavior}},
G_{\mathrm{uncertainty}},
G_{\mathrm{potential}}
\}.
$$

These components should initially be kept separate rather than combined
using arbitrary fixed weights.

For example, a candidate located inside a large sampling gap may have
high coverage value:

$$
G_{\mathrm{coverage}}(x\mid R).
$$

A candidate near an observed behavioral transition may have high
behavioral value:

$$
G_{\mathrm{behavior}}(x\mid R).
$$

Under a valid Lipschitz model, a candidate located where the certified
uncertainty is large may have high uncertainty-resolution value:

$$
G_{\mathrm{uncertainty}}(x\mid R).
$$

A candidate located in an area with substantial optimization potential
may have high potential-related value:

$$
G_{\mathrm{potential}}(x\mid R).
$$

The complete sampling-gain profile is therefore:

$$
\mathbf G_{\mathrm{sample}}(x\mid R)
=
\left[
G_{\mathrm{coverage}},
G_{\mathrm{behavior}},
G_{\mathrm{uncertainty}},
G_{\mathrm{potential}}
\right].
$$

Candidate sampling points can first be compared using the
multi-criteria dominance principle.

A candidate $x_1$ dominates candidate $x_2$ if:

$$
G_k(x_1\mid R)
\geq
G_k(x_2\mid R)
$$

for every information dimension $k$, with strict inequality in at least
one dimension.

The resulting non-dominated candidates form:

$$
\mathcal C_R^{\mathrm{ND}}.
$$

Only after this stage does ARRGO apply the current refinement objective
to select the final sampling point.

After the selected point is evaluated, the actual information gain can
be measured from the updated region state:

$$
G_{\mathrm{observed}}
=
\operatorname{Information}(S_R)
-
\operatorname{Information}(S_R^{+x}).
$$

Therefore:

$$
\boxed{
\widehat G_{\mathrm{sample}}
\neq
G_{\mathrm{observed}}
}
$$

The first is a prediction used for decision-making.

The second is the actual effect observed after the function evaluation.

This distinction allows ARRGO to continuously compare its predicted
refinement benefits with the effects actually obtained from previous
refinements.

## Estimated Gain of Splitting

Splitting refines the spatial structure of a region without immediately
requiring a new function evaluation.

For a region:

$$
R=[l,r],
$$

a candidate split point $s$ produces:

$$
R_L=[l,s],
$$

and:

$$
R_R=[s,r].
$$

The parent region remains in the region hierarchy.

Before performing the split, ARRGO estimates how much structural
information may be gained by introducing the proposed boundary.

For a candidate split point $s$, define:

$$
\widehat G_{\mathrm{split}}(s\mid R)
=
\operatorname{StructuralInformationGain}
(s,\mathcal D_R,B_R,\mathcal U_R,P_R).
$$

The estimated splitting gain can contain several distinct sources:

$$
\mathbf G_{\mathrm{split}}(s\mid R)
=
\left[
G_{\mathrm{direction}},
G_{\mathrm{slope}},
G_{\mathrm{coverage}},
G_{\mathrm{uncertainty}},
G_{\mathrm{potential}}
\right].
$$

The directional component measures whether the split is expected to
separate regions with different observed directional behavior.

The slope component measures whether the split separates regions with
different observed variation in secant slopes.

The coverage component measures whether the split improves the spatial
resolution of the region.

Under a valid Lipschitz uncertainty model, the uncertainty component
measures the potential reduction or redistribution of certified
uncertainty across the resulting subregions.

The potential component measures whether the split separates areas with
different remaining optimization potential.

These components describe different aspects of structural information
and should not initially be reduced to a single weighted score.

For feasible candidate split points, ARRGO first applies the
non-dominance principle.

A split $s_1$ dominates a split $s_2$ if:

$$
G_k(s_1\mid R)
\geq
G_k(s_2\mid R)
$$

for every information dimension $k$, with strict inequality in at
least one dimension.

The resulting set of non-dominated split candidates is:

$$
\mathcal S_R^{\mathrm{ND}}.
$$

The final split is then selected according to the current refinement
objective of the region.

Therefore:

$$
s^*
=
\operatorname{Select}
\left(
\mathcal S_R^{\mathrm{ND}},
O_R^*
\right).
$$

After the split is performed, the actual structural effect can be
measured from the resulting child regions.

Let:

$$
S_R
$$

be the state of the parent before splitting and:

$$
S_{R_L},S_{R_R}
$$

be the states of the resulting subregions.

The observed structural gain is:

$$
G_{\mathrm{observed}}
=
\operatorname{StructuralInformation}(S_{R_L},S_{R_R})
-
\operatorname{StructuralInformation}(S_R).
$$

Therefore:

$$
\boxed{
\widehat G_{\mathrm{split}}
\neq
G_{\mathrm{observed}}
}
$$

The estimated gain is used before the action to select a promising
structural refinement.

The observed gain is calculated after the split to update the region
state and evaluate the effectiveness of the decision.

Thus, splitting follows the same feedback principle as sampling:

$$
\boxed{
\text{Estimate}
\rightarrow
\text{Refine}
\rightarrow
\text{Observe}
\rightarrow
\text{Update}
}
$$

Splitting does not discard the parent region and does not prune either
of the resulting subregions.

The complete region hierarchy is preserved throughout the optimization
process.

## Common Refinement Objective

Sampling and splitting produce different types of refinement.

Sampling adds new information about the objective function:

$$
R(\mathcal D_R)
\rightarrow
R(\mathcal D_R^+).
$$

Splitting adds spatial structure:

$$
R
\rightarrow
R_L\cup R_R.
$$

Therefore, their raw information gains are not directly comparable.

For example, a reduction in sampling uncertainty and a change in
spatial resolution do not necessarily have the same numerical meaning.

ARRGO therefore does not compare the raw gain values directly.

Instead, both actions are evaluated with respect to the same underlying
question:

$$
\boxed{
\text{How effectively does this action resolve the current unresolved
information of the region?}
}
$$

Let the current unresolved-information state of a region be:

$$
\mathcal Q_R
=
[
Q_{\mathrm{coverage}},
Q_{\mathrm{behavior}},
Q_{\mathrm{uncertainty}},
Q_{\mathrm{potential}}
].
$$

For an action $A$, define its effect on the unresolved state as:

$$
\mathcal Q_R
\rightarrow
\mathcal Q_R^{A}.
$$

The action is useful when it is expected to reduce one or more
components of the unresolved state.

Conceptually, define the refinement effect as:

$$
\Delta\mathcal Q(A\mid R)
=
\mathcal Q_R-\mathcal Q_R^{A}.
$$

Because the exact post-action state is unknown before refinement, ARRGO
uses an estimated effect:

$$
\widehat{\Delta\mathcal Q}(A\mid R).
$$

For sampling:

$$
A=\text{SAMPLE}.
$$

For splitting:

$$
A=\text{SPLIT}.
$$

The two actions are therefore evaluated against the same refinement
objective rather than against their raw information-gain values.

The current dominant unresolved objective is determined from the region
state:

$$
O_R^*
=
\operatorname{DominantObjective}(\mathcal Q_R).
$$

ARRGO then asks which action is more relevant to that objective.

Conceptually:

$$
A_R^*
=
\operatorname{SelectAction}
\left(
\text{SAMPLE},
\text{SPLIT},
O_R^*
\right).
$$

If the dominant unresolved information concerns function values or
local behavior inside an already suitable spatial structure, sampling
becomes the natural refinement action.

If the dominant unresolved information concerns spatial separation or
structural heterogeneity, splitting becomes the natural refinement
action.

If neither action is expected to meaningfully change the relevant
decision, the region may become stable.

Thus, the comparison is not:

$$
\text{Which action has the larger arbitrary score?}
$$

Instead, it is:

$$
\boxed{
\text{Which action better addresses the current unresolved objective?}
}
$$

This design avoids introducing arbitrary fixed weights between
fundamentally different information dimensions.

The action-selection mechanism can later be formalized once the
resolution criteria for each objective have been fully defined.

The resulting decision process is:

$$
\text{Region State}
\rightarrow
\text{Unresolved Objective}
\rightarrow
\text{Evaluate Feasible Actions}
\rightarrow
\text{Select Action}
\rightarrow
\text{Refine}
\rightarrow
\text{Re-analyze}.
$$

## Dominant Unresolved Objective

A region may contain several types of unresolved information at the same
time.

For example, a region may have:

- poor spatial coverage,
- unresolved behavioral changes,
- high certified uncertainty,
- or substantial optimization potential.

Therefore, the region cannot be described by a single type of
uncertainty.

Let the unresolved-information state be:

$$
\mathcal Q_R
=
[
Q_{\mathrm{coverage}},
Q_{\mathrm{behavior}},
Q_{\mathrm{uncertainty}},
Q_{\mathrm{potential}}
].
$$

Each component represents the degree to which a specific aspect of the
region remains unresolved.

For example:

$$
Q_{\mathrm{coverage}}
$$

represents unresolved spatial coverage.

Similarly:

$$
Q_{\mathrm{behavior}}
$$

represents unresolved behavioral structure,

$$
Q_{\mathrm{uncertainty}}
$$

represents unresolved certified uncertainty, and

$$
Q_{\mathrm{potential}}
$$

represents unresolved optimization potential.

The dominant unresolved objective is the aspect that currently has the
greatest decision relevance.

Conceptually:

$$
O_R^*
=
\operatorname{DominantObjective}
(\mathcal Q_R).
$$

However, ARRGO does not assume that the components of
$\mathcal Q_R$ are directly comparable numerical quantities.

Instead, dominance can be determined through objective-specific
resolution criteria.

For example, if spatial coverage is insufficient while the other
objectives are already stable, then:

$$
O_R^*
=
O_{\mathrm{coverage}}.
$$

If the main unresolved issue is a behavioral transition, then:

$$
O_R^*
=
O_{\mathrm{behavior}}.
$$

Under a valid uncertainty model, if the certified bounds remain too
wide, then:

$$
O_R^*
=
O_{\mathrm{uncertainty}}.
$$

If the region still has substantial certified potential to improve the
current best solution, then:

$$
O_R^*
=
O_{\mathrm{potential}}.
$$

More than one objective may also remain unresolved simultaneously.

In this case, ARRGO does not need to force the region into a single
permanent category.

Instead, the unresolved state can be represented as:

$$
\mathcal Q_R
=
\{
Q_k(R):Q_k(R)>\text{ResolutionCriterion}_k
\}.
$$

The active refinement objective may then change after every new
observation or structural refinement.

Therefore:

$$
\mathcal Q_R
\rightarrow
\mathcal Q_R'
$$

after an action, and consequently:

$$
O_R^*
\rightarrow
O_R^{*\prime}.
$$

This makes the refinement strategy adaptive to the evolving information
state of the region.

The important principle is:

$$
\boxed{
\text{The refinement objective is a property of the current
information state, not a permanent property of the region.}
}
$$

Thus, a region that initially requires sampling may later require
splitting, and a region that initially requires splitting may later
become primarily uncertainty-limited.

This allows ARRGO to change its refinement strategy as information
accumulates.

## Action Selection Rule

After identifying the current unresolved objective of a region, ARRGO must
select an appropriate refinement action.

The available actions are:

$$
\mathcal A_R
=
\{
\text{SAMPLE},
\text{SPLIT},
\text{STABLE}
\}.
$$

Let the current refinement objective be:

$$
O_R^*
=
\operatorname{DominantObjective}(\mathcal Q_R).
$$

For each feasible action $A\in\mathcal A_R$, ARRGO evaluates whether the
action can meaningfully address the current unresolved objective.

The action-selection process can therefore be represented as:

$$
A_R^*
=
\operatorname{SelectAction}
\left(
R,O_R^*
\right).
$$

Sampling is considered when an additional function evaluation can provide
information relevant to the current unresolved objective.

Splitting is considered when increasing the spatial resolution can provide
information relevant to the current unresolved objective.

The STABLE action is considered only when the available information is
sufficiently decision-stable with respect to the relevant objectives.

Conceptually, the decision process is:

$$
\begin{aligned}
&\text{If unresolved information is primarily informational:}
&&\rightarrow \text{SAMPLE},\\
&\text{If unresolved information is primarily structural:}
&&\rightarrow \text{SPLIT},\\
&\text{If relevant decisions are stable:}
&&\rightarrow \text{STABLE}.
\end{aligned}
$$

However, ARRGO does not permanently associate one objective with one
action.

A region may require sampling at one stage and splitting at a later
stage.

Therefore, action selection is recomputed after every refinement:

$$
S_R
\rightarrow
S_R'
\rightarrow
A_R^{*\prime}.
$$

The selected action must also satisfy all structural and computational
constraints.

For splitting, the candidate split must satisfy the contraction
condition:

$$
\max(s-l,r-s)
\leq
\rho(r-l),
\qquad
\rho<1.
$$

For sampling, the selected point must belong to the region and must not
duplicate an existing observation.

The action-selection mechanism therefore separates three concepts:

$$
\boxed{
\text{Objective}
}
$$

identifies what information is unresolved,

$$
\boxed{
\text{Action}
}
$$

identifies how the algorithm attempts to resolve it,

and:

$$
\boxed{
\text{Constraint}
}
$$

defines what actions are admissible.

Thus:

$$
\boxed{
\text{Objective}
\rightarrow
\text{Action}
\rightarrow
\text{Admissibility}
\rightarrow
\text{Refinement}
}
$$

After refinement, ARRGO re-evaluates the region state rather than
assuming that the selected action was successful.

This creates a closed adaptive loop:

$$
\boxed{
\text{Analyze}
\rightarrow
\text{Identify Objective}
\rightarrow
\text{Select Action}
\rightarrow
\text{Refine}
\rightarrow
\text{Observe Effect}
\rightarrow
\text{Re-analyze}
}
$$

The effectiveness of an action is therefore determined by its observed
effect on the subsequent information state, rather than by assuming in
advance that a particular action is always superior.

## Stable State

A region is considered stable when the currently available information is
sufficiently stable for the decisions relevant to that region.

Stability is therefore not defined by the number of observations alone.

Let the state of a region be:

$$
S_R
=
(
\mathcal D_R,
B_R,
\mathcal Q_R
).
$$

For each relevant objective $k$, ARRGO evaluates whether the corresponding
decision remains stable under additional information.

Let:

$$
\operatorname{Decision}_k(S_R)
$$

represent the current decision associated with objective $k$.

A decision is considered stable when additional admissible information is
unlikely to change the decision in a meaningful way.

Conceptually:

$$
\operatorname{Decision}_k(S_R)
\approx
\operatorname{Decision}_k(S_R^{+}),
$$

where $S_R^{+}$ represents a sufficiently informative updated state.

Therefore, stability is objective-specific.

For coverage:

$$
\operatorname{Stable}_{\mathrm{coverage}}(R)
$$

means that the current spatial representation is sufficiently resolved
for the relevant refinement decision.

For behavior:

$$
\operatorname{Stable}_{\mathrm{behavior}}(R)
$$

means that the observed behavioral structure is sufficiently consistent
for the current decision.

Under a valid uncertainty model:

$$
\operatorname{Stable}_{\mathrm{uncertainty}}(R)
$$

means that the remaining certified uncertainty is sufficiently small for
the relevant decision.

Similarly:

$$
\operatorname{Stable}_{\mathrm{potential}}(R)
$$

means that the optimization potential is sufficiently resolved for the
current global optimization decision.

The overall stability state can therefore be represented as:

$$
\mathbf S_R
=
[
S_{\mathrm{coverage}},
S_{\mathrm{behavior}},
S_{\mathrm{uncertainty}},
S_{\mathrm{potential}}
].
$$

A region may be stable with respect to one objective while remaining
unstable with respect to another.

For example:

$$
S_{\mathrm{coverage}}=\text{stable}
$$

while:

$$
S_{\mathrm{behavior}}=\text{unstable}.
$$

In this situation, the region must remain active because unresolved
behavioral information can still affect the optimization process.

Thus, ARRGO does not define stability as:

$$
\text{Number of Samples}\geq N.
$$

Instead:

$$
\boxed{
\text{Stable}
=
\text{Sufficient Decision Stability}
}
$$

The region becomes globally stable only when all objectives that are
relevant to the current optimization decision satisfy their corresponding
resolution criteria.

Conceptually:

$$
\operatorname{Stable}(R)
=
\bigwedge_{k\in\mathcal K_R}
\operatorname{Stable}_k(R),
$$

where $\mathcal K_R$ is the set of objectives currently relevant to the
region.

When a region becomes stable, it is not removed from the region
hierarchy.

Instead, its state is changed:

$$
\text{ACTIVE}
\rightarrow
\text{STABLE}.
$$

The region remains available as part of the global optimization
structure.

If new global information later changes the relevance of the region,
ARRGO may re-evaluate its state.

Therefore, stability is not equivalent to permanent exclusion.

The complete state transition is:

$$
\boxed{
\text{Active}
\rightarrow
\text{Stable}
\rightarrow
\text{Re-evaluate if necessary}
}
$$

This preserves the no-pruning principle while allowing ARRGO to stop
performing unnecessary refinements on regions whose relevant decisions
are already sufficiently stable.

## Global Region Selection

As ARRGO refines the search domain, multiple regions may exist
simultaneously.

Let the current collection of regions be:

$$
\mathcal R
=
\{R_1,R_2,\ldots,R_m\}.
$$

Each region has its own information state, unresolved objectives, and
refinement history.

Therefore, ARRGO must determine which region should receive the next
refinement action.

This is a global selection problem over the current region hierarchy.

For each region $R$, define its unresolved-information state as:

$$
\mathcal Q_R
=
[
Q_{\mathrm{coverage}},
Q_{\mathrm{behavior}},
Q_{\mathrm{uncertainty}},
Q_{\mathrm{potential}}
].
$$

The global selection mechanism evaluates the regions according to the
importance of their unresolved information.

Conceptually, define:

$$
R^*
=
\operatorname{SelectRegion}
(\mathcal R).
$$

The selected region is the region for which further refinement is
currently most relevant to the global optimization process.

ARRGO does not require a single fixed numerical priority score.

Instead, region selection can use a hierarchical decision process.

First, regions that are currently stable are not selected for immediate
refinement.

Among the remaining active regions, ARRGO considers their unresolved
objectives.

Regions with unresolved optimization potential receive attention when
their current information still permits a meaningful possibility of
improving the incumbent.

Regions with unresolved uncertainty receive attention when additional
information can substantially clarify the current optimization state.

Regions with unresolved behavioral structure receive attention when the
observed function behavior remains insufficiently resolved.

Regions with poor spatial coverage receive attention when additional
spatial refinement is necessary to distinguish local structures.

The global selection process can therefore be represented as:

$$
\text{Active Regions}
\rightarrow
\text{Evaluate Unresolved Information}
\rightarrow
\text{Identify Relevant Objectives}
\rightarrow
\text{Compare Regions}
\rightarrow
R^*.
$$

The comparison between regions must preserve the distinction between
different types of information.

For example, a region with high uncertainty is not automatically more
important than a region with lower uncertainty but substantially greater
optimization potential.

Therefore, ARRGO evaluates region relevance with respect to the current
global optimization objective.

Let:

$$
f_{\mathrm{best}}
$$

be the best objective value observed so far.

Under a valid Lipschitz uncertainty model, each region can also have a
certified upper potential:

$$
P(R)
=
\max_{x\in R}U_R(x).
$$

The quantity:

$$
P(R)-f_{\mathrm{best}}
$$

indicates how much improvement the current information still permits in
that region.

This information can influence global region selection without
discarding regions whose current potential is low.

Thus:

$$
\boxed{
\text{Selection}
\neq
\text{Pruning}
}
$$

Region selection only determines which region receives the next
refinement action.

All regions remain part of the global hierarchy.

After refinement, the global state is updated:

$$
\mathcal R
\rightarrow
\mathcal R'.
$$

The next region is then selected again from the updated state.

Therefore, global region selection is dynamic:

$$
R_t^*
=
\operatorname{SelectRegion}
(\mathcal R_t),
$$

followed by:

$$
\mathcal R_t
\rightarrow
\mathcal R_{t+1}.
$$

This allows ARRGO to change its attention as new information becomes
available.

The complete global control loop is:

$$
\boxed{
\text{Select Region}
\rightarrow
\text{Analyze Region}
\rightarrow
\text{Select Action}
\rightarrow
\text{Refine}
\rightarrow
\text{Update Global State}
\rightarrow
\text{Select Region Again}
}
$$

The region-selection mechanism is therefore adaptive and state-dependent,
rather than based on a fixed traversal order such as left-to-right or
depth-first search.

## Global Region Priority

After generating multiple active regions, ARRGO must determine which
region should be refined next.

Let the active regions be:

$$
\mathcal R_{\mathrm{active}}
=
\{R_1,R_2,\ldots,R_m\}.
$$

Each region has an unresolved-information profile:

$$
\mathcal Q_R
=
[
Q_{\mathrm{coverage}},
Q_{\mathrm{behavior}},
Q_{\mathrm{uncertainty}},
Q_{\mathrm{potential}}
].
$$

The priority of a region represents its current relevance to the global
optimization process.

It is important to distinguish region priority from the local refinement
objective.

The local objective answers:

$$
\boxed{
\text{What information is unresolved inside this region?}
}
$$

while global priority answers:

$$
\boxed{
\text{Which region should receive attention next?}
}
$$

Therefore:

$$
\text{Local Objective}
\neq
\text{Global Priority}.
$$

A region may have high unresolved information but low global relevance,
while another region may have less uncertainty but substantially greater
optimization relevance.

Under a valid Lipschitz uncertainty model, optimization potential provides
one important source of global relevance:

$$
P(R)
=
\max_{x\in R}U_R(x).
$$

With the current incumbent:

$$
f_{\mathrm{best}}
=
\max_{x\in\mathcal D}f(x),
$$

the remaining certified improvement potential is:

$$
\Delta_R
=
P(R)-f_{\mathrm{best}}.
$$

A positive value indicates that the current information still permits the
possibility of improving the incumbent within the region.

However, optimization potential is not the only source of priority.

Other regions may require attention because their spatial or behavioral
structure remains unresolved.

Therefore, ARRGO represents region priority through a collection of
relevant criteria rather than a single arbitrary weighted score.

Define the region-priority profile as:

$$
\mathbf P_R
=
[
P_{\mathrm{optimization}},
P_{\mathrm{uncertainty}},
P_{\mathrm{behavior}},
P_{\mathrm{coverage}}
].
$$

These components describe different reasons why a region may deserve
attention.

A region can then be compared with another region using the same
multi-criteria dominance principle used for sampling points and split
locations.

Region $R_1$ dominates region $R_2$ if:

$$
P_k(R_1)
\geq
P_k(R_2)
$$

for every priority dimension $k$, with strict inequality in at least one
dimension.

The non-dominated active regions are:

$$
\mathcal R_{\mathrm{ND}}
=
\left\{
R\in\mathcal R_{\mathrm{active}}
\mid
\nexists Z\in\mathcal R_{\mathrm{active}}
\text{ such that }Z\succ R
\right\}.
$$

This prevents ARRGO from declaring one region globally superior merely
because of an arbitrary numerical weighting between fundamentally
different information dimensions.

If multiple non-dominated regions remain, ARRGO uses the current global
optimization objective and refinement context to determine which region
should receive the next action.

Thus, global region selection follows:

$$
\boxed{
\text{Active Regions}
\rightarrow
\text{Build Priority Profiles}
\rightarrow
\text{Remove Dominated Candidates}
\rightarrow
\text{Apply Global Objective}
\rightarrow
\text{Select Region}
}
$$

The term "remove" in the dominance stage refers only to removing a region
from the current selection candidate set.

It does not remove the region from the global region hierarchy.

Therefore:

$$
\boxed{
\text{Selection Filtering}
\neq
\text{Region Pruning}
}
$$

Every region remains part of the global search structure.

After the selected region is refined, its priority profile and the
priority profiles of other regions are recomputed because the global
incumbent and information state may have changed.

Thus:

$$
\mathbf P_R^{(t)}
\rightarrow
\mathbf P_R^{(t+1)}.
$$

Global priority is therefore dynamic and depends on the current state of
the entire optimization process.

The complete mechanism is:

$$
\boxed{
\text{Global State}
\rightarrow
\text{Region Priority}
\rightarrow
\text{Region Selection}
\rightarrow
\text{Local Refinement}
\rightarrow
\text{Global Update}
\rightarrow
\text{Re-prioritization}
}
$$

This allows ARRGO to continuously redirect computational effort toward
the regions that are currently most relevant to resolving the global
optimization problem.


## Global Optimization Objective

ARRGO operates over the entire search domain rather than optimizing each
region independently.

Let the search domain be:

$$
\Omega=[a,b].
$$

At iteration $t$, let the current collection of regions be:

$$
\mathcal R_t.
$$

Each region has its own local information state, but the ultimate
objective of ARRGO is global:

$$
\max_{x\in\Omega}f(x).
$$

Therefore, local refinement decisions must remain connected to the global
optimization objective.

Let the best currently observed solution be:

$$
x_{\mathrm{best}}
=
\arg\max_{x\in\mathcal D}f(x),
$$

with:

$$
f_{\mathrm{best}}
=
f(x_{\mathrm{best}}).
$$

The purpose of further refinement is to determine whether additional
information can:

1. improve the current best solution,
2. increase confidence in the current best solution,
3. or resolve spatial or behavioral information that may affect the
   identification of the global optimum.

Therefore, the global optimization objective can be viewed as a hierarchy
of information requirements.

The primary objective is:

$$
\boxed{
\text{Identify the best solution over the entire domain}
}
$$

while supporting objectives include:

$$
\boxed{
\text{Resolve information that can affect this decision}
}
$$

and:

$$
\boxed{
\text{Maintain sufficient spatial and behavioral resolution}
}
$$

A region is globally relevant when its unresolved information can still
affect the global optimization decision.

Under a valid Lipschitz uncertainty model, one important indicator of this
relevance is the region's certified upper potential:

$$
P(R)
=
\max_{x\in R}U_R(x).
$$

If:

$$
P(R)>f_{\mathrm{best}},
$$

then the current information does not rule out the existence of a point
inside the region that is better than the incumbent.

Such a region remains globally relevant.

If:

$$
P(R)\leq f_{\mathrm{best}},
$$

the current certified information does not permit an improvement over the
incumbent.

However, ARRGO does not remove the region from the search structure.

Instead, its current optimization relevance is reduced.

Thus:

$$
\boxed{
\text{Low Global Relevance}
\neq
\text{Pruned Region}
}
$$

The global objective therefore determines how local region priorities are
interpreted.

A region with high local uncertainty may still have lower global
importance than a region whose unresolved information directly concerns
the current best-solution decision.

Consequently, ARRGO separates:

$$
\text{Local Information Resolution}
$$

from:

$$
\text{Global Optimization Relevance}.
$$

The complete relationship is:

$$
\boxed{
\text{Global Objective}
\rightarrow
\text{Global Relevance}
\rightarrow
\text{Region Priority}
\rightarrow
\text{Local Objective}
\rightarrow
\text{Refinement Action}
}
$$

After every refinement, the global state is updated:

$$
(\mathcal R_t,f_{\mathrm{best}}^{(t)})
\rightarrow
(\mathcal R_{t+1},f_{\mathrm{best}}^{(t+1)}).
$$

Region priorities are then recomputed using the updated global state.

Therefore, the global optimization objective is not a one-time
criterion.

It continuously influences the allocation of refinement effort throughout
the search.

The resulting control structure is:

$$
\boxed{
\text{Global Objective}
\rightarrow
\text{Evaluate Global State}
\rightarrow
\text{Prioritize Regions}
\rightarrow
\text{Refine Selected Region}
\rightarrow
\text{Update Global State}
\rightarrow
\text{Repeat}
}
$$

This ensures that local adaptive refinement remains aligned with the
global optimization problem.

## Global Best Solution and Incumbent Update

ARRGO maintains the best solution found from all function evaluations
performed during the optimization process.

Let the set of all evaluated points up to iteration $t$ be:

$$
\mathcal D_t
=
\{(x_i,f(x_i))\}_{i=1}^{N_t}.
$$

The current best solution, also called the incumbent, is:

$$
x_{\mathrm{best}}^{(t)}
=
\arg\max_{x_i\in\mathcal D_t}f(x_i).
$$

The corresponding objective value is:

$$
f_{\mathrm{best}}^{(t)}
=
f(x_{\mathrm{best}}^{(t)}).
$$

The incumbent represents the best solution that ARRGO has actually
observed so far.

It is therefore important to distinguish the observed best value from
the unknown global optimum:

$$
f_{\mathrm{best}}^{(t)}
\leq
f^*,
$$

where:

$$
f^*
=
\max_{x\in\Omega}f(x).
$$

The inequality does not imply that the incumbent is globally optimal.

It only states that the current incumbent is the best value among the
points evaluated so far.

Whenever ARRGO performs a new function evaluation:

$$
x_{\mathrm{new}}\in\Omega,
$$

it obtains:

$$
y_{\mathrm{new}}
=
f(x_{\mathrm{new}}).
$$

The global observation set is then updated:

$$
\mathcal D_{t+1}
=
\mathcal D_t
\cup
\{(x_{\mathrm{new}},y_{\mathrm{new}})\}.
$$

The incumbent is recomputed as:

$$
x_{\mathrm{best}}^{(t+1)}
=
\arg\max_{x_i\in\mathcal D_{t+1}}f(x_i).
$$

and:

$$
f_{\mathrm{best}}^{(t+1)}
=
f(x_{\mathrm{best}}^{(t+1)}).
$$

If:

$$
y_{\mathrm{new}}
>
f_{\mathrm{best}}^{(t)},
$$

the new observation becomes the incumbent.

Otherwise, the incumbent remains unchanged.

The incumbent is global rather than region-specific.

Therefore, a sample obtained inside any region can improve the global
solution:

$$
R_j
\rightarrow
(x_{\mathrm{new}},y_{\mathrm{new}})
\rightarrow
f_{\mathrm{best}}^{(t+1)}.
$$

This global update can change the relevance of every region.

For example, under a valid Lipschitz uncertainty model, the optimization
potential of a region is:

$$
\Delta_R
=
P(R)-f_{\mathrm{best}}.
$$

When the incumbent increases:

$$
f_{\mathrm{best}}^{(t+1)}
>
f_{\mathrm{best}}^{(t)},
$$

the potential of multiple regions may decrease:

$$
\Delta_R^{(t+1)}
<
\Delta_R^{(t)}.
$$

Therefore, an observation in one region can change the priorities of
other regions.

This creates an important global feedback mechanism:

$$
\boxed{
\text{New Observation}
\rightarrow
\text{Incumbent Update}
\rightarrow
\text{Global Re-evaluation}
\rightarrow
\text{Region Re-prioritization}
}
$$

The incumbent is therefore not only a record of the best observed
solution.

It also acts as a global reference value against which the remaining
optimization potential of regions can be evaluated.

Under a valid uncertainty model, if a region satisfies:

$$
P(R)\leq f_{\mathrm{best}},
$$

then the current information certifies that the region cannot contain a
solution better than the incumbent.

Nevertheless, ARRGO does not delete the region.

Its state is simply updated according to the new global information.

Thus:

$$
\boxed{
\text{Incumbent Update}
\neq
\text{Region Pruning}
}
$$

The complete incumbent mechanism is:

$$
\boxed{
\text{Evaluate}
\rightarrow
\text{Update Global Dataset}
\rightarrow
\text{Update Incumbent}
\rightarrow
\text{Update Region Information}
\rightarrow
\text{Re-prioritize}
}
$$

This ensures that every new function evaluation can influence the global
optimization process, regardless of the region in which the evaluation
was performed.

## Global State Update

ARRGO is an adaptive optimization framework in which every refinement
can change both the local state of a region and the global state of the
optimization process.

Let the global state at iteration $t$ be represented by:

$$
\mathcal G_t
=
(
\mathcal R_t,
\mathcal D_t,
x_{\mathrm{best}}^{(t)},
f_{\mathrm{best}}^{(t)}
).
$$

Here:

$$
\mathcal R_t
$$

is the current region hierarchy,

$$
\mathcal D_t
$$

is the set of all function evaluations,

and:

$$
x_{\mathrm{best}}^{(t)},
f_{\mathrm{best}}^{(t)}
$$

represent the current incumbent solution and objective value.

After a refinement action, the global state becomes:

$$
\mathcal G_t
\rightarrow
\mathcal G_{t+1}.
$$

The update process depends on the type of refinement.

For sampling, a new function evaluation is obtained:

$$
x_{\mathrm{new}}\rightarrow y_{\mathrm{new}}=f(x_{\mathrm{new}}).
$$

The global observation set is updated:

$$
\mathcal D_{t+1}
=
\mathcal D_t
\cup
\{(x_{\mathrm{new}},y_{\mathrm{new}})\}.
$$

The incumbent is then updated:

$$
f_{\mathrm{best}}^{(t+1)}
=
\max_{(x,y)\in\mathcal D_{t+1}}y.
$$

The corresponding point becomes:

$$
x_{\mathrm{best}}^{(t+1)}
=
\arg\max_{(x,y)\in\mathcal D_{t+1}}y.
$$

For splitting, no new function evaluation is necessarily required.

Instead, the region hierarchy is updated:

$$
\mathcal R_t
\rightarrow
\mathcal R_{t+1}.
$$

The parent region remains part of the hierarchy, while the newly created
subregions are added to the active region structure.

Existing observations are reused whenever they belong to a child region.

After either type of refinement, the affected region information must be
recomputed.

This may include:

$$
B_R,
\mathcal Q_R,
\mathcal U(R),
P(R),
$$

and the corresponding refinement objectives.

Because the incumbent is global, a change in:

$$
f_{\mathrm{best}}
$$

may also affect the optimization potential of other regions.

Therefore, global priorities must be reconsidered after an incumbent
update.

The global update process can be represented as:

$$
\boxed{
\text{Local Refinement}
\rightarrow
\text{Update Observations or Structure}
\rightarrow
\text{Update Incumbent}
\rightarrow
\text{Recompute Affected Region States}
\rightarrow
\text{Update Global Priorities}
}
$$

Not every quantity must be recomputed from scratch after every action.

ARRGO can distinguish between:

$$
\text{Local Updates}
$$

and:

$$
\text{Global Updates}.
$$

A local update modifies information directly affected by the refinement.

A global update modifies quantities whose values depend on information
shared across the optimization process.

For example, adding a new observation directly changes the state of its
region, while an improved incumbent can change the optimization
potential of multiple regions.

This distinction allows ARRGO to remain adaptive without requiring every
internal quantity to be recomputed unnecessarily.

The global state therefore evolves through the following cycle:

$$
\boxed{
\mathcal G_t
\rightarrow
\text{Refinement}
\rightarrow
\mathcal G_{t+1}
\rightarrow
\text{Re-analysis}
}
$$

The updated global state becomes the basis for the next region-selection
decision.

Thus, no refinement decision is treated as permanently valid.

Every action can change the information landscape of the entire search
process.

The complete global feedback mechanism is:

$$
\boxed{
\text{Select Region}
\rightarrow
\text{Select Action}
\rightarrow
\text{Refine}
\rightarrow
\text{Update Global State}
\rightarrow
\text{Recompute Priorities}
\rightarrow
\text{Select Again}
}
$$

This feedback mechanism is a central component of the adaptive design of
ARRGO.

## Information Sharing Between Regions

When a region is split, the resulting subregions are not independent
optimization problems.

They are parts of the same global search domain and share information
obtained during the optimization process.

Let:

$$
R=[l,r]
$$

be a parent region that is split at $s$:

$$
R_L=[l,s],
$$

and:

$$
R_R=[s,r].
$$

The observations previously collected in the parent region are:

$$
\mathcal D_R
=
\{(x_i,y_i)\}.
$$

An observation belongs to a child region whenever its location satisfies
the corresponding child-region bounds.

Therefore, the local datasets of the children can be initialized from the
parent observations:

$$
\mathcal D_{R_L}
=
\{(x_i,y_i)\in\mathcal D_R:x_i\in R_L\},
$$

and:

$$
\mathcal D_{R_R}
=
\{(x_i,y_i)\in\mathcal D_R:x_i\in R_R\}.
$$

This allows ARRGO to reuse previously evaluated function values instead
of evaluating the same points again.

In addition to local observations, some information is inherently global.

The incumbent:

$$
x_{\mathrm{best}},
\qquad
f_{\mathrm{best}}
$$

is shared across all regions.

Therefore, every region is evaluated relative to the same current best
solution.

Under a valid global Lipschitz assumption, the same valid Lipschitz
constant $L$ is also available to all regions:

$$
|f(x)-f(z)|\leq L|x-z|.
$$

This allows each region to construct its own local uncertainty bounds
while remaining consistent with the same global assumption.

ARRGO therefore distinguishes between two types of information:

$$
\boxed{
\text{Local Information}
}
$$

and:

$$
\boxed{
\text{Global Information}
}
$$

Local information includes:

- observations located inside the region,
- local secant slopes,
- local directional behavior,
- local sampling gaps,
- local uncertainty,
- local optimization potential.

Global information includes:

- the complete set of evaluated observations,
- the current incumbent,
- the global evaluation budget,
- the global search domain,
- and any globally valid assumptions used by the algorithm.

A refinement in one region can therefore affect another region indirectly.

For example, suppose a new observation produces an improved incumbent:

$$
f_{\mathrm{best}}^{(t+1)}
>
f_{\mathrm{best}}^{(t)}.
$$

The optimization potential of other regions may then change:

$$
P(R_j)
\rightarrow
P(R_j)-f_{\mathrm{best}}^{(t+1)}.
$$

Thus, information does not need to be physically located inside a region
to affect its global decision state.

The information flow can be represented as:

$$
\boxed{
\text{Local Observation}
\rightarrow
\text{Global State}
\rightarrow
\text{Other Region Priorities}
}
$$

At the same time, local structural information should not be blindly
copied between unrelated regions.

For example, an observed directional pattern inside one region does not
imply that the same pattern exists in another region.

Therefore:

$$
\boxed{
\text{Global Information is Shared}
}
$$

while:

$$
\boxed{
\text{Local Behavioral Evidence remains Local}
}
$$

This distinction prevents ARRGO from treating observations from one
region as direct evidence about the unknown function behavior of another
region.

The resulting information architecture is:

$$
\boxed{
\text{Local Observations}
\rightarrow
\text{Local Analysis}
\rightarrow
\text{Global State}
\rightarrow
\text{Global Re-prioritization}
}
$$

while observations are reused structurally whenever a parent region is
split.

This allows ARRGO to preserve information, avoid redundant evaluations,
and maintain a consistent global optimization state throughout the
adaptive refinement process.

## Region Hierarchy and Parent-Child Relationship

ARRGO represents the adaptive search process as a hierarchy of regions.

The initial search domain is represented by a root region:

$$
R_0=\Omega.
$$

When a region is split, it produces two child regions.

For:

$$
R=[l,r],
$$

and split point:

$$
s\in(l,r),
$$

the resulting children are:

$$
R_L=[l,s],
$$

and:

$$
R_R=[s,r].
$$

The relationship can therefore be represented as:

$$
R
\rightarrow
\{R_L,R_R\}.
$$

The parent region remains part of the hierarchy after the split.

This means that ARRGO does not replace the parent with its children.

Instead, the hierarchy preserves both the historical region and the
spatial decomposition produced from it.

The resulting structure can be represented as a tree:

$$
R_0
\rightarrow
\{R_1,R_2\}
\rightarrow
\{R_3,R_4,\ldots\}.
$$

Each region therefore has a structural relationship with its parent and,
when applicable, with its children.

A region can be represented conceptually as:

$$
R=
(
[l,r],
\mathcal D_R,
B_R,
\mathcal Q_R,
\text{state},
\text{parent},
\text{children}
).
$$

The parent-child relationship provides several important capabilities.

First, observations can be reused after splitting.

If an observation:

$$
(x_i,y_i)
$$

was already evaluated inside a parent region, it does not need to be
evaluated again when the parent is divided.

Second, the hierarchy records how spatial resolution evolved during the
optimization process.

A deeper region represents a finer spatial decomposition than one of its
ancestors.

Third, the hierarchy preserves the history of refinement decisions.

For example, ARRGO can determine:

$$
\text{which region was split},
$$

$$
\text{where it was split},
$$

and:

$$
\text{what information was available at the time of splitting}.
$$

This information can later be used to analyze the effectiveness of
different refinement decisions.

The hierarchy also separates structural history from active computation.

A region may no longer be the current refinement target while still
remaining part of the global structure.

Thus:

$$
\boxed{
\text{Historical Region}
\neq
\text{Discarded Region}
}
$$

Similarly, when a parent region is split:

$$
R\rightarrow R_L\cup R_R,
$$

the children become new spatial units for subsequent refinement, while
the parent remains available as part of the hierarchy.

The no-pruning principle is therefore enforced structurally:

$$
\boxed{
\text{Every created region remains represented in the hierarchy.}
}
$$

This does not require every region to be refined indefinitely.

A region may enter the stable state:

$$
\text{ACTIVE}
\rightarrow
\text{STABLE},
$$

while remaining part of the hierarchy.

If later global information makes further refinement relevant, its state
can be reconsidered.

The hierarchy therefore acts as both:

$$
\boxed{
\text{Search Structure}
}
$$

and:

$$
\boxed{
\text{Refinement History}
}
$$

The complete structural evolution is:

$$
\boxed{
\text{Root Region}
\rightarrow
\text{Adaptive Splitting}
\rightarrow
\text{Parent-Child Hierarchy}
\rightarrow
\text{Finer Spatial Regions}
}
$$

This hierarchical representation allows ARRGO to preserve information,
reuse evaluations, track refinement history, and maintain the no-pruning
principle throughout the optimization process.

## Region Lifecycle

Each region in ARRGO evolves through a sequence of information and
structural states during the optimization process.

A newly created region begins as an active region:

$$
\text{ACTIVE}.
$$

An active region is a region that may still require additional
information or structural refinement.

The region is repeatedly analyzed to determine its current information
state.

The analysis produces:

$$
S_R
=
(
\mathcal D_R,
B_R,
\mathcal Q_R
).
$$

Based on this state, ARRGO selects one of the available refinement
actions:

$$
\mathcal A_R
=
\{
\text{SAMPLE},
\text{SPLIT},
\text{STABLE}
\}.
$$

If the selected action is sampling:

$$
\text{ACTIVE}
\xrightarrow{\text{SAMPLE}}
\text{ACTIVE}.
$$

The region remains active because its information state has changed and
must be re-analyzed.

The new observation updates:

$$
\mathcal D_R
\rightarrow
\mathcal D_R^{+}.
$$

The behavioral and information profiles are then recomputed.

If the selected action is splitting:

$$
\text{ACTIVE}
\xrightarrow{\text{SPLIT}}
\text{REFINED}.
$$

The parent region becomes structurally refined and produces child
regions:

$$
R
\rightarrow
\{R_L,R_R\}.
$$

The children begin their own active refinement processes:

$$
R_L,R_R
\rightarrow
\text{ACTIVE}.
$$

The parent remains in the hierarchy and retains its refinement history.

If the relevant information of a region becomes sufficiently
decision-stable:

$$
\text{ACTIVE}
\xrightarrow{\text{STABLE}}
\text{STABLE}.
$$

A stable region does not require immediate further refinement.

However, stable does not mean deleted.

The region remains part of the global hierarchy:

$$
\text{STABLE}
\in
\mathcal R.
$$

Its state may be reconsidered if later global information changes its
relevance.

Conceptually:

$$
\text{STABLE}
\xrightarrow{\text{Re-evaluate}}
\text{ACTIVE}
$$

is allowed when the global state changes sufficiently to make further
refinement relevant.

Therefore, the lifecycle is not necessarily a one-way sequence.

The general lifecycle can be represented as:

$$
\boxed{
\text{ACTIVE}
\rightarrow
\begin{cases}
\text{ACTIVE} & \text{after sampling},\\
\text{REFINED} & \text{after splitting},\\
\text{STABLE} & \text{after sufficient resolution}.
\end{cases}
}
$$

For a split:

$$
\boxed{
\text{REFINED Parent}
\rightarrow
\text{ACTIVE Children}
}
$$

The complete hierarchy may therefore contain regions in different states
simultaneously.

For example:

$$
\mathcal R
=
\{
R_1^{\mathrm{STABLE}},
R_2^{\mathrm{ACTIVE}},
R_3^{\mathrm{REFINED}},
R_4^{\mathrm{ACTIVE}}
\}.
$$

This is a natural consequence of adaptive refinement because different
regions can require different amounts and types of information.

The lifecycle separates three concepts:

$$
\boxed{
\text{Information Update}
}
$$

which occurs through sampling,

$$
\boxed{
\text{Structural Update}
}
$$

which occurs through splitting,

and:

$$
\boxed{
\text{Resolution}
}
$$

which determines whether a region currently requires further refinement.

Thus, the region lifecycle is driven by information state rather than
by a predetermined number of iterations.

The complete lifecycle is:

$$
\boxed{
\text{Create}
\rightarrow
\text{Analyze}
\rightarrow
\text{Refine or Stabilize}
\rightarrow
\text{Update}
\rightarrow
\text{Re-analyze}
}
$$

This lifecycle allows ARRGO to continuously adapt the treatment of each
region while preserving the complete search history and maintaining the
no-pruning principle.

## Termination Criteria

ARRGO must determine when the global optimization process should
terminate.

Termination should not depend solely on a fixed number of iterations or
a fixed number of samples.

Instead, termination should be related to whether further refinement can
still meaningfully affect the global optimization decision.

Let the global state at iteration $t$ be:

$$
\mathcal G_t
=
(
\mathcal R_t,
\mathcal D_t,
x_{\mathrm{best}}^{(t)},
f_{\mathrm{best}}^{(t)}
).
$$

The algorithm terminates when the remaining unresolved information across
the active regions is sufficiently small for the required optimization
decision.

Let the active regions be:

$$
\mathcal R_{\mathrm{active}}.
$$

For each active region $R$, define its unresolved-information state:

$$
\mathcal Q_R
=
[
Q_{\mathrm{coverage}},
Q_{\mathrm{behavior}},
Q_{\mathrm{uncertainty}},
Q_{\mathrm{potential}}
].
$$

The global unresolved state can therefore be represented collectively as:

$$
\mathcal Q_{\mathrm{global}}
=
\{
\mathcal Q_R
\mid
R\in\mathcal R_{\mathrm{active}}
\}.
$$

ARRGO should terminate when no active region contains unresolved
information that can meaningfully affect the global optimization
decision.

Conceptually:

$$
\boxed{
\text{Terminate}
\iff
\text{No Relevant Unresolved Information Remains}
}
$$

Under a valid certified uncertainty model, this idea can be expressed
more precisely using optimization potential.

For every active region:

$$
P(R)
=
\max_{x\in R}U_R(x).
$$

Let the current incumbent be:

$$
f_{\mathrm{best}}.
$$

If:

$$
P(R)\leq f_{\mathrm{best}}
$$

for every relevant region, then the current certified information rules
out any solution better than the incumbent.

In that case:

$$
f_{\mathrm{best}}
=
f^*
$$

can be certified under the validity of the uncertainty assumptions.

However, this condition is available only when the underlying uncertainty
model provides a valid mathematical upper bound.

Without such assumptions, ARRGO cannot claim exact global optimality from
finite function evaluations alone.

Therefore, ARRGO distinguishes between:

$$
\boxed{
\text{Certified Termination}
}
$$

and:

$$
\boxed{
\text{Budget-Based Termination}
}
$$

Certified termination occurs when the mathematical information available
to the algorithm is sufficient to establish the required optimization
guarantee.

Budget-based termination occurs when the available computational
resources have been exhausted before such a certificate is obtained.

Let the maximum number of allowed function evaluations be:

$$
N_{\max}.
$$

Then:

$$
N_t\geq N_{\max}
$$

provides a computational stopping condition.

Similarly, a maximum execution time or another externally imposed
resource constraint may also be used.

These conditions do not imply global optimality.

Therefore, when ARRGO terminates because of a computational budget, the
result should be reported as the best solution found:

$$
x_{\mathrm{best}},
\qquad
f_{\mathrm{best}},
$$

together with the fact that the optimization process stopped before a
global certificate was necessarily established.

The termination mechanism can therefore be represented as:

$$
\boxed{
\text{Global State}
\rightarrow
\text{Evaluate Remaining Relevance}
\rightarrow
\begin{cases}
\text{Certified Termination},\\
\text{Continue Refinement},\\
\text{Budget Termination}.
\end{cases}
}
$$

The distinction between these termination modes is essential.

ARRGO must never report an observed best solution as a mathematically
certified global optimum unless the required assumptions and bounds
actually justify that claim.

Thus:

$$
\boxed{
\text{Best Found}
\neq
\text{Certified Global Optimum}
}
$$

unless the termination conditions provide the corresponding guarantee.

The global optimization loop is therefore:

$$
\boxed{
\text{Select Region}
\rightarrow
\text{Select Action}
\rightarrow
\text{Refine}
\rightarrow
\text{Update Global State}
\rightarrow
\text{Check Termination}
\rightarrow
\text{Repeat}
}
$$

This ensures that termination is determined by the information state and
the available guarantees rather than by an arbitrary fixed iteration
count alone.

## Contraction Guarantee

Adaptive splitting allows ARRGO to refine the spatial structure of the
search domain.

However, adaptive split selection alone does not guarantee that the
resulting regions become sufficiently small.

Consider a region:

$$
R=[l,r].
$$

For a split point:

$$
s\in(l,r),
$$

the child regions are:

$$
R_L=[l,s],
$$

and:

$$
R_R=[s,r].
$$

Their lengths are:

$$
|R_L|=s-l,
$$

and:

$$
|R_R|=r-s.
$$

To guarantee progressive spatial refinement, ARRGO imposes a contraction
condition:

$$
\max(s-l,r-s)
\leq
\rho(r-l),
$$

where:

$$
0<\rho<1.
$$

This condition guarantees that the larger child region is strictly
smaller than the parent region.

Therefore:

$$
\max(|R_L|,|R_R|)
<
|R|.
$$

The contraction condition is a structural constraint rather than an
optimization heuristic.

It does not determine where the best split is.

Instead, it defines which split locations are admissible.

The feasible split set is therefore:

$$
\mathcal S_R^{\mathrm{feasible}}
=
\left\{
s\in(l,r)
\mid
\max(s-l,r-s)
\leq
\rho(r-l)
\right\}.
$$

ARRGO first generates candidate split locations based on the current
information state.

The contraction condition is then applied as an admissibility
constraint:

$$
\text{Candidate Splits}
\rightarrow
\text{Contraction Check}
\rightarrow
\mathcal S_R^{\mathrm{feasible}}.
$$

Only feasible candidates can be selected for structural refinement.

This separates two fundamentally different concepts:

$$
\boxed{
\text{Contraction}
=
\text{Structural Guarantee}
}
$$

and:

$$
\boxed{
\text{Split Value}
=
\text{Adaptive Preference}
}
$$

The algorithm therefore does not use the split criterion itself as a
substitute for the contraction guarantee.

If a candidate split has high structural value but violates the
contraction condition, it is not an admissible split.

The contraction condition also prevents pathological refinement in which
one child repeatedly receives almost the entire parent region.

For example, a split that produces:

$$
|R_L|\approx |R|
$$

and:

$$
|R_R|\approx0
$$

would provide very weak spatial refinement.

Such a split is excluded when it violates the contraction condition.

Under repeated admissible refinement, the diameter of a descendant region
decreases geometrically.

If a region has initial length:

$$
|R_0|,
$$

then after $k$ successive contractions along a descendant path:

$$
|R_k|
\leq
\rho^k|R_0|.
$$

Since:

$$
0<\rho<1,
$$

we have:

$$
\lim_{k\rightarrow\infty}|R_k|=0.
$$

Thus, repeated structural refinement provides increasing spatial
resolution.

The contraction condition does not by itself prove global optimality.

It only guarantees that the spatial scale of repeatedly refined regions
can decrease.

A global optimality guarantee requires additional assumptions and an
appropriate uncertainty or bounding mechanism.

Therefore:

$$
\boxed{
\text{Contraction}
\neq
\text{Optimality Certificate}
}
$$

Instead, contraction provides one of the structural conditions required
for analyzing the convergence behavior of the adaptive refinement
process.

The complete split-selection mechanism is therefore:

$$
\boxed{
\text{Generate Candidates}
\rightarrow
\text{Apply Contraction Constraint}
\rightarrow
\text{Evaluate Structural Value}
\rightarrow
\text{Select Split}
\rightarrow
\text{Refine}
}
$$

This guarantees that adaptive structural refinement remains spatially
meaningful while preserving the flexibility of information-driven split
selection.

## Global Convergence Mechanism

ARRGO performs global optimization through repeated information-driven
refinement of the search domain.

The convergence mechanism is based on the interaction between three
components:

$$
\boxed{
\text{Spatial Contraction}
+
\text{Information Refinement}
+
\text{Global Re-evaluation}
}
$$

Spatial contraction progressively reduces the size of refined regions.

Information refinement improves the available knowledge about the
objective function through additional function evaluations and structural
partitioning.

Global re-evaluation continuously updates the relative importance of all
regions using the information accumulated during the optimization process.

Let the search domain be:

$$
\Omega=[a,b].
$$

At iteration $t$, ARRGO maintains a collection of regions:

$$
\mathcal R_t=\{R_1,R_2,\ldots,R_m\}.
$$

Each region contains its own observations and behavioral information.

A refinement step transforms the current global state:

$$
\mathcal G_t
\rightarrow
\mathcal G_{t+1}.
$$

If the selected action is sampling, additional information about the
objective function is obtained:

$$
\mathcal D_R
\rightarrow
\mathcal D_R
\cup
\{(x_{\mathrm{new}},f(x_{\mathrm{new}}))\}.
$$

If the selected action is splitting, the spatial representation becomes
more refined:

$$
R
\rightarrow
\{R_L,R_R\}.
$$

In both cases, ARRGO re-analyzes the affected information state.

Therefore:

$$
\text{Refinement}
\rightarrow
\text{New Information}
\rightarrow
\text{Re-analysis}.
$$

The new information can change:

$$
B_R,
\qquad
\mathcal Q_R,
\qquad
P(R),
$$

and consequently the global priority of regions.

Thus, the optimization process is adaptive rather than following a fixed
predefined refinement sequence.

The global incumbent is updated after every new function evaluation:

$$
f_{\mathrm{best}}^{(t+1)}
=
\max
\left(
f_{\mathrm{best}}^{(t)},
f(x_{\mathrm{new}})
\right).
$$

An improved incumbent can change the optimization relevance of multiple
regions.

Consequently, information obtained in one region can affect decisions
about other regions through the global optimization state.

This creates the following feedback mechanism:

$$
\boxed{
\text{Local Refinement}
\rightarrow
\text{Global Update}
\rightarrow
\text{Region Re-prioritization}
\rightarrow
\text{Further Refinement}
}
$$

The mechanism is therefore not a collection of independent local
searches.

Instead, all regions participate in a single global information system.

For convergence analysis, the contraction property ensures that repeated
refinement can produce increasingly small regions:

$$
|R_k|
\leq
\rho^k|R_0|,
\qquad
0<\rho<1.
$$

Consequently:

$$
\lim_{k\rightarrow\infty}|R_k|=0.
$$

However, spatial contraction alone does not imply that the algorithm
approaches a global optimum.

A connection between spatial resolution and objective-value accuracy is
also required.

Such a connection may be obtained through a valid uncertainty or
regularity model for the objective function.

For example, under a valid Lipschitz assumption:

$$
|f(x)-f(y)|
\leq
L|x-y|.
$$

As the spatial diameter of a region decreases, the maximum possible
variation of the objective value within that region also decreases.

For a region $R$:

$$
\operatorname{diam}(R)=|R|.
$$

The Lipschitz condition implies:

$$
|f(x)-f(y)|
\leq
L\operatorname{diam}(R),
\qquad
x,y\in R.
$$

Therefore, sufficiently small regions provide increasingly precise
bounds on the possible variation of the objective function.

This establishes the conceptual chain:

$$
\boxed{
\text{Contraction}
\rightarrow
\text{Spatial Resolution}
\rightarrow
\text{Information Resolution}
\rightarrow
\text{Optimization Resolution}
}
$$

The final implication depends on the validity of the underlying
uncertainty model and the refinement strategy.

ARRGO therefore does not claim that contraction alone proves global
convergence.

Instead, contraction provides the structural mechanism through which
additional information can become increasingly localized.

The theoretical convergence result will consequently be established only
after specifying the required assumptions on the objective function,
uncertainty bounds, region-selection mechanism, and refinement process.

## Exact Uncertainty Computation

ARRGO requires a mechanism for quantifying the uncertainty associated
with the unknown behavior of the objective function inside each region.

Function evaluations provide information only at sampled points.

For a region:

$$
R=[l,r],
$$

suppose the available observations are:

$$
\mathcal D_R
=
\{(x_i,y_i)\}_{i=1}^{n},
\qquad
y_i=f(x_i).
$$

The values of the objective function between sampled points remain
unknown.

Therefore, the uncertainty model must distinguish between observed
function values and possible values at unsampled locations.

Without additional assumptions about the objective function, finite
function evaluations cannot provide a non-trivial rigorous bound on
unobserved values.

Consequently, ARRGO separates uncertainty estimation into two modes:

$$
\boxed{
\text{Certified Uncertainty}
}
$$

and:

$$
\boxed{
\text{Empirical Uncertainty}
}
$$

Certified uncertainty requires a mathematically valid assumption about
the objective function.

One example is a valid Lipschitz constant $L$ satisfying:

$$
|f(x)-f(y)|
\leq
L|x-y|.
$$

Under this assumption, every observation $(x_i,y_i)$ provides an upper
and lower bound for an arbitrary point $x\in R$:

$$
y_i-L|x-x_i|
\leq
f(x)
\leq
y_i+L|x-x_i|.
$$

Combining all observations gives the lower envelope:

$$
L_R(x)
=
\max_i
\left[
y_i-L|x-x_i|
\right],
$$

and the upper envelope:

$$
U_R(x)
=
\min_i
\left[
y_i+L|x-x_i|
\right].
$$

Therefore:

$$
L_R(x)
\leq
f(x)
\leq
U_R(x).
$$

The pointwise uncertainty is defined as:

$$
u_R(x)
=
U_R(x)-L_R(x).
$$

Thus:

$$
\boxed{
u_R(x)
=
\min_i
\left[
y_i+L|x-x_i|
\right]
-
\max_i
\left[
y_i-L|x-x_i|
\right]
}
$$

This quantity represents the range of function values that remain
consistent with the available observations and the assumed Lipschitz
constraint.

The uncertainty of the entire region is then:

$$
\mathcal U(R)
=
\max_{x\in R}
u_R(x).
$$

A large value of $\mathcal U(R)$ indicates that the current observations
leave substantial ambiguity about the possible behavior of the objective
function inside the region.

A small value indicates that the available information constrains the
possible objective behavior more strongly.

The uncertainty model therefore provides a direct connection between
sampling density and information resolution.

When a new sample is obtained:

$$
(x_{\mathrm{new}},y_{\mathrm{new}})
$$

the observation set becomes:

$$
\mathcal D_R'
=
\mathcal D_R
\cup
\{(x_{\mathrm{new}},y_{\mathrm{new}})\}.
$$

The envelopes are then recomputed:

$$
L_R(x)
\rightarrow
L_R'(x),
$$

$$
U_R(x)
\rightarrow
U_R'(x).
$$

Consequently:

$$
u_R(x)
\rightarrow
u_R'(x).
$$

This allows ARRGO to measure the effect of a new function evaluation on
the remaining uncertainty.

In particular, the expected structural role of sampling can be expressed
through the reduction of uncertainty:

$$
\Delta\mathcal U_{\mathrm{sample}}
=
\mathcal U(R)
-
\mathcal U'(R).
$$

This quantity is useful for evaluating whether additional sampling has
meaningfully improved the information state of the region.

However, uncertainty alone is not sufficient for global optimization.

A region may have high uncertainty but contain only low objective values.

Therefore, uncertainty must be considered together with optimization
potential.

The optimization potential is defined by the maximum valid upper bound:

$$
P(R)
=
\max_{x\in R}U_R(x).
$$

Thus, ARRGO distinguishes between two questions:

$$
\boxed{
\text{How uncertain is the region?}
}
$$

and:

$$
\boxed{
\text{How good could the region still be?}
}
$$

The first is measured by:

$$
\mathcal U(R),
$$

while the second is measured by:

$$
P(R).
$$

This distinction is essential for separating information acquisition from
optimization relevance.

Empirical uncertainty may also be used when no certified regularity
assumption is available.

However, empirical uncertainty is not a mathematical guarantee.

Therefore, ARRGO must explicitly identify whether an uncertainty value is
certified or empirical.

The complete uncertainty pipeline is:

$$
\boxed{
\text{Function Evaluations}
\rightarrow
\text{Uncertainty Model}
\rightarrow
\text{Pointwise Bounds}
\rightarrow
\text{Region Uncertainty}
\rightarrow
\text{Optimization Potential}
}
$$

This separation allows ARRGO to use uncertainty as an information
measure without incorrectly interpreting an empirical estimate as a
global optimality certificate.

## Exact Optimization Potential

Optimization potential measures how good the objective function could
still be inside a region according to the information currently available.

Consider a region:

$$
R=[l,r].
$$

Let the valid upper uncertainty envelope of the region be:

$$
U_R(x).
$$

For every point $x\in R$, the true objective value satisfies:

$$
f(x)\leq U_R(x).
$$

Therefore, the largest possible objective value consistent with the
available information is:

$$
P(R)
=
\max_{x\in R}U_R(x).
$$

Thus:

$$
\boxed{
P(R)
=
\max_{x\in R}U_R(x)
}
$$

This quantity is called the optimization potential of the region.

Optimization potential does not represent the objective value that is
actually observed in the region.

Instead, it represents the highest value that remains possible according
to the current valid information.

Let the best observed objective value in the region be:

$$
f_{\mathrm{best}}(R)
=
\max_{(x_i,y_i)\in\mathcal D_R}y_i.
$$

Then the difference between the potential value and the best observed
value can be expressed as:

$$
G_{\mathrm{potential}}(R)
=
P(R)-f_{\mathrm{best}}(R).
$$

This quantity represents the remaining optimization uncertainty relative
to the best value already observed inside the region.

A large value indicates that the region may still contain a substantially
better solution than the best observed point.

A small value indicates that the currently available information leaves
less room for improvement.

However, the local best value is not sufficient for global optimization.

ARRGO therefore compares the regional potential with the global
incumbent:

$$
f_{\mathrm{best}}
=
\max_{x_i\in\mathcal D}f(x_i).
$$

The global optimization gap of a region is defined as:

$$
\Delta_R
=
P(R)-f_{\mathrm{best}}.
$$

Therefore:

$$
\boxed{
\Delta_R
=
P(R)-f_{\mathrm{best}}
}
$$

This quantity has a direct global interpretation when $P(R)$ is a valid
upper bound.

If:

$$
\Delta_R>0,
$$

then the current information does not rule out the possibility that the
region contains a solution better than the current global incumbent.

If:

$$
\Delta_R\leq0,
$$

then the current certified information rules out a solution inside the
region that is better than the incumbent.

This does not mean that the region is deleted.

ARRGO does not use optimization potential as a pruning mechanism.

Instead, the region remains part of the global region hierarchy and its
state remains available for future analysis.

The global potential over the search domain can be represented as:

$$
P_{\mathrm{global}}
=
\max_{R\in\mathcal R}P(R).
$$

The corresponding global optimization gap is:

$$
\Delta_{\mathrm{global}}
=
P_{\mathrm{global}}
-
f_{\mathrm{best}}.
$$

When valid upper bounds are available, the global optimum satisfies:

$$
f^*
\leq
P_{\mathrm{global}}.
$$

Since the incumbent is an observed feasible value:

$$
f_{\mathrm{best}}
\leq
f^*.
$$

Therefore:

$$
0
\leq
f^*-f_{\mathrm{best}}
\leq
P_{\mathrm{global}}-f_{\mathrm{best}}.
$$

Hence:

$$
\boxed{
f^*-f_{\mathrm{best}}
\leq
\Delta_{\mathrm{global}}
}
$$

This gives the optimization gap a rigorous interpretation in certified
mode.

If:

$$
\Delta_{\mathrm{global}}\leq\epsilon,
$$

then:

$$
f^*-f_{\mathrm{best}}\leq\epsilon.
$$

Therefore, the current best solution is guaranteed to be
$\epsilon$-optimal under the validity of the upper bounds.

The distinction between uncertainty and optimization potential is
therefore:

$$
\boxed{
\mathcal U(R)
=
\text{How uncertain is the region?}
}
$$

while:

$$
\boxed{
P(R)
=
\text{How good could the region still be?}
}
$$

A region may have high uncertainty but low optimization potential.

Conversely, a region may have moderate uncertainty but high optimization
potential.

ARRGO therefore uses both quantities for different decisions.

The conceptual relationship is:

$$
\boxed{
\text{Uncertainty}
\rightarrow
\text{Possible Objective Values}
\rightarrow
\text{Optimization Potential}
\rightarrow
\text{Global Optimization Gap}
}
$$

After a new function evaluation, the upper envelope may change and the
global incumbent may also improve.

Therefore:

$$
P(R)
\rightarrow
P'(R),
$$

and:

$$
f_{\mathrm{best}}
\rightarrow
f_{\mathrm{best}}'.
$$

Consequently:

$$
\Delta_R
\rightarrow
\Delta_R'.
$$

This makes optimization potential a dynamic quantity rather than a fixed
property of a region.

The complete global interpretation is therefore:

$$
\boxed{
\text{Region Information}
\rightarrow
\text{Valid Upper Bound}
\rightarrow
\text{Optimization Potential}
\rightarrow
\text{Global Gap}
\rightarrow
\text{Termination Decision}
}
$$

Thus, optimization potential provides the mathematical connection between
local region analysis and the global optimization objective.

## Exact Candidate Generation

ARRGO does not select sampling points or split points arbitrarily.

Candidate locations are generated from the current information state of
the region.

Let:

$$
R=[l,r]
$$

be a region with observations:

$$
\mathcal D_R
=
\{(x_i,y_i)\}_{i=1}^{n}.
$$

The candidate-generation mechanism has two distinct purposes:

$$
\boxed{
\text{Sampling Candidate Generation}
}
$$

and:

$$
\boxed{
\text{Split Candidate Generation}
}
$$

These two candidate sets must not be treated as equivalent.

Sampling adds a new function evaluation and therefore aims to obtain
additional information at a specific location.

Splitting changes the spatial structure of the region and therefore aims
to create a more informative partition.

---

### Sampling Candidate Generation

Let the set of valid sampling locations be:

$$
\mathcal C_{\mathrm{sample}}(R)
\subseteq
(l,r).
$$

A valid sampling candidate must satisfy:

$$
x\in(l,r)
$$

and:

$$
x\notin\{x_1,\ldots,x_n\}.
$$

Therefore:

$$
\mathcal C_{\mathrm{sample}}(R)
=
\left\{
x\in(l,r)
\mid
x\notin\mathcal D_R
\right\}.
$$

In practice, ARRGO does not need to evaluate every possible point in
this continuous set.

Instead, it constructs a finite candidate set from locations that are
informative according to the current region state.

Candidate locations may be associated with different sources of
information, including:

$$
\boxed{
\text{Coverage}
}
$$

$$
\boxed{
\text{Uncertainty}
}
$$

$$
\boxed{
\text{Optimization Potential}
}
$$

and:

$$
\boxed{
\text{Behavioral Structure}
}
$$

For example, a candidate may be generated near a location where the
current uncertainty envelope is large.

Another candidate may be generated inside a large sampling gap.

Another candidate may be generated near a region where the observed
behavior suggests a possible extremum.

Thus, candidate generation is based on multiple independent information
sources.

The resulting candidate set can be represented as:

$$
\mathcal C_{\mathrm{sample}}(R)
=
\bigcup_k
\mathcal C_k(R),
$$

where each $\mathcal C_k(R)$ corresponds to a different information
source.

Duplicate candidates are merged before evaluation.

The candidates are then evaluated according to the current refinement
objective rather than using a fixed universal ordering.

---

### Split Candidate Generation

For a region:

$$
R=[l,r],
$$

a split candidate is a point:

$$
s\in(l,r).
$$

The corresponding children are:

$$
R_L=[l,s],
$$

and:

$$
R_R=[s,r].
$$

The initial split-candidate set is:

$$
\mathcal C_{\mathrm{split}}(R)
\subseteq
(l,r).
$$

A split candidate is generated where dividing the region may reveal a
meaningful structural distinction.

Possible structural signals include:

$$
\boxed{
\text{Directional Change}
}
$$

$$
\boxed{
\text{Slope Variation}
}
$$

$$
\boxed{
\text{Sampling-Density Difference}
}
$$

$$
\boxed{
\text{Uncertainty Difference}
}
$$

and:

$$
\boxed{
\text{Optimization-Potential Difference}
}
$$

For a candidate split $s$, these quantities are evaluated conceptually
by comparing the information on the two sides of the candidate.

The resulting structural profile can be written as:

$$
\mathbf S_{\mathrm{split}}(s\mid R)
=
[
S_{\mathrm{direction}},
S_{\mathrm{slope}},
S_{\mathrm{coverage}},
S_{\mathrm{uncertainty}},
S_{\mathrm{potential}}
].
$$

The split candidate is not selected solely because it maximizes one of
these quantities.

Instead, candidate splits are compared using the current refinement
objective and the non-dominance principle defined earlier.

Before a split candidate becomes eligible for selection, it must satisfy
the contraction constraint:

$$
\max(s-l,r-s)
\leq
\rho(r-l),
\qquad
0<\rho<1.
$$

Therefore, the admissible split candidates are:

$$
\mathcal C_{\mathrm{split}}^{\mathrm{feasible}}(R)
=
\left\{
s\in\mathcal C_{\mathrm{split}}(R)
\mid
\max(s-l,r-s)
\leq
\rho(r-l)
\right\}.
$$

The candidate-generation process can therefore be represented as:

$$
\boxed{
\text{Region Information}
\rightarrow
\text{Generate Sampling Candidates}
}
$$

and:

$$
\boxed{
\text{Region Information}
\rightarrow
\text{Generate Split Candidates}
\rightarrow
\text{Apply Structural Constraints}
}
$$

Candidate generation and candidate selection are deliberately separated.

Candidate generation answers:

$$
\boxed{
\text{Where could refinement be useful?}
}
$$

Candidate selection answers:

$$
\boxed{
\text{Which useful candidate should be selected now?}
}
$$

This separation prevents the refinement mechanism from becoming dependent
on a single predefined sampling pattern or a fixed split rule.

After every refinement action, the region information changes.

Therefore, the candidate sets must also be regenerated:

$$
\mathcal C_R^{(t)}
\rightarrow
\mathcal C_R^{(t+1)}.
$$

This makes candidate generation adaptive to newly acquired information.

The complete process is:

$$
\boxed{
\text{Analyze Region}
\rightarrow
\text{Generate Candidates}
\rightarrow
\text{Filter Invalid Candidates}
\rightarrow
\text{Evaluate Candidate Information}
\rightarrow
\text{Select Candidate}
\rightarrow
\text{Refine}
}
$$

Thus, ARRGO does not rely on a fixed grid or predetermined sequence of
evaluation points.

Instead, the set of possible refinement locations evolves with the
information accumulated by the algorithm.

## Exact Action Selection

After analyzing a region and generating candidate locations, ARRGO must
select the refinement action that provides the most relevant resolution
of the current unresolved information.

The available actions are:

$$
\mathcal A_R
=
\{
\mathrm{SAMPLE},
\mathrm{SPLIT},
\mathrm{STABLE}
\}.
$$

The action is determined by the current information state of the region.

Let:

$$
O_R^*
=
\operatorname{DominantObjective}(\mathcal Q_R)
$$

represent the dominant unresolved objective of region $R$.

The dominant objective is not necessarily permanent.

It may change after every new observation or structural refinement.

Therefore:

$$
O_R^{*(t)}
\neq
O_R^{*(t+1)}
$$

is possible.

For a sampling action, let the candidate set be:

$$
\mathcal C_{\mathrm{sample}}(R).
$$

For every candidate:

$$
x\in\mathcal C_{\mathrm{sample}}(R),
$$

ARRGO evaluates its predicted effect on the unresolved information
state.

Conceptually:

$$
\widehat{\Delta\mathcal Q}_{\mathrm{sample}}
(x\mid R).
$$

This is represented as a profile rather than a single arbitrary weighted
score:

$$
\widehat{\Delta\mathcal Q}_{\mathrm{sample}}
(x\mid R)
=
[
\Delta Q_{\mathrm{coverage}},
\Delta Q_{\mathrm{behavior}},
\Delta Q_{\mathrm{uncertainty}},
\Delta Q_{\mathrm{potential}}
].
$$

For a split candidate:

$$
s\in
\mathcal C_{\mathrm{split}}^{\mathrm{feasible}}(R),
$$

ARRGO similarly estimates the structural effect of the split:

$$
\widehat{\Delta\mathcal Q}_{\mathrm{split}}
(s\mid R).
$$

The action-selection mechanism first identifies which information
dimension is currently limiting the decision.

Therefore:

$$
O_R^*
\rightarrow
\text{Relevant Resolution Criterion}.
$$

If the dominant unresolved objective is primarily associated with missing
pointwise information, sampling candidates are considered first.

If the dominant unresolved objective is primarily associated with
insufficient spatial separation of different behaviors, split candidates
are considered first.

If the relevant objectives are sufficiently stable, the region may enter
the stable state.

This gives the following decision structure:

$$
\boxed{
\text{Region State}
\rightarrow
\text{Dominant Unresolved Objective}
\rightarrow
\text{Candidate Class}
\rightarrow
\text{Candidate Selection}
}
$$

Candidate selection itself is based on dominance rather than an
arbitrary weighted sum.

For two sampling candidates $x_1$ and $x_2$, candidate $x_1$ dominates
$x_2$ if it is at least as effective as $x_2$ in every relevant
resolution dimension and strictly more effective in at least one:

$$
x_1\succ x_2
$$

if:

$$
\Delta Q_k(x_1)
\geq
\Delta Q_k(x_2)
\quad
\forall k\in\mathcal K_R
$$

and:

$$
\Delta Q_j(x_1)
>
\Delta Q_j(x_2)
$$

for at least one:

$$
j\in\mathcal K_R.
$$

The same dominance principle is applied to feasible split candidates.

The resulting non-dominated candidate set is:

$$
\mathcal C_R^{ND}.
$$

If multiple candidates remain non-dominated, ARRGO applies deterministic
tie-breaking rules.

The tie-breaking mechanism should prefer candidates that provide
greater resolution of the currently dominant objective.

If candidates remain equivalent with respect to the relevant objective,
the algorithm may use secondary structural constraints such as improved
spatial coverage or balanced contraction.

This prevents arbitrary weighted combinations of fundamentally different
information quantities.

The final action can therefore be represented as:

$$
A_R^*
=
\operatorname{SelectAction}
\left(
O_R^*,
\mathcal C_{\mathrm{sample}}^{ND},
\mathcal C_{\mathrm{split}}^{ND}
\right).
$$

The selected action must also satisfy all validity constraints.

For sampling:

$$
x_{\mathrm{new}}\in(l,r)
$$

and:

$$
x_{\mathrm{new}}\notin\{x_1,\ldots,x_n\}.
$$

For splitting:

$$
s\in(l,r)
$$

and:

$$
\max(s-l,r-s)
\leq
\rho(r-l).
$$

Therefore, candidate generation, candidate evaluation, and action
selection remain separate stages:

$$
\boxed{
\text{Generate}
\rightarrow
\text{Evaluate}
\rightarrow
\text{Compare}
\rightarrow
\text{Select}
}
$$

After the selected action is executed, ARRGO does not assume that the
previous decision remains optimal.

Instead, the region is re-analyzed:

$$
\text{Action}
\rightarrow
\text{New Information}
\rightarrow
\text{Re-analysis}
\rightarrow
\text{New Decision}.
$$

This creates a closed adaptive loop:

$$
\boxed{
\text{Analyze}
\rightarrow
\text{Identify Objective}
\rightarrow
\text{Generate Candidates}
\rightarrow
\text{Select Action}
\rightarrow
\text{Refine}
\rightarrow
\text{Re-analyze}
}
$$

The action-selection mechanism therefore remains adaptive throughout
the optimization process.

No action is permanently assigned to a region, and no region is removed
from the global search structure merely because another action currently
has higher priority.

## Numerical Robustness

ARRGO relies on numerical function evaluations and spatial comparisons.

Therefore, the refinement mechanism must remain robust when numerical
precision affects the observed information.

Even when the objective function is deterministic, numerical issues may
arise from floating-point representation, very small spatial distances,
nearly equal function values, or nearly equal observed slopes.

Let two sampled locations be:

$$
x_i
\quad\text{and}\quad
x_j.
$$

A direct comparison such as:

$$
x_i=x_j
$$

may be unreliable when the values are produced through floating-point
computations.

ARRGO therefore uses a numerical tolerance when determining whether two
locations should be considered equivalent.

Conceptually:

$$
|x_i-x_j|\leq\varepsilon_x
\quad
\Longrightarrow
\quad
x_i\equiv x_j.
$$

Similarly, when comparing two objective values:

$$
y_i
\quad\text{and}\quad
y_j,
$$

small numerical differences should not automatically be interpreted as
meaningful behavioral differences.

A numerical equivalence condition can be defined as:

$$
|y_i-y_j|\leq\varepsilon_f.
$$

The same principle applies to observed secant slopes.

For:

$$
s_i
=
\frac{y_{i+1}-y_i}
{x_{i+1}-x_i},
$$

a slope whose magnitude is sufficiently close to zero should not be
treated as definitively positive or negative.

Instead, ARRGO distinguishes between:

$$
s_i>\varepsilon_s,
$$

$$
|s_i|\leq\varepsilon_s,
$$

and:

$$
s_i<-\varepsilon_s.
$$

Thus, the observed directional behavior becomes:

$$
\operatorname{Sign}_{\varepsilon_s}(s_i)
=
\begin{cases}
+1 & s_i>\varepsilon_s,\\
0 & |s_i|\leq\varepsilon_s,\\
-1 & s_i<-\varepsilon_s.
\end{cases}
$$

This prevents very small numerical fluctuations from being interpreted
as genuine directional changes.

The same principle applies to slope variation.

For:

$$
\Delta s_i=s_{i+1}-s_i,
$$

small values of:

$$
|\Delta s_i|
$$

may represent numerical variation rather than meaningful behavioral
change.

Therefore, behavioral analysis should distinguish between significant
and numerically negligible changes.

Numerical robustness is also required when generating candidate points.

A candidate sampling location must not be accepted if it is numerically
indistinguishable from an existing observation.

Thus:

$$
|x_{\mathrm{candidate}}-x_i|
\leq
\varepsilon_x
$$

for any existing sample invalidates that candidate.

The same principle applies to split points.

A split must create two numerically meaningful subregions.

Therefore, a candidate $s$ must satisfy not only:

$$
s\in(l,r),
$$

but also:

$$
s-l>\varepsilon_x
$$

and:

$$
r-s>\varepsilon_x.
$$

Otherwise, the resulting child region may be too small to represent
reliably in floating-point arithmetic.

Numerical robustness must also be considered when evaluating uncertainty
bounds.

If two mathematically equivalent bounds differ only by floating-point
rounding, ARRGO should not interpret the difference as meaningful
information.

Therefore, comparisons involving uncertainty, optimization potential,
and termination gaps should use appropriate numerical tolerances.

For example, the certified global gap:

$$
\Delta_{\mathrm{global}}
=
P_{\mathrm{global}}
-
f_{\mathrm{best}}
$$

should be interpreted with numerical precision taken into account.

The distinction between mathematical uncertainty and numerical error is
important.

Let:

$$
\mathcal U_{\mathrm{model}}
$$

represent uncertainty arising from incomplete knowledge of the objective
function, and let:

$$
\mathcal E_{\mathrm{num}}
$$

represent numerical error.

These are conceptually different quantities:

$$
\boxed{
\text{Model Uncertainty}
\neq
\text{Numerical Error}
}
$$

Model uncertainty arises because the objective function is not fully
known between observations.

Numerical error arises because computations are performed with finite
precision.

ARRGO therefore treats numerical robustness as a supporting layer around
the information and refinement mechanisms rather than as a replacement
for them.

The numerical processing pipeline is:

$$
\boxed{
\text{Function Evaluation}
\rightarrow
\text{Numerical Validation}
\rightarrow
\text{Information Analysis}
\rightarrow
\text{Refinement Decision}
}
$$

This ensures that adaptive decisions are based on meaningful numerical
differences rather than artifacts caused by finite-precision computation.

Numerical tolerances should not be used as arbitrary substitutes for
optimization tolerances.

In particular:

$$
\varepsilon_x,
\qquad
\varepsilon_f,
\qquad
\varepsilon_s
$$

serve different purposes and should be defined according to the
corresponding numerical operation.

Their values should be configurable rather than hard-coded into the
mathematical definition of ARRGO.

The final implementation should therefore keep numerical robustness
parameters separate from the conceptual optimization criteria.

## Duplicate and Boundary Handling

ARRGO generates new sampling and splitting locations adaptively.

Therefore, the algorithm must explicitly handle duplicate locations,
boundary locations, and numerically invalid refinement points.

These cases must be resolved before a refinement action is executed.

---

### Duplicate Sampling Locations

Let the existing observations of a region be:

$$
\mathcal D_R
=
\{(x_i,y_i)\}_{i=1}^{n}.
$$

A newly generated sampling candidate $x$ is valid only if it is
numerically distinct from every existing observation.

Using the spatial tolerance $\varepsilon_x$:

$$
|x-x_i|>\varepsilon_x
\qquad
\forall i.
$$

Therefore, a candidate satisfying:

$$
|x-x_i|\leq\varepsilon_x
$$

is considered a duplicate and must not trigger another function
evaluation.

The candidate-generation mechanism should therefore remove duplicate
locations before candidate selection.

Conceptually:

$$
\boxed{
\text{Generate}
\rightarrow
\text{Duplicate Check}
\rightarrow
\text{Valid Candidates}
}
$$

This prevents ARRGO from spending evaluation budget on numerically
equivalent points.

---

### Boundary Sampling Locations

For a region:

$$
R=[l,r],
$$

the boundaries $l$ and $r$ are already part of the region definition.

If their function values have already been evaluated, they should be
reused rather than evaluated again.

Therefore, a boundary point is a valid candidate only when its value is
not already available.

This allows boundary observations to contribute to region analysis
without unnecessary repeated evaluations.

Boundary observations are particularly important because a global maximum
may occur at the boundary of the search domain.

ARRGO must therefore never exclude boundary points from the optimization
analysis merely because they are not interior points.

---

### Duplicate Split Locations

A split point must create two distinct child regions.

For:

$$
R=[l,r]
$$

and split point $s$, the children are:

$$
R_L=[l,s],
$$

and:

$$
R_R=[s,r].
$$

A valid split must satisfy:

$$
l<s<r.
$$

In addition, numerical separation must be maintained:

$$
s-l>\varepsilon_x,
$$

and:

$$
r-s>\varepsilon_x.
$$

A split candidate that violates either condition is rejected.

This prevents the creation of empty or numerically meaningless regions.

---

### Boundary and Contraction Constraints

Split validity must satisfy both numerical separation and the contraction
condition.

Therefore:

$$
s-l>\varepsilon_x,
$$

$$
r-s>\varepsilon_x,
$$

and:

$$
\max(s-l,r-s)
\leq
\rho(r-l).
$$

Thus, a feasible split belongs to:

$$
\mathcal C_{\mathrm{split}}^{\mathrm{valid}}(R)
=
\left\{
s\in(l,r)
\mid
\begin{array}{l}
s-l>\varepsilon_x,\\
r-s>\varepsilon_x,\\
\max(s-l,r-s)\leq\rho(r-l)
\end{array}
\right\}.
$$

This combines numerical validity with structural refinement guarantees.

---

### Shared Boundary Observations

When a region is split:

$$
R=[l,r]
\rightarrow
R_L=[l,s],
\quad
R_R=[s,r],
$$

the split point $s$ belongs to both child regions geometrically.

If $s$ has already been evaluated, the corresponding observation:

$$
(s,f(s))
$$

can be reused by both children.

Thus, function evaluations are not duplicated merely because a point lies
on the boundary between two regions.

This provides information sharing between neighboring regions while
preserving their independent local analyses.

---

### Global Evaluation Uniqueness

ARRGO maintains a global collection of evaluated points:

$$
\mathcal D
=
\{(x_i,f(x_i))\}_{i=1}^{N}.
$$

Before evaluating a new candidate, ARRGO first checks the global
observation set.

If a numerically equivalent location already exists:

$$
|x_{\mathrm{new}}-x_i|
\leq
\varepsilon_x,
$$

the stored function value is reused.

Therefore, the same physical location should not be evaluated multiple
times unless the objective-function model explicitly permits or requires
repeated evaluations.

This is especially important when multiple regions share boundaries.

---

### Candidate Validity Pipeline

Every generated refinement candidate passes through a validation stage.

For sampling:

$$
\boxed{
\text{Generate}
\rightarrow
\text{Boundary Check}
\rightarrow
\text{Duplicate Check}
\rightarrow
\text{Valid Sample}
}
$$

For splitting:

$$
\boxed{
\text{Generate}
\rightarrow
\text{Interior Check}
\rightarrow
\text{Numerical Separation}
\rightarrow
\text{Contraction Check}
\rightarrow
\text{Valid Split}
}
$$

Invalid candidates are not treated as failed optimization regions.

They are simply excluded from the current candidate set.

The region itself remains unchanged.

---

### Preservation of Global Coverage

Boundary handling must preserve the original search domain:

$$
\Omega=[a,b].
$$

After any valid split:

$$
R_L\cup R_R=R,
$$

and:

$$
R_L\cap R_R=\{s\}.
$$

Therefore, the union of the child regions completely covers the parent
region.

Repeated splitting consequently preserves coverage of the original
search domain.

Thus:

$$
\boxed{
\text{Refinement}
\neq
\text{Loss of Search Domain}
}
$$

The refinement process changes the resolution of the search structure
without removing any part of the original domain.

This property is essential for global optimization because potential
optima must remain represented throughout the refinement process.

## Evaluation Budget

Function evaluations are the primary source of new information in ARRGO.

In many optimization problems, evaluating the objective function may be
computationally expensive.

Therefore, ARRGO must explicitly manage the number of function
evaluations performed during the optimization process.

Let:

$$
N_t
$$

denote the total number of function evaluations performed up to iteration
$t$.

Let the maximum available evaluation budget be:

$$
N_{\max}.
$$

The fundamental budget constraint is:

$$
N_t\leq N_{\max}.
$$

A function evaluation may only be performed when sufficient evaluation
budget remains.

Therefore:

$$
N_t<N_{\max}
$$

is a necessary condition for executing a new function evaluation.

The evaluation budget is global rather than belonging independently to
each region.

Thus:

$$
\boxed{
N_{\mathrm{total}}
=
\sum_R N_R
}
$$

where $N_R$ represents the number of evaluations associated with region
$R$.

Because neighboring regions may share previously evaluated boundary
points, the same function evaluation should not be counted multiple times.

ARRGO therefore maintains a global evaluation history:

$$
\mathcal D
=
\{(x_i,f(x_i))\}_{i=1}^{N_t}.
$$

Before spending budget on a candidate point, ARRGO checks whether an
equivalent observation already exists.

If:

$$
|x_{\mathrm{candidate}}-x_i|
\leq
\varepsilon_x,
$$

the existing observation is reused.

Consequently, evaluation budget is spent only when genuinely new
function information is required.

---

### Budget-Aware Refinement

The availability of evaluation budget affects the choice between
sampling and structural refinement.

Sampling normally requires a new function evaluation:

$$
\mathrm{SAMPLE}
\rightarrow
N_t+1.
$$

Splitting does not necessarily require a new function evaluation if the
existing observations can be reused:

$$
\mathrm{SPLIT}
\rightarrow
N_t.
$$

Therefore, splitting can refine the spatial structure without
automatically consuming additional evaluation budget.

This creates an important distinction:

$$
\boxed{
\text{Evaluation Cost}
\neq
\text{Structural Refinement Cost}
}
$$

ARRGO can therefore continue refining the spatial representation even
when additional function evaluations are limited.

However, splitting without acquiring useful new information should not be
performed indefinitely.

The decision must remain connected to the current unresolved information
state of the region.

---

### Budget-Aware Action Selection

Let:

$$
B_t
=
N_{\max}-N_t
$$

denote the remaining evaluation budget.

If:

$$
B_t>0,
$$

both sampling and splitting may remain available, subject to their other
validity conditions.

If:

$$
B_t=0,
$$

no new function evaluation may be performed.

In this case, ARRGO may still perform purely structural operations when
they provide meaningful refinement, but it cannot claim that such
operations create new objective-function information.

Therefore:

$$
\boxed{
B_t=0
\Rightarrow
\text{No New Function Evaluation}
}
$$

The remaining budget should influence action selection without replacing
the information-based refinement objective.

Thus, ARRGO does not simply distribute the budget uniformly among
regions.

Instead, the available budget is allocated dynamically according to the
current global information state.

Conceptually:

$$
\boxed{
\text{Global Information State}
\rightarrow
\text{Remaining Budget}
\rightarrow
\text{Budget-Aware Refinement}
}
$$

A region with high unresolved optimization relevance may receive an
additional evaluation when sufficient budget exists.

Another region may instead be structurally refined if splitting can
resolve important spatial information without requiring an immediate new
evaluation.

---

### Budget Exhaustion

When:

$$
N_t=N_{\max},
$$

the evaluation budget is exhausted.

ARRGO then enters a budget-constrained state.

The algorithm must not silently perform additional objective-function
evaluations.

The final incumbent is:

$$
x_{\mathrm{best}}
=
\arg\max_{x_i\in\mathcal D}f(x_i),
$$

with:

$$
f_{\mathrm{best}}
=
f(x_{\mathrm{best}}).
$$

If a valid global certificate has not been established, the result must
be reported as the best solution found under the available evaluation
budget.

Therefore:

$$
\boxed{
\text{Budget Exhaustion}
\neq
\text{Global Optimality}
}
$$

A budget limit is a computational constraint, not a mathematical proof
of optimality.

---

### Evaluation Efficiency

The objective of budget management is not simply to minimize the number
of evaluations.

Instead, ARRGO seeks to maximize the information obtained from each
expensive evaluation.

Conceptually:

$$
\boxed{
\text{Evaluation Efficiency}
=
\text{Information Obtained per Function Evaluation}
}
$$

This is consistent with the central philosophy of ARRGO.

A new evaluation is valuable when it significantly reduces unresolved
information or improves the global optimization decision.

Therefore, candidate selection and evaluation-budget management remain
connected:

$$
\text{Candidate Information Value}
\rightarrow
\text{Budget Allocation}
\rightarrow
\text{Function Evaluation}.
$$

The evaluation budget consequently acts as a resource constraint around
the adaptive refinement mechanism rather than as the mechanism itself.

The complete budget-aware loop is:

$$
\boxed{
\text{Analyze}
\rightarrow
\text{Generate Candidates}
\rightarrow
\text{Estimate Information Value}
\rightarrow
\text{Check Budget}
\rightarrow
\text{Evaluate or Refine}
\rightarrow
\text{Update Budget}
}
$$

This allows ARRGO to remain adaptive under both inexpensive and expensive
objective functions while preserving the distinction between
computational limitations and mathematical guarantees.

## Algorithm Formalization

ARRGO can be formalized as an iterative adaptive global optimization
framework.

The optimization problem is:

$$
\max_{x\in\Omega}f(x),
$$

where:

$$
\Omega=[a,b].
$$

The objective function is accessed through function evaluations:

$$
y=f(x).
$$

ARRGO maintains a global state consisting of:

$$
\mathcal G_t
=
(
\mathcal R_t,
\mathcal D_t,
x_{\mathrm{best}}^{(t)},
f_{\mathrm{best}}^{(t)},
N_t
).
$$

Here:

- $\mathcal R_t$ is the collection of regions,
- $\mathcal D_t$ is the global set of observations,
- $x_{\mathrm{best}}^{(t)}$ is the current incumbent location,
- $f_{\mathrm{best}}^{(t)}$ is the current best observed value,
- $N_t$ is the number of function evaluations performed.

Each region contains:

$$
R
=
(
[l,r],
\mathcal D_R,
B_R,
\mathcal Q_R,
\mathcal U_R,
P_R,
\mathrm{state},
\mathrm{parent},
\mathrm{children}
).
$$

The algorithm begins by creating the root region:

$$
R_0=\Omega.
$$

Initial observations are obtained inside the root region according to
the initialization strategy.

The root region is then analyzed to construct its initial information
state:

$$
\mathcal D_{R_0}
\rightarrow
B_{R_0}
\rightarrow
\mathcal Q_{R_0}.
$$

The initial global state is therefore:

$$
\mathcal G_0
=
(
\mathcal R_0,
\mathcal D_0,
x_{\mathrm{best}}^{(0)},
f_{\mathrm{best}}^{(0)},
N_0
).
$$

The main optimization process proceeds iteratively.

At iteration $t$, ARRGO first evaluates the current global state and
determines which region should receive attention.

The selected region is:

$$
R_t^*
=
\operatorname{SelectRegion}(\mathcal R_t).
$$

The selected region is then analyzed:

$$
R_t^*
\rightarrow
B_{R_t^*}
\rightarrow
\mathcal Q_{R_t^*}.
$$

The dominant unresolved objective is identified:

$$
O_{R_t^*}^*
=
\operatorname{DominantObjective}
(
\mathcal Q_{R_t^*}
).
$$

ARRGO then generates candidate refinement locations.

For sampling:

$$
\mathcal C_{\mathrm{sample}}(R_t^*).
$$

For splitting:

$$
\mathcal C_{\mathrm{split}}^{\mathrm{valid}}(R_t^*).
$$

The candidate sets are evaluated according to their expected ability to
resolve the dominant unresolved information.

The available actions are:

$$
\mathcal A_R
=
\{
\mathrm{SAMPLE},
\mathrm{SPLIT},
\mathrm{STABLE}
\}.
$$

ARRGO selects the action:

$$
A_t^*
=
\operatorname{SelectAction}
(
R_t^*,
O_{R_t^*}^*,
\mathcal A_{R_t^*}
).
$$

If the selected action is sampling, ARRGO selects a valid candidate:

$$
x_t^*
\in
\mathcal C_{\mathrm{sample}}(R_t^*).
$$

The objective function is evaluated:

$$
y_t^*=f(x_t^*).
$$

The new observation is added to the global observation set:

$$
\mathcal D_{t+1}
=
\mathcal D_t
\cup
\{(x_t^*,y_t^*)\}.
$$

The corresponding regional observation set is also updated.

The affected region is then re-analyzed.

If the selected action is splitting, ARRGO selects a valid split point:

$$
s_t^*
\in
\mathcal C_{\mathrm{split}}^{\mathrm{valid}}(R_t^*).
$$

The selected region is structurally refined:

$$
R_t^*
\rightarrow
\{
R_L,R_R
\}.
$$

The parent region remains in the hierarchy.

Existing observations are transferred to the appropriate child regions:

$$
\mathcal D_{R_L}
=
\{
(x_i,y_i)\in\mathcal D_{R_t^*}
\mid
x_i\in R_L
\},
$$

and:

$$
\mathcal D_{R_R}
=
\{
(x_i,y_i)\in\mathcal D_{R_t^*}
\mid
x_i\in R_R
\}.
$$

If an observation lies exactly on the shared boundary, it may be reused
by both children.

The children are then analyzed independently.

If the selected action is STABLE, the region is marked as stable:

$$
R_t^*
\rightarrow
\mathrm{STABLE}.
$$

A stable region remains part of the global hierarchy and is not deleted.

After every refinement action, ARRGO updates the global incumbent.

Let:

$$
\mathcal D_{t+1}
=
\{(x_i,f(x_i))\}_{i=1}^{N_{t+1}}.
$$

Then:

$$
x_{\mathrm{best}}^{(t+1)}
=
\arg\max_{x_i\in\mathcal D_{t+1}}
f(x_i),
$$

and:

$$
f_{\mathrm{best}}^{(t+1)}
=
f(x_{\mathrm{best}}^{(t+1)}).
$$

The affected regional states are then recomputed.

Global region priorities are also updated because a new observation or
an improved incumbent can change the optimization relevance of multiple
regions.

The global state therefore evolves as:

$$
\mathcal G_t
\rightarrow
\mathcal G_{t+1}.
$$

Before beginning the next iteration, ARRGO checks the termination
conditions.

Conceptually:

$$
\operatorname{Terminate}(\mathcal G_{t+1})
=
\begin{cases}
\mathrm{TRUE},
&
\text{if the required termination condition is satisfied},
\\
\mathrm{FALSE},
&
\text{otherwise}.
\end{cases}
$$

The complete optimization cycle is therefore:

$$
\boxed{
\begin{aligned}
&\text{Initialize}
\\
&\downarrow
\\
&\text{Analyze Global State}
\\
&\downarrow
\\
&\text{Select Region}
\\
&\downarrow
\\
&\text{Analyze Region}
\\
&\downarrow
\\
&\text{Identify Dominant Objective}
\\
&\downarrow
\\
&\text{Generate Candidates}
\\
&\downarrow
\\
&\text{Select Action}
\\
&\downarrow
\\
&\text{Refine}
\\
&\downarrow
\\
&\text{Update Observations and Structure}
\\
&\downarrow
\\
&\text{Update Incumbent}
\\
&\downarrow
\\
&\text{Update Global Priorities}
\\
&\downarrow
\\
&\text{Check Termination}
\\
&\downarrow
\\
&\text{Repeat or Return Best Solution}.
\end{aligned}
}
$$

The formal pseudocode of ARRGO is therefore:

$$
\boxed{
\begin{array}{l}
\textbf{Input: }
\Omega,\ f,\ N_{\max},\ \text{algorithm parameters}
\\[4pt]

\textbf{Initialize root region }R_0=\Omega
\\
\textbf{Initialize observations and global state}
\\[4pt]

\textbf{while termination condition is false:}
\\
\qquad
R^*
\leftarrow
\operatorname{SelectRegion}(\mathcal R)
\\
\qquad
\operatorname{Analyze}(R^*)
\\
\qquad
O^*
\leftarrow
\operatorname{DominantObjective}(\mathcal Q_{R^*})
\\
\qquad
\text{Generate valid candidates}
\\
\qquad
A^*
\leftarrow
\operatorname{SelectAction}(R^*,O^*)
\\[4pt]

\qquad
\textbf{if }A^*=\mathrm{SAMPLE:}
\\
\qquad\qquad
x^*
\leftarrow
\operatorname{SelectSampleCandidate}(R^*)
\\
\qquad\qquad
y^*
\leftarrow
f(x^*)
\\
\qquad\qquad
\text{Update observations}
\\[4pt]

\qquad
\textbf{else if }A^*=\mathrm{SPLIT:}
\\
\qquad\qquad
s^*
\leftarrow
\operatorname{SelectSplitCandidate}(R^*)
\\
\qquad\qquad
R^*\rightarrow\{R_L,R_R\}
\\
\qquad\qquad
\text{Transfer reusable observations}
\\
\qquad\qquad
\text{Analyze child regions}
\\[4pt]

\qquad
\textbf{else if }A^*=\mathrm{STABLE:}
\\
\qquad\qquad
\operatorname{MarkStable}(R^*)
\\[4pt]

\qquad
\text{Update incumbent}
\\
\qquad
\text{Update affected region states}
\\
\qquad
\text{Update global priorities}
\\
\qquad
\text{Check termination}
\\[4pt]

\textbf{return }
(x_{\mathrm{best}},f_{\mathrm{best}})
\end{array}
}
$$

This formalization separates the core responsibilities of ARRGO:

$$
\boxed{
\text{Region Analysis}
}
$$

$$
\boxed{
\text{Candidate Generation}
}
$$

$$
\boxed{
\text{Action Selection}
}
$$

$$
\boxed{
\text{Refinement}
}
$$

$$
\boxed{
\text{Global State Update}
}
$$

and:

$$
\boxed{
\text{Termination}
}
$$

These components form the algorithmic architecture that will later be
translated into an implementation.

The formalization intentionally does not define implementation-specific
details such as data structures, numerical libraries, or programming
language constructs.

Those details belong to the implementation architecture rather than the
mathematical definition of ARRGO.
